```python
# =============================================================================
# STEP 1: CAR MICS6 — LOAD THE FOUR SOURCE DATASETS
# =============================================================================

import os
import numpy as np
import pandas as pd

print("=" * 100)
print("STEP 1: CAR MICS6 — SOURCE DATA LOADING")
print("=" * 100)

# -------------------------------------------------------------------------
# 1. DATA DIRECTORY
# -------------------------------------------------------------------------

DATA_DIR = r"C:\Users\user\Downloads\CAR_MICS6_Undernutrition\data"

CH_PATH = os.path.join(DATA_DIR, "ch.sav")
HH_PATH = os.path.join(DATA_DIR, "hh.sav")
WM_PATH = os.path.join(DATA_DIR, "wm.sav")
BH_PATH = os.path.join(DATA_DIR, "bh.sav")

print("\nDATA DIRECTORY")
print("-" * 100)
print(DATA_DIR)

# -------------------------------------------------------------------------
# 2. CHECK FILES
# -------------------------------------------------------------------------

print("\nSOURCE FILE CHECK")
print("-" * 100)

required_files = {
    "CH": CH_PATH,
    "HH": HH_PATH,
    "WM": WM_PATH,
    "BH": BH_PATH
}

for name, path in required_files.items():

    if os.path.exists(path):
        print(f"✓ {name}.sav found")
        print(f"  {path}")
    else:
        raise FileNotFoundError(
            f"{name}.sav was not found at:\n{path}"
        )

# -------------------------------------------------------------------------
# 3. LOAD SOURCE DATA
# -------------------------------------------------------------------------

print("\nLOADING SOURCE DATASETS")
print("-" * 100)

ch = pd.read_spss(CH_PATH)
hh = pd.read_spss(HH_PATH)
wm = pd.read_spss(WM_PATH)
bh = pd.read_spss(BH_PATH)

print(f"✓ CH loaded : {ch.shape}")
print(f"✓ HH loaded : {hh.shape}")
print(f"✓ WM loaded : {wm.shape}")
print(f"✓ BH loaded : {bh.shape}")

# -------------------------------------------------------------------------
# 4. EXPECTED SIZE CHECK
# -------------------------------------------------------------------------

print("\nDATASET SIZE VALIDATION")
print("-" * 100)

expected_sizes = {
    "CH": 9037,
    "HH": 8994,
    "WM": 9778,
    "BH": 26536
}

actual_sizes = {
    "CH": len(ch),
    "HH": len(hh),
    "WM": len(wm),
    "BH": len(bh)
}

for name in expected_sizes:

    expected = expected_sizes[name]
    actual = actual_sizes[name]

    if actual == expected:
        print(f"✓ {name}: {actual:,} records")
    else:
        print(
            f"⚠ {name}: expected {expected:,}, "
            f"found {actual:,}"
        )

# -------------------------------------------------------------------------
# 5. REQUIRED IDENTIFIERS
# -------------------------------------------------------------------------

print("\nIDENTIFIER CHECK")
print("-" * 100)

required_identifiers = {
    "CH": ["HH1", "HH2", "LN"],
    "HH": ["HH1", "HH2"],
    "WM": ["WM1", "WM2", "WM3"],
    "BH": ["HH1", "HH2", "BH8"]
}

datasets = {
    "CH": ch,
    "HH": hh,
    "WM": wm,
    "BH": bh
}

for dataset_name, columns in required_identifiers.items():

    df = datasets[dataset_name]

    for column in columns:

        if column in df.columns:
            print(f"✓ {dataset_name}.{column}")
        else:
            raise KeyError(
                f"Required identifier {dataset_name}.{column} "
                f"is missing."
            )

# -------------------------------------------------------------------------
# 6. REQUIRED ANTHROPOMETRIC VARIABLES
# -------------------------------------------------------------------------

print("\nANTHROPOMETRIC VARIABLE CHECK")
print("-" * 100)

anthro_variables = [
    "CAGE",
    "HAZ",
    "WAZ",
    "HAZFLAG",
    "WAZFLAG"
]

for variable in anthro_variables:

    if variable in ch.columns:
        print(f"✓ {variable}")
    else:
        raise KeyError(
            f"Required anthropometric variable {variable} "
            f"is missing from CH."
        )

# -------------------------------------------------------------------------
# 7. BASIC CHILD IDENTIFIER DUPLICATE CHECK
# -------------------------------------------------------------------------

print("\nCHILD IDENTIFIER DUPLICATE CHECK")
print("-" * 100)

duplicate_children = ch.duplicated(
    subset=["HH1", "HH2", "LN"],
    keep=False
).sum()

print(
    f"Duplicate HH1 + HH2 + LN records: "
    f"{duplicate_children}"
)

if duplicate_children != 0:
    raise ValueError(
        "Duplicate child identifiers detected."
    )

print("✓ Child identifiers are unique")

# -------------------------------------------------------------------------
# 8. FINAL STATUS
# -------------------------------------------------------------------------

print("\n" + "=" * 100)
print("STEP 1 STATUS")
print("=" * 100)

print("✓ CH.sav loaded")
print("✓ HH.sav loaded")
print("✓ WM.sav loaded")
print("✓ BH.sav loaded")
print("✓ Source dataset sizes checked")
print("✓ Child identifiers checked")
print("✓ Required linkage identifiers checked")
print("✓ Anthropometric variables checked")
print("✓ No duplicate child identifiers")

print("\nSTEP 1 COMPLETE")
print("=" * 100)

print("\nNEXT STEP:")
print("STEP 2 — VALIDATE AND MERGE CH + HH + WM + BH")
print("Do NOT construct the train/validation/test split yet.")
```

```python
# =============================================================================
# STEP 2: CAR MICS6 — CREATE / RECOVER CANONICAL INTEGRATED DATASET
# =============================================================================
#
# Canonical structure:
# CH + HH + BH + WM
#
# Child-level unit of analysis = 1 row per child
# Total expected children = 9,037
#
# DO NOT split the data here.
# DO NOT preprocess here.
# DO NOT select final ML features here.
# =============================================================================

import os
import numpy as np
import pandas as pd
import pyreadstat

print("=" * 100)
print("STEP 2: CAR MICS6 — CANONICAL INTEGRATED DATASET")
print("=" * 100)


# =============================================================================
# 1. DATA DIRECTORY
# =============================================================================

DATA_DIR = r"C:\Users\user\Downloads\CAR_MICS6_Undernutrition\data"

print("\n1. DATA DIRECTORY")
print("-" * 100)
print(DATA_DIR)


# =============================================================================
# 2. CHECK SOURCE FILES
# =============================================================================

files_required = {
    "CH": os.path.join(DATA_DIR, "ch.sav"),
    "HH": os.path.join(DATA_DIR, "hh.sav"),
    "WM": os.path.join(DATA_DIR, "wm.sav"),
    "BH": os.path.join(DATA_DIR, "bh.sav"),
}

print("\n2. SOURCE FILE CHECK")
print("-" * 100)

for name, path in files_required.items():

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"{name}.sav not found:\n{path}"
        )

    print(f"✓ {name}.sav found")


# =============================================================================
# 3. LOAD SOURCE DATA
# =============================================================================

print("\n3. LOADING SOURCE DATA")
print("-" * 100)

ch, ch_meta = pyreadstat.read_sav(files_required["CH"])
hh, hh_meta = pyreadstat.read_sav(files_required["HH"])
wm, wm_meta = pyreadstat.read_sav(files_required["WM"])
bh, bh_meta = pyreadstat.read_sav(files_required["BH"])

print(f"CH : {ch.shape}")
print(f"HH : {hh.shape}")
print(f"WM : {wm.shape}")
print(f"BH : {bh.shape}")


# =============================================================================
# 4. CHILD IDENTIFIER VALIDATION
# =============================================================================

print("\n4. CHILD IDENTIFIER VALIDATION")
print("-" * 100)

required_child_ids = ["HH1", "HH2", "LN"]

for col in required_child_ids:

    if col not in ch.columns:
        raise KeyError(f"CH missing required identifier: {col}")

child_duplicates = ch.duplicated(
    subset=["HH1", "HH2", "LN"]
).sum()

print(f"Duplicate HH1 + HH2 + LN: {child_duplicates}")

if child_duplicates != 0:
    raise ValueError(
        "CH contains duplicate child identifiers."
    )

print("✓ One record per child confirmed")


# =============================================================================
# 5. CONSTRUCT OUTCOMES
# =============================================================================

print("\n5. CONSTRUCTING TARGET VARIABLES")
print("-" * 100)

required_anthro = [
    "HAZ",
    "WAZ",
    "HAZFLAG",
    "WAZFLAG"
]

for col in required_anthro:

    if col not in ch.columns:
        raise KeyError(
            f"CH missing anthropometric variable: {col}"
        )


# Start with missing outcomes
ch["stunted"] = np.nan
ch["underweight"] = np.nan


# Valid HAZ observations
valid_haz = (
    ch["HAZ"].notna()
    & ch["HAZFLAG"].notna()
    & (ch["HAZFLAG"] == 0)
)

ch.loc[valid_haz, "stunted"] = (
    ch.loc[valid_haz, "HAZ"] < -2
).astype(int)


# Valid WAZ observations
valid_waz = (
    ch["WAZ"].notna()
    & ch["WAZFLAG"].notna()
    & (ch["WAZFLAG"] == 0)
)

ch.loc[valid_waz, "underweight"] = (
    ch.loc[valid_waz, "WAZ"] < -2
).astype(int)


print(
    f"Stunting valid N   : {ch['stunted'].notna().sum():,}"
)

print(
    f"Underweight valid N: {ch['underweight'].notna().sum():,}"
)


# =============================================================================
# 6. CH → HH MERGE
# =============================================================================

print("\n6. CH → HH HOUSEHOLD LINKAGE")
print("-" * 100)

hh_key = ["HH1", "HH2"]

hh_duplicates = hh.duplicated(
    subset=hh_key
).sum()

print(
    f"Duplicate HH1 + HH2 household keys: {hh_duplicates}"
)

if hh_duplicates != 0:
    raise ValueError(
        "HH household key is not unique."
    )


# Household predictors already established in the project
hh_predictors = [

    "HH6",
    "windex5",
    "WS1",
    "WS11",
    "WS15",

]


missing_hh = [
    v for v in hh_predictors
    if v not in hh.columns
]

if missing_hh:

    raise KeyError(
        "Required HH predictors missing:\n"
        + str(missing_hh)
    )


# Keep only variables not already in CH
hh_to_merge = [
    "HH1",
    "HH2"
] + [
    v for v in hh_predictors
    if v not in ch.columns
]


integrated = ch.merge(
    hh[hh_to_merge],
    on=["HH1", "HH2"],
    how="left",
    validate="many_to_one",
    indicator=True
)

hh_match_rate = (
    (integrated["_merge"] == "both").mean()
    * 100
)

print(
    f"Household match rate: {hh_match_rate:.2f}%"
)

integrated.drop(
    columns="_merge",
    inplace=True
)


# =============================================================================
# 7. CH + HH ROW PRESERVATION
# =============================================================================

if len(integrated) != len(ch):

    raise ValueError(
        "CH + HH merge changed the number of child records."
    )

print(
    f"✓ Rows preserved: {len(integrated):,}"
)


# =============================================================================
# 8. CH → WM MATERNAL LINKAGE
# =============================================================================

print("\n8. CH → WM MATERNAL LINKAGE")
print("-" * 100)

required_ch_wm = [
    "UF1",
    "UF2",
    "UF4"
]

required_wm = [
    "WM1",
    "WM2",
    "WM3"
]

for col in required_ch_wm:

    if col not in integrated.columns:
        raise KeyError(
            f"Integrated data missing maternal linkage variable: {col}"
        )

for col in required_wm:

    if col not in wm.columns:
        raise KeyError(
            f"WM missing maternal linkage variable: {col}"
        )


# Check WM maternal key
wm_duplicates = wm.duplicated(
    subset=["WM1", "WM2", "WM3"]
).sum()

print(
    f"Duplicate WM mother keys: {wm_duplicates}"
)

if wm_duplicates != 0:

    raise ValueError(
        "WM1 + WM2 + WM3 is not unique."
    )


# Maternal predictors established in the project
wm_predictors = [

    "WB4",
    "WAGEM",
    "welevel",
    "MSTATUS",
    "CEB",
    "CSURV",
    "CDEAD",
    "CM11",
    "CM17",
    "insurance",

]


missing_wm = [
    v for v in wm_predictors
    if v not in wm.columns
]

if missing_wm:

    raise KeyError(
        "Required WM predictors missing:\n"
        + str(missing_wm)
    )


wm_to_merge = [
    "WM1",
    "WM2",
    "WM3"
] + [
    v for v in wm_predictors
    if v not in integrated.columns
]


integrated = integrated.merge(

    wm[wm_to_merge],

    left_on=[
        "UF1",
        "UF2",
        "UF4"
    ],

    right_on=[
        "WM1",
        "WM2",
        "WM3"
    ],

    how="left",

    validate="many_to_one",

    indicator=True

)


wm_match_rate = (
    (integrated["_merge"] == "both").mean()
    * 100
)

wm_matched = (
    integrated["_merge"] == "both"
).sum()

wm_unmatched = (
    integrated["_merge"] == "left_only"
).sum()


print(
    f"Matched maternal records  : {wm_matched:,}"
)

print(
    f"Unmatched maternal records: {wm_unmatched:,}"
)

print(
    f"Maternal match rate        : {wm_match_rate:.2f}%"
)

integrated.drop(
    columns="_merge",
    inplace=True
)


# =============================================================================
# 9. CH + HH + WM ROW PRESERVATION
# =============================================================================

if len(integrated) != len(ch):

    raise ValueError(
        "Maternal merge changed the number of child records."
    )

print(
    f"✓ Rows preserved after maternal merge: {len(integrated):,}"
)


# =============================================================================
# 10. BH CHILD-LEVEL LINKAGE
# =============================================================================

print("\n10. CH + HH + WM → BH BIRTH-HISTORY LINKAGE")
print("-" * 100)

required_bh = [

    "HH1",
    "HH2",
    "BH8",
    "brthord",
    "magebrt",
    "birthint",
    "BH2"

]

missing_bh = [
    v for v in required_bh
    if v not in bh.columns
]

if missing_bh:

    raise KeyError(
        "Required BH variables missing:\n"
        + str(missing_bh)
    )


# Create child-level BH table
bh_child = bh[required_bh].copy()


# BH8 corresponds to child line number
bh_child = bh_child.rename(

    columns={
        "BH8": "LN",
        "BH2": "twin"
    }

)


# Remove non-child birth-history records
bh_child = bh_child[
    bh_child["LN"].notna()
    & (bh_child["LN"] > 0)
].copy()


print(
    f"BH child-level records after LN filtering: "
    f"{len(bh_child):,}"
)


# Check duplicate child keys
bh_duplicates = bh_child.duplicated(

    subset=[
        "HH1",
        "HH2",
        "LN"
    ]

).sum()


print(
    f"Duplicate BH child keys: {bh_duplicates}"
)


if bh_duplicates != 0:

    raise ValueError(
        "Duplicate BH child identifiers detected."
    )


# Merge BH
integrated = integrated.merge(

    bh_child[
        [
            "HH1",
            "HH2",
            "LN",
            "brthord",
            "magebrt",
            "birthint",
            "twin"
        ]
    ],

    on=[
        "HH1",
        "HH2",
        "LN"
    ],

    how="left",

    validate="one_to_one",

    indicator=True

)


bh_matched = (
    integrated["_merge"] == "both"
).sum()

bh_unmatched = (
    integrated["_merge"] == "left_only"
).sum()

bh_match_rate = (
    bh_matched / len(integrated)
) * 100


print(
    f"Matched birth-history records  : {bh_matched:,}"
)

print(
    f"Unmatched birth-history records: {bh_unmatched:,}"
)

print(
    f"Birth-history match rate        : {bh_match_rate:.2f}%"
)

integrated.drop(
    columns="_merge",
    inplace=True
)


# =============================================================================
# 11. FINAL ROW PRESERVATION
# =============================================================================

print("\n11. FINAL CHILD-LEVEL VALIDATION")
print("-" * 100)

if len(integrated) != 9037:

    raise ValueError(
        f"Expected 9,037 children, got {len(integrated):,}"
    )

final_duplicates = integrated.duplicated(

    subset=[
        "HH1",
        "HH2",
        "LN"
    ]

).sum()


print(
    f"Final rows: {len(integrated):,}"
)

print(
    f"Final duplicate child IDs: {final_duplicates}"
)

if final_duplicates != 0:

    raise ValueError(
        "Final integrated dataset contains duplicate children."
    )

print("✓ Exactly one record per child")


# =============================================================================
# 12. CANONICAL DATASET NAME
# =============================================================================

final_integrated = integrated.copy()

analysis_df = final_integrated.copy()
merged = final_integrated.copy()


# =============================================================================
# 13. FINAL PREDICTOR INVENTORY — CURRENT 21-PREDICTOR ML SET
# =============================================================================

ml_predictors_21 = [

    # Child
    "CAGE",
    "HL4",

    # Household
    "HH6",
    "windex5",
    "WS1",
    "WS11",
    "WS15",
    "WB4",

    # Maternal
    "welevel",
    "WAGEM",
    "MSTATUS",
    "CEB",
    "CSURV",
    "CDEAD",
    "CM11",
    "CM17",
    "insurance",

    # Birth history
    "brthord",
    "birthint",
    "twin",
    "magebrt",

]


missing_final_predictors = [

    v for v in ml_predictors_21
    if v not in final_integrated.columns

]


print("\n12. CURRENT ML PREDICTOR CHECK")
print("-" * 100)

print(
    f"Expected ML predictors: {len(ml_predictors_21)}"
)

print(
    f"Available: "
    f"{len(ml_predictors_21) - len(missing_final_predictors)}"
)


if missing_final_predictors:

    print("\n❌ MISSING PREDICTORS:")

    for v in missing_final_predictors:
        print("  -", v)

    raise KeyError(
        "Some approved ML predictors are missing."
    )

else:

    print("✓ All 21 approved ML predictors are available")


# =============================================================================
# 14. OUTCOME CHECK
# =============================================================================

print("\n13. OUTCOME VALIDATION")
print("-" * 100)

print("\nSTUNTING")

print(
    final_integrated["stunted"]
    .value_counts(dropna=False)
    .sort_index()
)


print("\nUNDERWEIGHT")

print(
    final_integrated["underweight"]
    .value_counts(dropna=False)
    .sort_index()
)


# =============================================================================
# 15. SAVE CANONICAL DATASET
# =============================================================================

FINAL_PKL = os.path.join(
    DATA_DIR,
    "CAR_MICS6_FINAL_INTEGRATED_DATASET_WITH_OUTCOMES.pkl"
)

FINAL_CSV = os.path.join(
    DATA_DIR,
    "CAR_MICS6_FINAL_INTEGRATED_DATASET_WITH_OUTCOMES.csv"
)


final_integrated.to_pickle(FINAL_PKL)
final_integrated.to_csv(
    FINAL_CSV,
    index=False
)


# =============================================================================
# 16. FINAL STATUS
# =============================================================================

print("\n" + "=" * 100)
print("STEP 2 COMPLETE")
print("=" * 100)

print(
    f"Final integrated dataset: {final_integrated.shape}"
)

print("✓ CH retained")
print("✓ HH linked")
print("✓ WM maternal information linked")
print("✓ BH birth history linked")
print("✓ 9,037 child records preserved")
print("✓ Child identifiers unique")
print("✓ Stunting outcome constructed")
print("✓ Underweight outcome constructed")
print("✓ 21-predictor ML set verified")
print("✓ Canonical object created: final_integrated")
print("✓ Canonical pickle saved")
print("✓ Canonical CSV saved")

print("\nCANONICAL OBJECT:")
print("final_integrated")

print("\nSaved:")
print(FINAL_PKL)
print(FINAL_CSV)

print("=" * 100)
```

```python
# =============================================================================
# STEP 3 — FINAL SELF-CONTAINED STATISTICAL SUMMARY
# =============================================================================

print("=" * 100)
print("STEP 3: CAR MICS6 — FINAL DESCRIPTIVE STATISTICAL ANALYSIS")
print("=" * 100)

# =============================================================================
# 1. CANONICAL DATASET
# =============================================================================

print("\n1. CANONICAL DATASET")
print("-" * 100)

print(f"Rows    : {final_integrated.shape[0]:,}")
print(f"Columns : {final_integrated.shape[1]:,}")

required = ["stunted", "underweight", "chweight"]

missing = [
    v for v in required
    if v not in final_integrated.columns
]

if missing:
    raise KeyError(f"Missing required variables: {missing}")

print("✓ Required variables available")


# =============================================================================
# 2. POSITIVE SURVEY-WEIGHT SAMPLE
# =============================================================================

print("\n" + "=" * 100)
print("2. SURVEY WEIGHT HANDLING")
print("=" * 100)

positive_weight = final_integrated["chweight"] > 0
negative_weight = final_integrated["chweight"] < 0
zero_weight = final_integrated["chweight"] == 0

print(f"Total observations       : {len(final_integrated):,}")
print(f"Positive weights         : {positive_weight.sum():,}")
print(f"Zero weights             : {zero_weight.sum():,}")
print(f"Negative weights         : {negative_weight.sum():,}")

if negative_weight.sum() > 0:
    raise ValueError(
        "Negative survey weights detected."
    )

print("✓ No negative survey weights")


# =============================================================================
# 3. STUNTING — WEIGHTED PREVALENCE
# =============================================================================

print("\n" + "=" * 100)
print("3. STUNTING — WEIGHTED PREVALENCE")
print("=" * 100)

stunting_weighted_sample = final_integrated.loc[
    positive_weight &
    final_integrated["stunted"].notna(),
    ["stunted", "chweight"]
].copy()

stunting_weighted_cases = (
    stunting_weighted_sample["stunted"] *
    stunting_weighted_sample["chweight"]
).sum()

stunting_weight_total = (
    stunting_weighted_sample["chweight"]
).sum()

stunting_weighted_prevalence = (
    stunting_weighted_cases /
    stunting_weight_total
) * 100

print(
    f"Valid observations : "
    f"{len(stunting_weighted_sample):,}"
)

print(
    f"Weighted cases     : "
    f"{stunting_weighted_cases:.2f}"
)

print(
    f"Weighted prevalence: "
    f"{stunting_weighted_prevalence:.2f}%"
)


# =============================================================================
# 4. UNDERWEIGHT — WEIGHTED PREVALENCE
# =============================================================================

print("\n" + "=" * 100)
print("4. UNDERWEIGHT — WEIGHTED PREVALENCE")
print("=" * 100)

underweight_weighted_sample = final_integrated.loc[
    positive_weight &
    final_integrated["underweight"].notna(),
    ["underweight", "chweight"]
].copy()

underweight_weighted_cases = (
    underweight_weighted_sample["underweight"] *
    underweight_weighted_sample["chweight"]
).sum()

underweight_weight_total = (
    underweight_weighted_sample["chweight"]
).sum()

underweight_weighted_prevalence = (
    underweight_weighted_cases /
    underweight_weight_total
) * 100

print(
    f"Valid observations : "
    f"{len(underweight_weighted_sample):,}"
)

print(
    f"Weighted cases     : "
    f"{underweight_weighted_cases:.2f}"
)

print(
    f"Weighted prevalence: "
    f"{underweight_weighted_prevalence:.2f}%"
)


# =============================================================================
# 5. UNWEIGHTED DISTRIBUTIONS
# =============================================================================

print("\n" + "=" * 100)
print("5. UNWEIGHTED OUTCOME DISTRIBUTIONS")
print("=" * 100)

# ----- STUNTING -----

stunting_unweighted = final_integrated.loc[
    final_integrated["stunted"].notna(),
    "stunted"
]

stunting_counts = (
    stunting_unweighted
    .value_counts()
    .sort_index()
)

stunting_percentages = (
    stunting_unweighted
    .value_counts(normalize=True)
    .sort_index() * 100
)

print("\nSTUNTING")
print("-" * 100)

for cls in [0, 1]:

    if cls in stunting_counts.index:

        label = (
            "Not Stunted"
            if cls == 0
            else "Stunted"
        )

        print(
            f"Class {cls} ({label:<14}) : "
            f"{stunting_counts[cls]:,} "
            f"({stunting_percentages[cls]:.2f}%)"
        )


# ----- UNDERWEIGHT -----

underweight_unweighted = final_integrated.loc[
    final_integrated["underweight"].notna(),
    "underweight"
]

underweight_counts = (
    underweight_unweighted
    .value_counts()
    .sort_index()
)

underweight_percentages = (
    underweight_unweighted
    .value_counts(normalize=True)
    .sort_index() * 100
)

print("\nUNDERWEIGHT")
print("-" * 100)

for cls in [0, 1]:

    if cls in underweight_counts.index:

        label = (
            "Not Underweight"
            if cls == 0
            else "Underweight"
        )

        print(
            f"Class {cls} ({label:<17}) : "
            f"{underweight_counts[cls]:,} "
            f"({underweight_percentages[cls]:.2f}%)"
        )


# =============================================================================
# 6. WEIGHTED VS UNWEIGHTED
# =============================================================================

print("\n" + "=" * 100)
print("6. WEIGHTED VS UNWEIGHTED PREVALENCE")
print("=" * 100)

unweighted_stunting_prevalence = (
    stunting_unweighted.mean() * 100
)

unweighted_underweight_prevalence = (
    underweight_unweighted.mean() * 100
)

print(
    f"Stunting    : "
    f"{unweighted_stunting_prevalence:.2f}% unweighted | "
    f"{stunting_weighted_prevalence:.2f}% weighted"
)

print(
    f"Underweight : "
    f"{unweighted_underweight_prevalence:.2f}% unweighted | "
    f"{underweight_weighted_prevalence:.2f}% weighted"
)


# =============================================================================
# 7. FINAL INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("7. FINAL STEP 3 INTEGRITY CHECK")
print("=" * 100)

print(
    f"Canonical dataset             : "
    f"{len(final_integrated):,}"
)

print(
    f"Stunting valid                : "
    f"{stunting_unweighted.notna().sum():,}"
)

print(
    f"Underweight valid             : "
    f"{underweight_unweighted.notna().sum():,}"
)

print(
    f"Zero-weight records           : "
    f"{zero_weight.sum():,}"
)

print(
    f"Negative-weight records       : "
    f"{negative_weight.sum():,}"
)

print("\n✓ No temporary analysis objects required")
print("✓ No undefined variables required")
print("✓ Zero-weight observations handled explicitly")
print("✓ Weighted prevalence calculated")
print("✓ Unweighted prevalence calculated")
print("✓ Canonical final_integrated object preserved")

print("\n" + "=" * 100)
print("STEP 3 COMPLETE")
print("=" * 100)
```

```python
# =============================================================================
# STEP 4.2: CAR MICS6 — MISSING-VALUE AND SPECIAL-CODE HANDLING
# =============================================================================

import numpy as np
import pandas as pd

print("=" * 100)
print("STEP 4.2: CAR MICS6 — MISSING-VALUE AND SPECIAL-CODE HANDLING")
print("=" * 100)

# =============================================================================
# 1. CANONICAL DATASET CHECK
# =============================================================================

print("\n1. CANONICAL DATASET CHECK")
print("-" * 100)

if "final_integrated" not in globals():
    raise NameError(
        "final_integrated is not available. "
        "Run STEP 2 first to create the canonical integrated dataset."
    )

print(f"Rows    : {final_integrated.shape[0]:,}")
print(f"Columns : {final_integrated.shape[1]:,}")

# =============================================================================
# 2. APPROVED PREDICTORS
# =============================================================================

approved_predictors = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cinsurance",
    "cdisability",
    "CA1",
    "CA14",
    "CA16",
    "CA17",
    "BD2",
    "TN3",
    "EC1",
    "IM2",
    "CA31"
]

required_outcomes = [
    "stunted",
    "underweight"
]

required_columns = approved_predictors + required_outcomes

missing_columns = [
    col for col in required_columns
    if col not in final_integrated.columns
]

if missing_columns:
    raise KeyError(
        f"Required columns are missing from final_integrated: {missing_columns}"
    )

print("\n2. APPROVED PREDICTOR CHECK")
print("-" * 100)

print(f"Number of approved predictors: {len(approved_predictors)}")

for i, col in enumerate(approved_predictors, 1):
    print(f"{i:2d}. ✓ {col}")

print("\n✓ All 20 approved predictors are available")

# =============================================================================
# 3. IMPORTANT: DO NOT MODIFY THE CANONICAL DATASET
# =============================================================================

print("\n3. CANONICAL DATASET PROTECTION")
print("-" * 100)

# Create a working copy.
# final_integrated itself will remain unchanged.
analysis_data = final_integrated.copy()

print("✓ Working copy created: analysis_data")
print("✓ final_integrated remains unchanged")

# =============================================================================
# 4. RAW MISSINGNESS CHECK
# =============================================================================

print("\n4. RAW MISSINGNESS CHECK")
print("-" * 100)

missing_summary = pd.DataFrame({
    "Variable": approved_predictors,
    "Missing_N": [
        analysis_data[col].isna().sum()
        for col in approved_predictors
    ]
})

missing_summary["Missing_%"] = (
    missing_summary["Missing_N"]
    / len(analysis_data)
    * 100
)

missing_summary = missing_summary.sort_values(
    "Missing_%",
    ascending=False
)

print(missing_summary.to_string(index=False))

# =============================================================================
# 5. SPECIAL-CODE AUDIT
# =============================================================================

print("\n5. SPECIAL-CODE AUDIT")
print("-" * 100)

special_codes = [8, 9, 96, 97, 98, 99]

special_code_records = []

for variable in approved_predictors:

    series = analysis_data[variable]

    for code in special_codes:

        frequency = (series == code).sum()

        if frequency > 0:

            special_code_records.append({
                "Variable": variable,
                "Code": code,
                "Frequency": int(frequency),
                "Percentage": round(
                    frequency / len(analysis_data) * 100,
                    3
                )
            })

special_code_table = pd.DataFrame(special_code_records)

if len(special_code_table) > 0:
    print(
        special_code_table.to_string(index=False)
    )
else:
    print("No potential special codes detected.")

print("\n✓ Special codes are being flagged for review only")
print("✓ No special code has been automatically converted to missing")

# =============================================================================
# 6. MISSINGNESS AMONG VALID OUTCOMES
# =============================================================================

print("\n6. MISSINGNESS AMONG VALID OUTCOMES")
print("-" * 100)

stunting_valid_mask = analysis_data["stunted"].notna()
underweight_valid_mask = analysis_data["underweight"].notna()

missingness_outcome = []

for variable in approved_predictors:

    missing_mask = analysis_data[variable].isna()

    stunting_valid_missing = (
        missing_mask & stunting_valid_mask
    ).sum()

    underweight_valid_missing = (
        missing_mask & underweight_valid_mask
    ).sum()

    missingness_outcome.append({
        "Variable": variable,
        "Missing_N": int(missing_mask.sum()),
        "Stunting_valid_among_missing": int(
            stunting_valid_missing
        ),
        "Underweight_valid_among_missing": int(
            underweight_valid_missing
        )
    })

missingness_outcome_df = pd.DataFrame(
    missingness_outcome
)

print(
    missingness_outcome_df.to_string(index=False)
)

# =============================================================================
# 7. HIGH-MISSINGNESS PREDICTORS
# =============================================================================

print("\n7. HIGH-MISSINGNESS PREDICTORS")
print("-" * 100)

high_missing = missing_summary[
    missing_summary["Missing_%"] >= 20
].copy()

if len(high_missing) > 0:
    print(
        high_missing.to_string(index=False)
    )
else:
    print("No predictor has >=20% missing values.")

# =============================================================================
# 8. MISSING-VALUE DECISION
# =============================================================================

print("\n8. MISSING-VALUE DECISION")
print("-" * 100)

print("✓ Do NOT delete observations because of predictor missingness")
print("✓ Do NOT perform complete-case deletion")
print("✓ Preserve all eligible children")
print("✓ Missing predictor values will be handled during preprocessing")
print("✓ Imputation parameters will be learned from TRAINING data only")
print("✓ Validation and test data will never determine imputation values")

# =============================================================================
# 9. SPECIAL-CODE DECISION
# =============================================================================

print("\n9. SPECIAL-CODE DECISION")
print("-" * 100)

print("✓ Potential special codes remain unchanged at this stage")
print("✓ No blanket conversion of 8/9/96/97/98/99")
print("✓ Variable-specific coding will not be guessed")
print("✓ Preprocessing will treat the observed values consistently")
print("✓ final_integrated remains unchanged")

# =============================================================================
# 10. OUTCOME VALIDITY CHECK
# =============================================================================

print("\n10. OUTCOME VALIDITY CHECK")
print("-" * 100)

stunting_values = sorted(
    analysis_data.loc[
        analysis_data["stunted"].notna(),
        "stunted"
    ].unique()
)

underweight_values = sorted(
    analysis_data.loc[
        analysis_data["underweight"].notna(),
        "underweight"
    ].unique()
)

print("Stunting target values    :", stunting_values)
print("Underweight target values:", underweight_values)

if not set(stunting_values).issubset({0.0, 1.0}):
    raise ValueError(
        "Stunting contains values other than 0 and 1."
    )

if not set(underweight_values).issubset({0.0, 1.0}):
    raise ValueError(
        "Underweight contains values other than 0 and 1."
    )

print("✓ Both outcomes are valid binary targets")

# =============================================================================
# 11. DATASET INTEGRITY CHECK
# =============================================================================

print("\n11. DATASET INTEGRITY CHECK")
print("-" * 100)

print(
    f"Canonical dataset rows : {len(final_integrated):,}"
)

print(
    f"Working dataset rows   : {len(analysis_data):,}"
)

if len(final_integrated) != len(analysis_data):
    raise ValueError(
        "Working copy changed the number of observations."
    )

print("✓ Number of observations preserved")

# =============================================================================
# 12. STEP 4.2 SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("STEP 4.2 COMPLETE")
print("=" * 100)

print(f"Canonical records          : {len(final_integrated):,}")
print(f"Approved ML predictors     : {len(approved_predictors)}")
print(f"High-missingness predictors: {len(high_missing)}")
print("Special codes              : Flagged, not automatically converted")
print("Missing values             : Preserved for training-only preprocessing")
print("Complete-case deletion     : NOT performed")
print("Canonical dataset          : UNCHANGED")

print("\nNEXT STEP:")
print("STEP 4.3 — CREATE THE STRATIFIED 70/15/15 DEVELOPMENT SPLIT")
print("Training = 70%")
print("Validation = 15%")
print("Test = 15%")
print("Test data will remain completely unseen.")
```

```python
# =============================================================================
# STEP 4.3: CAR MICS6 — 70/15/15 STRATIFIED DEVELOPMENT SPLIT
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

print("=" * 100)
print("STEP 4.3: CAR MICS6 — 70/15/15 STRATIFIED DEVELOPMENT SPLIT")
print("=" * 100)


# =============================================================================
# 1. CANONICAL WORKING DATASET CHECK
# =============================================================================

print("\n1. WORKING DATASET CHECK")
print("-" * 100)

if "analysis_data" not in globals():
    raise NameError(
        "analysis_data is not available. "
        "Run STEP 4.2 first."
    )

print(f"Rows    : {analysis_data.shape[0]:,}")
print(f"Columns : {analysis_data.shape[1]:,}")

# =============================================================================
# 2. APPROVED PREDICTORS
# =============================================================================

approved_predictors = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cinsurance",
    "cdisability",
    "CA1",
    "CA14",
    "CA16",
    "CA17",
    "BD2",
    "TN3",
    "EC1",
    "IM2",
    "CA31"
]

print("\n2. PREDICTOR CHECK")
print("-" * 100)

missing_predictors = [
    v for v in approved_predictors
    if v not in analysis_data.columns
]

if missing_predictors:
    raise KeyError(
        f"Missing approved predictors: {missing_predictors}"
    )

print(f"Approved predictors: {len(approved_predictors)}")
print("✓ All 20 predictors available")


# =============================================================================
# 3. OUTCOME-SPECIFIC VALID DATA
# =============================================================================

print("\n3. OUTCOME-SPECIFIC VALID DATA")
print("-" * 100)

# STUNTING
stunting_data = analysis_data.loc[
    analysis_data["stunted"].notna()
].copy()

# UNDERWEIGHT
underweight_data = analysis_data.loc[
    analysis_data["underweight"].notna()
].copy()

print(
    f"Stunting valid observations    : "
    f"{len(stunting_data):,}"
)

print(
    f"Underweight valid observations : "
    f"{len(underweight_data):,}"
)

# =============================================================================
# 4. STUNTING — FIRST SPLIT
# =============================================================================

print("\n" + "=" * 100)
print("4. STUNTING — 70/15/15 SPLIT")
print("=" * 100)

stunt_train, stunt_temp = train_test_split(
    stunting_data,
    test_size=0.30,
    stratify=stunting_data["stunted"],
    random_state=42
)

# Split remaining 30% equally:
# 15% validation + 15% test
stunt_val, stunt_test = train_test_split(
    stunt_temp,
    test_size=0.50,
    stratify=stunt_temp["stunted"],
    random_state=42
)

print(
    f"Training   : {len(stunt_train):,} "
    f"({len(stunt_train)/len(stunting_data)*100:.2f}%)"
)

print(
    f"Validation : {len(stunt_val):,} "
    f"({len(stunt_val)/len(stunting_data)*100:.2f}%)"
)

print(
    f"Test       : {len(stunt_test):,} "
    f"({len(stunt_test)/len(stunting_data)*100:.2f}%)"
)

# =============================================================================
# 5. UNDERWEIGHT — FIRST SPLIT
# =============================================================================

print("\n" + "=" * 100)
print("5. UNDERWEIGHT — 70/15/15 SPLIT")
print("=" * 100)

under_train, under_temp = train_test_split(
    underweight_data,
    test_size=0.30,
    stratify=underweight_data["underweight"],
    random_state=42
)

under_val, under_test = train_test_split(
    under_temp,
    test_size=0.50,
    stratify=under_temp["underweight"],
    random_state=42
)

print(
    f"Training   : {len(under_train):,} "
    f"({len(under_train)/len(underweight_data)*100:.2f}%)"
)

print(
    f"Validation : {len(under_val):,} "
    f"({len(under_val)/len(underweight_data)*100:.2f}%)"
)

print(
    f"Test       : {len(under_test):,} "
    f"({len(under_test)/len(underweight_data)*100:.2f}%)"
)

# =============================================================================
# 6. STUNTING CLASS DISTRIBUTION
# =============================================================================

print("\n" + "=" * 100)
print("6. STUNTING — CLASS DISTRIBUTION")
print("=" * 100)

def print_class_distribution(df, target, name):

    counts = df[target].value_counts().sort_index()
    percentages = (
        df[target]
        .value_counts(normalize=True)
        .sort_index()
        * 100
    )

    print(f"\n{name}")
    print("-" * 70)

    for cls in counts.index:
        print(
            f"Class {int(cls)}: "
            f"{counts[cls]:,} "
            f"({percentages[cls]:.2f}%)"
        )


print_class_distribution(
    stunt_train,
    "stunted",
    "Training"
)

print_class_distribution(
    stunt_val,
    "stunted",
    "Validation"
)

print_class_distribution(
    stunt_test,
    "stunted",
    "Test"
)

# =============================================================================
# 7. UNDERWEIGHT CLASS DISTRIBUTION
# =============================================================================

print("\n" + "=" * 100)
print("7. UNDERWEIGHT — CLASS DISTRIBUTION")
print("=" * 100)

print_class_distribution(
    under_train,
    "underweight",
    "Training"
)

print_class_distribution(
    under_val,
    "underweight",
    "Validation"
)

print_class_distribution(
    under_test,
    "underweight",
    "Test"
)

# =============================================================================
# 8. CLASS IMBALANCE
# =============================================================================

print("\n" + "=" * 100)
print("8. CLASS IMBALANCE")
print("=" * 100)

stunt_negative = (
    stunt_train["stunted"] == 0
).sum()

stunt_positive = (
    stunt_train["stunted"] == 1
).sum()

under_negative = (
    under_train["underweight"] == 0
).sum()

under_positive = (
    under_train["underweight"] == 1
).sum()

stunt_scale_pos_weight = (
    stunt_negative / stunt_positive
)

under_scale_pos_weight = (
    under_negative / under_positive
)

print(
    f"Stunting scale_pos_weight   : "
    f"{stunt_scale_pos_weight:.4f}"
)

print(
    f"Underweight scale_pos_weight: "
    f"{under_scale_pos_weight:.4f}"
)

# =============================================================================
# 9. CREATE X AND y OBJECTS
# =============================================================================

print("\n" + "=" * 100)
print("9. FINAL DEVELOPMENT MATRICES")
print("=" * 100)

# STUNTING
X_train_stunt = stunt_train[approved_predictors].copy()
X_val_stunt = stunt_val[approved_predictors].copy()
X_test_stunt = stunt_test[approved_predictors].copy()

y_train_stunt = stunt_train["stunted"].astype(int).copy()
y_val_stunt = stunt_val["stunted"].astype(int).copy()
y_test_stunt = stunt_test["stunted"].astype(int).copy()

# UNDERWEIGHT
X_train_under = under_train[approved_predictors].copy()
X_val_under = under_val[approved_predictors].copy()
X_test_under = under_test[approved_predictors].copy()

y_train_under = under_train["underweight"].astype(int).copy()
y_val_under = under_val["underweight"].astype(int).copy()
y_test_under = under_test["underweight"].astype(int).copy()

print("\nSTUNTING")
print("-" * 70)

print(f"X_train : {X_train_stunt.shape}")
print(f"X_val   : {X_val_stunt.shape}")
print(f"X_test  : {X_test_stunt.shape}")
print(f"y_train : {y_train_stunt.shape}")
print(f"y_val   : {y_val_stunt.shape}")
print(f"y_test  : {y_test_stunt.shape}")

print("\nUNDERWEIGHT")
print("-" * 70)

print(f"X_train : {X_train_under.shape}")
print(f"X_val   : {X_val_under.shape}")
print(f"X_test  : {X_test_under.shape}")
print(f"y_train : {y_train_under.shape}")
print(f"y_val   : {y_val_under.shape}")
print(f"y_test  : {y_test_under.shape}")

# =============================================================================
# 10. INDEX OVERLAP CHECK
# =============================================================================

print("\n" + "=" * 100)
print("10. SPLIT INDEPENDENCE CHECK")
print("=" * 100)

stunt_train_test_overlap = len(
    set(stunt_train.index) &
    set(stunt_test.index)
)

stunt_train_val_overlap = len(
    set(stunt_train.index) &
    set(stunt_val.index)
)

stunt_val_test_overlap = len(
    set(stunt_val.index) &
    set(stunt_test.index)
)

under_train_test_overlap = len(
    set(under_train.index) &
    set(under_test.index)
)

under_train_val_overlap = len(
    set(under_train.index) &
    set(under_val.index)
)

under_val_test_overlap = len(
    set(under_val.index) &
    set(under_test.index)
)

print(
    f"Stunting train/test overlap : "
    f"{stunt_train_test_overlap}"
)

print(
    f"Stunting train/val overlap  : "
    f"{stunt_train_val_overlap}"
)

print(
    f"Stunting val/test overlap   : "
    f"{stunt_val_test_overlap}"
)

print(
    f"Underweight train/test overlap : "
    f"{under_train_test_overlap}"
)

print(
    f"Underweight train/val overlap  : "
    f"{under_train_val_overlap}"
)

print(
    f"Underweight val/test overlap   : "
    f"{under_val_test_overlap}"
)

if any([
    stunt_train_test_overlap,
    stunt_train_val_overlap,
    stunt_val_test_overlap,
    under_train_test_overlap,
    under_train_val_overlap,
    under_val_test_overlap
]):
    raise ValueError(
        "Overlapping observations detected between splits."
    )

print("✓ No observation overlap detected")

# =============================================================================
# 11. TEST-SET ISOLATION
# =============================================================================

print("\n" + "=" * 100)
print("11. TEST-SET ISOLATION")
print("=" * 100)

print("✓ Test predictors created")
print("✓ Test targets created")
print("✓ Test data has not been used for preprocessing")
print("✓ Test data has not been used for feature selection")
print("✓ Test data has not been used for hyperparameter tuning")
print("✓ Test data has not been used for threshold selection")

# =============================================================================
# 12. FINAL CHECK
# =============================================================================

print("\n" + "=" * 100)
print("12. STEP 4.3 FINAL CHECK")
print("=" * 100)

expected_stunt_total = (
    len(stunt_train) +
    len(stunt_val) +
    len(stunt_test)
)

expected_under_total = (
    len(under_train) +
    len(under_val) +
    len(under_test)
)

if expected_stunt_total != len(stunting_data):
    raise ValueError(
        "Stunting split does not preserve all valid observations."
    )

if expected_under_total != len(underweight_data):
    raise ValueError(
        "Underweight split does not preserve all valid observations."
    )

print(
    f"Stunting valid records preserved    : "
    f"{expected_stunt_total:,} / {len(stunting_data):,}"
)

print(
    f"Underweight valid records preserved : "
    f"{expected_under_total:,} / {len(underweight_data):,}"
)

print("\n✓ 70/15/15 stratified development split complete")
print("✓ Training, validation and test sets are mutually exclusive")
print("✓ Class proportions preserved")
print("✓ Test set remains untouched")

print("\n" + "=" * 100)
print("STEP 4.3 COMPLETE")
print("=" * 100)

print("\nNEXT STEP:")
print("STEP 5 — TRAINING-ONLY PREPROCESSING")
```

```python
# =============================================================================
# STEP 5: CAR MICS6 — TRAINING-ONLY PREPROCESSING
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

print("=" * 100)
print("STEP 5: CAR MICS6 — TRAINING-ONLY PREPROCESSING")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

print("\n1. CHECKING STEP 4.3 OBJECTS")
print("-" * 100)

required_objects = [
    "X_train_stunt",
    "X_val_stunt",
    "X_test_stunt",
    "y_train_stunt",
    "y_val_stunt",
    "y_test_stunt",
    "X_train_under",
    "X_val_under",
    "X_test_under",
    "y_train_under",
    "y_val_under",
    "y_test_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "The following Step 4.3 objects are missing:\n"
        + "\n".join(missing_objects)
        + "\n\nRun STEP 4.3 successfully before running STEP 5."
    )

print("✓ All Step 4.3 objects are available")


# =============================================================================
# 2. LOCK APPROVED PREDICTORS
# =============================================================================

approved_predictors = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cinsurance",
    "cdisability",
    "CA1",
    "CA14",
    "CA16",
    "CA17",
    "BD2",
    "TN3",
    "EC1",
    "IM2",
    "CA31"
]

print("\n2. APPROVED PREDICTORS")
print("-" * 100)

print(f"Number of predictors: {len(approved_predictors)}")

for i, variable in enumerate(approved_predictors, 1):
    print(f"{i:2d}. {variable}")

# Check predictor order in every matrix
for name in [
    "X_train_stunt",
    "X_val_stunt",
    "X_test_stunt",
    "X_train_under",
    "X_val_under",
    "X_test_under"
]:
    obj = globals()[name]

    if list(obj.columns) != approved_predictors:
        raise ValueError(
            f"{name} does not contain the approved predictors "
            "in the expected order."
        )

print("\n✓ Predictor set and order confirmed")


# =============================================================================
# 3. PREDICTOR TYPES
# =============================================================================

print("\n" + "=" * 100)
print("3. PREDICTOR TYPE ASSIGNMENT")
print("=" * 100)

# CAGE is the only continuous/numeric predictor.
numeric_features = [
    "CAGE"
]

# All remaining approved predictors are categorical MICS variables.
categorical_features = [
    v for v in approved_predictors
    if v not in numeric_features
]

print("\nNumeric predictor:")
for v in numeric_features:
    print(f"✓ {v}")

print("\nCategorical predictors:")
for v in categorical_features:
    print(f"✓ {v}")

print(
    f"\nNumeric count     : {len(numeric_features)}"
)

print(
    f"Categorical count : {len(categorical_features)}"
)

print(
    f"Total             : "
    f"{len(numeric_features) + len(categorical_features)}"
)


# =============================================================================
# 4. CREATE PREPROCESSING COMPONENTS
# =============================================================================

print("\n" + "=" * 100)
print("4. PREPROCESSING DESIGN")
print("=" * 100)

print("Numeric pipeline:")
print("  • Median imputation")
print("  • StandardScaler")

print("\nCategorical pipeline:")
print("  • Most-frequent imputation")
print("  • OneHotEncoder")
print("  • handle_unknown='ignore'")

print(
    "\n✓ All preprocessing parameters will be learned from training data only"
)


# =============================================================================
# 5. FUNCTION TO CREATE PREPROCESSOR
# =============================================================================

def create_preprocessor():

    numeric_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),
            (
                "scaler",
                StandardScaler()
            )
        ]
    )

    categorical_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(
                    strategy="most_frequent"
                )
            ),
            (
                "encoder",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                )
            )
        ]
    )

    preprocessor = ColumnTransformer(
        transformers=[
            (
                "numeric",
                numeric_pipeline,
                numeric_features
            ),
            (
                "categorical",
                categorical_pipeline,
                categorical_features
            )
        ],
        remainder="drop"
    )

    return preprocessor


# =============================================================================
# 6. STUNTING PREPROCESSING
# =============================================================================

print("\n" + "=" * 100)
print("6. STUNTING — TRAINING-ONLY PREPROCESSING")
print("=" * 100)

preprocessor_stunt = create_preprocessor()

print("Fitting preprocessor using STUNTING TRAINING data only...")

X_train_stunt_processed = (
    preprocessor_stunt.fit_transform(
        X_train_stunt
    )
)

print("✓ Preprocessor fitted on stunting training data")

# Validation transformation
X_val_stunt_processed = (
    preprocessor_stunt.transform(
        X_val_stunt
    )
)

print("✓ Stunting validation data transformed")

# Test transformation
X_test_stunt_processed = (
    preprocessor_stunt.transform(
        X_test_stunt
    )
)

print("✓ Stunting test data transformed")


# =============================================================================
# 7. UNDERWEIGHT PREPROCESSING
# =============================================================================

print("\n" + "=" * 100)
print("7. UNDERWEIGHT — TRAINING-ONLY PREPROCESSING")
print("=" * 100)

preprocessor_under = create_preprocessor()

print("Fitting preprocessor using UNDERWEIGHT TRAINING data only...")

X_train_under_processed = (
    preprocessor_under.fit_transform(
        X_train_under
    )
)

print("✓ Preprocessor fitted on underweight training data")

# Validation transformation
X_val_under_processed = (
    preprocessor_under.transform(
        X_val_under
    )
)

print("✓ Underweight validation data transformed")

# Test transformation
X_test_under_processed = (
    preprocessor_under.transform(
        X_test_under
    )
)

print("✓ Underweight test data transformed")


# =============================================================================
# 8. CONVERT TO NUMPY ARRAYS
# =============================================================================

X_train_stunt_processed = np.asarray(
    X_train_stunt_processed,
    dtype=np.float32
)

X_val_stunt_processed = np.asarray(
    X_val_stunt_processed,
    dtype=np.float32
)

X_test_stunt_processed = np.asarray(
    X_test_stunt_processed,
    dtype=np.float32
)

X_train_under_processed = np.asarray(
    X_train_under_processed,
    dtype=np.float32
)

X_val_under_processed = np.asarray(
    X_val_under_processed,
    dtype=np.float32
)

X_test_under_processed = np.asarray(
    X_test_under_processed,
    dtype=np.float32
)


# =============================================================================
# 9. PROCESSED MATRIX SHAPES
# =============================================================================

print("\n" + "=" * 100)
print("9. PROCESSED MATRIX SHAPES")
print("=" * 100)

print("\nSTUNTING")
print("-" * 70)
print(
    f"Training   : {X_train_stunt_processed.shape}"
)
print(
    f"Validation : {X_val_stunt_processed.shape}"
)
print(
    f"Test       : {X_test_stunt_processed.shape}"
)

print("\nUNDERWEIGHT")
print("-" * 70)
print(
    f"Training   : {X_train_under_processed.shape}"
)
print(
    f"Validation : {X_val_under_processed.shape}"
)
print(
    f"Test       : {X_test_under_processed.shape}"
)


# =============================================================================
# 10. TRANSFORMED FEATURE NAMES
# =============================================================================

print("\n" + "=" * 100)
print("10. TRANSFORMED FEATURE NAMES")
print("=" * 100)

stunt_feature_names = (
    preprocessor_stunt
    .get_feature_names_out()
)

under_feature_names = (
    preprocessor_under
    .get_feature_names_out()
)

print(
    f"Stunting transformed features   : "
    f"{len(stunt_feature_names)}"
)

print(
    f"Underweight transformed features: "
    f"{len(under_feature_names)}"
)

print("\nFirst 20 stunting transformed features:")

for i, feature in enumerate(
    stunt_feature_names[:20],
    1
):
    print(f"{i:2d}. {feature}")


# =============================================================================
# 11. MISSING-VALUE CHECK
# =============================================================================

print("\n" + "=" * 100)
print("11. POST-PROCESSING MISSING-VALUE CHECK")
print("=" * 100)

processed_arrays = {
    "Stunting training": X_train_stunt_processed,
    "Stunting validation": X_val_stunt_processed,
    "Stunting test": X_test_stunt_processed,
    "Underweight training": X_train_under_processed,
    "Underweight validation": X_val_under_processed,
    "Underweight test": X_test_under_processed
}

total_nan = 0
total_inf = 0

for name, array in processed_arrays.items():

    nan_count = int(np.isnan(array).sum())
    inf_count = int(np.isinf(array).sum())

    total_nan += nan_count
    total_inf += inf_count

    print(
        f"{name:<25} "
        f"NaN: {nan_count:<6,} "
        f"Inf: {inf_count:<6,}"
    )

if total_nan > 0:
    raise ValueError(
        "NaN values remain after preprocessing."
    )

if total_inf > 0:
    raise ValueError(
        "Infinite values remain after preprocessing."
    )

print("\n✓ No NaN values remain")
print("✓ No infinite values remain")


# =============================================================================
# 12. TRAINING-ONLY FIT CHECK
# =============================================================================

print("\n" + "=" * 100)
print("12. PREPROCESSING INTEGRITY")
print("=" * 100)

print("✓ Stunting preprocessor fitted on training data only")
print("✓ Underweight preprocessor fitted on training data only")
print("✓ Validation data only transformed")
print("✓ Test data only transformed")
print("✓ Test statistics were not used")
print("✓ Test categories were not used to fit encoders")
print("✓ Missing-value imputation learned from training data only")
print("✓ No feature selection performed")
print("✓ No model tuning performed")
print("✓ No threshold tuning performed")


# =============================================================================
# 13. PRESERVE TARGETS
# =============================================================================

print("\n" + "=" * 100)
print("13. TARGET CHECK")
print("=" * 100)

y_train_stunt = np.asarray(
    y_train_stunt,
    dtype=np.int8
)

y_val_stunt = np.asarray(
    y_val_stunt,
    dtype=np.int8
)

y_test_stunt = np.asarray(
    y_test_stunt,
    dtype=np.int8
)

y_train_under = np.asarray(
    y_train_under,
    dtype=np.int8
)

y_val_under = np.asarray(
    y_val_under,
    dtype=np.int8
)

y_test_under = np.asarray(
    y_test_under,
    dtype=np.int8
)

print(
    f"Stunting targets    : "
    f"{y_train_stunt.shape}, "
    f"{y_val_stunt.shape}, "
    f"{y_test_stunt.shape}"
)

print(
    f"Underweight targets : "
    f"{y_train_under.shape}, "
    f"{y_val_under.shape}, "
    f"{y_test_under.shape}"
)

if not set(np.unique(y_train_stunt)).issubset({0, 1}):
    raise ValueError("Invalid stunting training target.")

if not set(np.unique(y_val_stunt)).issubset({0, 1}):
    raise ValueError("Invalid stunting validation target.")

if not set(np.unique(y_test_stunt)).issubset({0, 1}):
    raise ValueError("Invalid stunting test target.")

if not set(np.unique(y_train_under)).issubset({0, 1}):
    raise ValueError("Invalid underweight training target.")

if not set(np.unique(y_val_under)).issubset({0, 1}):
    raise ValueError("Invalid underweight validation target.")

if not set(np.unique(y_test_under)).issubset({0, 1}):
    raise ValueError("Invalid underweight test target.")

print("✓ All targets contain only 0 and 1")


# =============================================================================
# 14. FINAL STEP 5 SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("STEP 5 COMPLETE")
print("=" * 100)

print("✓ 20 approved predictors retained")
print("✓ CAGE treated as numeric")
print("✓ Remaining 19 predictors treated as categorical")
print("✓ Missing values imputed")
print("✓ Categorical variables one-hot encoded")
print("✓ CAGE standardized")
print("✓ Stunting preprocessor fitted on training data only")
print("✓ Underweight preprocessor fitted on training data only")
print("✓ Validation data transformed")
print("✓ Test data transformed")
print("✓ No test information used to fit preprocessing")
print("✓ No feature selection performed")
print("✓ No model tuning performed")
print("✓ No threshold tuning performed")
print("✓ Test set remains reserved for final evaluation")

print("\nNEXT STEP:")
print("STEP 6 — BASELINE MODEL COMPARISON")
```

```python
# =============================================================================
# STEP 6: CAR MICS6 — BASELINE MODEL COMPARISON
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

print("=" * 100)
print("STEP 6: CAR MICS6 — BASELINE MODEL COMPARISON")
print("=" * 100)


# =============================================================================
# 1. CHECK STEP 5 OBJECTS
# =============================================================================

print("\n1. CHECKING STEP 5 OBJECTS")
print("-" * 100)

required_objects = [
    "X_train_stunt_processed",
    "X_val_stunt_processed",
    "X_test_stunt_processed",
    "y_train_stunt",
    "y_val_stunt",
    "y_test_stunt",

    "X_train_under_processed",
    "X_val_under_processed",
    "X_test_under_processed",
    "y_train_under",
    "y_val_under",
    "y_test_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "Missing Step 5 objects:\n"
        + "\n".join(missing_objects)
        + "\n\nRun Step 5 successfully before Step 6."
    )

print("✓ All Step 5 objects are available")


# =============================================================================
# 2. DATA SHAPE CHECK
# =============================================================================

print("\n" + "=" * 100)
print("2. DATA SHAPE CHECK")
print("=" * 100)

print("\nSTUNTING")
print("-" * 70)
print(f"Training   : {X_train_stunt_processed.shape}")
print(f"Validation : {X_val_stunt_processed.shape}")
print(f"Test       : {X_test_stunt_processed.shape}")
print(f"y_train    : {y_train_stunt.shape}")
print(f"y_val      : {y_val_stunt.shape}")
print(f"y_test     : {y_test_stunt.shape}")

print("\nUNDERWEIGHT")
print("-" * 70)
print(f"Training   : {X_train_under_processed.shape}")
print(f"Validation : {X_val_under_processed.shape}")
print(f"Test       : {X_test_under_processed.shape}")
print(f"y_train    : {y_train_under.shape}")
print(f"y_val      : {y_val_under.shape}")
print(f"y_test     : {y_test_under.shape}")


# =============================================================================
# 3. CLASS IMBALANCE
# =============================================================================

print("\n" + "=" * 100)
print("3. CLASS IMBALANCE")
print("=" * 100)

stunt_counts = np.bincount(y_train_stunt)
under_counts = np.bincount(y_train_under)

stunt_scale_pos_weight = (
    stunt_counts[0] / stunt_counts[1]
)

under_scale_pos_weight = (
    under_counts[0] / under_counts[1]
)

print("\nSTUNTING")
print(f"Class 0 : {stunt_counts[0]:,}")
print(f"Class 1 : {stunt_counts[1]:,}")
print(
    f"scale_pos_weight : "
    f"{stunt_scale_pos_weight:.4f}"
)

print("\nUNDERWEIGHT")
print(f"Class 0 : {under_counts[0]:,}")
print(f"Class 1 : {under_counts[1]:,}")
print(
    f"scale_pos_weight : "
    f"{under_scale_pos_weight:.4f}"
)


# =============================================================================
# 4. DEFINE BASELINE MODELS
# =============================================================================

print("\n" + "=" * 100)
print("4. DEFINING BASELINE MODELS")
print("=" * 100)

models_stunt = {

    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=5,
        min_samples_leaf=10,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        min_samples_leaf=10,
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        n_estimators=200,
        max_depth=3,
        learning_rate=0.05,
        min_child_weight=5,
        subsample=0.85,
        colsample_bytree=0.85,
        scale_pos_weight=stunt_scale_pos_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )
}


models_under = {

    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=5,
        min_samples_leaf=10,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        min_samples_leaf=10,
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        n_estimators=200,
        max_depth=3,
        learning_rate=0.05,
        min_child_weight=5,
        subsample=0.85,
        colsample_bytree=0.85,
        scale_pos_weight=under_scale_pos_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )
}

print("✓ Logistic Regression")
print("✓ Decision Tree")
print("✓ Random Forest")
print("✓ Gradient Boosting")
print("✓ XGBoost")


# =============================================================================
# 5. EVALUATION FUNCTION
# =============================================================================

def evaluate_baseline_model(
    model,
    X_train,
    y_train,
    X_val,
    y_val
):

    model.fit(X_train, y_train)

    probabilities = model.predict_proba(X_val)[:, 1]

    predictions = (
        probabilities >= 0.50
    ).astype(int)

    results = {
        "Accuracy": accuracy_score(
            y_val,
            predictions
        ),

        "Precision": precision_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "Recall": recall_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "F1": f1_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "ROC_AUC": roc_auc_score(
            y_val,
            probabilities
        ),

        "PR_AUC": average_precision_score(
            y_val,
            probabilities
        )
    }

    return model, probabilities, predictions, results


# =============================================================================
# 6. STUNTING BASELINES
# =============================================================================

print("\n" + "=" * 100)
print("5. BASELINE MODELING — STUNTING")
print("=" * 100)

stunting_results = []
stunting_models = {}
stunting_probabilities = {}
stunting_predictions = {}

for model_name, model in models_stunt.items():

    print(
        f"\nTraining: {model_name}"
    )

    (
        fitted_model,
        probabilities,
        predictions,
        results
    ) = evaluate_baseline_model(
        model,
        X_train_stunt_processed,
        y_train_stunt,
        X_val_stunt_processed,
        y_val_stunt
    )

    stunting_models[model_name] = fitted_model
    stunting_probabilities[model_name] = probabilities
    stunting_predictions[model_name] = predictions

    row = {
        "Target": "Stunting",
        "Model": model_name,
        **results
    }

    stunting_results.append(row)

    print(
        f"Accuracy={results['Accuracy']:.4f} | "
        f"Precision={results['Precision']:.4f} | "
        f"Recall={results['Recall']:.4f} | "
        f"F1={results['F1']:.4f} | "
        f"ROC-AUC={results['ROC_AUC']:.4f} | "
        f"PR-AUC={results['PR_AUC']:.4f}"
    )


# =============================================================================
# 7. UNDERWEIGHT BASELINES
# =============================================================================

print("\n" + "=" * 100)
print("6. BASELINE MODELING — UNDERWEIGHT")
print("=" * 100)

underweight_results = []
underweight_models = {}
underweight_probabilities = {}
underweight_predictions = {}

for model_name, model in models_under.items():

    print(
        f"\nTraining: {model_name}"
    )

    (
        fitted_model,
        probabilities,
        predictions,
        results
    ) = evaluate_baseline_model(
        model,
        X_train_under_processed,
        y_train_under,
        X_val_under_processed,
        y_val_under
    )

    underweight_models[model_name] = fitted_model
    underweight_probabilities[model_name] = probabilities
    underweight_predictions[model_name] = predictions

    row = {
        "Target": "Underweight",
        "Model": model_name,
        **results
    }

    underweight_results.append(row)

    print(
        f"Accuracy={results['Accuracy']:.4f} | "
        f"Precision={results['Precision']:.4f} | "
        f"Recall={results['Recall']:.4f} | "
        f"F1={results['F1']:.4f} | "
        f"ROC-AUC={results['ROC_AUC']:.4f} | "
        f"PR-AUC={results['PR_AUC']:.4f}"
    )


# =============================================================================
# 8. COMBINED VALIDATION RESULTS
# =============================================================================

print("\n" + "=" * 100)
print("7. BASELINE MODEL COMPARISON — VALIDATION SET")
print("=" * 100)

baseline_results = pd.DataFrame(
    stunting_results + underweight_results
)

baseline_results = baseline_results.sort_values(
    ["Target", "F1"],
    ascending=[True, False]
).reset_index(drop=True)

print(
    baseline_results.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 9. BEST BASELINE BY F1
# =============================================================================

print("\n" + "=" * 100)
print("8. BEST BASELINE MODEL BY F1-SCORE")
print("=" * 100)

best_stunt_row = (
    baseline_results[
        baseline_results["Target"] == "Stunting"
    ]
    .sort_values("F1", ascending=False)
    .iloc[0]
)

best_under_row = (
    baseline_results[
        baseline_results["Target"] == "Underweight"
    ]
    .sort_values("F1", ascending=False)
    .iloc[0]
)

print("\nSTUNTING")
print("-" * 70)
print(f"Model     : {best_stunt_row['Model']}")
print(f"Accuracy  : {best_stunt_row['Accuracy']:.4f}")
print(f"Precision : {best_stunt_row['Precision']:.4f}")
print(f"Recall    : {best_stunt_row['Recall']:.4f}")
print(f"F1        : {best_stunt_row['F1']:.4f}")
print(f"ROC-AUC   : {best_stunt_row['ROC_AUC']:.4f}")
print(f"PR-AUC    : {best_stunt_row['PR_AUC']:.4f}")

print("\nUNDERWEIGHT")
print("-" * 70)
print(f"Model     : {best_under_row['Model']}")
print(f"Accuracy  : {best_under_row['Accuracy']:.4f}")
print(f"Precision : {best_under_row['Precision']:.4f}")
print(f"Recall    : {best_under_row['Recall']:.4f}")
print(f"F1        : {best_under_row['F1']:.4f}")
print(f"ROC-AUC   : {best_under_row['ROC_AUC']:.4f}")
print(f"PR-AUC    : {best_under_row['PR_AUC']:.4f}")


# =============================================================================
# 10. CONFUSION MATRICES
# =============================================================================

print("\n" + "=" * 100)
print("9. CONFUSION MATRICES — BEST BASELINE MODELS")
print("=" * 100)

best_stunt_model_name = best_stunt_row["Model"]
best_under_model_name = best_under_row["Model"]

stunt_cm = confusion_matrix(
    y_val_stunt,
    stunting_predictions[best_stunt_model_name]
)

under_cm = confusion_matrix(
    y_val_under,
    underweight_predictions[best_under_model_name]
)

print("\nSTUNTING")
print("-" * 70)
print(stunt_cm)

print("\nClassification Report:")
print(
    classification_report(
        y_val_stunt,
        stunting_predictions[best_stunt_model_name],
        target_names=[
            "Not Stunted",
            "Stunted"
        ],
        zero_division=0
    )
)

print("\nUNDERWEIGHT")
print("-" * 70)
print(under_cm)

print("\nClassification Report:")
print(
    classification_report(
        y_val_under,
        underweight_predictions[best_under_model_name],
        target_names=[
            "Not Underweight",
            "Underweight"
        ],
        zero_division=0
    )
)


# =============================================================================
# 11. TEST-SET PROTECTION CHECK
# =============================================================================

print("\n" + "=" * 100)
print("10. TEST-SET PROTECTION")
print("=" * 100)

print("✓ Test predictors were NOT used for model fitting")
print("✓ Test targets were NOT used for model fitting")
print("✓ Test data was NOT used for model comparison")
print("✓ Test data was NOT used for threshold selection")
print("✓ Test data remains completely unseen")


# =============================================================================
# 12. STEP 6 SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("STEP 6 COMPLETE")
print("=" * 100)

print("✓ Five baseline classifiers evaluated")
print("✓ Training data used for fitting")
print("✓ Validation data used for comparison")
print("✓ Default threshold = 0.50")
print("✓ ROC-AUC and PR-AUC calculated from probabilities")
print("✓ F1 used for baseline ranking")
print("✓ Confusion matrices generated")
print("✓ Test set remains untouched")

print("\nNEXT STEP:")
print("STEP 7 — VALIDATION-ONLY MODEL OPTIMIZATION")
```

```python
# =============================================================================
# STEP 7: CAR MICS6 — CONTROLLED XGBOOST MODEL OPTIMIZATION
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

print("=" * 100)
print("STEP 7: CAR MICS6 — CONTROLLED XGBOOST MODEL OPTIMIZATION")
print("=" * 100)


# =============================================================================
# 1. CHECK REQUIRED STEP 5 OBJECTS
# =============================================================================

print("\n1. CHECKING STEP 5 OBJECTS")
print("-" * 100)

required_objects = [
    "X_train_stunt_processed",
    "X_val_stunt_processed",
    "X_test_stunt_processed",
    "y_train_stunt",
    "y_val_stunt",
    "y_test_stunt",
    "X_train_under_processed",
    "X_val_under_processed",
    "X_test_under_processed",
    "y_train_under",
    "y_val_under",
    "y_test_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "Missing Step 5 objects:\n"
        + "\n".join(missing_objects)
        + "\n\nRun Step 5 successfully before Step 7."
    )

print("✓ All required Step 5 objects are available")


# =============================================================================
# 2. TEST-SET LOCK
# =============================================================================

print("\n" + "=" * 100)
print("2. TEST-SET LOCK")
print("=" * 100)

print("✓ X_test_stunt_processed will NOT be used")
print("✓ y_test_stunt will NOT be used")
print("✓ X_test_under_processed will NOT be used")
print("✓ y_test_under will NOT be used")
print("✓ Test set remains completely unseen")


# =============================================================================
# 3. CLASS IMBALANCE
# =============================================================================

print("\n" + "=" * 100)
print("3. CLASS IMBALANCE")
print("=" * 100)

stunt_counts = np.bincount(y_train_stunt)
under_counts = np.bincount(y_train_under)

stunt_spw = stunt_counts[0] / stunt_counts[1]
under_spw = under_counts[0] / under_counts[1]

print(
    f"Stunting    : {stunt_spw:.4f}"
)

print(
    f"Underweight : {under_spw:.4f}"
)


# =============================================================================
# 4. CONTROLLED PARAMETER GRID
# =============================================================================

print("\n" + "=" * 100)
print("4. CONTROLLED XGBOOST SEARCH SPACE")
print("=" * 100)

parameter_grid = [

    # Conservative shallow model
    {
        "n_estimators": 300,
        "max_depth": 3,
        "learning_rate": 0.03,
        "min_child_weight": 5
    },

    # More trees, lower learning rate
    {
        "n_estimators": 500,
        "max_depth": 3,
        "learning_rate": 0.02,
        "min_child_weight": 5
    },

    # Slightly deeper
    {
        "n_estimators": 300,
        "max_depth": 4,
        "learning_rate": 0.03,
        "min_child_weight": 5
    },

    {
        "n_estimators": 500,
        "max_depth": 4,
        "learning_rate": 0.02,
        "min_child_weight": 5
    },

    # Stronger regularization
    {
        "n_estimators": 400,
        "max_depth": 3,
        "learning_rate": 0.03,
        "min_child_weight": 8
    },

    {
        "n_estimators": 500,
        "max_depth": 3,
        "learning_rate": 0.02,
        "min_child_weight": 8
    },

    # Less conservative
    {
        "n_estimators": 400,
        "max_depth": 4,
        "learning_rate": 0.03,
        "min_child_weight": 8
    },

    {
        "n_estimators": 500,
        "max_depth": 4,
        "learning_rate": 0.02,
        "min_child_weight": 8
    }
]

print(
    f"Number of configurations: {len(parameter_grid)}"
)


# =============================================================================
# 5. THRESHOLD GRID
# =============================================================================

thresholds = np.round(
    np.arange(0.25, 0.71, 0.01),
    2
)

print(
    f"Thresholds evaluated: "
    f"{thresholds[0]:.2f} to {thresholds[-1]:.2f}"
)


# =============================================================================
# 6. MODEL EVALUATION FUNCTION
# =============================================================================

def evaluate_thresholds(
    y_true,
    probabilities,
    thresholds
):

    rows = []

    for threshold in thresholds:

        predictions = (
            probabilities >= threshold
        ).astype(int)

        rows.append({

            "Threshold": threshold,

            "Accuracy": accuracy_score(
                y_true,
                predictions
            ),

            "Precision": precision_score(
                y_true,
                predictions,
                zero_division=0
            ),

            "Recall": recall_score(
                y_true,
                predictions,
                zero_division=0
            ),

            "F1": f1_score(
                y_true,
                predictions,
                zero_division=0
            )
        })

    return pd.DataFrame(rows)


# =============================================================================
# 7. STUNTING OPTIMIZATION
# =============================================================================

print("\n" + "=" * 100)
print("5. XGBOOST OPTIMIZATION — STUNTING")
print("=" * 100)

stunting_optimization = []

stunting_probability_store = {}

for i, params in enumerate(
    parameter_grid,
    start=1
):

    print(
        f"\nConfiguration {i}/{len(parameter_grid)}"
    )

    print(
        f"n_estimators={params['n_estimators']} | "
        f"max_depth={params['max_depth']} | "
        f"learning_rate={params['learning_rate']} | "
        f"min_child_weight={params['min_child_weight']}"
    )

    model = XGBClassifier(

        n_estimators=params["n_estimators"],

        max_depth=params["max_depth"],

        learning_rate=params["learning_rate"],

        min_child_weight=params["min_child_weight"],

        subsample=0.85,

        colsample_bytree=0.85,

        gamma=0,

        reg_alpha=0.0,

        reg_lambda=1.0,

        scale_pos_weight=stunt_spw,

        objective="binary:logistic",

        eval_metric="logloss",

        random_state=42,

        n_jobs=-1
    )

    model.fit(
        X_train_stunt_processed,
        y_train_stunt
    )

    probabilities = model.predict_proba(
        X_val_stunt_processed
    )[:, 1]

    roc_auc = roc_auc_score(
        y_val_stunt,
        probabilities
    )

    pr_auc = average_precision_score(
        y_val_stunt,
        probabilities
    )

    threshold_results = evaluate_thresholds(
        y_val_stunt,
        probabilities,
        thresholds
    )

    best_threshold_row = (
        threshold_results
        .sort_values(
            ["F1", "Recall", "Precision"],
            ascending=False
        )
        .iloc[0]
    )

    result = {

        **params,

        "ROC_AUC": roc_auc,

        "PR_AUC": pr_auc,

        "Threshold": best_threshold_row[
            "Threshold"
        ],

        "Accuracy": best_threshold_row[
            "Accuracy"
        ],

        "Precision": best_threshold_row[
            "Precision"
        ],

        "Recall": best_threshold_row[
            "Recall"
        ],

        "F1": best_threshold_row[
            "F1"
        ]
    }

    stunting_optimization.append(
        result
    )

    stunting_probability_store[
        i
    ] = {
        "model": model,
        "probabilities": probabilities
    }

    print(
        f"ROC-AUC={roc_auc:.4f} | "
        f"PR-AUC={pr_auc:.4f} | "
        f"Threshold={result['Threshold']:.2f} | "
        f"Precision={result['Precision']:.4f} | "
        f"Recall={result['Recall']:.4f} | "
        f"F1={result['F1']:.4f}"
    )


# =============================================================================
# 8. STUNTING RESULTS
# =============================================================================

stunting_optimization_df = pd.DataFrame(
    stunting_optimization
)

stunting_optimization_df = (
    stunting_optimization_df
    .sort_values(
        ["F1", "PR_AUC", "ROC_AUC"],
        ascending=False
    )
    .reset_index(drop=True)
)

print("\n" + "=" * 100)
print("6. STUNTING — OPTIMIZATION RESULTS")
print("=" * 100)

print(
    stunting_optimization_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 9. BEST STUNTING MODEL
# =============================================================================

best_stunting = (
    stunting_optimization_df
    .iloc[0]
)

print("\n" + "=" * 100)
print("7. BEST OPTIMIZED STUNTING CONFIGURATION")
print("=" * 100)

print(
    f"n_estimators     : "
    f"{int(best_stunting['n_estimators'])}"
)

print(
    f"max_depth        : "
    f"{int(best_stunting['max_depth'])}"
)

print(
    f"learning_rate    : "
    f"{best_stunting['learning_rate']}"
)

print(
    f"min_child_weight : "
    f"{int(best_stunting['min_child_weight'])}"
)

print(
    f"Threshold        : "
    f"{best_stunting['Threshold']:.2f}"
)

print(
    f"ROC-AUC          : "
    f"{best_stunting['ROC_AUC']:.4f}"
)

print(
    f"PR-AUC           : "
    f"{best_stunting['PR_AUC']:.4f}"
)

print(
    f"Accuracy         : "
    f"{best_stunting['Accuracy']:.4f}"
)

print(
    f"Precision        : "
    f"{best_stunting['Precision']:.4f}"
)

print(
    f"Recall           : "
    f"{best_stunting['Recall']:.4f}"
)

print(
    f"F1               : "
    f"{best_stunting['F1']:.4f}"
)


# =============================================================================
# 10. UNDERWEIGHT OPTIMIZATION
# =============================================================================

print("\n" + "=" * 100)
print("8. XGBOOST OPTIMIZATION — UNDERWEIGHT")
print("=" * 100)

underweight_optimization = []

underweight_probability_store = {}

for i, params in enumerate(
    parameter_grid,
    start=1
):

    print(
        f"\nConfiguration {i}/{len(parameter_grid)}"
    )

    print(
        f"n_estimators={params['n_estimators']} | "
        f"max_depth={params['max_depth']} | "
        f"learning_rate={params['learning_rate']} | "
        f"min_child_weight={params['min_child_weight']}"
    )

    model = XGBClassifier(

        n_estimators=params["n_estimators"],

        max_depth=params["max_depth"],

        learning_rate=params["learning_rate"],

        min_child_weight=params["min_child_weight"],

        subsample=0.85,

        colsample_bytree=0.85,

        gamma=0,

        reg_alpha=0.0,

        reg_lambda=1.0,

        scale_pos_weight=under_spw,

        objective="binary:logistic",

        eval_metric="logloss",

        random_state=42,

        n_jobs=-1
    )

    model.fit(
        X_train_under_processed,
        y_train_under
    )

    probabilities = model.predict_proba(
        X_val_under_processed
    )[:, 1]

    roc_auc = roc_auc_score(
        y_val_under,
        probabilities
    )

    pr_auc = average_precision_score(
        y_val_under,
        probabilities
    )

    threshold_results = evaluate_thresholds(
        y_val_under,
        probabilities,
        thresholds
    )

    best_threshold_row = (
        threshold_results
        .sort_values(
            ["F1", "Recall", "Precision"],
            ascending=False
        )
        .iloc[0]
    )

    result = {

        **params,

        "ROC_AUC": roc_auc,

        "PR_AUC": pr_auc,

        "Threshold": best_threshold_row[
            "Threshold"
        ],

        "Accuracy": best_threshold_row[
            "Accuracy"
        ],

        "Precision": best_threshold_row[
            "Precision"
        ],

        "Recall": best_threshold_row[
            "Recall"
        ],

        "F1": best_threshold_row[
            "F1"
        ]
    }

    underweight_optimization.append(
        result
    )

    underweight_probability_store[
        i
    ] = {
        "model": model,
        "probabilities": probabilities
    }

    print(
        f"ROC-AUC={roc_auc:.4f} | "
        f"PR-AUC={pr_auc:.4f} | "
        f"Threshold={result['Threshold']:.2f} | "
        f"Precision={result['Precision']:.4f} | "
        f"Recall={result['Recall']:.4f} | "
        f"F1={result['F1']:.4f}"
    )


# =============================================================================
# 11. UNDERWEIGHT RESULTS
# =============================================================================

underweight_optimization_df = pd.DataFrame(
    underweight_optimization
)

underweight_optimization_df = (
    underweight_optimization_df
    .sort_values(
        ["F1", "PR_AUC", "ROC_AUC"],
        ascending=False
    )
    .reset_index(drop=True)
)

print("\n" + "=" * 100)
print("9. UNDERWEIGHT — OPTIMIZATION RESULTS")
print("=" * 100)

print(
    underweight_optimization_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 12. BEST UNDERWEIGHT MODEL
# =============================================================================

best_underweight = (
    underweight_optimization_df
    .iloc[0]
)

print("\n" + "=" * 100)
print("10. BEST OPTIMIZED UNDERWEIGHT CONFIGURATION")
print("=" * 100)

print(
    f"n_estimators     : "
    f"{int(best_underweight['n_estimators'])}"
)

print(
    f"max_depth        : "
    f"{int(best_underweight['max_depth'])}"
)

print(
    f"learning_rate    : "
    f"{best_underweight['learning_rate']}"
)

print(
    f"min_child_weight : "
    f"{int(best_underweight['min_child_weight'])}"
)

print(
    f"Threshold        : "
    f"{best_underweight['Threshold']:.2f}"
)

print(
    f"ROC-AUC          : "
    f"{best_underweight['ROC_AUC']:.4f}"
)

print(
    f"PR-AUC           : "
    f"{best_underweight['PR_AUC']:.4f}"
)

print(
    f"Accuracy         : "
    f"{best_underweight['Accuracy']:.4f}"
)

print(
    f"Precision        : "
    f"{best_underweight['Precision']:.4f}"
)

print(
    f"Recall           : "
    f"{best_underweight['Recall']:.4f}"
)

print(
    f"F1               : "
    f"{best_underweight['F1']:.4f}"
)


# =============================================================================
# 13. BASELINE COMPARISON
# =============================================================================

print("\n" + "=" * 100)
print("11. BASELINE VS OPTIMIZED PERFORMANCE")
print("=" * 100)

baseline_stunt = (
    baseline_results[
        baseline_results["Target"] == "Stunting"
    ]
    .sort_values("F1", ascending=False)
    .iloc[0]
)

baseline_under = (
    baseline_results[
        baseline_results["Target"] == "Underweight"
    ]
    .sort_values("F1", ascending=False)
    .iloc[0]
)

comparison = pd.DataFrame({

    "Target": [
        "Stunting",
        "Underweight"
    ],

    "Baseline_Model": [
        baseline_stunt["Model"],
        baseline_under["Model"]
    ],

    "Baseline_F1": [
        baseline_stunt["F1"],
        baseline_under["F1"]
    ],

    "Optimized_F1": [
        best_stunting["F1"],
        best_underweight["F1"]
    ],

    "Baseline_ROC_AUC": [
        baseline_stunt["ROC_AUC"],
        baseline_under["ROC_AUC"]
    ],

    "Optimized_ROC_AUC": [
        best_stunting["ROC_AUC"],
        best_underweight["ROC_AUC"]
    ],

    "Baseline_PR_AUC": [
        baseline_stunt["PR_AUC"],
        baseline_under["PR_AUC"]
    ],

    "Optimized_PR_AUC": [
        best_stunting["PR_AUC"],
        best_underweight["PR_AUC"]
    ]
})

comparison["F1_Improvement"] = (
    comparison["Optimized_F1"]
    - comparison["Baseline_F1"]
)

comparison["ROC_AUC_Improvement"] = (
    comparison["Optimized_ROC_AUC"]
    - comparison["Baseline_ROC_AUC"]
)

comparison["PR_AUC_Improvement"] = (
    comparison["Optimized_PR_AUC"]
    - comparison["Baseline_PR_AUC"]
)

print(
    comparison.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 14. CONFUSION MATRICES FOR OPTIMIZED MODELS
# =============================================================================

print("\n" + "=" * 100)
print("12. OPTIMIZED VALIDATION CONFUSION MATRICES")
print("=" * 100)

# Recover the actual best model from the stored configurations
best_stunting_index = (
    stunting_optimization_df.index[0]
)

best_underweight_index = (
    underweight_optimization_df.index[0]
)

# Find matching configuration by parameter values
def recover_best_model(
    optimization_df,
    probability_store,
    target_row
):

    for key, stored in probability_store.items():

        candidate_model = stored["model"]

        if (
            candidate_model.n_estimators
            == int(target_row["n_estimators"])
            and
            candidate_model.max_depth
            == int(target_row["max_depth"])
            and
            abs(
                candidate_model.learning_rate
                - target_row["learning_rate"]
            ) < 1e-12
            and
            candidate_model.min_child_weight
            == int(target_row["min_child_weight"])
        ):
            return candidate_model, stored["probabilities"]

    raise RuntimeError(
        "Could not recover best model."
    )


best_stunting_model, best_stunting_probabilities = (
    recover_best_model(
        stunting_optimization_df,
        stunting_probability_store,
        best_stunting
    )
)

best_underweight_model, best_underweight_probabilities = (
    recover_best_model(
        underweight_optimization_df,
        underweight_probability_store,
        best_underweight
    )
)

stunting_final_val_predictions = (
    best_stunting_probabilities
    >= best_stunting["Threshold"]
).astype(int)

underweight_final_val_predictions = (
    best_underweight_probabilities
    >= best_underweight["Threshold"]
).astype(int
)

print("\nSTUNTING")
print("-" * 70)

print(
    confusion_matrix(
        y_val_stunt,
        stunting_final_val_predictions
    )
)

print(
    classification_report(
        y_val_stunt,
        stunting_final_val_predictions,
        target_names=[
            "Not Stunted",
            "Stunted"
        ],
        zero_division=0
    )
)

print("\nUNDERWEIGHT")
print("-" * 70)

print(
    confusion_matrix(
        y_val_under,
        underweight_final_val_predictions
    )
)

print(
    classification_report(
        y_val_under,
        underweight_final_val_predictions,
        target_names=[
            "Not Underweight",
            "Underweight"
        ],
        zero_division=0
    )
)


# =============================================================================
# 15. TARGET REGION CHECK
# =============================================================================

print("\n" + "=" * 100)
print("13. TARGET REGION CHECK")
print("=" * 100)

def target_check(row):

    precision_ok = row["Precision"] >= 0.60
    recall_ok = row["Recall"] >= 0.70
    f1_ok = row["F1"] >= 0.60

    return precision_ok, recall_ok, f1_ok


stunt_checks = target_check(best_stunting)
under_checks = target_check(best_underweight)

print("\nSTUNTING")

print(
    f"Precision >= 60% : "
    f"{'✓' if stunt_checks[0] else '✗'}"
)

print(
    f"Recall >= 70%    : "
    f"{'✓' if stunt_checks[1] else '✗'}"
)

print(
    f"F1 >= 60%        : "
    f"{'✓' if stunt_checks[2] else '✗'}"
)

print("\nUNDERWEIGHT")

print(
    f"Precision >= 60% : "
    f"{'✓' if under_checks[0] else '✗'}"
)

print(
    f"Recall >= 70%    : "
    f"{'✓' if under_checks[1] else '✗'}"
)

print(
    f"F1 >= 60%        : "
    f"{'✓' if under_checks[2] else '✗'}"
)


# =============================================================================
# 16. FINAL STEP 7 INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("14. STEP 7 INTEGRITY CHECK")
print("=" * 100)

print("✓ Models fitted using training data only")
print("✓ Hyperparameter selection based on validation performance")
print("✓ Threshold selection based on validation predictions")
print("✓ Test predictors were NOT used")
print("✓ Test targets were NOT used")
print("✓ Test-set performance was NOT inspected")
print("✓ No test-based tuning performed")
print("✓ Baseline comparison completed")


# =============================================================================
# 17. SAVE BEST OBJECTS IN MEMORY
# =============================================================================

final_stunting_optimized_model = best_stunting_model
final_underweight_optimized_model = best_underweight_model

final_stunting_threshold = float(
    best_stunting["Threshold"]
)

final_underweight_threshold = float(
    best_underweight["Threshold"]
)

print("\n" + "=" * 100)
print("STEP 7 COMPLETE")
print("=" * 100)

print("✓ Stunting optimized model selected")
print("✓ Underweight optimized model selected")
print("✓ Validation thresholds selected")
print("✓ Best models stored in memory")
print("✓ Test set remains locked")

print("\nNEXT STEP:")
print("STEP 8 — FINAL MODEL FITTING AND VALIDATION CONFIRMATION")
```

```python
# =============================================================================
# STEP 7B: CAR MICS6 — SECOND XGBOOST OPTIMIZATION ROUND
# Focus:
#   • Wider learning-rate search
#   • More estimators for lower learning rates
#   • Stronger regularization
#   • Subsampling
#   • Feature subsampling
#   • Validation-only threshold optimization
#   • TEST SET REMAINS LOCKED
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

print("=" * 100)
print("STEP 7B: CAR MICS6 — SECOND XGBOOST OPTIMIZATION ROUND")
print("=" * 100)


# =============================================================================
# 1. CHECK STEP 5 OBJECTS
# =============================================================================

print("\n1. CHECKING STEP 5 OBJECTS")
print("-" * 100)

required_objects = [
    "X_train_stunt_processed",
    "X_val_stunt_processed",
    "X_test_stunt_processed",
    "y_train_stunt",
    "y_val_stunt",
    "y_test_stunt",
    "X_train_under_processed",
    "X_val_under_processed",
    "X_test_under_processed",
    "y_train_under",
    "y_val_under",
    "y_test_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required Step 5 objects:\n"
        + "\n".join(missing_objects)
        + "\n\nPlease run Steps 1–5 successfully before Step 7B."
    )

print("✓ All Step 5 objects available")


# =============================================================================
# 2. LOCK TEST SET
# =============================================================================

print("\n" + "=" * 100)
print("2. TEST-SET LOCK")
print("=" * 100)

print("✓ Test predictors will NOT be used")
print("✓ Test targets will NOT be used")
print("✓ Test performance will NOT be inspected")
print("✓ Test set remains completely unseen")


# =============================================================================
# 3. CLASS WEIGHTS
# =============================================================================

stunt_counts = np.bincount(y_train_stunt)
under_counts = np.bincount(y_train_under)

stunt_spw = stunt_counts[0] / stunt_counts[1]
under_spw = under_counts[0] / under_counts[1]

print("\n" + "=" * 100)
print("3. CLASS IMBALANCE")
print("=" * 100)

print(f"Stunting scale_pos_weight    : {stunt_spw:.4f}")
print(f"Underweight scale_pos_weight : {under_spw:.4f}")


# =============================================================================
# 4. WIDER SEARCH SPACE
# =============================================================================

print("\n" + "=" * 100)
print("4. SECOND-ROUND SEARCH SPACE")
print("=" * 100)

parameter_grid_2 = [

    # -------------------------------------------------------------------------
    # LOW LEARNING RATE
    # -------------------------------------------------------------------------

    {
        "n_estimators": 800,
        "max_depth": 2,
        "learning_rate": 0.005,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.0,
        "reg_lambda": 1.0
    },

    {
        "n_estimators": 600,
        "max_depth": 3,
        "learning_rate": 0.01,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.0,
        "reg_lambda": 1.0
    },

    {
        "n_estimators": 800,
        "max_depth": 3,
        "learning_rate": 0.01,
        "min_child_weight": 8,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.0,
        "reg_lambda": 1.0
    },

    # -------------------------------------------------------------------------
    # MODERATE LEARNING RATE
    # -------------------------------------------------------------------------

    {
        "n_estimators": 500,
        "max_depth": 3,
        "learning_rate": 0.015,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.0,
        "reg_lambda": 1.0
    },

    {
        "n_estimators": 600,
        "max_depth": 3,
        "learning_rate": 0.015,
        "min_child_weight": 8,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.0,
        "reg_lambda": 1.0
    },

    {
        "n_estimators": 500,
        "max_depth": 3,
        "learning_rate": 0.025,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.0,
        "reg_lambda": 1.0
    },

    # -------------------------------------------------------------------------
    # STRONGER REGULARIZATION
    # -------------------------------------------------------------------------

    {
        "n_estimators": 500,
        "max_depth": 3,
        "learning_rate": 0.02,
        "min_child_weight": 8,
        "subsample": 0.75,
        "colsample_bytree": 0.75,
        "reg_alpha": 0.10,
        "reg_lambda": 2.0
    },

    {
        "n_estimators": 500,
        "max_depth": 3,
        "learning_rate": 0.02,
        "min_child_weight": 10,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.20,
        "reg_lambda": 2.0
    },

    # -------------------------------------------------------------------------
    # DEEPER MODEL WITH REGULARIZATION
    # -------------------------------------------------------------------------

    {
        "n_estimators": 400,
        "max_depth": 4,
        "learning_rate": 0.015,
        "min_child_weight": 8,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.10,
        "reg_lambda": 2.0
    },

    {
        "n_estimators": 500,
        "max_depth": 4,
        "learning_rate": 0.02,
        "min_child_weight": 10,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.20,
        "reg_lambda": 2.0
    }
]

print(f"Configurations: {len(parameter_grid_2)}")
print("Learning rates tested:")
print("0.005, 0.010, 0.015, 0.020, 0.025")


# =============================================================================
# 5. THRESHOLD SEARCH
# =============================================================================

thresholds_2 = np.round(
    np.arange(0.30, 0.71, 0.01),
    2
)

print(
    f"Threshold range: "
    f"{thresholds_2.min():.2f}–{thresholds_2.max():.2f}"
)


# =============================================================================
# 6. THRESHOLD EVALUATION
# =============================================================================

def threshold_search(
    y_true,
    probabilities,
    thresholds
):

    results = []

    for threshold in thresholds:

        pred = (
            probabilities >= threshold
        ).astype(int)

        results.append({

            "Threshold": threshold,

            "Accuracy": accuracy_score(
                y_true,
                pred
            ),

            "Precision": precision_score(
                y_true,
                pred,
                zero_division=0
            ),

            "Recall": recall_score(
                y_true,
                pred,
                zero_division=0
            ),

            "F1": f1_score(
                y_true,
                pred,
                zero_division=0
            )
        })

    return pd.DataFrame(results)


# =============================================================================
# 7. STUNTING
# =============================================================================

print("\n" + "=" * 100)
print("5. SECOND-ROUND OPTIMIZATION — STUNTING")
print("=" * 100)

stunting_results_2 = []

for i, params in enumerate(
    parameter_grid_2,
    start=1
):

    print(
        f"\nConfiguration {i}/{len(parameter_grid_2)}"
    )

    model = XGBClassifier(

        n_estimators=params["n_estimators"],
        max_depth=params["max_depth"],
        learning_rate=params["learning_rate"],
        min_child_weight=params["min_child_weight"],

        subsample=params["subsample"],
        colsample_bytree=params["colsample_bytree"],

        reg_alpha=params["reg_alpha"],
        reg_lambda=params["reg_lambda"],

        gamma=0,

        scale_pos_weight=stunt_spw,

        objective="binary:logistic",
        eval_metric="logloss",

        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_stunt_processed,
        y_train_stunt
    )

    val_prob = model.predict_proba(
        X_val_stunt_processed
    )[:, 1]

    roc_auc = roc_auc_score(
        y_val_stunt,
        val_prob
    )

    pr_auc = average_precision_score(
        y_val_stunt,
        val_prob
    )

    threshold_results = threshold_search(
        y_val_stunt,
        val_prob,
        thresholds_2
    )

    best = (
        threshold_results
        .sort_values(
            ["F1", "Precision", "Recall"],
            ascending=False
        )
        .iloc[0]
    )

    result = {
        **params,
        "ROC_AUC": roc_auc,
        "PR_AUC": pr_auc,
        "Threshold": best["Threshold"],
        "Accuracy": best["Accuracy"],
        "Precision": best["Precision"],
        "Recall": best["Recall"],
        "F1": best["F1"]
    }

    stunting_results_2.append(result)

    print(
        f"LR={params['learning_rate']:.3f} | "
        f"Depth={params['max_depth']} | "
        f"Estimators={params['n_estimators']} | "
        f"ROC-AUC={roc_auc:.4f} | "
        f"PR-AUC={pr_auc:.4f} | "
        f"Threshold={best['Threshold']:.2f} | "
        f"Precision={best['Precision']:.4f} | "
        f"Recall={best['Recall']:.4f} | "
        f"F1={best['F1']:.4f}"
    )


stunting_results_2_df = (
    pd.DataFrame(stunting_results_2)
    .sort_values(
        ["F1", "PR_AUC", "ROC_AUC"],
        ascending=False
    )
    .reset_index(drop=True)
)


# =============================================================================
# 8. UNDERWEIGHT
# =============================================================================

print("\n" + "=" * 100)
print("6. SECOND-ROUND OPTIMIZATION — UNDERWEIGHT")
print("=" * 100)

underweight_results_2 = []

for i, params in enumerate(
    parameter_grid_2,
    start=1
):

    print(
        f"\nConfiguration {i}/{len(parameter_grid_2)}"
    )

    model = XGBClassifier(

        n_estimators=params["n_estimators"],
        max_depth=params["max_depth"],
        learning_rate=params["learning_rate"],
        min_child_weight=params["min_child_weight"],

        subsample=params["subsample"],
        colsample_bytree=params["colsample_bytree"],

        reg_alpha=params["reg_alpha"],
        reg_lambda=params["reg_lambda"],

        gamma=0,

        scale_pos_weight=under_spw,

        objective="binary:logistic",
        eval_metric="logloss",

        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_under_processed,
        y_train_under
    )

    val_prob = model.predict_proba(
        X_val_under_processed
    )[:, 1]

    roc_auc = roc_auc_score(
        y_val_under,
        val_prob
    )

    pr_auc = average_precision_score(
        y_val_under,
        val_prob
    )

    threshold_results = threshold_search(
        y_val_under,
        val_prob,
        thresholds_2
    )

    best = (
        threshold_results
        .sort_values(
            ["F1", "Precision", "Recall"],
            ascending=False
        )
        .iloc[0]
    )

    result = {
        **params,
        "ROC_AUC": roc_auc,
        "PR_AUC": pr_auc,
        "Threshold": best["Threshold"],
        "Accuracy": best["Accuracy"],
        "Precision": best["Precision"],
        "Recall": best["Recall"],
        "F1": best["F1"]
    }

    underweight_results_2.append(result)

    print(
        f"LR={params['learning_rate']:.3f} | "
        f"Depth={params['max_depth']} | "
        f"Estimators={params['n_estimators']} | "
        f"ROC-AUC={roc_auc:.4f} | "
        f"PR-AUC={pr_auc:.4f} | "
        f"Threshold={best['Threshold']:.2f} | "
        f"Precision={best['Precision']:.4f} | "
        f"Recall={best['Recall']:.4f} | "
        f"F1={best['F1']:.4f}"
    )


underweight_results_2_df = (
    pd.DataFrame(underweight_results_2)
    .sort_values(
        ["F1", "PR_AUC", "ROC_AUC"],
        ascending=False
    )
    .reset_index(drop=True)
)


# =============================================================================
# 9. DISPLAY RESULTS
# =============================================================================

print("\n" + "=" * 100)
print("7. STUNTING — SECOND ROUND RESULTS")
print("=" * 100)

print(
    stunting_results_2_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\n" + "=" * 100)
print("8. UNDERWEIGHT — SECOND ROUND RESULTS")
print("=" * 100)

print(
    underweight_results_2_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 10. BEST MODELS
# =============================================================================

best_stunting_2 = stunting_results_2_df.iloc[0]
best_underweight_2 = underweight_results_2_df.iloc[0]

print("\n" + "=" * 100)
print("9. BEST SECOND-ROUND MODELS")
print("=" * 100)

print("\nSTUNTING")
print("-" * 70)
print(
    f"Estimators     : {int(best_stunting_2['n_estimators'])}"
)
print(
    f"Max depth      : {int(best_stunting_2['max_depth'])}"
)
print(
    f"Learning rate  : {best_stunting_2['learning_rate']:.3f}"
)
print(
    f"Min child wt   : {int(best_stunting_2['min_child_weight'])}"
)
print(
    f"Subsample      : {best_stunting_2['subsample']:.2f}"
)
print(
    f"Colsample      : {best_stunting_2['colsample_bytree']:.2f}"
)
print(
    f"Reg alpha      : {best_stunting_2['reg_alpha']:.2f}"
)
print(
    f"Reg lambda     : {best_stunting_2['reg_lambda']:.2f}"
)
print(
    f"Threshold      : {best_stunting_2['Threshold']:.2f}"
)
print(
    f"ROC-AUC        : {best_stunting_2['ROC_AUC']:.4f}"
)
print(
    f"PR-AUC         : {best_stunting_2['PR_AUC']:.4f}"
)
print(
    f"Precision      : {best_stunting_2['Precision']:.4f}"
)
print(
    f"Recall         : {best_stunting_2['Recall']:.4f}"
)
print(
    f"F1             : {best_stunting_2['F1']:.4f}"
)


print("\nUNDERWEIGHT")
print("-" * 70)
print(
    f"Estimators     : {int(best_underweight_2['n_estimators'])}"
)
print(
    f"Max depth      : {int(best_underweight_2['max_depth'])}"
)
print(
    f"Learning rate  : {best_underweight_2['learning_rate']:.3f}"
)
print(
    f"Min child wt   : {int(best_underweight_2['min_child_weight'])}"
)
print(
    f"Subsample      : {best_underweight_2['subsample']:.2f}"
)
print(
    f"Colsample      : {best_underweight_2['colsample_bytree']:.2f}"
)
print(
    f"Reg alpha      : {best_underweight_2['reg_alpha']:.2f}"
)
print(
    f"Reg lambda     : {best_underweight_2['reg_lambda']:.2f}"
)
print(
    f"Threshold      : {best_underweight_2['Threshold']:.2f}"
)
print(
    f"ROC-AUC        : {best_underweight_2['ROC_AUC']:.4f}"
)
print(
    f"PR-AUC         : {best_underweight_2['PR_AUC']:.4f}"
)
print(
    f"Precision      : {best_underweight_2['Precision']:.4f}"
)
print(
    f"Recall         : {best_underweight_2['Recall']:.4f}"
)
print(
    f"F1             : {best_underweight_2['F1']:.4f}"
)


# =============================================================================
# 11. COMPARE AGAINST STEP 6 AND STEP 7
# =============================================================================

print("\n" + "=" * 100)
print("10. PERFORMANCE IMPROVEMENT")
print("=" * 100)

comparison_2 = pd.DataFrame({

    "Target": [
        "Stunting",
        "Underweight"
    ],

    "Step6_Baseline_F1": [
        0.5714,
        0.4508
    ],

    "Step7_F1": [
        0.5875,
        0.4718
    ],

    "Step7B_F1": [
        best_stunting_2["F1"],
        best_underweight_2["F1"]
    ],

    "Step6_ROC_AUC": [
        0.7008,
        0.6730
    ],

    "Step7_ROC_AUC": [
        0.7080,
        0.6765
    ],

    "Step7B_ROC_AUC": [
        best_stunting_2["ROC_AUC"],
        best_underweight_2["ROC_AUC"]
    ],

    "Step6_PR_AUC": [
        0.4878,
        0.3811
    ],

    "Step7_PR_AUC": [
        0.4985,
        0.3840
    ],

    "Step7B_PR_AUC": [
        best_stunting_2["PR_AUC"],
        best_underweight_2["PR_AUC"]
    ]
})

comparison_2["F1_vs_Step6"] = (
    comparison_2["Step7B_F1"]
    - comparison_2["Step6_Baseline_F1"]
)

comparison_2["F1_vs_Step7"] = (
    comparison_2["Step7B_F1"]
    - comparison_2["Step7_F1"]
)

comparison_2["ROC_AUC_vs_Step6"] = (
    comparison_2["Step7B_ROC_AUC"]
    - comparison_2["Step6_ROC_AUC"]
)

comparison_2["PR_AUC_vs_Step6"] = (
    comparison_2["Step7B_PR_AUC"]
    - comparison_2["Step6_PR_AUC"]
)

print(
    comparison_2.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 12. FINAL DECISION
# =============================================================================

print("\n" + "=" * 100)
print("11. STEP 7B DECISION")
print("=" * 100)

print("\nSTUNTING")

if best_stunting_2["F1"] > 0.5875:
    print("✓ Step 7B improved on the previous optimized stunting model")
else:
    print("✗ Step 7B did not improve stunting F1")

print("\nUNDERWEIGHT")

if best_underweight_2["F1"] > 0.4718:
    print("✓ Step 7B improved on the previous optimized underweight model")
else:
    print("✗ Step 7B did not improve underweight F1")


# =============================================================================
# 13. INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("12. STEP 7B INTEGRITY CHECK")
print("=" * 100)

print("✓ Training data used for model fitting")
print("✓ Validation data used for model selection")
print("✓ Validation data used for threshold selection")
print("✓ Test predictors NOT used")
print("✓ Test targets NOT used")
print("✓ No test performance inspected")
print("✓ No test-based tuning")
print("✓ Learning-rate search expanded")
print("✓ Regularization search expanded")
print("✓ Subsampling considered")
print("✓ Feature subsampling considered")


print("\n" + "=" * 100)
print("STEP 7B COMPLETE")
print("=" * 100)

print("""
DO NOT RUN STEP 8 YET.

First review the Step 7B results.
""")
```

```python
# =============================================================================
# STEP 7C: CAR MICS6 — PRECISION-FOCUSED OPTIMIZATION
# =============================================================================
#
# OBJECTIVE:
#   Improve precision and reduce false positives.
#
# IMPORTANT:
#   - Training data: MODEL FITTING ONLY
#   - Validation data: MODEL/THRESHOLD COMPARISON
#   - Test data: COMPLETELY LOCKED
#
# We do NOT select the threshold using F1 alone.
# We explicitly examine precision-recall trade-offs.
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

print("=" * 100)
print("STEP 7C: CAR MICS6 — PRECISION-FOCUSED OPTIMIZATION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

print("\n1. CHECKING REQUIRED OBJECTS")
print("-" * 100)

required_objects = [
    "X_train_stunt_processed",
    "X_val_stunt_processed",
    "y_train_stunt",
    "y_val_stunt",

    "X_train_under_processed",
    "X_val_under_processed",
    "y_train_under",
    "y_val_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:

    raise NameError(
        "The following required objects are missing:\n\n"
        + "\n".join(missing_objects)
        + "\n\n"
        "Please run Steps 1–7B successfully before running Step 7C."
    )

print("✓ All required training/validation objects are available")


# =============================================================================
# 2. EXPLICIT TEST-SET PROTECTION
# =============================================================================

print("\n" + "=" * 100)
print("2. TEST-SET PROTECTION")
print("=" * 100)

print("✓ Test predictors are NOT used")
print("✓ Test targets are NOT used")
print("✓ Test thresholds are NOT evaluated")
print("✓ Test performance is NOT inspected")
print("✓ Test set remains completely unseen")


# =============================================================================
# 3. CLASS WEIGHTS
# =============================================================================

print("\n" + "=" * 100)
print("3. TRAINING CLASS DISTRIBUTION")
print("=" * 100)

stunt_counts = np.bincount(
    np.asarray(y_train_stunt).astype(int)
)

under_counts = np.bincount(
    np.asarray(y_train_under).astype(int)
)

stunt_spw = (
    stunt_counts[0] / stunt_counts[1]
)

under_spw = (
    under_counts[0] / under_counts[1]
)

print("\nSTUNTING")
print(f"Class 0 : {stunt_counts[0]:,}")
print(f"Class 1 : {stunt_counts[1]:,}")
print(f"scale_pos_weight : {stunt_spw:.4f}")

print("\nUNDERWEIGHT")
print(f"Class 0 : {under_counts[0]:,}")
print(f"Class 1 : {under_counts[1]:,}")
print(f"scale_pos_weight : {under_spw:.4f}")


# =============================================================================
# 4. MODELS TO COMPARE
# =============================================================================
#
# We are deliberately comparing models rather than assuming XGBoost must win.
# A model with a better precision-recall operating point may be preferable.
# =============================================================================

print("\n" + "=" * 100)
print("4. PRECISION-FOCUSED MODEL SET")
print("=" * 100)

models_stunting = {

    "XGBoost_Optimized": XGBClassifier(
        n_estimators=500,
        max_depth=3,
        learning_rate=0.02,
        min_child_weight=8,
        subsample=0.75,
        colsample_bytree=0.75,
        reg_alpha=0.10,
        reg_lambda=2.0,
        gamma=0,
        scale_pos_weight=stunt_spw,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    ),

    "Logistic_Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Random_Forest": RandomForestClassifier(
        n_estimators=500,
        max_depth=8,
        min_samples_leaf=5,
        max_features="sqrt",
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}


models_underweight = {

    "XGBoost_Optimized": XGBClassifier(
        n_estimators=500,
        max_depth=3,
        learning_rate=0.015,
        min_child_weight=5,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_alpha=0.0,
        reg_lambda=1.0,
        gamma=0,
        scale_pos_weight=under_spw,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    ),

    "Logistic_Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Random_Forest": RandomForestClassifier(
        n_estimators=500,
        max_depth=8,
        min_samples_leaf=5,
        max_features="sqrt",
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}

print("✓ XGBoost")
print("✓ Logistic Regression")
print("✓ Random Forest")


# =============================================================================
# 5. THRESHOLD GRID
# =============================================================================
#
# Higher threshold generally means fewer predicted positives.
# This can increase precision and reduce false positives.
# =============================================================================

thresholds = np.round(
    np.arange(0.30, 0.91, 0.01),
    2
)

print("\n" + "=" * 100)
print("5. THRESHOLD SEARCH")
print("=" * 100)

print(
    f"Threshold range : "
    f"{thresholds.min():.2f} – {thresholds.max():.2f}"
)

print("✓ Thresholds will be selected using VALIDATION data only")


# =============================================================================
# 6. EVALUATION FUNCTION
# =============================================================================

def evaluate_thresholds(
    y_true,
    probabilities,
    thresholds,
    model_name,
    target_name
):

    rows = []

    roc_auc = roc_auc_score(
        y_true,
        probabilities
    )

    pr_auc = average_precision_score(
        y_true,
        probabilities
    )

    for threshold in thresholds:

        y_pred = (
            probabilities >= threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_true,
            y_pred,
            labels=[0, 1]
        ).ravel()

        precision = precision_score(
            y_true,
            y_pred,
            zero_division=0
        )

        recall = recall_score(
            y_true,
            y_pred,
            zero_division=0
        )

        f1 = f1_score(
            y_true,
            y_pred,
            zero_division=0
        )

        accuracy = accuracy_score(
            y_true,
            y_pred
        )

        rows.append({

            "Target": target_name,
            "Model": model_name,

            "Threshold": threshold,

            "Accuracy": accuracy,
            "Precision": precision,
            "Recall": recall,
            "F1": f1,

            "ROC_AUC": roc_auc,
            "PR_AUC": pr_auc,

            "TN": tn,
            "FP": fp,
            "FN": fn,
            "TP": tp,

            "FPR": (
                fp / (fp + tn)
                if (fp + tn) > 0
                else 0
            )
        })

    return pd.DataFrame(rows)


# =============================================================================
# 7. STUNTING MODEL EVALUATION
# =============================================================================

print("\n" + "=" * 100)
print("6. PRECISION-FOCUSED MODELING — STUNTING")
print("=" * 100)

stunting_precision_results = []

for model_name, model in models_stunting.items():

    print(f"\nTraining: {model_name}")

    model.fit(
        X_train_stunt_processed,
        y_train_stunt
    )

    val_prob = model.predict_proba(
        X_val_stunt_processed
    )[:, 1]

    result = evaluate_thresholds(
        y_val_stunt,
        val_prob,
        thresholds,
        model_name,
        "Stunting"
    )

    stunting_precision_results.append(result)

    print(
        f"ROC-AUC = "
        f"{roc_auc_score(y_val_stunt, val_prob):.4f}"
    )

    print(
        f"PR-AUC  = "
        f"{average_precision_score(y_val_stunt, val_prob):.4f}"
    )


stunting_precision_df = pd.concat(
    stunting_precision_results,
    ignore_index=True
)


# =============================================================================
# 8. UNDERWEIGHT MODEL EVALUATION
# =============================================================================

print("\n" + "=" * 100)
print("7. PRECISION-FOCUSED MODELING — UNDERWEIGHT")
print("=" * 100)

underweight_precision_results = []

for model_name, model in models_underweight.items():

    print(f"\nTraining: {model_name}")

    model.fit(
        X_train_under_processed,
        y_train_under
    )

    val_prob = model.predict_proba(
        X_val_under_processed
    )[:, 1]

    result = evaluate_thresholds(
        y_val_under,
        val_prob,
        thresholds,
        model_name,
        "Underweight"
    )

    underweight_precision_results.append(result)

    print(
        f"ROC-AUC = "
        f"{roc_auc_score(y_val_under, val_prob):.4f}"
    )

    print(
        f"PR-AUC  = "
        f"{average_precision_score(y_val_under, val_prob):.4f}"
    )


underweight_precision_df = pd.concat(
    underweight_precision_results,
    ignore_index=True
)


# =============================================================================
# 9. PRECISION TARGET TABLE
# =============================================================================

print("\n" + "=" * 100)
print("8. PRECISION TARGET ANALYSIS")
print("=" * 100)

precision_targets = [
    0.50,
    0.55,
    0.60,
    0.65,
    0.70
]


def best_at_precision_target(
    results_df,
    target_precision
):

    eligible = results_df[
        results_df["Precision"] >= target_precision
    ].copy()

    if eligible.empty:
        return None

    # Among thresholds meeting precision target,
    # retain the greatest recall.
    eligible = eligible.sort_values(
        [
            "Recall",
            "F1",
            "Precision"
        ],
        ascending=False
    )

    return eligible.iloc[0]


# -----------------------------------------------------------------------------
# STUNTING
# -----------------------------------------------------------------------------

stunting_target_rows = []

for model_name in models_stunting.keys():

    model_df = stunting_precision_df[
        stunting_precision_df["Model"] == model_name
    ]

    for target_precision in precision_targets:

        best = best_at_precision_target(
            model_df,
            target_precision
        )

        if best is None:

            stunting_target_rows.append({

                "Model": model_name,
                "Required_Precision":
                    target_precision,

                "Achievable": "NO",

                "Threshold": np.nan,
                "Precision": np.nan,
                "Recall": np.nan,
                "F1": np.nan,
                "FPR": np.nan
            })

        else:

            stunting_target_rows.append({

                "Model": model_name,
                "Required_Precision":
                    target_precision,

                "Achievable": "YES",

                "Threshold":
                    best["Threshold"],

                "Precision":
                    best["Precision"],

                "Recall":
                    best["Recall"],

                "F1":
                    best["F1"],

                "FPR":
                    best["FPR"]
            })


stunting_precision_targets_df = pd.DataFrame(
    stunting_target_rows
)


# -----------------------------------------------------------------------------
# UNDERWEIGHT
# -----------------------------------------------------------------------------

underweight_target_rows = []

for model_name in models_underweight.keys():

    model_df = underweight_precision_df[
        underweight_precision_df["Model"] == model_name
    ]

    for target_precision in precision_targets:

        best = best_at_precision_target(
            model_df,
            target_precision
        )

        if best is None:

            underweight_target_rows.append({

                "Model": model_name,
                "Required_Precision":
                    target_precision,

                "Achievable": "NO",

                "Threshold": np.nan,
                "Precision": np.nan,
                "Recall": np.nan,
                "F1": np.nan,
                "FPR": np.nan
            })

        else:

            underweight_target_rows.append({

                "Model": model_name,
                "Required_Precision":
                    target_precision,

                "Achievable": "YES",

                "Threshold":
                    best["Threshold"],

                "Precision":
                    best["Precision"],

                "Recall":
                    best["Recall"],

                "F1":
                    best["F1"],

                "FPR":
                    best["FPR"]
            })


underweight_precision_targets_df = pd.DataFrame(
    underweight_target_rows
)


# =============================================================================
# 10. DISPLAY PRECISION TARGET RESULTS
# =============================================================================

print("\nSTUNTING — PRECISION TARGETS")
print("-" * 100)

print(
    stunting_precision_targets_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\nUNDERWEIGHT — PRECISION TARGETS")
print("-" * 100)

print(
    underweight_precision_targets_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 11. BEST PRECISION-RECALL OPERATING POINTS
# =============================================================================

print("\n" + "=" * 100)
print("9. BEST PRECISION-RECALL OPERATING POINTS")
print("=" * 100)


def select_precision_priority_model(
    results_df
):

    # First try precision >= 60%.
    eligible = results_df[
        results_df["Precision"] >= 0.60
    ].copy()

    if not eligible.empty:

        # Among models reaching 60% precision,
        # maximize recall, then F1.
        best = eligible.sort_values(
            [
                "Recall",
                "F1",
                "Precision"
            ],
            ascending=False
        ).iloc[0]

        return best, "Precision >= 60%"


    # If 60% is not achievable,
    # find the maximum precision available
    # while avoiding extremely tiny positive predictions.
    eligible = results_df[
        results_df["Recall"] >= 0.10
    ].copy()

    if eligible.empty:

        eligible = results_df.copy()

    best = eligible.sort_values(
        [
            "Precision",
            "Recall",
            "F1"
        ],
        ascending=False
    ).iloc[0]

    return best, "Maximum achievable precision"


best_stunting_precision, stunt_rule = (
    select_precision_priority_model(
        stunting_precision_df
    )
)

best_underweight_precision, under_rule = (
    select_precision_priority_model(
        underweight_precision_df
    )
)


print("\nSTUNTING")
print("-" * 80)
print(
    f"Selection rule : {stunt_rule}"
)
print(
    f"Model          : "
    f"{best_stunting_precision['Model']}"
)
print(
    f"Threshold      : "
    f"{best_stunting_precision['Threshold']:.2f}"
)
print(
    f"Precision      : "
    f"{best_stunting_precision['Precision']:.4f}"
)
print(
    f"Recall         : "
    f"{best_stunting_precision['Recall']:.4f}"
)
print(
    f"F1             : "
    f"{best_stunting_precision['F1']:.4f}"
)
print(
    f"FPR            : "
    f"{best_stunting_precision['FPR']:.4f}"
)
print(
    f"ROC-AUC        : "
    f"{best_stunting_precision['ROC_AUC']:.4f}"
)
print(
    f"PR-AUC         : "
    f"{best_stunting_precision['PR_AUC']:.4f}"
)


print("\nUNDERWEIGHT")
print("-" * 80)
print(
    f"Selection rule : {under_rule}"
)
print(
    f"Model          : "
    f"{best_underweight_precision['Model']}"
)
print(
    f"Threshold      : "
    f"{best_underweight_precision['Threshold']:.2f}"
)
print(
    f"Precision      : "
    f"{best_underweight_precision['Precision']:.4f}"
)
print(
    f"Recall         : "
    f"{best_underweight_precision['Recall']:.4f}"
)
print(
    f"F1             : "
    f"{best_underweight_precision['F1']:.4f}"
)
print(
    f"FPR            : "
    f"{best_underweight_precision['FPR']:.4f}"
)
print(
    f"ROC-AUC        : "
    f"{best_underweight_precision['ROC_AUC']:.4f}"
)
print(
    f"PR-AUC         : "
    f"{best_underweight_precision['PR_AUC']:.4f}"
)


# =============================================================================
# 12. CONFUSION MATRICES AT SELECTED OPERATING POINTS
# =============================================================================

print("\n" + "=" * 100)
print("10. CONFUSION MATRICES AT PRECISION-FOCUSED POINTS")
print("=" * 100)


def print_selected_confusion(
    y_true,
    results_df,
    selected_row,
    target_name
):

    model_name = selected_row["Model"]
    threshold = selected_row["Threshold"]

    if target_name == "Stunting":

        model = models_stunting[model_name]

        X_val = X_val_stunt_processed

    else:

        model = models_underweight[model_name]

        X_val = X_val_under_processed

    probabilities = model.predict_proba(
        X_val
    )[:, 1]

    predictions = (
        probabilities >= threshold
    ).astype(int)

    cm = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    )

    tn, fp, fn, tp = cm.ravel()

    print(f"\n{target_name}")
    print("-" * 80)

    print(
        f"Model     : {model_name}"
    )

    print(
        f"Threshold : {threshold:.2f}"
    )

    print(
        f"TN        : {tn}"
    )

    print(
        f"FP        : {fp}"
    )

    print(
        f"FN        : {fn}"
    )

    print(
        f"TP        : {tp}"
    )

    print(
        f"Precision : "
        f"{precision_score(y_true, predictions, zero_division=0):.4f}"
    )

    print(
        f"Recall    : "
        f"{recall_score(y_true, predictions, zero_division=0):.4f}"
    )

    print(
        f"F1        : "
        f"{f1_score(y_true, predictions, zero_division=0):.4f}"
    )


print_selected_confusion(
    y_val_stunt,
    stunting_precision_df,
    best_stunting_precision,
    "Stunting"
)

print_selected_confusion(
    y_val_under,
    underweight_precision_df,
    best_underweight_precision,
    "Underweight"
)


# =============================================================================
# 13. FALSE-POSITIVE REDUCTION CHECK
# =============================================================================

print("\n" + "=" * 100)
print("11. FALSE-POSITIVE REDUCTION")
print("=" * 100)


def false_positive_summary(
    y_true,
    selected_row,
    target_name
):

    threshold = selected_row["Threshold"]

    if target_name == "Stunting":

        model = models_stunting[
            selected_row["Model"]
        ]

        X_val = X_val_stunt_processed

    else:

        model = models_underweight[
            selected_row["Model"]
        ]

        X_val = X_val_under_processed

    probabilities = model.predict_proba(
        X_val
    )[:, 1]

    predictions = (
        probabilities >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    ).ravel()

    total_actual_negative = tn + fp

    false_positive_rate = (
        fp / total_actual_negative
        if total_actual_negative > 0
        else 0
    )

    predicted_positive = tp + fp

    print(f"\n{target_name}")

    print(
        f"Predicted positive : "
        f"{predicted_positive}"
    )

    print(
        f"True positive      : "
        f"{tp}"
    )

    print(
        f"False positive     : "
        f"{fp}"
    )

    print(
        f"False-positive rate: "
        f"{false_positive_rate:.4f}"
    )


false_positive_summary(
    y_val_stunt,
    best_stunting_precision,
    "Stunting"
)

false_positive_summary(
    y_val_under,
    best_underweight_precision,
    "Underweight"
)


# =============================================================================
# 14. SAVE RESULTS IN MEMORY
# =============================================================================

precision_optimization_summary = pd.DataFrame({

    "Target": [
        "Stunting",
        "Underweight"
    ],

    "Model": [
        best_stunting_precision["Model"],
        best_underweight_precision["Model"]
    ],

    "Threshold": [
        best_stunting_precision["Threshold"],
        best_underweight_precision["Threshold"]
    ],

    "Precision": [
        best_stunting_precision["Precision"],
        best_underweight_precision["Precision"]
    ],

    "Recall": [
        best_stunting_precision["Recall"],
        best_underweight_precision["Recall"]
    ],

    "F1": [
        best_stunting_precision["F1"],
        best_underweight_precision["F1"]
    ],

    "ROC_AUC": [
        best_stunting_precision["ROC_AUC"],
        best_underweight_precision["ROC_AUC"]
    ],

    "PR_AUC": [
        best_stunting_precision["PR_AUC"],
        best_underweight_precision["PR_AUC"]
    ],

    "FPR": [
        best_stunting_precision["FPR"],
        best_underweight_precision["FPR"]
    ]
})


# =============================================================================
# 15. FINAL INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("12. STEP 7C INTEGRITY CHECK")
print("=" * 100)

print("✓ Models fitted using training data")
print("✓ Thresholds evaluated using validation data")
print("✓ Precision explicitly prioritized")
print("✓ False-positive rate calculated")
print("✓ ROC-AUC calculated")
print("✓ PR-AUC calculated")
print("✓ Test predictors NOT used")
print("✓ Test targets NOT used")
print("✓ Test performance NOT inspected")
print("✓ No test-based threshold selection")
print("✓ No test-based model selection")


print("\n" + "=" * 100)
print("STEP 7C COMPLETE")
print("=" * 100)

print("""
IMPORTANT:
Do NOT proceed to Step 8 yet.

Review the precision-target tables first.
We need to see what precision is actually achievable
before fixing the final thresholds.
""")
```

```python
# =============================================================================
# STEP 7D: CAR MICS6 — CLASS-WEIGHT OPTIMIZATION FOR PRECISION
# =============================================================================
#
# OBJECTIVE:
#   Reduce false positives by testing different scale_pos_weight values.
#
# IMPORTANT:
#   Training data -> model fitting
#   Validation data -> model comparison and threshold selection
#   Test data -> COMPLETELY LOCKED
#
# We are NOT selecting the model simply by maximum precision.
# A precision of 100% with almost zero recall is not useful.
#
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

print("=" * 100)
print("STEP 7D: CAR MICS6 — CLASS-WEIGHT OPTIMIZATION FOR PRECISION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

print("\n1. CHECKING REQUIRED OBJECTS")
print("-" * 100)

required_objects = [
    "X_train_stunt_processed",
    "X_val_stunt_processed",
    "y_train_stunt",
    "y_val_stunt",

    "X_train_under_processed",
    "X_val_under_processed",
    "y_train_under",
    "y_val_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:

    raise NameError(
        "Missing required objects:\n\n"
        + "\n".join(missing_objects)
        + "\n\n"
        "Run Steps 1–7C before Step 7D."
    )

print("✓ All required training/validation objects are available")


# =============================================================================
# 2. TEST-SET PROTECTION
# =============================================================================

print("\n" + "=" * 100)
print("2. TEST-SET PROTECTION")
print("=" * 100)

print("✓ Test predictors are NOT used")
print("✓ Test targets are NOT used")
print("✓ Test thresholds are NOT evaluated")
print("✓ Test performance is NOT inspected")
print("✓ Test set remains completely unseen")


# =============================================================================
# 3. SEARCH SPACE
# =============================================================================

print("\n" + "=" * 100)
print("3. SCALE_POS_WEIGHT SEARCH")
print("=" * 100)

scale_weights = [
    1.00,
    1.25,
    1.50,
    1.75,
    2.00,
    2.50,
    3.00
]

print(
    "Values tested:",
    ", ".join(f"{x:.2f}" for x in scale_weights)
)

print("""
Interpretation:
1.00 = no positive-class weighting
Higher values = progressively stronger emphasis on positive cases
""")


# =============================================================================
# 4. THRESHOLD GRID
# =============================================================================

thresholds = np.round(
    np.arange(0.40, 0.81, 0.01),
    2
)

print("\n" + "=" * 100)
print("4. THRESHOLD SEARCH")
print("=" * 100)

print(
    f"Threshold range: "
    f"{thresholds.min():.2f} – {thresholds.max():.2f}"
)

print("✓ Thresholds evaluated on VALIDATION data only")


# =============================================================================
# 5. XGBOOST BASE CONFIGURATION
# =============================================================================

base_params = {

    "n_estimators": 500,
    "max_depth": 3,
    "learning_rate": 0.02,

    "min_child_weight": 8,

    "subsample": 0.80,
    "colsample_bytree": 0.80,

    "reg_alpha": 0.10,
    "reg_lambda": 2.0,

    "gamma": 0,

    "objective": "binary:logistic",
    "eval_metric": "logloss",

    "random_state": 42,
    "n_jobs": -1
}

print("\nBase XGBoost configuration:")
for key, value in base_params.items():
    print(f"{key:20s}: {value}")


# =============================================================================
# 6. EVALUATION FUNCTION
# =============================================================================

def evaluate_model_thresholds(
    y_true,
    probabilities,
    threshold_values,
    target,
    scale_weight
):

    rows = []

    roc_auc = roc_auc_score(
        y_true,
        probabilities
    )

    pr_auc = average_precision_score(
        y_true,
        probabilities
    )

    for threshold in threshold_values:

        predictions = (
            probabilities >= threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_true,
            predictions,
            labels=[0, 1]
        ).ravel()

        precision = precision_score(
            y_true,
            predictions,
            zero_division=0
        )

        recall = recall_score(
            y_true,
            predictions,
            zero_division=0
        )

        f1 = f1_score(
            y_true,
            predictions,
            zero_division=0
        )

        accuracy = accuracy_score(
            y_true,
            predictions
        )

        fpr = (
            fp / (fp + tn)
            if (fp + tn) > 0
            else 0
        )

        predicted_positive = tp + fp

        rows.append({

            "Target": target,

            "Scale_Pos_Weight":
                scale_weight,

            "Threshold":
                threshold,

            "Accuracy":
                accuracy,

            "Precision":
                precision,

            "Recall":
                recall,

            "F1":
                f1,

            "ROC_AUC":
                roc_auc,

            "PR_AUC":
                pr_auc,

            "TN":
                tn,

            "FP":
                fp,

            "FN":
                fn,

            "TP":
                tp,

            "FPR":
                fpr,

            "Predicted_Positive":
                predicted_positive
        })

    return pd.DataFrame(rows)


# =============================================================================
# 7. STUNTING SEARCH
# =============================================================================

print("\n" + "=" * 100)
print("5. STUNTING — SCALE_POS_WEIGHT OPTIMIZATION")
print("=" * 100)

stunting_weight_results = []

for weight in scale_weights:

    print(
        f"\nTesting scale_pos_weight = {weight:.2f}"
    )

    params = base_params.copy()

    params["scale_pos_weight"] = weight

    model = XGBClassifier(**params)

    model.fit(
        X_train_stunt_processed,
        y_train_stunt
    )

    val_prob = model.predict_proba(
        X_val_stunt_processed
    )[:, 1]

    results = evaluate_model_thresholds(
        y_val_stunt,
        val_prob,
        thresholds,
        "Stunting",
        weight
    )

    stunting_weight_results.append(results)

    print(
        f"ROC-AUC: "
        f"{roc_auc_score(y_val_stunt, val_prob):.4f}"
    )

    print(
        f"PR-AUC : "
        f"{average_precision_score(y_val_stunt, val_prob):.4f}"
    )


stunting_weight_df = pd.concat(
    stunting_weight_results,
    ignore_index=True
)


# =============================================================================
# 8. UNDERWEIGHT SEARCH
# =============================================================================

print("\n" + "=" * 100)
print("6. UNDERWEIGHT — SCALE_POS_WEIGHT OPTIMIZATION")
print("=" * 100)

underweight_weight_results = []

for weight in scale_weights:

    print(
        f"\nTesting scale_pos_weight = {weight:.2f}"
    )

    params = base_params.copy()

    params["scale_pos_weight"] = weight

    model = XGBClassifier(**params)

    model.fit(
        X_train_under_processed,
        y_train_under
    )

    val_prob = model.predict_proba(
        X_val_under_processed
    )[:, 1]

    results = evaluate_model_thresholds(
        y_val_under,
        val_prob,
        thresholds,
        "Underweight",
        weight
    )

    underweight_weight_results.append(results)

    print(
        f"ROC-AUC: "
        f"{roc_auc_score(y_val_under, val_prob):.4f}"
    )

    print(
        f"PR-AUC : "
        f"{average_precision_score(y_val_under, val_prob):.4f}"
    )


underweight_weight_df = pd.concat(
    underweight_weight_results,
    ignore_index=True
)


# =============================================================================
# 9. DEFINE USEFUL PRECISION/RECALL REGION
# =============================================================================

print("\n" + "=" * 100)
print("7. USEFUL PRECISION–RECALL REGION")
print("=" * 100)

print("""
Primary screening region:

Precision >= 50%
Recall    >= 40%

This is NOT an automatic final criterion.

It is used to identify whether a useful operating
region exists without sacrificing too much recall.
""")


# =============================================================================
# 10. STUNTING — MODELS MEETING BOTH CONDITIONS
# =============================================================================

stunting_useful = stunting_weight_df[
    (stunting_weight_df["Precision"] >= 0.50) &
    (stunting_weight_df["Recall"] >= 0.40)
].copy()

print("\nSTUNTING")
print("-" * 100)

if stunting_useful.empty:

    print(
        "No configuration simultaneously achieved "
        "Precision >= 50% and Recall >= 40%."
    )

else:

    stunting_useful = stunting_useful.sort_values(
        [
            "Precision",
            "Recall",
            "F1",
            "PR_AUC"
        ],
        ascending=False
    )

    print(
        stunting_useful[
            [
                "Scale_Pos_Weight",
                "Threshold",
                "Precision",
                "Recall",
                "F1",
                "FPR",
                "FP",
                "FN",
                "TP",
                "PR_AUC"
            ]
        ].head(20).to_string(
            index=False,
            float_format=lambda x: f"{x:.4f}"
        )
    )


# =============================================================================
# 11. UNDERWEIGHT — MODELS MEETING BOTH CONDITIONS
# =============================================================================

underweight_useful = underweight_weight_df[
    (underweight_weight_df["Precision"] >= 0.50) &
    (underweight_weight_df["Recall"] >= 0.40)
].copy()

print("\nUNDERWEIGHT")
print("-" * 100)

if underweight_useful.empty:

    print(
        "No configuration simultaneously achieved "
        "Precision >= 50% and Recall >= 40%."
    )

else:

    underweight_useful = underweight_useful.sort_values(
        [
            "Precision",
            "Recall",
            "F1",
            "PR_AUC"
        ],
        ascending=False
    )

    print(
        underweight_useful[
            [
                "Scale_Pos_Weight",
                "Threshold",
                "Precision",
                "Recall",
                "F1",
                "FPR",
                "FP",
                "FN",
                "TP",
                "PR_AUC"
            ]
        ].head(20).to_string(
            index=False,
            float_format=lambda x: f"{x:.4f}"
        )
    )


# =============================================================================
# 12. PRECISION-BY-RECALL ANALYSIS
# =============================================================================
#
# This is particularly important.
#
# Rather than saying "give me the highest precision", we ask:
#
# What is the highest precision achievable while retaining
# at least 40%, 50%, 60% and 70% recall?
# =============================================================================

print("\n" + "=" * 100)
print("8. PRECISION AT DIFFERENT MINIMUM RECALL LEVELS")
print("=" * 100)

recall_targets = [
    0.40,
    0.50,
    0.60,
    0.70
]


def best_precision_at_recall(
    dataframe,
    minimum_recall
):

    eligible = dataframe[
        dataframe["Recall"] >= minimum_recall
    ].copy()

    if eligible.empty:

        return None

    return eligible.sort_values(
        [
            "Precision",
            "F1",
            "Recall"
        ],
        ascending=False
    ).iloc[0]


def create_recall_constraint_table(
    dataframe
):

    rows = []

    for weight in scale_weights:

        weight_df = dataframe[
            dataframe["Scale_Pos_Weight"] == weight
        ]

        for recall_target in recall_targets:

            best = best_precision_at_recall(
                weight_df,
                recall_target
            )

            if best is None:

                rows.append({

                    "Scale_Pos_Weight":
                        weight,

                    "Minimum_Recall":
                        recall_target,

                    "Achievable":
                        "NO",

                    "Threshold":
                        np.nan,

                    "Precision":
                        np.nan,

                    "Recall":
                        np.nan,

                    "F1":
                        np.nan,

                    "FPR":
                        np.nan
                })

            else:

                rows.append({

                    "Scale_Pos_Weight":
                        weight,

                    "Minimum_Recall":
                        recall_target,

                    "Achievable":
                        "YES",

                    "Threshold":
                        best["Threshold"],

                    "Precision":
                        best["Precision"],

                    "Recall":
                        best["Recall"],

                    "F1":
                        best["F1"],

                    "FPR":
                        best["FPR"]
                })

    return pd.DataFrame(rows)


stunting_recall_constraints = (
    create_recall_constraint_table(
        stunting_weight_df
    )
)

underweight_recall_constraints = (
    create_recall_constraint_table(
        underweight_weight_df
    )
)


print("\nSTUNTING")
print("-" * 100)

print(
    stunting_recall_constraints.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


print("\nUNDERWEIGHT")
print("-" * 100)

print(
    underweight_recall_constraints.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 13. BEST BALANCED CONFIGURATION
# =============================================================================
#
# We require:
#
#   Precision >= 50%
#   Recall    >= 40%
#
# Among those, choose highest F1.
#
# If none exists, report that honestly.
# =============================================================================

print("\n" + "=" * 100)
print("9. BEST BALANCED CONFIGURATION")
print("=" * 100)


def best_balanced_configuration(
    dataframe
):

    eligible = dataframe[
        (dataframe["Precision"] >= 0.50) &
        (dataframe["Recall"] >= 0.40)
    ].copy()

    if eligible.empty:

        return None

    return eligible.sort_values(
        [
            "F1",
            "Precision",
            "Recall",
            "PR_AUC"
        ],
        ascending=False
    ).iloc[0]


best_stunting_7d = (
    best_balanced_configuration(
        stunting_weight_df
    )
)

best_underweight_7d = (
    best_balanced_configuration(
        underweight_weight_df
    )


)

print("\nSTUNTING")

if best_stunting_7d is None:

    print(
        "No configuration met both "
        "Precision >= 50% and Recall >= 40%."
    )

else:

    for column in [
        "Scale_Pos_Weight",
        "Threshold",
        "Precision",
        "Recall",
        "F1",
        "FPR",
        "FP",
        "FN",
        "TP",
        "ROC_AUC",
        "PR_AUC"
    ]:

        print(
            f"{column:20s}: "
            f"{best_stunting_7d[column]:.4f}"
            if isinstance(
                best_stunting_7d[column],
                (float, np.floating)
            )
            else
            f"{column:20s}: "
            f"{best_stunting_7d[column]}"
        )


print("\nUNDERWEIGHT")

if best_underweight_7d is None:

    print(
        "No configuration met both "
        "Precision >= 50% and Recall >= 40%."
    )

else:

    for column in [
        "Scale_Pos_Weight",
        "Threshold",
        "Precision",
        "Recall",
        "F1",
        "FPR",
        "FP",
        "FN",
        "TP",
        "ROC_AUC",
        "PR_AUC"
    ]:

        print(
            f"{column:20s}: "
            f"{best_underweight_7d[column]:.4f}"
            if isinstance(
                best_underweight_7d[column],
                (float, np.floating)
            )
            else
            f"{column:20s}: "
            f"{best_underweight_7d[column]}"
        )


# =============================================================================
# 14. COMPARE WEIGHTING STRATEGIES
# =============================================================================

print("\n" + "=" * 100)
print("10. OVERALL WEIGHTING COMPARISON")
print("=" * 100)


def best_f1_by_weight(dataframe):

    rows = []

    for weight in scale_weights:

        subset = dataframe[
            dataframe["Scale_Pos_Weight"] == weight
        ]

        best = subset.sort_values(
            [
                "F1",
                "Precision",
                "Recall"
            ],
            ascending=False
        ).iloc[0]

        rows.append({

            "Scale_Pos_Weight":
                weight,

            "Threshold":
                best["Threshold"],

            "Precision":
                best["Precision"],

            "Recall":
                best["Recall"],

            "F1":
                best["F1"],

            "FPR":
                best["FPR"],

            "PR_AUC":
                best["PR_AUC"]
        })

    return pd.DataFrame(rows)


stunting_weight_summary = (
    best_f1_by_weight(
        stunting_weight_df
    )
)

underweight_weight_summary = (
    best_f1_by_weight(
        underweight_weight_df
    )
)


print("\nSTUNTING")
print(
    stunting_weight_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\nUNDERWEIGHT")
print(
    underweight_weight_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 15. SAVE STEP 7D RESULTS IN MEMORY
# =============================================================================

step7d_results = {

    "stunting_all":
        stunting_weight_df,

    "underweight_all":
        underweight_weight_df,

    "stunting_recall_constraints":
        stunting_recall_constraints,

    "underweight_recall_constraints":
        underweight_recall_constraints,

    "stunting_weight_summary":
        stunting_weight_summary,

    "underweight_weight_summary":
        underweight_weight_summary
}


# =============================================================================
# 16. INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("11. STEP 7D INTEGRITY CHECK")
print("=" * 100)

print("✓ Models fitted using training data only")
print("✓ Scale_pos_weight varied")
print("✓ Thresholds evaluated using validation data only")
print("✓ Precision explicitly considered")
print("✓ Recall explicitly constrained")
print("✓ False-positive rate calculated")
print("✓ False-positive counts calculated")
print("✓ Test predictors NOT used")
print("✓ Test targets NOT used")
print("✓ Test performance NOT inspected")
print("✓ No test-based model selection")
print("✓ No test-based threshold selection")


print("\n" + "=" * 100)
print("STEP 7D COMPLETE")
print("=" * 100)

print("""
DO NOT RUN STEP 8 YET.

Review the Step 7D results first.

The key question is:

Can we obtain substantially better precision
WITHOUT reducing recall to an unusable level?

If yes, we will take that configuration forward.

If no, the next improvement should focus on
the predictor set/model discrimination rather
than simply increasing the threshold.
""")
```

```python
# =============================================================================
# STEP 7E: CAR MICS6 — XGBOOST HYPERPARAMETER OPTIMIZATION
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

print("=" * 100)
print("STEP 7E: CAR MICS6 — XGBOOST HYPERPARAMETER OPTIMIZATION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

required_objects = [
    "X_train_stunt_processed",
    "X_val_stunt_processed",
    "y_train_stunt",
    "y_val_stunt",
    "X_train_under_processed",
    "X_val_under_processed",
    "y_train_under",
    "y_val_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "The following required Step 5 objects are missing:\n"
        + "\n".join(missing_objects)
    )

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)
print("✓ All required training and validation objects are available")


# =============================================================================
# 2. TEST-SET PROTECTION
# =============================================================================

print("\n2. TEST-SET PROTECTION")
print("-" * 100)

print("✓ Test predictors will NOT be used")
print("✓ Test targets will NOT be used")
print("✓ Test performance will NOT be evaluated")
print("✓ Test thresholds will NOT be selected")
print("✓ Test set remains completely unseen")


# =============================================================================
# 3. DATA SHAPE CHECK
# =============================================================================

print("\n3. DEVELOPMENT DATA SHAPES")
print("-" * 100)

print("\nSTUNTING")
print(f"Training predictors : {X_train_stunt_processed.shape}")
print(f"Validation predictors: {X_val_stunt_processed.shape}")
print(f"Training target     : {y_train_stunt.shape}")
print(f"Validation target   : {y_val_stunt.shape}")

print("\nUNDERWEIGHT")
print(f"Training predictors : {X_train_under_processed.shape}")
print(f"Validation predictors: {X_val_under_processed.shape}")
print(f"Training target     : {y_train_under.shape}")
print(f"Validation target   : {y_val_under.shape}")


# =============================================================================
# 4. CLASS WEIGHTS
# =============================================================================

stunt_neg = (y_train_stunt == 0).sum()
stunt_pos = (y_train_stunt == 1).sum()

under_neg = (y_train_under == 0).sum()
under_pos = (y_train_under == 1).sum()

stunt_weight = stunt_neg / stunt_pos
under_weight = under_neg / under_pos

print("\n4. CLASS WEIGHTS")
print("-" * 100)

print(f"Stunting scale_pos_weight   : {stunt_weight:.4f}")
print(f"Underweight scale_pos_weight: {under_weight:.4f}")


# =============================================================================
# 5. HYPERPARAMETER GRID
# =============================================================================
#
# We deliberately vary:
#   - learning_rate
#   - max_depth
#   - min_child_weight
#   - subsample
#   - colsample_bytree
#   - reg_alpha
#   - reg_lambda
#
# The grid is deliberately moderate so that the notebook does not become
# unnecessarily slow.
# =============================================================================

param_grid = [
    {
        "learning_rate": 0.01,
        "max_depth": 2,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.1,
        "reg_lambda": 2.0
    },
    {
        "learning_rate": 0.02,
        "max_depth": 2,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.1,
        "reg_lambda": 2.0
    },
    {
        "learning_rate": 0.03,
        "max_depth": 2,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.1,
        "reg_lambda": 2.0
    },
    {
        "learning_rate": 0.05,
        "max_depth": 2,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.1,
        "reg_lambda": 2.0
    },
    {
        "learning_rate": 0.02,
        "max_depth": 3,
        "min_child_weight": 3,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.1,
        "reg_lambda": 2.0
    },
    {
        "learning_rate": 0.03,
        "max_depth": 3,
        "min_child_weight": 3,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.1,
        "reg_lambda": 2.0
    },
    {
        "learning_rate": 0.05,
        "max_depth": 3,
        "min_child_weight": 3,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.1,
        "reg_lambda": 2.0
    },
    {
        "learning_rate": 0.02,
        "max_depth": 4,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.1,
        "reg_lambda": 2.0
    },
    {
        "learning_rate": 0.03,
        "max_depth": 4,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.1,
        "reg_lambda": 2.0
    },
    {
        "learning_rate": 0.05,
        "max_depth": 4,
        "min_child_weight": 5,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "reg_alpha": 0.1,
        "reg_lambda": 2.0
    },
    {
        "learning_rate": 0.03,
        "max_depth": 3,
        "min_child_weight": 8,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.5,
        "reg_lambda": 3.0
    },
    {
        "learning_rate": 0.05,
        "max_depth": 3,
        "min_child_weight": 8,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.5,
        "reg_lambda": 3.0
    }
]

print("\n5. HYPERPARAMETER SEARCH")
print("-" * 100)
print(f"Configurations to test: {len(param_grid)}")
print("✓ Learning rate will be varied")
print("✓ Tree depth will be varied")
print("✓ Minimum child weight will be varied")
print("✓ Row/column subsampling will be varied")
print("✓ Regularization will be varied")


# =============================================================================
# 6. EVALUATION FUNCTION
# =============================================================================

def evaluate_xgb_model(
    model,
    X_val,
    y_val,
    threshold=0.50
):
    probabilities = model.predict_proba(X_val)[:, 1]

    predictions = (
        probabilities >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        predictions,
        labels=[0, 1]
    ).ravel()

    return {
        "Accuracy": accuracy_score(y_val, predictions),
        "Precision": precision_score(
            y_val,
            predictions,
            zero_division=0
        ),
        "Recall": recall_score(
            y_val,
            predictions,
            zero_division=0
        ),
        "F1": f1_score(
            y_val,
            predictions,
            zero_division=0
        ),
        "ROC_AUC": roc_auc_score(
            y_val,
            probabilities
        ),
        "PR_AUC": average_precision_score(
            y_val,
            probabilities
        ),
        "FPR": fp / (fp + tn) if (fp + tn) > 0 else 0,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }


# =============================================================================
# 7. STUNTING OPTIMIZATION
# =============================================================================

print("\n" + "=" * 100)
print("6. STUNTING — HYPERPARAMETER OPTIMIZATION")
print("=" * 100)

stunting_results = []

for i, params in enumerate(param_grid, start=1):

    print(
        f"\nConfiguration {i}/{len(param_grid)}"
    )

    print(
        f"learning_rate={params['learning_rate']} | "
        f"max_depth={params['max_depth']} | "
        f"min_child_weight={params['min_child_weight']}"
    )

    model = XGBClassifier(
        n_estimators=500,
        learning_rate=params["learning_rate"],
        max_depth=params["max_depth"],
        min_child_weight=params["min_child_weight"],
        subsample=params["subsample"],
        colsample_bytree=params["colsample_bytree"],
        reg_alpha=params["reg_alpha"],
        reg_lambda=params["reg_lambda"],
        gamma=0,
        scale_pos_weight=stunt_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_stunt_processed,
        y_train_stunt
    )

    metrics = evaluate_xgb_model(
        model,
        X_val_stunt_processed,
        y_val_stunt,
        threshold=0.50
    )

    result = params.copy()
    result.update(metrics)

    stunting_results.append(result)

    print(
        f"ROC-AUC={metrics['ROC_AUC']:.4f} | "
        f"PR-AUC={metrics['PR_AUC']:.4f} | "
        f"Precision={metrics['Precision']:.4f} | "
        f"Recall={metrics['Recall']:.4f} | "
        f"F1={metrics['F1']:.4f}"
    )


stunting_results_df = pd.DataFrame(
    stunting_results
).sort_values(
    by=["PR_AUC", "F1"],
    ascending=False
).reset_index(drop=True)


# =============================================================================
# 8. UNDERWEIGHT OPTIMIZATION
# =============================================================================

print("\n" + "=" * 100)
print("7. UNDERWEIGHT — HYPERPARAMETER OPTIMIZATION")
print("=" * 100)

underweight_results = []

for i, params in enumerate(param_grid, start=1):

    print(
        f"\nConfiguration {i}/{len(param_grid)}"
    )

    print(
        f"learning_rate={params['learning_rate']} | "
        f"max_depth={params['max_depth']} | "
        f"min_child_weight={params['min_child_weight']}"
    )

    model = XGBClassifier(
        n_estimators=500,
        learning_rate=params["learning_rate"],
        max_depth=params["max_depth"],
        min_child_weight=params["min_child_weight"],
        subsample=params["subsample"],
        colsample_bytree=params["colsample_bytree"],
        reg_alpha=params["reg_alpha"],
        reg_lambda=params["reg_lambda"],
        gamma=0,
        scale_pos_weight=under_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_under_processed,
        y_train_under
    )

    metrics = evaluate_xgb_model(
        model,
        X_val_under_processed,
        y_val_under,
        threshold=0.50
    )

    result = params.copy()
    result.update(metrics)

    underweight_results.append(result)

    print(
        f"ROC-AUC={metrics['ROC_AUC']:.4f} | "
        f"PR-AUC={metrics['PR_AUC']:.4f} | "
        f"Precision={metrics['Precision']:.4f} | "
        f"Recall={metrics['Recall']:.4f} | "
        f"F1={metrics['F1']:.4f}"
    )


underweight_results_df = pd.DataFrame(
    underweight_results
).sort_values(
    by=["PR_AUC", "F1"],
    ascending=False
).reset_index(drop=True)


# =============================================================================
# 9. DISPLAY STUNTING RESULTS
# =============================================================================

print("\n" + "=" * 100)
print("8. STUNTING OPTIMIZATION RESULTS")
print("=" * 100)

display(
    stunting_results_df[
        [
            "learning_rate",
            "max_depth",
            "min_child_weight",
            "subsample",
            "colsample_bytree",
            "reg_alpha",
            "reg_lambda",
            "Precision",
            "Recall",
            "F1",
            "ROC_AUC",
            "PR_AUC",
            "FPR"
        ]
    ].round(4)
)


# =============================================================================
# 10. DISPLAY UNDERWEIGHT RESULTS
# =============================================================================

print("\n" + "=" * 100)
print("9. UNDERWEIGHT OPTIMIZATION RESULTS")
print("=" * 100)

display(
    underweight_results_df[
        [
            "learning_rate",
            "max_depth",
            "min_child_weight",
            "subsample",
            "colsample_bytree",
            "reg_alpha",
            "reg_lambda",
            "Precision",
            "Recall",
            "F1",
            "ROC_AUC",
            "PR_AUC",
            "FPR"
        ]
    ].round(4)
)


# =============================================================================
# 11. BEST CONFIGURATION BY PR-AUC
# =============================================================================

best_stunting_params = stunting_results_df.iloc[0].to_dict()
best_underweight_params = underweight_results_df.iloc[0].to_dict()

print("\n" + "=" * 100)
print("10. BEST CONFIGURATION BY VALIDATION PR-AUC")
print("=" * 100)

print("\nSTUNTING")
print("-" * 100)

for key in param_grid[0].keys():
    print(f"{key:20s}: {best_stunting_params[key]}")

print(
    f"Precision : {best_stunting_params['Precision']:.4f}"
)
print(
    f"Recall    : {best_stunting_params['Recall']:.4f}"
)
print(
    f"F1        : {best_stunting_params['F1']:.4f}"
)
print(
    f"ROC-AUC   : {best_stunting_params['ROC_AUC']:.4f}"
)
print(
    f"PR-AUC    : {best_stunting_params['PR_AUC']:.4f}"
)
print(
    f"FPR       : {best_stunting_params['FPR']:.4f}"
)

print("\nUNDERWEIGHT")
print("-" * 100)

for key in param_grid[0].keys():
    print(f"{key:20s}: {best_underweight_params[key]}")

print(
    f"Precision : {best_underweight_params['Precision']:.4f}"
)
print(
    f"Recall    : {best_underweight_params['Recall']:.4f}"
)
print(
    f"F1        : {best_underweight_params['F1']:.4f}"
)
print(
    f"ROC-AUC   : {best_underweight_params['ROC_AUC']:.4f}"
)
print(
    f"PR-AUC    : {best_underweight_params['PR_AUC']:.4f}"
)
print(
    f"FPR       : {best_underweight_params['FPR']:.4f}"
)


# =============================================================================
# 12. BEST CONFIGURATION BY PRECISION WITH RECALL >= 40%
# =============================================================================

print("\n" + "=" * 100)
print("11. PRECISION-FOCUSED CONFIGURATION")
print("=" * 100)

print("\nSTUNTING")
print("-" * 100)

stunt_precision_candidates = stunting_results_df[
    stunting_results_df["Recall"] >= 0.40
]

if len(stunt_precision_candidates) > 0:

    best_stunt_precision = (
        stunt_precision_candidates
        .sort_values(
            by=["Precision", "F1"],
            ascending=False
        )
        .iloc[0]
    )

    print(
        f"Precision : {best_stunt_precision['Precision']:.4f}"
    )
    print(
        f"Recall    : {best_stunt_precision['Recall']:.4f}"
    )
    print(
        f"F1        : {best_stunt_precision['F1']:.4f}"
    )
    print(
        f"PR-AUC    : {best_stunt_precision['PR_AUC']:.4f}"
    )

else:
    print(
        "No configuration achieved Recall >= 40%."
    )


print("\nUNDERWEIGHT")
print("-" * 100)

under_precision_candidates = underweight_results_df[
    underweight_results_df["Recall"] >= 0.40
]

if len(under_precision_candidates) > 0:

    best_under_precision = (
        under_precision_candidates
        .sort_values(
            by=["Precision", "F1"],
            ascending=False
        )
        .iloc[0]
    )

    print(
        f"Precision : {best_under_precision['Precision']:.4f}"
    )
    print(
        f"Recall    : {best_under_precision['Recall']:.4f}"
    )
    print(
        f"F1        : {best_under_precision['F1']:.4f}"
    )
    print(
        f"PR-AUC    : {best_under_precision['PR_AUC']:.4f}"
    )

else:
    print(
        "No configuration achieved Recall >= 40%."
    )


# =============================================================================
# 13. STEP 7E INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("STEP 7E COMPLETE")
print("=" * 100)

print("✓ Hyperparameter optimization performed")
print("✓ Learning rate tested")
print("✓ Tree depth tested")
print("✓ Minimum child weight tested")
print("✓ Subsampling tested")
print("✓ Column sampling tested")
print("✓ Regularization tested")
print("✓ Validation data used for model comparison")
print("✓ Test data remained completely untouched")
print("✓ No test-based model selection performed")

print("\nNEXT STEP:")
print("STEP 7F — VALIDATION THRESHOLD OPTIMIZATION")
print("The best optimized models will be evaluated across thresholds")
print("to find operating points that reduce false positives while")
print("retaining clinically meaningful recall.")
```

```python
# =============================================================================
# STEP 7F: CAR MICS6 — PRECISION-FOCUSED THRESHOLD OPTIMIZATION
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

print("=" * 100)
print("STEP 7F: CAR MICS6 — PRECISION-FOCUSED THRESHOLD OPTIMIZATION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

required_objects = [
    "X_train_stunt_processed",
    "X_val_stunt_processed",
    "y_train_stunt",
    "y_val_stunt",
    "X_train_under_processed",
    "X_val_under_processed",
    "y_train_under",
    "y_val_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required objects:\n"
        + "\n".join(missing_objects)
    )

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)
print("✓ All required training and validation objects are available")


# =============================================================================
# 2. TEST-SET PROTECTION
# =============================================================================

print("\n2. TEST-SET PROTECTION")
print("-" * 100)
print("✓ Test predictors will NOT be used")
print("✓ Test targets will NOT be used")
print("✓ Test thresholds will NOT be evaluated")
print("✓ Test performance will NOT be inspected")
print("✓ Test set remains completely unseen")


# =============================================================================
# 3. BEST CONFIGURATIONS FROM STEP 7E
# =============================================================================
#
# STUNTING:
# Highest PR-AUC configuration
#
# UNDERWEIGHT:
# Highest PR-AUC configuration
#
# These are selected from validation performance only.
# =============================================================================

stunting_params = {
    "learning_rate": 0.03,
    "max_depth": 3,
    "min_child_weight": 8,
    "subsample": 0.80,
    "colsample_bytree": 0.80,
    "reg_alpha": 0.5,
    "reg_lambda": 3.0
}

underweight_params = {
    "learning_rate": 0.01,
    "max_depth": 2,
    "min_child_weight": 5,
    "subsample": 0.85,
    "colsample_bytree": 0.85,
    "reg_alpha": 0.1,
    "reg_lambda": 2.0
}

stunt_weight = (
    (y_train_stunt == 0).sum() /
    (y_train_stunt == 1).sum()
)

under_weight = (
    (y_train_under == 0).sum() /
    (y_train_under == 1).sum()
)


# =============================================================================
# 4. FIT OPTIMIZED STUNTING MODEL
# =============================================================================

print("\n3. FITTING OPTIMIZED STUNTING MODEL")
print("-" * 100)

xgb_stunt_optimized = XGBClassifier(
    n_estimators=500,
    learning_rate=stunting_params["learning_rate"],
    max_depth=stunting_params["max_depth"],
    min_child_weight=stunting_params["min_child_weight"],
    subsample=stunting_params["subsample"],
    colsample_bytree=stunting_params["colsample_bytree"],
    reg_alpha=stunting_params["reg_alpha"],
    reg_lambda=stunting_params["reg_lambda"],
    gamma=0,
    scale_pos_weight=stunt_weight,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_stunt_optimized.fit(
    X_train_stunt_processed,
    y_train_stunt
)

stunt_val_prob = xgb_stunt_optimized.predict_proba(
    X_val_stunt_processed
)[:, 1]

print("✓ Optimized stunting model fitted")
print("✓ Training data used only for model fitting")


# =============================================================================
# 5. FIT OPTIMIZED UNDERWEIGHT MODEL
# =============================================================================

print("\n4. FITTING OPTIMIZED UNDERWEIGHT MODEL")
print("-" * 100)

xgb_under_optimized = XGBClassifier(
    n_estimators=500,
    learning_rate=underweight_params["learning_rate"],
    max_depth=underweight_params["max_depth"],
    min_child_weight=underweight_params["min_child_weight"],
    subsample=underweight_params["subsample"],
    colsample_bytree=underweight_params["colsample_bytree"],
    reg_alpha=underweight_params["reg_alpha"],
    reg_lambda=underweight_params["reg_lambda"],
    gamma=0,
    scale_pos_weight=under_weight,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_under_optimized.fit(
    X_train_under_processed,
    y_train_under
)

under_val_prob = xgb_under_optimized.predict_proba(
    X_val_under_processed
)[:, 1]

print("✓ Optimized underweight model fitted")
print("✓ Training data used only for model fitting")


# =============================================================================
# 6. PROBABILITY-BASED METRICS
# =============================================================================

print("\n5. VALIDATION PROBABILITY METRICS")
print("-" * 100)

print("\nSTUNTING")
print(
    f"ROC-AUC : "
    f"{roc_auc_score(y_val_stunt, stunt_val_prob):.4f}"
)
print(
    f"PR-AUC  : "
    f"{average_precision_score(y_val_stunt, stunt_val_prob):.4f}"
)

print("\nUNDERWEIGHT")
print(
    f"ROC-AUC : "
    f"{roc_auc_score(y_val_under, under_val_prob):.4f}"
)
print(
    f"PR-AUC  : "
    f"{average_precision_score(y_val_under, under_val_prob):.4f}"
)


# =============================================================================
# 7. THRESHOLD EVALUATION FUNCTION
# =============================================================================

def threshold_analysis(
    y_true,
    probabilities,
    thresholds
):

    results = []

    for threshold in thresholds:

        predictions = (
            probabilities >= threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_true,
            predictions,
            labels=[0, 1]
        ).ravel()

        precision = precision_score(
            y_true,
            predictions,
            zero_division=0
        )

        recall = recall_score(
            y_true,
            predictions,
            zero_division=0
        )

        f1 = f1_score(
            y_true,
            predictions,
            zero_division=0
        )

        fpr = (
            fp / (fp + tn)
            if (fp + tn) > 0
            else 0
        )

        results.append({
            "Threshold": threshold,
            "Accuracy": accuracy_score(
                y_true,
                predictions
            ),
            "Precision": precision,
            "Recall": recall,
            "F1": f1,
            "FPR": fpr,
            "TN": tn,
            "FP": fp,
            "FN": fn,
            "TP": tp
        })

    return pd.DataFrame(results)


# =============================================================================
# 8. THRESHOLD GRID
# =============================================================================

thresholds = np.round(
    np.arange(0.30, 0.81, 0.01),
    2
)

print("\n6. THRESHOLD SEARCH")
print("-" * 100)
print("Threshold range : 0.30 – 0.80")
print("Step size       : 0.01")
print(f"Thresholds tested: {len(thresholds)}")
print("✓ Validation data only")


# =============================================================================
# 9. STUNTING THRESHOLD ANALYSIS
# =============================================================================

stunting_threshold_results = threshold_analysis(
    y_val_stunt,
    stunt_val_prob,
    thresholds
)

print("\n" + "=" * 100)
print("7. STUNTING — THRESHOLD RESULTS")
print("=" * 100)

display(
    stunting_threshold_results[
        [
            "Threshold",
            "Precision",
            "Recall",
            "F1",
            "FPR",
            "TP",
            "FP",
            "FN",
            "TN"
        ]
    ].round(4)
)


# =============================================================================
# 10. UNDERWEIGHT THRESHOLD ANALYSIS
# =============================================================================

underweight_threshold_results = threshold_analysis(
    y_val_under,
    under_val_prob,
    thresholds
)

print("\n" + "=" * 100)
print("8. UNDERWEIGHT — THRESHOLD RESULTS")
print("=" * 100)

display(
    underweight_threshold_results[
        [
            "Threshold",
            "Precision",
            "Recall",
            "F1",
            "FPR",
            "TP",
            "FP",
            "FN",
            "TN"
        ]
    ].round(4)
)


# =============================================================================
# 11. PRECISION >= 50%
# =============================================================================

print("\n" + "=" * 100)
print("9. OPERATING POINTS WITH PRECISION >= 50%")
print("=" * 100)

print("\nSTUNTING")
print("-" * 100)

stunt_p50 = stunting_threshold_results[
    stunting_threshold_results["Precision"] >= 0.50
]

if len(stunt_p50) > 0:

    display(
        stunt_p50[
            [
                "Threshold",
                "Precision",
                "Recall",
                "F1",
                "FPR",
                "TP",
                "FP",
                "FN",
                "TN"
            ]
        ].round(4)
    )

else:
    print("No threshold achieved precision >= 50%.")


print("\nUNDERWEIGHT")
print("-" * 100)

under_p50 = underweight_threshold_results[
    underweight_threshold_results["Precision"] >= 0.50
]

if len(under_p50) > 0:

    display(
        under_p50[
            [
                "Threshold",
                "Precision",
                "Recall",
                "F1",
                "FPR",
                "TP",
                "FP",
                "FN",
                "TN"
            ]
        ].round(4)
    )

else:
    print("No threshold achieved precision >= 50%.")


# =============================================================================
# 12. PRECISION >= 50% WITH RECALL >= 30%
# =============================================================================

print("\n" + "=" * 100)
print("10. BALANCED PRECISION-FOCUSED OPERATING POINTS")
print("=" * 100)

print("\nSTUNTING")
print("-" * 100)

stunt_balanced = stunting_threshold_results[
    (stunting_threshold_results["Precision"] >= 0.50) &
    (stunting_threshold_results["Recall"] >= 0.30)
]

if len(stunt_balanced) > 0:

    display(
        stunt_balanced[
            [
                "Threshold",
                "Precision",
                "Recall",
                "F1",
                "FPR",
                "TP",
                "FP",
                "FN",
                "TN"
            ]
        ].sort_values(
            ["Precision", "F1"],
            ascending=False
        ).round(4)
    )

else:
    print(
        "No stunting threshold achieved both "
        "Precision >= 50% and Recall >= 30%."
    )


print("\nUNDERWEIGHT")
print("-" * 100)

under_balanced = underweight_threshold_results[
    (underweight_threshold_results["Precision"] >= 0.50) &
    (underweight_threshold_results["Recall"] >= 0.30)
]

if len(under_balanced) > 0:

    display(
        under_balanced[
            [
                "Threshold",
                "Precision",
                "Recall",
                "F1",
                "FPR",
                "TP",
                "FP",
                "FN",
                "TN"
            ]
        ].sort_values(
            ["Precision", "F1"],
            ascending=False
        ).round(4)
    )

else:
    print(
        "No underweight threshold achieved both "
        "Precision >= 50% and Recall >= 30%."
    )


# =============================================================================
# 13. BEST F1 OPERATING POINT
# =============================================================================

best_stunt_f1 = (
    stunting_threshold_results
    .sort_values(
        ["F1", "Precision"],
        ascending=False
    )
    .iloc[0]
)

best_under_f1 = (
    underweight_threshold_results
    .sort_values(
        ["F1", "Precision"],
        ascending=False
    )
    .iloc[0]
)

print("\n" + "=" * 100)
print("11. BEST F1 OPERATING POINTS")
print("=" * 100)

print("\nSTUNTING")
print("-" * 100)
print(f"Threshold : {best_stunt_f1['Threshold']:.2f}")
print(f"Precision : {best_stunt_f1['Precision']:.4f}")
print(f"Recall    : {best_stunt_f1['Recall']:.4f}")
print(f"F1        : {best_stunt_f1['F1']:.4f}")
print(f"FPR       : {best_stunt_f1['FPR']:.4f}")
print(f"TP        : {int(best_stunt_f1['TP'])}")
print(f"FP        : {int(best_stunt_f1['FP'])}")

print("\nUNDERWEIGHT")
print("-" * 100)
print(f"Threshold : {best_under_f1['Threshold']:.2f}")
print(f"Precision : {best_under_f1['Precision']:.4f}")
print(f"Recall    : {best_under_f1['Recall']:.4f}")
print(f"F1        : {best_under_f1['F1']:.4f}")
print(f"FPR       : {best_under_f1['FPR']:.4f}")
print(f"TP        : {int(best_under_f1['TP'])}")
print(f"FP        : {int(best_under_f1['FP'])}")


# =============================================================================
# 14. BEST PRECISION WITH RECALL >= 30%
# =============================================================================

print("\n" + "=" * 100)
print("12. BEST PRECISION WHILE RETAINING RECALL >= 30%")
print("=" * 100)

stunt_recall30 = stunting_threshold_results[
    stunting_threshold_results["Recall"] >= 0.30
]

under_recall30 = underweight_threshold_results[
    underweight_threshold_results["Recall"] >= 0.30
]


print("\nSTUNTING")
print("-" * 100)

if len(stunt_recall30) > 0:

    best_stunt_precision = (
        stunt_recall30
        .sort_values(
            ["Precision", "F1"],
            ascending=False
        )
        .iloc[0]
    )

    print(
        f"Threshold : "
        f"{best_stunt_precision['Threshold']:.2f}"
    )
    print(
        f"Precision : "
        f"{best_stunt_precision['Precision']:.4f}"
    )
    print(
        f"Recall    : "
        f"{best_stunt_precision['Recall']:.4f}"
    )
    print(
        f"F1        : "
        f"{best_stunt_precision['F1']:.4f}"
    )
    print(
        f"FPR       : "
        f"{best_stunt_precision['FPR']:.4f}"
    )
    print(
        f"TP        : "
        f"{int(best_stunt_precision['TP'])}"
    )
    print(
        f"FP        : "
        f"{int(best_stunt_precision['FP'])}"
    )

else:
    print("No valid threshold.")


print("\nUNDERWEIGHT")
print("-" * 100)

if len(under_recall30) > 0:

    best_under_precision = (
        under_recall30
        .sort_values(
            ["Precision", "F1"],
            ascending=False
        )
        .iloc[0]
    )

    print(
        f"Threshold : "
        f"{best_under_precision['Threshold']:.2f}"
    )
    print(
        f"Precision : "
        f"{best_under_precision['Precision']:.4f}"
    )
    print(
        f"Recall    : "
        f"{best_under_precision['Recall']:.4f}"
    )
    print(
        f"F1        : "
        f"{best_under_precision['F1']:.4f}"
    )
    print(
        f"FPR       : "
        f"{best_under_precision['FPR']:.4f}"
    )
    print(
        f"TP        : "
        f"{int(best_under_precision['TP'])}"
    )
    print(
        f"FP        : "
        f"{int(best_under_precision['FP'])}"
    )

else:
    print("No valid threshold.")


# =============================================================================
# 15. FINAL INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("STEP 7F COMPLETE")
print("=" * 100)

print("✓ Optimized XGBoost models fitted")
print("✓ Validation probabilities generated")
print("✓ Thresholds evaluated using validation data only")
print("✓ Precision calculated")
print("✓ Recall calculated")
print("✓ F1 calculated")
print("✓ False-positive rate calculated")
print("✓ Confusion matrices calculated")
print("✓ Test data remained completely untouched")

print("\nNEXT STEP:")
print("STEP 7G — SELECT FINAL VALIDATION OPERATING POINTS")
print("We will compare the candidate thresholds and choose")
print("the final threshold based on the research objective.")
```

```python
# =============================================================================
# STEP 7G: CAR MICS6 — FEATURE IMPORTANCE + PREDICTOR IMPROVEMENT
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.inspection import permutation_importance

print("=" * 100)
print("STEP 7G: CAR MICS6 — FEATURE IMPORTANCE + PREDICTOR IMPROVEMENT")
print("=" * 100)


# =============================================================================
# 1. CHECK EXISTING OBJECTS
# =============================================================================

required_objects = [
    "xgb_stunt_optimized",
    "xgb_under_optimized",
    "X_val_stunt_processed",
    "X_val_under_processed",
    "y_val_stunt",
    "y_val_under",
    "stunting_threshold_results",
    "underweight_threshold_results"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "The following required Step 7F objects are missing:\n"
        + "\n".join(missing_objects)
        + "\n\nPlease run Step 7F first."
    )

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)
print("✓ Step 7F models available")
print("✓ Validation matrices available")
print("✓ Validation targets available")
print("✓ Test data is not required")


# =============================================================================
# 2. FEATURE NAME CHECK
# =============================================================================

if hasattr(X_val_stunt_processed, "columns"):
    stunt_feature_names = list(
        X_val_stunt_processed.columns
    )
else:
    if "stunt_feature_names" in globals():
        stunt_feature_names = list(stunt_feature_names)
    elif "stunting_feature_names" in globals():
        stunt_feature_names = list(stunting_feature_names)
    else:
        raise NameError(
            "Stunting transformed feature names are unavailable."
        )


if hasattr(X_val_under_processed, "columns"):
    under_feature_names = list(
        X_val_under_processed.columns
    )
else:
    if "under_feature_names" in globals():
        under_feature_names = list(under_feature_names)
    elif "underweight_feature_names" in globals():
        under_feature_names = list(underweight_feature_names)
    else:
        raise NameError(
            "Underweight transformed feature names are unavailable."
        )


print("\n2. TRANSFORMED FEATURE CHECK")
print("-" * 100)

print(
    f"Stunting transformed features   : "
    f"{len(stunt_feature_names)}"
)

print(
    f"Underweight transformed features: "
    f"{len(under_feature_names)}"
)


# =============================================================================
# 3. XGBOOST FEATURE IMPORTANCE
# =============================================================================

print("\n" + "=" * 100)
print("3. XGBOOST FEATURE IMPORTANCE")
print("=" * 100)


# -----------------------------------------------------------------------------
# STUNTING
# -----------------------------------------------------------------------------

stunt_importance = pd.DataFrame({
    "Feature": stunt_feature_names,
    "Importance": xgb_stunt_optimized.feature_importances_
})

stunt_importance = (
    stunt_importance
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

stunt_importance["Rank"] = (
    np.arange(len(stunt_importance)) + 1
)

print("\nSTUNTING — TOP 30 FEATURES")
print("-" * 100)

display(
    stunt_importance[
        [
            "Rank",
            "Feature",
            "Importance"
        ]
    ]
    .head(30)
    .round(6)
)


# -----------------------------------------------------------------------------
# UNDERWEIGHT
# -----------------------------------------------------------------------------

under_importance = pd.DataFrame({
    "Feature": under_feature_names,
    "Importance": xgb_under_optimized.feature_importances_
})

under_importance = (
    under_importance
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

under_importance["Rank"] = (
    np.arange(len(under_importance)) + 1
)

print("\nUNDERWEIGHT — TOP 30 FEATURES")
print("-" * 100)

display(
    under_importance[
        [
            "Rank",
            "Feature",
            "Importance"
        ]
    ]
    .head(30)
    .round(6)
)


# =============================================================================
# 4. ZERO-IMPORTANCE FEATURES
# =============================================================================

print("\n" + "=" * 100)
print("4. ZERO-IMPORTANCE FEATURE CHECK")
print("=" * 100)

stunt_zero = stunt_importance[
    stunt_importance["Importance"] == 0
]

under_zero = under_importance[
    under_importance["Importance"] == 0
]

print(
    f"\nStunting zero-importance transformed features   : "
    f"{len(stunt_zero)}"
)

print(
    f"Underweight zero-importance transformed features: "
    f"{len(under_zero)}"
)

if len(stunt_zero) > 0:
    print("\nStunting zero-importance features:")
    print(stunt_zero["Feature"].tolist())

if len(under_zero) > 0:
    print("\nUnderweight zero-importance features:")
    print(under_zero["Feature"].tolist())


# =============================================================================
# 5. PERMUTATION IMPORTANCE
# =============================================================================
#
# This is more useful than relying only on XGBoost's built-in importance.
#
# It evaluates how much validation performance changes when a transformed
# feature is randomly shuffled.
#
# IMPORTANT:
# Validation data is being used only for diagnostic feature analysis.
# Test data remains untouched.
# =============================================================================

print("\n" + "=" * 100)
print("5. PERMUTATION IMPORTANCE — VALIDATION DATA")
print("=" * 100)

print("\nCalculating stunting permutation importance...")

stunt_perm = permutation_importance(
    xgb_stunt_optimized,
    X_val_stunt_processed,
    y_val_stunt,
    scoring="average_precision",
    n_repeats=10,
    random_state=42,
    n_jobs=-1
)

stunt_perm_df = pd.DataFrame({
    "Feature": stunt_feature_names,
    "Permutation_Mean": stunt_perm.importances_mean,
    "Permutation_SD": stunt_perm.importances_std
})

stunt_perm_df = (
    stunt_perm_df
    .sort_values(
        "Permutation_Mean",
        ascending=False
    )
    .reset_index(drop=True)
)

stunt_perm_df["Rank"] = (
    np.arange(len(stunt_perm_df)) + 1
)

print("\nSTUNTING — TOP 30 PERMUTATION FEATURES")
print("-" * 100)

display(
    stunt_perm_df[
        [
            "Rank",
            "Feature",
            "Permutation_Mean",
            "Permutation_SD"
        ]
    ]
    .head(30)
    .round(6)
)


print("\nCalculating underweight permutation importance...")

under_perm = permutation_importance(
    xgb_under_optimized,
    X_val_under_processed,
    y_val_under,
    scoring="average_precision",
    n_repeats=10,
    random_state=42,
    n_jobs=-1
)

under_perm_df = pd.DataFrame({
    "Feature": under_feature_names,
    "Permutation_Mean": under_perm.importances_mean,
    "Permutation_SD": under_perm.importances_std
})

under_perm_df = (
    under_perm_df
    .sort_values(
        "Permutation_Mean",
        ascending=False
    )
    .reset_index(drop=True)
)

under_perm_df["Rank"] = (
    np.arange(len(under_perm_df)) + 1
)

print("\nUNDERWEIGHT — TOP 30 PERMUTATION FEATURES")
print("-" * 100)

display(
    under_perm_df[
        [
            "Rank",
            "Feature",
            "Permutation_Mean",
            "Permutation_SD"
        ]
    ]
    .head(30)
    .round(6)
)


# =============================================================================
# 6. AGGREGATE ONE-HOT FEATURES BACK TO ORIGINAL PREDICTORS
# =============================================================================
#
# This is important.
#
# One original predictor can generate many one-hot variables.
# Therefore we should not judge predictor usefulness only from individual
# encoded columns.
# =============================================================================

print("\n" + "=" * 100)
print("6. ORIGINAL-PREDICTOR IMPORTANCE")
print("=" * 100)


approved_predictors = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cinsurance",
    "cdisability",
    "CA1",
    "CA14",
    "CA16",
    "CA17",
    "BD2",
    "TN3",
    "EC1",
    "IM2",
    "CA31"
]


def original_predictor(feature_name, approved):

    # Remove common sklearn transformer prefixes
    cleaned = feature_name

    for prefix in [
        "numeric__",
        "categorical__"
    ]:
        if cleaned.startswith(prefix):
            cleaned = cleaned[len(prefix):]

    # Match the longest approved predictor first
    matches = [
        p for p in approved
        if cleaned == p
        or cleaned.startswith(p + "_")
    ]

    if matches:
        return max(matches, key=len)

    return cleaned


# -----------------------------------------------------------------------------
# STUNTING AGGREGATED IMPORTANCE
# -----------------------------------------------------------------------------

stunt_importance["Original_Predictor"] = (
    stunt_importance["Feature"]
    .apply(
        lambda x:
        original_predictor(
            x,
            approved_predictors
        )
    )
)

stunt_original = (
    stunt_importance
    .groupby(
        "Original_Predictor",
        as_index=False
    )["Importance"]
    .sum()
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

stunt_original["Rank"] = (
    np.arange(len(stunt_original)) + 1
)

print("\nSTUNTING — AGGREGATED PREDICTOR IMPORTANCE")
print("-" * 100)

display(
    stunt_original[
        [
            "Rank",
            "Original_Predictor",
            "Importance"
        ]
    ].round(6)
)


# -----------------------------------------------------------------------------
# UNDERWEIGHT AGGREGATED IMPORTANCE
# -----------------------------------------------------------------------------

under_importance["Original_Predictor"] = (
    under_importance["Feature"]
    .apply(
        lambda x:
        original_predictor(
            x,
            approved_predictors
        )
    )
)

under_original = (
    under_importance
    .groupby(
        "Original_Predictor",
        as_index=False
    )["Importance"]
    .sum()
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

under_original["Rank"] = (
    np.arange(len(under_original)) + 1
)

print("\nUNDERWEIGHT — AGGREGATED PREDICTOR IMPORTANCE")
print("-" * 100)

display(
    under_original[
        [
            "Rank",
            "Original_Predictor",
            "Importance"
        ]
    ].round(6)
)


# =============================================================================
# 7. LOW-VALUE PREDICTOR SCREEN
# =============================================================================

print("\n" + "=" * 100)
print("7. LOW-VALUE PREDICTOR SCREEN")
print("=" * 100)

print(
    "\nIMPORTANT: Low importance does NOT automatically mean "
    "the predictor should be removed."
)

print(
    "Predictors will only be considered for removal after "
    "validation comparison."
)

stunt_low = stunt_original[
    stunt_original["Importance"] <= 0.01
]

under_low = under_original[
    under_original["Importance"] <= 0.01
]

print("\nStunting predictors with aggregated importance <= 0.01:")
print(
    stunt_low["Original_Predictor"].tolist()
)

print("\nUnderweight predictors with aggregated importance <= 0.01:")
print(
    under_low["Original_Predictor"].tolist()
)


# =============================================================================
# 8. VALIDATION PERFORMANCE OF CURRENT MODELS
# =============================================================================

print("\n" + "=" * 100)
print("8. CURRENT VALIDATION BASELINE")
print("=" * 100)


stunt_prob = xgb_stunt_optimized.predict_proba(
    X_val_stunt_processed
)[:, 1]

under_prob = xgb_under_optimized.predict_proba(
    X_val_under_processed
)[:, 1]


from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)

print("\nSTUNTING")
print("-" * 100)
print(
    f"ROC-AUC : "
    f"{roc_auc_score(y_val_stunt, stunt_prob):.4f}"
)
print(
    f"PR-AUC  : "
    f"{average_precision_score(y_val_stunt, stunt_prob):.4f}"
)

print("\nUNDERWEIGHT")
print("-" * 100)
print(
    f"ROC-AUC : "
    f"{roc_auc_score(y_val_under, under_prob):.4f}"
)
print(
    f"PR-AUC  : "
    f"{average_precision_score(y_val_under, under_prob):.4f}"
)


# =============================================================================
# 9. FINAL DIAGNOSTIC SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("STEP 7G COMPLETE")
print("=" * 100)

print("✓ XGBoost feature importance calculated")
print("✓ Validation permutation importance calculated")
print("✓ One-hot features aggregated to original predictors")
print("✓ Low-importance predictors identified")
print("✓ Current validation PR-AUC recorded")
print("✓ Test set remained completely untouched")

print("\nIMPORTANT:")
print("No predictors have been removed yet.")
print("No new predictors have been added yet.")
print("No test-set evaluation has been performed.")

print("\nNEXT STEP:")
print("STEP 7H — CONTROLLED PREDICTOR-SET EXPERIMENT")
print(
    "Compare the current 20-predictor model against carefully "
    "defined predictor sets using validation data only."
)
```

```python
# =============================================================================
# STEP 7H: CAR MICS6 — CONTROLLED PREDICTOR-SET ABLATION
# =============================================================================
#
# PURPOSE
# -------
# Determine whether removing weak predictors identified in Step 7G improves
# validation performance.
#
# IMPORTANT
# ---------
# • The 20-predictor model is the reference model.
# • Only TRAINING and VALIDATION data are used.
# • TEST data are NOT used.
# • No final threshold is selected here.
# • Predictors are removed only when the validation results support removal.
#
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder


print("=" * 100)
print("STEP 7H: CAR MICS6 — CONTROLLED PREDICTOR-SET ABLATION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

required_objects = [
    "X_train_stunt",
    "X_val_stunt",
    "y_train_stunt",
    "y_val_stunt",
    "X_train_under",
    "X_val_under",
    "y_train_under",
    "y_val_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)

if missing_objects:
    print("Missing objects:")
    for obj in missing_objects:
        print(" -", obj)

    raise NameError(
        "\nStep 7H requires the raw training/validation predictor objects "
        "from Step 4.3."
    )

print("✓ Raw training/validation predictor objects available")


# =============================================================================
# 2. APPROVED 20-PREDICTOR SET
# =============================================================================

approved_predictors = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cinsurance",
    "cdisability",
    "CA1",
    "CA14",
    "CA16",
    "CA17",
    "BD2",
    "TN3",
    "EC1",
    "IM2",
    "CA31"
]

print("\n2. APPROVED PREDICTORS")
print("-" * 100)

print("Number of approved predictors:", len(approved_predictors))

for i, predictor in enumerate(
    approved_predictors,
    start=1
):
    print(f"{i:2d}. {predictor}")


# =============================================================================
# 3. VERIFY PREDICTOR AVAILABILITY
# =============================================================================

missing_stunt = [
    p for p in approved_predictors
    if p not in X_train_stunt.columns
]

missing_under = [
    p for p in approved_predictors
    if p not in X_train_under.columns
]

if missing_stunt:
    raise KeyError(
        "Missing stunting predictors: "
        + str(missing_stunt)
    )

if missing_under:
    raise KeyError(
        "Missing underweight predictors: "
        + str(missing_under)
    )

print("\n✓ All 20 predictors available in stunting data")
print("✓ All 20 predictors available in underweight data")


# =============================================================================
# 4. TEST-SET PROTECTION
# =============================================================================

print("\n" + "=" * 100)
print("3. TEST-SET PROTECTION")
print("=" * 100)

print("✓ Test predictors are NOT accessed")
print("✓ Test targets are NOT accessed")
print("✓ Test performance is NOT calculated")
print("✓ Test threshold is NOT selected")
print("✓ Test set remains completely unseen")


# =============================================================================
# 5. PREDICTOR SETS TO COMPARE
# =============================================================================
#
# SET A = all 20 predictors
#
# SET B = remove cinsurance
#
# SET C = remove cinsurance + CA16 + TN3 + CA17
#
# SET D = remove cinsurance + CA16 + TN3 + CA17 + EC1
#
# We deliberately do NOT remove predictors based solely on importance.
# Each set will be empirically evaluated.
# =============================================================================

predictor_sets = {

    "A_All_20": approved_predictors.copy(),

    "B_Remove_cinsurance": [
        p for p in approved_predictors
        if p not in [
            "cinsurance"
        ]
    ],

    "C_Remove_4_Low_Value": [
        p for p in approved_predictors
        if p not in [
            "cinsurance",
            "CA16",
            "TN3",
            "CA17"
        ]
    ],

    "D_Remove_5_Low_Value": [
        p for p in approved_predictors
        if p not in [
            "cinsurance",
            "CA16",
            "TN3",
            "CA17",
            "EC1"
        ]
    ]
}


print("\n" + "=" * 100)
print("4. CONTROLLED PREDICTOR SETS")
print("=" * 100)

for name, predictors in predictor_sets.items():

    print(
        f"\n{name}"
    )

    print(
        f"Number of predictors: {len(predictors)}"
    )

    print(
        "Predictors:",
        ", ".join(predictors)
    )


# =============================================================================
# 6. PREPROCESSOR FUNCTION
# =============================================================================

numeric_predictors = [
    "CAGE"
]

categorical_predictors = [
    p for p in approved_predictors
    if p != "CAGE"
]


def create_preprocessor(predictors):

    numeric_features = [
        p for p in predictors
        if p in numeric_predictors
    ]

    categorical_features = [
        p for p in predictors
        if p in categorical_predictors
    ]

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ])

    preprocessor = ColumnTransformer(
        transformers=[
            (
                "numeric",
                numeric_pipeline,
                numeric_features
            ),
            (
                "categorical",
                categorical_pipeline,
                categorical_features
            )
        ],
        remainder="drop"
    )

    return preprocessor


# =============================================================================
# 7. MODEL FUNCTION
# =============================================================================

def train_and_evaluate(
    X_train,
    X_val,
    y_train,
    y_val,
    predictors,
    target_name
):

    # -------------------------------------------------------------------------
    # Select predictors
    # -------------------------------------------------------------------------

    Xtr = X_train[predictors].copy()
    Xv = X_val[predictors].copy()

    # -------------------------------------------------------------------------
    # Training-only preprocessing
    # -------------------------------------------------------------------------

    preprocessor = create_preprocessor(
        predictors
    )

    Xtr_processed = preprocessor.fit_transform(
        Xtr
    )

    Xv_processed = preprocessor.transform(
        Xv
    )

    # -------------------------------------------------------------------------
    # Class imbalance
    # -------------------------------------------------------------------------

    negative = (
        y_train == 0
    ).sum()

    positive = (
        y_train == 1
    ).sum()

    scale_pos_weight = (
        negative / positive
    )

    # -------------------------------------------------------------------------
    # XGBoost configuration
    #
    # Same optimized configuration used in Step 7G.
    # This isolates the effect of predictor removal.
    # -------------------------------------------------------------------------

    model = XGBClassifier(
        n_estimators=500,
        learning_rate=(
            0.03
            if target_name == "Stunting"
            else 0.01
        ),
        max_depth=(
            3
            if target_name == "Stunting"
            else 2
        ),
        min_child_weight=(
            8
            if target_name == "Stunting"
            else 5
        ),
        subsample=(
            0.80
            if target_name == "Stunting"
            else 0.85
        ),
        colsample_bytree=(
            0.80
            if target_name == "Stunting"
            else 0.85
        ),
        reg_alpha=(
            0.5
            if target_name == "Stunting"
            else 0.1
        ),
        reg_lambda=(
            3.0
            if target_name == "Stunting"
            else 2.0
        ),
        gamma=0,
        scale_pos_weight=scale_pos_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    # -------------------------------------------------------------------------
    # Fit on TRAINING only
    # -------------------------------------------------------------------------

    model.fit(
        Xtr_processed,
        y_train
    )

    # -------------------------------------------------------------------------
    # Validation probabilities
    # -------------------------------------------------------------------------

    probabilities = model.predict_proba(
        Xv_processed
    )[:, 1]

    # Default threshold ONLY for model comparison
    predictions = (
        probabilities >= 0.50
    ).astype(int)

    # -------------------------------------------------------------------------
    # Metrics
    # -------------------------------------------------------------------------

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        predictions,
        labels=[0, 1]
    ).ravel()

    return {
        "Target": target_name,
        "Predictor_Set": None,
        "N_Predictors": len(predictors),

        "Accuracy": accuracy_score(
            y_val,
            predictions
        ),

        "Precision": precision_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "Recall": recall_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "F1": f1_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "ROC_AUC": roc_auc_score(
            y_val,
            probabilities
        ),

        "PR_AUC": average_precision_score(
            y_val,
            probabilities
        ),

        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }


# =============================================================================
# 8. RUN STUNTING ABLATION
# =============================================================================

print("\n" + "=" * 100)
print("5. STUNTING — CONTROLLED ABLATION")
print("=" * 100)

stunting_results = []

for set_name, predictors in predictor_sets.items():

    print(
        f"\nEvaluating {set_name}..."
    )

    result = train_and_evaluate(
        X_train_stunt,
        X_val_stunt,
        y_train_stunt,
        y_val_stunt,
        predictors,
        "Stunting"
    )

    result["Predictor_Set"] = set_name

    stunting_results.append(
        result
    )

    print(
        f"Precision = {result['Precision']:.4f} | "
        f"Recall = {result['Recall']:.4f} | "
        f"F1 = {result['F1']:.4f} | "
        f"ROC-AUC = {result['ROC_AUC']:.4f} | "
        f"PR-AUC = {result['PR_AUC']:.4f} | "
        f"FP = {result['FP']}"
    )


stunting_ablation = pd.DataFrame(
    stunting_results
)


# =============================================================================
# 9. RUN UNDERWEIGHT ABLATION
# =============================================================================

print("\n" + "=" * 100)
print("6. UNDERWEIGHT — CONTROLLED ABLATION")
print("=" * 100)

underweight_results = []

for set_name, predictors in predictor_sets.items():

    print(
        f"\nEvaluating {set_name}..."
    )

    result = train_and_evaluate(
        X_train_under,
        X_val_under,
        y_train_under,
        y_val_under,
        predictors,
        "Underweight"
    )

    result["Predictor_Set"] = set_name

    underweight_results.append(
        result
    )

    print(
        f"Precision = {result['Precision']:.4f} | "
        f"Recall = {result['Recall']:.4f} | "
        f"F1 = {result['F1']:.4f} | "
        f"ROC-AUC = {result['ROC_AUC']:.4f} | "
        f"PR-AUC = {result['PR_AUC']:.4f} | "
        f"FP = {result['FP']}"
    )


underweight_ablation = pd.DataFrame(
    underweight_results
)


# =============================================================================
# 10. STUNTING COMPARISON
# =============================================================================

print("\n" + "=" * 100)
print("7. STUNTING — ABLATION COMPARISON")
print("=" * 100)

display(
    stunting_ablation[
        [
            "Predictor_Set",
            "N_Predictors",
            "Accuracy",
            "Precision",
            "Recall",
            "F1",
            "ROC_AUC",
            "PR_AUC",
            "TN",
            "FP",
            "FN",
            "TP"
        ]
    ]
    .sort_values(
        "F1",
        ascending=False
    )
    .round(4)
)


# =============================================================================
# 11. UNDERWEIGHT COMPARISON
# =============================================================================

print("\n" + "=" * 100)
print("8. UNDERWEIGHT — ABLATION COMPARISON")
print("=" * 100)

display(
    underweight_ablation[
        [
            "Predictor_Set",
            "N_Predictors",
            "Accuracy",
            "Precision",
            "Recall",
            "F1",
            "ROC_AUC",
            "PR_AUC",
            "TN",
            "FP",
            "FN",
            "TP"
        ]
    ]
    .sort_values(
        "F1",
        ascending=False
    )
    .round(4)
)


# =============================================================================
# 12. PRECISION-FOCUSED COMPARISON
# =============================================================================

print("\n" + "=" * 100)
print("9. PRECISION-FOCUSED COMPARISON")
print("=" * 100)

print(
    "\nSTUNTING — ranked by Precision"
)

display(
    stunting_ablation[
        [
            "Predictor_Set",
            "N_Predictors",
            "Precision",
            "Recall",
            "F1",
            "FPR",
            "FP"
        ]
    ]
    if "FPR" in stunting_ablation.columns
    else stunting_ablation[
        [
            "Predictor_Set",
            "N_Predictors",
            "Precision",
            "Recall",
            "F1",
            "FP"
        ]
    ]
    .sort_values(
        "Precision",
        ascending=False
    )
    .round(4)
)


print(
    "\nUNDERWEIGHT — ranked by Precision"
)

display(
    underweight_ablation[
        [
            "Predictor_Set",
            "N_Predictors",
            "Precision",
            "Recall",
            "F1",
            "FP"
        ]
    ]
    .sort_values(
        "Precision",
        ascending=False
    )
    .round(4)
)


# =============================================================================
# 13. BEST SET BY F1
# =============================================================================

best_stunting = (
    stunting_ablation
    .sort_values(
        [
            "F1",
            "Precision"
        ],
        ascending=False
    )
    .iloc[0]
)

best_underweight = (
    underweight_ablation
    .sort_values(
        [
            "F1",
            "Precision"
        ],
        ascending=False
    )
    .iloc[0]
)


# =============================================================================
# 14. BEST SET BY PRECISION WITH RECALL >= 30%
# =============================================================================

stunt_candidates = stunting_ablation[
    stunting_ablation["Recall"] >= 0.30
]

under_candidates = underweight_ablation[
    underweight_ablation["Recall"] >= 0.30
]


print("\n" + "=" * 100)
print("10. BEST PRECISION WHILE RETAINING RECALL >= 30%")
print("=" * 100)


print("\nSTUNTING")
print("-" * 100)

if len(stunt_candidates) > 0:

    best_stunt_precision = (
        stunt_candidates
        .sort_values(
            [
                "Precision",
                "F1"
            ],
            ascending=False
        )
        .iloc[0]
    )

    print(
        f"Predictor set : "
        f"{best_stunt_precision['Predictor_Set']}"
    )

    print(
        f"Predictors    : "
        f"{int(best_stunt_precision['N_Predictors'])}"
    )

    print(
        f"Precision     : "
        f"{best_stunt_precision['Precision']:.4f}"
    )

    print(
        f"Recall        : "
        f"{best_stunt_precision['Recall']:.4f}"
    )

    print(
        f"F1            : "
        f"{best_stunt_precision['F1']:.4f}"
    )

    print(
        f"FP            : "
        f"{int(best_stunt_precision['FP'])}"
    )

else:

    print(
        "No predictor set achieved recall >= 30%."
    )


print("\nUNDERWEIGHT")
print("-" * 100)

if len(under_candidates) > 0:

    best_under_precision = (
        under_candidates
        .sort_values(
            [
                "Precision",
                "F1"
            ],
            ascending=False
        )
        .iloc[0]
    )

    print(
        f"Predictor set : "
        f"{best_under_precision['Predictor_Set']}"
    )

    print(
        f"Predictors    : "
        f"{int(best_under_precision['N_Predictors'])}"
    )

    print(
        f"Precision     : "
        f"{best_under_precision['Precision']:.4f}"
    )

    print(
        f"Recall        : "
        f"{best_under_precision['Recall']:.4f}"
    )

    print(
        f"F1            : "
        f"{best_under_precision['F1']:.4f}"
    )

    print(
        f"FP            : "
        f"{int(best_under_precision['FP'])}"
    )

else:

    print(
        "No predictor set achieved recall >= 30%."
    )


# =============================================================================
# 15. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("11. STEP 7H SUMMARY")
print("=" * 100)

print("\nSTUNTING — BEST F1 SET")
print(
    f"Set       : {best_stunting['Predictor_Set']}"
)
print(
    f"Predictors: {int(best_stunting['N_Predictors'])}"
)
print(
    f"Precision : {best_stunting['Precision']:.4f}"
)
print(
    f"Recall    : {best_stunting['Recall']:.4f}"
)
print(
    f"F1        : {best_stunting['F1']:.4f}"
)
print(
    f"PR-AUC    : {best_stunting['PR_AUC']:.4f}"
)


print("\nUNDERWEIGHT — BEST F1 SET")
print(
    f"Set       : {best_underweight['Predictor_Set']}"
)
print(
    f"Predictors: {int(best_underweight['N_Predictors'])}"
)
print(
    f"Precision : {best_underweight['Precision']:.4f}"
)
print(
    f"Recall    : {best_underweight['Recall']:.4f}"
)
print(
    f"F1        : {best_underweight['F1']:.4f}"
)
print(
    f"PR-AUC    : {best_underweight['PR_AUC']:.4f}"
)


# =============================================================================
# 16. INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("STEP 7H COMPLETE")
print("=" * 100)

print("✓ Four controlled predictor sets compared")
print("✓ Original 20-predictor model retained as reference")
print("✓ Predictor removal evaluated empirically")
print("✓ Training-only preprocessing used")
print("✓ Models fitted using training data only")
print("✓ Validation data used for comparison")
print("✓ Test data remained completely untouched")
print("✓ No final test threshold selected")
print("✓ No predictors permanently removed yet")

print("\nNEXT STEP:")
print("STEP 7I — SELECT THE BEST VALIDATED PREDICTOR SET")
print(
    "Only after reviewing these results will we decide "
    "which predictor set should proceed to final optimization."
)
```

```python
# =============================================================================
# STEP 7I: CAR MICS6 — CONTROLLED XGBOOST OPTIMIZATION
# =============================================================================
#
# OBJECTIVE
# ---------
# Improve precision and reduce false positives while retaining useful recall.
#
# IMPORTANT
# ---------
# • Uses TRAINING + VALIDATION only
# • TEST data are NOT accessed
# • Predictor set is fixed
# • One optimization dimension is changed at a time
# • Threshold selection is NOT performed here
#
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder


print("=" * 100)
print("STEP 7I: CAR MICS6 — CONTROLLED XGBOOST OPTIMIZATION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

required_objects = [
    "X_train_stunt",
    "X_val_stunt",
    "y_train_stunt",
    "y_val_stunt",
    "X_train_under",
    "X_val_under",
    "y_train_under",
    "y_val_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)

if missing_objects:
    print("Missing objects:")
    for obj in missing_objects:
        print(" -", obj)

    raise NameError(
        "Required Step 4.3 training/validation objects are missing. "
        "Run Step 4.3 before Step 7I."
    )

print("✓ Training and validation objects available")


# =============================================================================
# 2. FIXED PREDICTOR SET
# =============================================================================
#
# Set D from Step 7H.
#
# Removed:
# cinsurance
# CA16
# TN3
# CA17
# EC1
#
# =============================================================================

predictors_15 = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cdisability",
    "CA1",
    "CA14",
    "BD2",
    "IM2",
    "CA31"
]

print("\n2. FIXED PREDICTOR SET")
print("-" * 100)

print("Number of predictors:", len(predictors_15))

for i, predictor in enumerate(
    predictors_15,
    start=1
):
    print(f"{i:2d}. {predictor}")


# =============================================================================
# 3. PREDICTOR AVAILABILITY
# =============================================================================

missing_stunt = [
    p for p in predictors_15
    if p not in X_train_stunt.columns
]

missing_under = [
    p for p in predictors_15
    if p not in X_train_under.columns
]

if missing_stunt:
    raise KeyError(
        "Missing stunting predictors: "
        + str(missing_stunt)
    )

if missing_under:
    raise KeyError(
        "Missing underweight predictors: "
        + str(missing_under)
    )

print("\n✓ All 15 predictors available for stunting")
print("✓ All 15 predictors available for underweight")


# =============================================================================
# 4. TEST-SET PROTECTION
# =============================================================================

print("\n" + "=" * 100)
print("3. TEST-SET PROTECTION")
print("=" * 100)

print("✓ Test predictors are NOT accessed")
print("✓ Test targets are NOT accessed")
print("✓ Test probabilities are NOT calculated")
print("✓ Test thresholds are NOT evaluated")
print("✓ Test set remains completely unseen")


# =============================================================================
# 5. PREPROCESSOR
# =============================================================================

numeric_predictors = [
    "CAGE"
]

categorical_predictors = [
    p for p in predictors_15
    if p != "CAGE"
]


def make_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ])

    return ColumnTransformer(
        transformers=[
            (
                "numeric",
                numeric_pipeline,
                numeric_predictors
            ),
            (
                "categorical",
                categorical_pipeline,
                categorical_predictors
            )
        ],
        remainder="drop"
    )


# =============================================================================
# 6. PREPROCESS TRAINING + VALIDATION
# =============================================================================

print("\n" + "=" * 100)
print("4. TRAINING-ONLY PREPROCESSING")
print("=" * 100)


stunt_preprocessor = make_preprocessor()

Xtr_stunt = stunt_preprocessor.fit_transform(
    X_train_stunt[predictors_15]
)

Xv_stunt = stunt_preprocessor.transform(
    X_val_stunt[predictors_15]
)


under_preprocessor = make_preprocessor()

Xtr_under = under_preprocessor.fit_transform(
    X_train_under[predictors_15]
)

Xv_under = under_preprocessor.transform(
    X_val_under[predictors_15]
)


print(
    "Stunting processed training shape   :",
    Xtr_stunt.shape
)

print(
    "Stunting processed validation shape :",
    Xv_stunt.shape
)

print(
    "Underweight processed training shape:",
    Xtr_under.shape
)

print(
    "Underweight processed validation shape:",
    Xv_under.shape
)

print("✓ Preprocessors fitted using training data only")


# =============================================================================
# 7. CLASS WEIGHTS
# =============================================================================

stunt_negative = (
    y_train_stunt == 0
).sum()

stunt_positive = (
    y_train_stunt == 1
).sum()

under_negative = (
    y_train_under == 0
).sum()

under_positive = (
    y_train_under == 1
).sum()


stunt_weight = (
    stunt_negative / stunt_positive
)

under_weight = (
    under_negative / under_positive
)


print("\n" + "=" * 100)
print("5. CLASS IMBALANCE")
print("=" * 100)

print(
    f"Stunting scale_pos_weight   : "
    f"{stunt_weight:.4f}"
)

print(
    f"Underweight scale_pos_weight: "
    f"{under_weight:.4f}"
)


# =============================================================================
# 8. MODEL CONFIGURATIONS
# =============================================================================
#
# We deliberately test controlled combinations.
#
# Focus:
#   • learning rate
#   • tree depth
#   • min_child_weight
#   • regularization
#   • subsampling
#
# The test set is never involved.
# =============================================================================

configurations = [

    # -------------------------------------------------------------------------
    # BASELINE / REFERENCE
    # -------------------------------------------------------------------------

    {
        "name": "M1_Reference",
        "n_estimators": 500,
        "learning_rate": 0.03,
        "max_depth": 3,
        "min_child_weight": 8,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.5,
        "reg_lambda": 3.0
    },

    # -------------------------------------------------------------------------
    # LOWER LEARNING RATE
    # -------------------------------------------------------------------------

    {
        "name": "M2_LR_0.02",
        "n_estimators": 700,
        "learning_rate": 0.02,
        "max_depth": 3,
        "min_child_weight": 8,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.5,
        "reg_lambda": 3.0
    },

    {
        "name": "M3_LR_0.01",
        "n_estimators": 900,
        "learning_rate": 0.01,
        "max_depth": 3,
        "min_child_weight": 8,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.5,
        "reg_lambda": 3.0
    },

    # -------------------------------------------------------------------------
    # SHALLOWER TREES
    # -------------------------------------------------------------------------

    {
        "name": "M4_Depth_2",
        "n_estimators": 700,
        "learning_rate": 0.02,
        "max_depth": 2,
        "min_child_weight": 8,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.5,
        "reg_lambda": 3.0
    },

    # -------------------------------------------------------------------------
    # STRONGER MIN CHILD WEIGHT
    # -------------------------------------------------------------------------

    {
        "name": "M5_MinChild_12",
        "n_estimators": 700,
        "learning_rate": 0.02,
        "max_depth": 3,
        "min_child_weight": 12,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.5,
        "reg_lambda": 3.0
    },

    # -------------------------------------------------------------------------
    # STRONGER L1 REGULARIZATION
    # -------------------------------------------------------------------------

    {
        "name": "M6_L1_Strong",
        "n_estimators": 700,
        "learning_rate": 0.02,
        "max_depth": 3,
        "min_child_weight": 8,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 1.0,
        "reg_lambda": 3.0
    },

    # -------------------------------------------------------------------------
    # STRONGER L2 REGULARIZATION
    # -------------------------------------------------------------------------

    {
        "name": "M7_L2_Strong",
        "n_estimators": 700,
        "learning_rate": 0.02,
        "max_depth": 3,
        "min_child_weight": 8,
        "subsample": 0.80,
        "colsample_bytree": 0.80,
        "reg_alpha": 0.5,
        "reg_lambda": 5.0
    },

    # -------------------------------------------------------------------------
    # STRONGER SUBSAMPLING
    # -------------------------------------------------------------------------

    {
        "name": "M8_SubSample_0.70",
        "n_estimators": 700,
        "learning_rate": 0.02,
        "max_depth": 3,
        "min_child_weight": 8,
        "subsample": 0.70,
        "colsample_bytree": 0.70,
        "reg_alpha": 0.5,
        "reg_lambda": 3.0
    }
]


print("\n" + "=" * 100)
print("6. OPTIMIZATION CONFIGURATIONS")
print("=" * 100)

print(
    f"Number of configurations: "
    f"{len(configurations)}"
)

for config in configurations:

    print(
        f"\n{config['name']}"
    )

    print(
        f"  estimators       : {config['n_estimators']}"
    )

    print(
        f"  learning_rate    : {config['learning_rate']}"
    )

    print(
        f"  max_depth       : {config['max_depth']}"
    )

    print(
        f"  min_child_weight: {config['min_child_weight']}"
    )

    print(
        f"  subsample       : {config['subsample']}"
    )

    print(
        f"  colsample       : {config['colsample_bytree']}"
    )

    print(
        f"  reg_alpha       : {config['reg_alpha']}"
    )

    print(
        f"  reg_lambda      : {config['reg_lambda']}"
    )


# =============================================================================
# 9. EVALUATION FUNCTION
# =============================================================================

def evaluate_configuration(
    X_train_processed,
    X_val_processed,
    y_train,
    y_val,
    scale_pos_weight,
    config
):

    model = XGBClassifier(
        n_estimators=config["n_estimators"],
        learning_rate=config["learning_rate"],
        max_depth=config["max_depth"],
        min_child_weight=config["min_child_weight"],
        subsample=config["subsample"],
        colsample_bytree=config["colsample_bytree"],
        reg_alpha=config["reg_alpha"],
        reg_lambda=config["reg_lambda"],
        objective="binary:logistic",
        eval_metric="logloss",
        scale_pos_weight=scale_pos_weight,
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_processed,
        y_train
    )

    probabilities = model.predict_proba(
        X_val_processed
    )[:, 1]

    predictions = (
        probabilities >= 0.50
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        predictions,
        labels=[0, 1]
    ).ravel()

    return {
        "Accuracy": accuracy_score(
            y_val,
            predictions
        ),

        "Precision": precision_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "Recall": recall_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "F1": f1_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "ROC_AUC": roc_auc_score(
            y_val,
            probabilities
        ),

        "PR_AUC": average_precision_score(
            y_val,
            probabilities
        ),

        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }


# =============================================================================
# 10. STUNTING OPTIMIZATION
# =============================================================================

print("\n" + "=" * 100)
print("7. STUNTING — XGBOOST OPTIMIZATION")
print("=" * 100)

stunting_results = []

for config in configurations:

    print(
        f"\nRunning {config['name']}..."
    )

    metrics = evaluate_configuration(
        Xtr_stunt,
        Xv_stunt,
        y_train_stunt,
        y_val_stunt,
        stunt_weight,
        config
    )

    metrics["Model"] = config["name"]

    stunting_results.append(
        metrics
    )

    print(
        f"Precision={metrics['Precision']:.4f} | "
        f"Recall={metrics['Recall']:.4f} | "
        f"F1={metrics['F1']:.4f} | "
        f"ROC-AUC={metrics['ROC_AUC']:.4f} | "
        f"PR-AUC={metrics['PR_AUC']:.4f} | "
        f"FP={metrics['FP']}"
    )


stunting_optimization = pd.DataFrame(
    stunting_results
)


# =============================================================================
# 11. UNDERWEIGHT OPTIMIZATION
# =============================================================================

print("\n" + "=" * 100)
print("8. UNDERWEIGHT — XGBOOST OPTIMIZATION")
print("=" * 100)

underweight_results = []

for config in configurations:

    print(
        f"\nRunning {config['name']}..."
    )

    metrics = evaluate_configuration(
        Xtr_under,
        Xv_under,
        y_train_under,
        y_val_under,
        under_weight,
        config
    )

    metrics["Model"] = config["name"]

    underweight_results.append(
        metrics
    )

    print(
        f"Precision={metrics['Precision']:.4f} | "
        f"Recall={metrics['Recall']:.4f} | "
        f"F1={metrics['F1']:.4f} | "
        f"ROC-AUC={metrics['ROC_AUC']:.4f} | "
        f"PR-AUC={metrics['PR_AUC']:.4f} | "
        f"FP={metrics['FP']}"
    )


underweight_optimization = pd.DataFrame(
    underweight_results
)


# =============================================================================
# 12. STUNTING RESULTS
# =============================================================================

print("\n" + "=" * 100)
print("9. STUNTING — OPTIMIZATION RESULTS")
print("=" * 100)

display(
    stunting_optimization[
        [
            "Model",
            "Accuracy",
            "Precision",
            "Recall",
            "F1",
            "ROC_AUC",
            "PR_AUC",
            "TN",
            "FP",
            "FN",
            "TP"
        ]
    ]
    .sort_values(
        "F1",
        ascending=False
    )
    .round(4)
)


# =============================================================================
# 13. UNDERWEIGHT RESULTS
# =============================================================================

print("\n" + "=" * 100)
print("10. UNDERWEIGHT — OPTIMIZATION RESULTS")
print("=" * 100)

display(
    underweight_optimization[
        [
            "Model",
            "Accuracy",
            "Precision",
            "Recall",
            "F1",
            "ROC_AUC",
            "PR_AUC",
            "TN",
            "FP",
            "FN",
            "TP"
        ]
    ]
    .sort_values(
        "F1",
        ascending=False
    )
    .round(4)
)


# =============================================================================
# 14. PRECISION-FOCUSED RANKING
# =============================================================================

print("\n" + "=" * 100)
print("11. PRECISION-FOCUSED RANKING")
print("=" * 100)


print("\nSTUNTING — PRECISION RANKING")
print("-" * 100)

display(
    stunting_optimization[
        [
            "Model",
            "Precision",
            "Recall",
            "F1",
            "PR_AUC",
            "FP"
        ]
    ]
    .sort_values(
        [
            "Precision",
            "F1"
        ],
        ascending=False
    )
    .round(4)
)


print("\nUNDERWEIGHT — PRECISION RANKING")
print("-" * 100)

display(
    underweight_optimization[
        [
            "Model",
            "Precision",
            "Recall",
            "F1",
            "PR_AUC",
            "FP"
        ]
    ]
    .sort_values(
        [
            "Precision",
            "F1"
        ],
        ascending=False
    )
    .round(4)
)


# =============================================================================
# 15. BEST MODEL WITH RECALL >= 30%
# =============================================================================

print("\n" + "=" * 100)
print("12. BEST MODEL WITH RECALL >= 30%")
print("=" * 100)


stunt_candidates = stunting_optimization[
    stunting_optimization["Recall"] >= 0.30
]

under_candidates = underweight_optimization[
    underweight_optimization["Recall"] >= 0.30
]


print("\nSTUNTING")
print("-" * 100)

if len(stunt_candidates) > 0:

    best_stunt = (
        stunt_candidates
        .sort_values(
            [
                "Precision",
                "F1"
            ],
            ascending=False
        )
        .iloc[0]
    )

    print(
        "Model     :",
        best_stunt["Model"]
    )

    print(
        f"Precision : "
        f"{best_stunt['Precision']:.4f}"
    )

    print(
        f"Recall    : "
        f"{best_stunt['Recall']:.4f}"
    )

    print(
        f"F1        : "
        f"{best_stunt['F1']:.4f}"
    )

    print(
        f"PR-AUC    : "
        f"{best_stunt['PR_AUC']:.4f}"
    )

    print(
        f"FP        : "
        f"{int(best_stunt['FP'])}"
    )

else:

    print(
        "No model achieved recall >= 30%."
    )


print("\nUNDERWEIGHT")
print("-" * 100)

if len(under_candidates) > 0:

    best_under = (
        under_candidates
        .sort_values(
            [
                "Precision",
                "F1"
            ],
            ascending=False
        )
        .iloc[0]
    )

    print(
        "Model     :",
        best_under["Model"]
    )

    print(
        f"Precision : "
        f"{best_under['Precision']:.4f}"
    )

    print(
        f"Recall    : "
        f"{best_under['Recall']:.4f}"
    )

    print(
        f"F1        : "
        f"{best_under['F1']:.4f}"
    )

    print(
        f"PR-AUC    : "
        f"{best_under['PR_AUC']:.4f}"
    )

    print(
        f"FP        : "
        f"{int(best_under['FP'])}"
    )

else:

    print(
        "No model achieved recall >= 30%."
    )


# =============================================================================
# 16. BEST F1 MODEL
# =============================================================================

best_stunt_f1 = (
    stunting_optimization
    .sort_values(
        "F1",
        ascending=False
    )
    .iloc[0]
)

best_under_f1 = (
    underweight_optimization
    .sort_values(
        "F1",
        ascending=False
    )
    .iloc[0]
)


print("\n" + "=" * 100)
print("13. BEST F1 MODEL")
print("=" * 100)


print("\nSTUNTING")
print(
    f"Model     : {best_stunt_f1['Model']}"
)
print(
    f"Precision : {best_stunt_f1['Precision']:.4f}"
)
print(
    f"Recall    : {best_stunt_f1['Recall']:.4f}"
)
print(
    f"F1        : {best_stunt_f1['F1']:.4f}"
)
print(
    f"PR-AUC    : {best_stunt_f1['PR_AUC']:.4f}"
)


print("\nUNDERWEIGHT")
print(
    f"Model     : {best_under_f1['Model']}"
)
print(
    f"Precision : {best_under_f1['Precision']:.4f}"
)
print(
    f"Recall    : {best_under_f1['Recall']:.4f}"
)
print(
    f"F1        : {best_under_f1['F1']:.4f}"
)
print(
    f"PR-AUC    : {best_under_f1['PR_AUC']:.4f}"
)


# =============================================================================
# 17. INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("STEP 7I COMPLETE")
print("=" * 100)

print("✓ Fixed 15-predictor set used")
print("✓ Training-only preprocessing used")
print("✓ Multiple controlled XGBoost configurations tested")
print("✓ Learning rate investigated")
print("✓ Tree depth investigated")
print("✓ Minimum child weight investigated")
print("✓ L1 regularization investigated")
print("✓ L2 regularization investigated")
print("✓ Subsampling investigated")
print("✓ Validation data used for model comparison")
print("✓ Test set remained completely untouched")
print("✓ No final threshold selected")

print("\nNEXT STEP:")
print("STEP 7J — VALIDATION THRESHOLD OPTIMIZATION")
print(
    "Use the best validated model configuration from Step 7I "
    "to identify a precision-focused operating threshold."
)
```

```python
# =============================================================================
# STEP 7J: CAR MICS6 — CONTROLLED VALIDATION THRESHOLD OPTIMIZATION
# =============================================================================
#
# PURPOSE:
#   Select validation-based operating thresholds for the winning Step 7I
#   XGBoost configurations, with emphasis on:
#
#       1. Increasing precision
#       2. Reducing false positives
#       3. Retaining useful recall
#       4. Avoiding a trivial "precision = 1.00, recall ≈ 0" solution
#
# IMPORTANT:
#   • TRAINING + VALIDATION ONLY
#   • TEST SET IS NOT ACCESSED
#   • TEST SET IS NOT USED FOR THRESHOLD SELECTION
#   • Predictor set remains the 15 predictors from Step 7H
#   • Models are the winners from Step 7I
#
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)


print("=" * 100)
print("STEP 7J: CAR MICS6 — CONTROLLED VALIDATION THRESHOLD OPTIMIZATION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

required_objects = [
    "X_train_stunt",
    "X_val_stunt",
    "y_train_stunt",
    "y_val_stunt",
    "X_train_under",
    "X_val_under",
    "y_train_under",
    "y_val_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)

if missing_objects:
    print("Missing objects:")
    for obj in missing_objects:
        print(" -", obj)

    raise NameError(
        "Required Step 4.3 training/validation objects are missing."
    )

print("✓ Training and validation objects available")


# =============================================================================
# 2. FIXED 15-PREDICTOR SET
# =============================================================================

predictors_15 = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cdisability",
    "CA1",
    "CA14",
    "BD2",
    "IM2",
    "CA31"
]

print("\n2. FIXED PREDICTOR SET")
print("-" * 100)

print("Number of predictors:", len(predictors_15))

for i, p in enumerate(predictors_15, start=1):
    print(f"{i:2d}. {p}")


# =============================================================================
# 3. PREDICTOR AVAILABILITY
# =============================================================================

missing_stunt = [
    p for p in predictors_15
    if p not in X_train_stunt.columns
]

missing_under = [
    p for p in predictors_15
    if p not in X_train_under.columns
]

if missing_stunt:
    raise KeyError(
        "Missing stunting predictors: " + str(missing_stunt)
    )

if missing_under:
    raise KeyError(
        "Missing underweight predictors: " + str(missing_under)
    )

print("\n✓ All 15 predictors available for stunting")
print("✓ All 15 predictors available for underweight")


# =============================================================================
# 4. TEST-SET PROTECTION
# =============================================================================

print("\n" + "=" * 100)
print("3. TEST-SET PROTECTION")
print("=" * 100)

print("✓ Test predictors are NOT accessed")
print("✓ Test targets are NOT accessed")
print("✓ Test probabilities are NOT calculated")
print("✓ Test thresholds are NOT evaluated")
print("✓ Test performance is NOT calculated")
print("✓ Test set remains completely unseen")


# =============================================================================
# 5. PREPROCESSING
# =============================================================================

numeric_predictors = [
    "CAGE"
]

categorical_predictors = [
    p for p in predictors_15
    if p != "CAGE"
]


def make_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            numeric_predictors
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_predictors
        )
    ])


print("\n" + "=" * 100)
print("4. TRAINING-ONLY PREPROCESSING")
print("=" * 100)


# -------------------------------------------------------------------------
# STUNTING
# -------------------------------------------------------------------------

stunt_preprocessor_7j = make_preprocessor()

Xtr_stunt_7j = stunt_preprocessor_7j.fit_transform(
    X_train_stunt[predictors_15]
)

Xv_stunt_7j = stunt_preprocessor_7j.transform(
    X_val_stunt[predictors_15]
)


# -------------------------------------------------------------------------
# UNDERWEIGHT
# -------------------------------------------------------------------------

under_preprocessor_7j = make_preprocessor()

Xtr_under_7j = under_preprocessor_7j.fit_transform(
    X_train_under[predictors_15]
)

Xv_under_7j = under_preprocessor_7j.transform(
    X_val_under[predictors_15]
)


print(
    "Stunting training processed   :",
    Xtr_stunt_7j.shape
)

print(
    "Stunting validation processed :",
    Xv_stunt_7j.shape
)

print(
    "Underweight training processed:",
    Xtr_under_7j.shape
)

print(
    "Underweight validation processed:",
    Xv_under_7j.shape
)

print("✓ Preprocessors fitted using training data only")


# =============================================================================
# 6. CLASS WEIGHTS
# =============================================================================

stunt_negative = (
    y_train_stunt == 0
).sum()

stunt_positive = (
    y_train_stunt == 1
).sum()

under_negative = (
    y_train_under == 0
).sum()

under_positive = (
    y_train_under == 1
).sum()


stunt_weight = (
    stunt_negative / stunt_positive
)

under_weight = (
    under_negative / under_positive
)


print("\n" + "=" * 100)
print("5. CLASS IMBALANCE")
print("=" * 100)

print(
    f"Stunting scale_pos_weight   : {stunt_weight:.4f}"
)

print(
    f"Underweight scale_pos_weight: {under_weight:.4f}"
)


# =============================================================================
# 7. WINNING STEP 7I CONFIGURATIONS
# =============================================================================
#
# Stunting winner:
#   M7_L2_Strong
#
# Underweight winner:
#   M6_L1_Strong
#
# These are taken directly from Step 7I.
# =============================================================================

stunting_config = {
    "name": "M7_L2_Strong",
    "n_estimators": 700,
    "learning_rate": 0.02,
    "max_depth": 3,
    "min_child_weight": 8,
    "subsample": 0.80,
    "colsample_bytree": 0.80,
    "reg_alpha": 0.5,
    "reg_lambda": 5.0
}


underweight_config = {
    "name": "M6_L1_Strong",
    "n_estimators": 700,
    "learning_rate": 0.02,
    "max_depth": 3,
    "min_child_weight": 8,
    "subsample": 0.80,
    "colsample_bytree": 0.80,
    "reg_alpha": 1.0,
    "reg_lambda": 3.0
}


print("\n" + "=" * 100)
print("6. SELECTED STEP 7I MODEL CONFIGURATIONS")
print("=" * 100)

print("\nSTUNTING")
print("-" * 100)

for key, value in stunting_config.items():
    print(f"{key:20s}: {value}")


print("\nUNDERWEIGHT")
print("-" * 100)

for key, value in underweight_config.items():
    print(f"{key:20s}: {value}")


# =============================================================================
# 8. TRAIN FINAL DEVELOPMENT MODELS
# =============================================================================

print("\n" + "=" * 100)
print("7. FITTING STEP 7I WINNING MODELS")
print("=" * 100)


# -------------------------------------------------------------------------
# STUNTING MODEL
# -------------------------------------------------------------------------

stunting_model_7j = XGBClassifier(
    n_estimators=stunting_config["n_estimators"],
    learning_rate=stunting_config["learning_rate"],
    max_depth=stunting_config["max_depth"],
    min_child_weight=stunting_config["min_child_weight"],
    subsample=stunting_config["subsample"],
    colsample_bytree=stunting_config["colsample_bytree"],
    reg_alpha=stunting_config["reg_alpha"],
    reg_lambda=stunting_config["reg_lambda"],
    objective="binary:logistic",
    eval_metric="logloss",
    scale_pos_weight=stunt_weight,
    random_state=42,
    n_jobs=-1
)


stunting_model_7j.fit(
    Xtr_stunt_7j,
    y_train_stunt
)

print("✓ Stunting M7_L2_Strong fitted")


# -------------------------------------------------------------------------
# UNDERWEIGHT MODEL
# -------------------------------------------------------------------------

underweight_model_7j = XGBClassifier(
    n_estimators=underweight_config["n_estimators"],
    learning_rate=underweight_config["learning_rate"],
    max_depth=underweight_config["max_depth"],
    min_child_weight=underweight_config["min_child_weight"],
    subsample=underweight_config["subsample"],
    colsample_bytree=underweight_config["colsample_bytree"],
    reg_alpha=underweight_config["reg_alpha"],
    reg_lambda=underweight_config["reg_lambda"],
    objective="binary:logistic",
    eval_metric="logloss",
    scale_pos_weight=under_weight,
    random_state=42,
    n_jobs=-1
)


underweight_model_7j.fit(
    Xtr_under_7j,
    y_train_under
)

print("✓ Underweight M6_L1_Strong fitted")


# =============================================================================
# 9. VALIDATION PROBABILITIES
# =============================================================================

print("\n" + "=" * 100)
print("8. VALIDATION PROBABILITY GENERATION")
print("=" * 100)


stunting_val_prob_7j = (
    stunting_model_7j
    .predict_proba(Xv_stunt_7j)[:, 1]
)

underweight_val_prob_7j = (
    underweight_model_7j
    .predict_proba(Xv_under_7j)[:, 1]
)


print(
    "Stunting validation probabilities :",
    len(stunting_val_prob_7j)
)

print(
    "Underweight validation probabilities:",
    len(underweight_val_prob_7j)
)

print("✓ Validation probabilities generated")
print("✓ No test probabilities generated")


# =============================================================================
# 10. THRESHOLD GRID
# =============================================================================
#
# We begin at 0.50 because Step 7I used the default threshold.
#
# Then increase the threshold to reduce false positives.
#
# =============================================================================

thresholds = np.round(
    np.arange(
        0.50,
        0.91,
        0.01
    ),
    2
)

print("\n" + "=" * 100)
print("9. VALIDATION THRESHOLD GRID")
print("=" * 100)

print(
    "Threshold range:",
    f"{thresholds.min():.2f}",
    "to",
    f"{thresholds.max():.2f}"
)

print(
    "Number of thresholds:",
    len(thresholds)
)

print("✓ Thresholds will be evaluated using validation data only")


# =============================================================================
# 11. THRESHOLD EVALUATION FUNCTION
# =============================================================================

def evaluate_thresholds(
    y_true,
    probabilities,
    thresholds
):

    results = []

    for threshold in thresholds:

        predictions = (
            probabilities >= threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_true,
            predictions,
            labels=[0, 1]
        ).ravel()

        precision = precision_score(
            y_true,
            predictions,
            zero_division=0
        )

        recall = recall_score(
            y_true,
            predictions,
            zero_division=0
        )

        f1 = f1_score(
            y_true,
            predictions,
            zero_division=0
        )

        accuracy = accuracy_score(
            y_true,
            predictions
        )

        fpr = (
            fp / (fp + tn)
            if (fp + tn) > 0
            else 0
        )

        results.append({
            "Threshold": threshold,
            "Accuracy": accuracy,
            "Precision": precision,
            "Recall": recall,
            "F1": f1,
            "FPR": fpr,
            "TN": tn,
            "FP": fp,
            "FN": fn,
            "TP": tp
        })

    return pd.DataFrame(results)


# =============================================================================
# 12. STUNTING THRESHOLD ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("10. STUNTING — VALIDATION THRESHOLD ANALYSIS")
print("=" * 100)


stunting_threshold_results = evaluate_thresholds(
    y_val_stunt,
    stunting_val_prob_7j,
    thresholds
)

stunting_threshold_results["ROC_AUC"] = roc_auc_score(
    y_val_stunt,
    stunting_val_prob_7j
)

stunting_threshold_results["PR_AUC"] = average_precision_score(
    y_val_stunt,
    stunting_val_prob_7j
)


display(
    stunting_threshold_results.round(4)
)


# =============================================================================
# 13. UNDERWEIGHT THRESHOLD ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("11. UNDERWEIGHT — VALIDATION THRESHOLD ANALYSIS")
print("=" * 100)


underweight_threshold_results = evaluate_thresholds(
    y_val_under,
    underweight_val_prob_7j,
    thresholds
)

underweight_threshold_results["ROC_AUC"] = roc_auc_score(
    y_val_under,
    underweight_val_prob_7j
)

underweight_threshold_results["PR_AUC"] = average_precision_score(
    y_val_under,
    underweight_val_prob_7j
)


display(
    underweight_threshold_results.round(4)
)


# =============================================================================
# 14. PRECISION TARGET ANALYSIS
# =============================================================================
#
# IMPORTANT:
# Do not select the highest precision blindly.
#
# We require minimum recall to avoid a trivial classifier.
#
# =============================================================================

precision_targets = [
    0.50,
    0.55,
    0.60,
    0.65,
    0.70
]

minimum_recall = 0.30


def precision_target_table(
    results,
    targets,
    min_recall=0.30
):

    rows = []

    for target in targets:

        eligible = results[
            (results["Precision"] >= target) &
            (results["Recall"] >= min_recall)
        ].copy()

        if len(eligible) == 0:

            rows.append({
                "Required_Precision": target,
                "Achievable": "NO",
                "Threshold": np.nan,
                "Precision": np.nan,
                "Recall": np.nan,
                "F1": np.nan,
                "FPR": np.nan,
                "FP": np.nan
            })

        else:

            # Among thresholds satisfying the precision and recall
            # constraints, choose the one with the highest F1.
            best = (
                eligible
                .sort_values(
                    [
                        "F1",
                        "Precision"
                    ],
                    ascending=False
                )
                .iloc[0]
            )

            rows.append({
                "Required_Precision": target,
                "Achievable": "YES",
                "Threshold": best["Threshold"],
                "Precision": best["Precision"],
                "Recall": best["Recall"],
                "F1": best["F1"],
                "FPR": best["FPR"],
                "FP": best["FP"]
            })

    return pd.DataFrame(rows)


print("\n" + "=" * 100)
print("12. STUNTING — PRECISION TARGETS WITH RECALL >= 30%")
print("=" * 100)


stunting_precision_targets = precision_target_table(
    stunting_threshold_results,
    precision_targets,
    minimum_recall
)

display(
    stunting_precision_targets.round(4)
)


print("\n" + "=" * 100)
print("13. UNDERWEIGHT — PRECISION TARGETS WITH RECALL >= 30%")
print("=" * 100)


underweight_precision_targets = precision_target_table(
    underweight_threshold_results,
    precision_targets,
    minimum_recall
)

display(
    underweight_precision_targets.round(4)
)


# =============================================================================
# 15. BEST F1 OPERATING POINT
# =============================================================================

print("\n" + "=" * 100)
print("14. BEST F1 OPERATING POINTS")
print("=" * 100)


stunting_best_f1 = (
    stunting_threshold_results[
        stunting_threshold_results["Recall"] >= minimum_recall
    ]
    .sort_values(
        [
            "F1",
            "Precision"
        ],
        ascending=False
    )
    .iloc[0]
)


underweight_best_f1 = (
    underweight_threshold_results[
        underweight_threshold_results["Recall"] >= minimum_recall
    ]
    .sort_values(
        [
            "F1",
            "Precision"
        ],
        ascending=False
    )
    .iloc[0]
)


print("\nSTUNTING")
print("-" * 100)

print(
    f"Threshold : {stunting_best_f1['Threshold']:.2f}"
)

print(
    f"Precision : {stunting_best_f1['Precision']:.4f}"
)

print(
    f"Recall    : {stunting_best_f1['Recall']:.4f}"
)

print(
    f"F1        : {stunting_best_f1['F1']:.4f}"
)

print(
    f"FPR       : {stunting_best_f1['FPR']:.4f}"
)

print(
    f"FP        : {int(stunting_best_f1['FP'])}"
)


print("\nUNDERWEIGHT")
print("-" * 100)

print(
    f"Threshold : {underweight_best_f1['Threshold']:.2f}"
)

print(
    f"Precision : {underweight_best_f1['Precision']:.4f}"
)

print(
    f"Recall    : {underweight_best_f1['Recall']:.4f}"
)

print(
    f"F1        : {underweight_best_f1['F1']:.4f}"
)

print(
    f"FPR       : {underweight_best_f1['FPR']:.4f}"
)

print(
    f"FP        : {int(underweight_best_f1['FP'])}"
)


# =============================================================================
# 16. BEST PRECISION WITH RECALL >= 30%
# =============================================================================

print("\n" + "=" * 100)
print("15. BEST PRECISION WHILE RETAINING RECALL >= 30%")
print("=" * 100)


stunting_recall_constrained = (
    stunting_threshold_results[
        stunting_threshold_results["Recall"] >= minimum_recall
    ]
    .sort_values(
        [
            "Precision",
            "F1"
        ],
        ascending=False
    )
    .iloc[0]
)


underweight_recall_constrained = (
    underweight_threshold_results[
        underweight_threshold_results["Recall"] >= minimum_recall
    ]
    .sort_values(
        [
            "Precision",
            "F1"
        ],
        ascending=False
    )
    .iloc[0]
)


print("\nSTUNTING")
print("-" * 100)

print(
    f"Threshold : {stunting_recall_constrained['Threshold']:.2f}"
)

print(
    f"Precision : {stunting_recall_constrained['Precision']:.4f}"
)

print(
    f"Recall    : {stunting_recall_constrained['Recall']:.4f}"
)

print(
    f"F1        : {stunting_recall_constrained['F1']:.4f}"
)

print(
    f"FPR       : {stunting_recall_constrained['FPR']:.4f}"
)

print(
    f"FP        : {int(stunting_recall_constrained['FP'])}"
)


print("\nUNDERWEIGHT")
print("-" * 100)

print(
    f"Threshold : {underweight_recall_constrained['Threshold']:.2f}"
)

print(
    f"Precision : {underweight_recall_constrained['Precision']:.4f}"
)

print(
    f"Recall    : {underweight_recall_constrained['Recall']:.4f}"
)

print(
    f"F1        : {underweight_recall_constrained['F1']:.4f}"
)

print(
    f"FPR       : {underweight_recall_constrained['FPR']:.4f}"
)

print(
    f"FP        : {int(underweight_recall_constrained['FP'])}"
)


# =============================================================================
# 17. COMPARE DEFAULT 0.50 WITH SELECTED OPERATING POINTS
# =============================================================================

print("\n" + "=" * 100)
print("16. DEFAULT THRESHOLD VS PRECISION-FOCUSED OPERATING POINT")
print("=" * 100)


def get_threshold_row(
    results,
    threshold
):

    row = results[
        results["Threshold"] == threshold
    ]

    if len(row) == 0:
        return None

    return row.iloc[0]


stunt_default = get_threshold_row(
    stunting_threshold_results,
    0.50
)

under_default = get_threshold_row(
    underweight_threshold_results,
    0.50
)


comparison = pd.DataFrame([
    {
        "Target": "Stunting",
        "Operating_Point": "Default_0.50",
        "Threshold": stunt_default["Threshold"],
        "Precision": stunt_default["Precision"],
        "Recall": stunt_default["Recall"],
        "F1": stunt_default["F1"],
        "FPR": stunt_default["FPR"],
        "FP": stunt_default["FP"]
    },
    {
        "Target": "Stunting",
        "Operating_Point": "Precision_Focused",
        "Threshold": stunting_recall_constrained["Threshold"],
        "Precision": stunting_recall_constrained["Precision"],
        "Recall": stunting_recall_constrained["Recall"],
        "F1": stunting_recall_constrained["F1"],
        "FPR": stunting_recall_constrained["FPR"],
        "FP": stunting_recall_constrained["FP"]
    },
    {
        "Target": "Underweight",
        "Operating_Point": "Default_0.50",
        "Threshold": under_default["Threshold"],
        "Precision": under_default["Precision"],
        "Recall": under_default["Recall"],
        "F1": under_default["F1"],
        "FPR": under_default["FPR"],
        "FP": under_default["FP"]
    },
    {
        "Target": "Underweight",
        "Operating_Point": "Precision_Focused",
        "Threshold": underweight_recall_constrained["Threshold"],
        "Precision": underweight_recall_constrained["Precision"],
        "Recall": underweight_recall_constrained["Recall"],
        "F1": underweight_recall_constrained["F1"],
        "FPR": underweight_recall_constrained["FPR"],
        "FP": underweight_recall_constrained["FP"]
    }
])


display(
    comparison.round(4)
)


# =============================================================================
# 18. ROC-AUC AND PR-AUC
# =============================================================================

stunting_roc_auc_7j = roc_auc_score(
    y_val_stunt,
    stunting_val_prob_7j
)

stunting_pr_auc_7j = average_precision_score(
    y_val_stunt,
    stunting_val_prob_7j
)

underweight_roc_auc_7j = roc_auc_score(
    y_val_under,
    underweight_val_prob_7j
)

underweight_pr_auc_7j = average_precision_score(
    y_val_under,
    underweight_val_prob_7j
)


print("\n" + "=" * 100)
print("17. DISCRIMINATION PERFORMANCE")
print("=" * 100)

print("\nSTUNTING")
print(
    f"ROC-AUC : {stunting_roc_auc_7j:.4f}"
)
print(
    f"PR-AUC  : {stunting_pr_auc_7j:.4f}"
)

print("\nUNDERWEIGHT")
print(
    f"ROC-AUC : {underweight_roc_auc_7j:.4f}"
)
print(
    f"PR-AUC  : {underweight_pr_auc_7j:.4f}"
)


# =============================================================================
# 19. FINAL STEP 7J SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("18. STEP 7J SUMMARY")
print("=" * 100)

print("\nSTUNTING")
print("-" * 100)

print("Model     : M7_L2_Strong")

print(
    f"ROC-AUC   : {stunting_roc_auc_7j:.4f}"
)

print(
    f"PR-AUC    : {stunting_pr_auc_7j:.4f}"
)

print(
    f"Selected threshold candidate : "
    f"{stunting_recall_constrained['Threshold']:.2f}"
)

print(
    f"Precision : "
    f"{stunting_recall_constrained['Precision']:.4f}"
)

print(
    f"Recall    : "
    f"{stunting_recall_constrained['Recall']:.4f}"
)

print(
    f"F1        : "
    f"{stunting_recall_constrained['F1']:.4f}"
)

print(
    f"FP        : "
    f"{int(stunting_recall_constrained['FP'])}"
)


print("\nUNDERWEIGHT")
print("-" * 100)

print("Model     : M6_L1_Strong")

print(
    f"ROC-AUC   : {underweight_roc_auc_7j:.4f}"
)

print(
    f"PR-AUC    : {underweight_pr_auc_7j:.4f}"
)

print(
    f"Selected threshold candidate : "
    f"{underweight_recall_constrained['Threshold']:.2f}"
)

print(
    f"Precision : "
    f"{underweight_recall_constrained['Precision']:.4f}"
)

print(
    f"Recall    : "
    f"{underweight_recall_constrained['Recall']:.4f}"
)

print(
    f"F1        : "
    f"{underweight_recall_constrained['F1']:.4f}"
)

print(
    f"FP        : "
    f"{int(underweight_recall_constrained['FP'])}"
)


# =============================================================================
# 20. INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("STEP 7J COMPLETE")
print("=" * 100)

print("✓ Step 7I winning models used")
print("✓ Stunting: M7_L2_Strong")
print("✓ Underweight: M6_L1_Strong")
print("✓ Fixed 15-predictor set retained")
print("✓ Training-only preprocessing used")
print("✓ Validation probabilities generated")
print("✓ Thresholds evaluated using validation data only")
print("✓ Precision targets evaluated")
print("✓ Minimum recall constraint applied")
print("✓ False-positive rate calculated")
print("✓ False-positive counts calculated")
print("✓ ROC-AUC calculated")
print("✓ PR-AUC calculated")
print("✓ Test predictors NOT accessed")
print("✓ Test targets NOT accessed")
print("✓ Test probabilities NOT generated")
print("✓ Test threshold NOT selected")
print("✓ Test set remains completely unseen")

print("\nNEXT STEP:")
print("STEP 8 — FINAL MODEL FITTING AND VALIDATION CONFIRMATION")
print(
    "After reviewing the Step 7J output, "
    "the final validation-based thresholds will be fixed "
    "before the unseen test evaluation."
)
```

```python
# =============================================================================
# STEP 7K: CAR MICS6 — CONTROLLED CLASS-WEIGHT OPTIMIZATION
# =============================================================================
#
# OBJECTIVE:
#   Reduce false positives and improve precision without destroying recall.
#
# DESIGN:
#   • Same 15-predictor set selected in Step 7H
#   • Same strong XGBoost configurations selected in Step 7I
#   • Vary scale_pos_weight only
#   • Fit using TRAINING data only
#   • Evaluate using VALIDATION data only
#   • No TEST data accessed
#   • No test threshold selection
#
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

print("=" * 100)
print("STEP 7K: CAR MICS6 — CONTROLLED CLASS-WEIGHT OPTIMIZATION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)

required_objects = [
    "X_train_stunt",
    "X_val_stunt",
    "y_train_stunt",
    "y_val_stunt",
    "X_train_under",
    "X_val_under",
    "y_train_under",
    "y_val_under"
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    print("Missing objects:")
    for name in missing_objects:
        print(" -", name)

    raise NameError(
        "Step 7K cannot continue because required Step 4.3 "
        "training/validation objects are missing."
    )

print("✓ All required training/validation objects are available")


# =============================================================================
# 2. FIXED 15-PREDICTOR SET
# =============================================================================

predictors_15 = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cdisability",
    "CA1",
    "CA14",
    "BD2",
    "IM2",
    "CA31"
]

print("\n2. FIXED PREDICTOR SET")
print("-" * 100)

print("Number of predictors:", len(predictors_15))

for i, predictor in enumerate(predictors_15, 1):
    print(f"{i:2d}. {predictor}")


# =============================================================================
# 3. PREDICTOR AVAILABILITY CHECK
# =============================================================================

missing_stunt_predictors = [
    p for p in predictors_15
    if p not in X_train_stunt.columns
]

missing_under_predictors = [
    p for p in predictors_15
    if p not in X_train_under.columns
]

if missing_stunt_predictors:
    raise KeyError(
        "Missing stunting predictors: "
        + str(missing_stunt_predictors)
    )

if missing_under_predictors:
    raise KeyError(
        "Missing underweight predictors: "
        + str(missing_under_predictors)
    )

print("✓ All 15 predictors available for stunting")
print("✓ All 15 predictors available for underweight")


# =============================================================================
# 4. TEST-SET PROTECTION
# =============================================================================

print("\n" + "=" * 100)
print("3. TEST-SET PROTECTION")
print("=" * 100)

print("✓ Test predictors are NOT accessed")
print("✓ Test targets are NOT accessed")
print("✓ Test probabilities are NOT generated")
print("✓ Test performance is NOT calculated")
print("✓ Test thresholds are NOT selected")
print("✓ Test set remains completely unseen")


# =============================================================================
# 5. PREDICTOR TYPES
# =============================================================================

numeric_predictors = [
    "CAGE"
]

categorical_predictors = [
    p for p in predictors_15
    if p != "CAGE"
]


# =============================================================================
# 6. PREPROCESSOR FUNCTION
# =============================================================================

def create_step7k_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            numeric_predictors
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_predictors
        )
    ])


# =============================================================================
# 7. TRAINING-ONLY PREPROCESSING
# =============================================================================

print("\n" + "=" * 100)
print("4. TRAINING-ONLY PREPROCESSING")
print("=" * 100)


# STUNTING
stunt_preprocessor_7k = create_step7k_preprocessor()

X_train_stunt_7k = stunt_preprocessor_7k.fit_transform(
    X_train_stunt[predictors_15]
)

X_val_stunt_7k = stunt_preprocessor_7k.transform(
    X_val_stunt[predictors_15]
)


# UNDERWEIGHT
under_preprocessor_7k = create_step7k_preprocessor()

X_train_under_7k = under_preprocessor_7k.fit_transform(
    X_train_under[predictors_15]
)

X_val_under_7k = under_preprocessor_7k.transform(
    X_val_under[predictors_15]
)


print(
    "Stunting training   :",
    X_train_stunt_7k.shape
)

print(
    "Stunting validation :",
    X_val_stunt_7k.shape
)

print(
    "Underweight training   :",
    X_train_under_7k.shape
)

print(
    "Underweight validation :",
    X_val_under_7k.shape
)

print("✓ Preprocessors fitted on training data only")


# =============================================================================
# 8. BASELINE CLASS WEIGHTS
# =============================================================================

stunt_negative = int(
    (y_train_stunt == 0).sum()
)

stunt_positive = int(
    (y_train_stunt == 1).sum()
)

under_negative = int(
    (y_train_under == 0).sum()
)

under_positive = int(
    (y_train_under == 1).sum()
)


stunt_base_weight = (
    stunt_negative / stunt_positive
)

under_base_weight = (
    under_negative / under_positive
)


print("\n" + "=" * 100)
print("5. BASELINE CLASS WEIGHTS")
print("=" * 100)

print(
    f"Stunting   : {stunt_base_weight:.4f}"
)

print(
    f"Underweight: {under_base_weight:.4f}"
)


# =============================================================================
# 9. CLASS-WEIGHT GRIDS
# =============================================================================
#
# The purpose is to determine whether the previous weighting was making
# the models too aggressive in predicting class 1.
#
# Lower values should generally make positive predictions less aggressive,
# which may reduce FP and improve precision.
#
# These are VALIDATION experiments only.
# =============================================================================

stunting_weights = [
    1.00,
    1.25,
    1.50,
    1.75,
    2.00
]

underweight_weights = [
    1.00,
    1.25,
    1.50,
    1.75,
    2.00,
    2.25,
    2.50,
    2.75
]


print("\n" + "=" * 100)
print("6. CLASS-WEIGHT SEARCH GRID")
print("=" * 100)

print(
    "Stunting weights   :",
    stunting_weights
)

print(
    "Underweight weights:",
    underweight_weights
)


# =============================================================================
# 10. WINNING STEP 7I MODEL PARAMETERS
# =============================================================================

stunting_model_parameters = {
    "n_estimators": 700,
    "learning_rate": 0.02,
    "max_depth": 3,
    "min_child_weight": 8,
    "subsample": 0.80,
    "colsample_bytree": 0.80,
    "reg_alpha": 0.5,
    "reg_lambda": 5.0
}


underweight_model_parameters = {
    "n_estimators": 700,
    "learning_rate": 0.02,
    "max_depth": 3,
    "min_child_weight": 8,
    "subsample": 0.80,
    "colsample_bytree": 0.80,
    "reg_alpha": 1.0,
    "reg_lambda": 3.0
}


# =============================================================================
# 11. TRAIN/EVALUATE STUNTING WEIGHTS
# =============================================================================

print("\n" + "=" * 100)
print("7. STUNTING — CLASS-WEIGHT OPTIMIZATION")
print("=" * 100)

stunting_weight_results = []

for weight in stunting_weights:

    print(
        f"\nTesting scale_pos_weight = {weight:.2f}"
    )

    model = XGBClassifier(
        n_estimators=stunting_model_parameters["n_estimators"],
        learning_rate=stunting_model_parameters["learning_rate"],
        max_depth=stunting_model_parameters["max_depth"],
        min_child_weight=stunting_model_parameters["min_child_weight"],
        subsample=stunting_model_parameters["subsample"],
        colsample_bytree=stunting_model_parameters["colsample_bytree"],
        reg_alpha=stunting_model_parameters["reg_alpha"],
        reg_lambda=stunting_model_parameters["reg_lambda"],
        scale_pos_weight=weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_stunt_7k,
        y_train_stunt
    )

    probabilities = model.predict_proba(
        X_val_stunt_7k
    )[:, 1]

    predictions = (
        probabilities >= 0.50
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val_stunt,
        predictions,
        labels=[0, 1]
    ).ravel()

    precision = precision_score(
        y_val_stunt,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_val_stunt,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_val_stunt,
        predictions,
        zero_division=0
    )

    accuracy = accuracy_score(
        y_val_stunt,
        predictions
    )

    fpr = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0
    )

    roc_auc = roc_auc_score(
        y_val_stunt,
        probabilities
    )

    pr_auc = average_precision_score(
        y_val_stunt,
        probabilities
    )

    stunting_weight_results.append({
        "Scale_Pos_Weight": weight,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "ROC_AUC": roc_auc,
        "PR_AUC": pr_auc,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })

    print(
        f"Precision={precision:.4f} | "
        f"Recall={recall:.4f} | "
        f"F1={f1:.4f} | "
        f"ROC-AUC={roc_auc:.4f} | "
        f"PR-AUC={pr_auc:.4f} | "
        f"FP={fp}"
    )


stunting_weight_results = pd.DataFrame(
    stunting_weight_results
)


# =============================================================================
# 12. TRAIN/EVALUATE UNDERWEIGHT WEIGHTS
# =============================================================================

print("\n" + "=" * 100)
print("8. UNDERWEIGHT — CLASS-WEIGHT OPTIMIZATION")
print("=" * 100)

underweight_weight_results = []

for weight in underweight_weights:

    print(
        f"\nTesting scale_pos_weight = {weight:.2f}"
    )

    model = XGBClassifier(
        n_estimators=underweight_model_parameters["n_estimators"],
        learning_rate=underweight_model_parameters["learning_rate"],
        max_depth=underweight_model_parameters["max_depth"],
        min_child_weight=underweight_model_parameters["min_child_weight"],
        subsample=underweight_model_parameters["subsample"],
        colsample_bytree=underweight_model_parameters["colsample_bytree"],
        reg_alpha=underweight_model_parameters["reg_alpha"],
        reg_lambda=underweight_model_parameters["reg_lambda"],
        scale_pos_weight=weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_under_7k,
        y_train_under
    )

    probabilities = model.predict_proba(
        X_val_under_7k
    )[:, 1]

    predictions = (
        probabilities >= 0.50
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val_under,
        predictions,
        labels=[0, 1]
    ).ravel()

    precision = precision_score(
        y_val_under,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_val_under,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_val_under,
        predictions,
        zero_division=0
    )

    accuracy = accuracy_score(
        y_val_under,
        predictions
    )

    fpr = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0
    )

    roc_auc = roc_auc_score(
        y_val_under,
        probabilities
    )

    pr_auc = average_precision_score(
        y_val_under,
        probabilities
    )

    underweight_weight_results.append({
        "Scale_Pos_Weight": weight,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "ROC_AUC": roc_auc,
        "PR_AUC": pr_auc,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })

    print(
        f"Precision={precision:.4f} | "
        f"Recall={recall:.4f} | "
        f"F1={f1:.4f} | "
        f"ROC-AUC={roc_auc:.4f} | "
        f"PR-AUC={pr_auc:.4f} | "
        f"FP={fp}"
    )


underweight_weight_results = pd.DataFrame(
    underweight_weight_results
)


# =============================================================================
# 13. STUNTING COMPARISON
# =============================================================================

print("\n" + "=" * 100)
print("9. STUNTING — CLASS-WEIGHT COMPARISON")
print("=" * 100)

display(
    stunting_weight_results.round(4)
)


# =============================================================================
# 14. UNDERWEIGHT COMPARISON
# =============================================================================

print("\n" + "=" * 100)
print("10. UNDERWEIGHT — CLASS-WEIGHT COMPARISON")
print("=" * 100)

display(
    underweight_weight_results.round(4)
)


# =============================================================================
# 15. BEST PRECISION WITH RECALL >= 30%
# =============================================================================

minimum_recall = 0.30

print("\n" + "=" * 100)
print("11. BEST PRECISION WITH RECALL >= 30%")
print("=" * 100)


stunting_eligible = stunting_weight_results[
    stunting_weight_results["Recall"] >= minimum_recall
].copy()


underweight_eligible = underweight_weight_results[
    underweight_weight_results["Recall"] >= minimum_recall
].copy()


if len(stunting_eligible) == 0:

    stunting_best_precision = None

    print(
        "STUNTING: No class weight achieved recall >= 30%."
    )

else:

    stunting_best_precision = (
        stunting_eligible
        .sort_values(
            ["Precision", "F1"],
            ascending=False
        )
        .iloc[0]
    )

    print("\nSTUNTING")
    print("-" * 100)

    print(
        f"Weight    : "
        f"{stunting_best_precision['Scale_Pos_Weight']:.2f}"
    )

    print(
        f"Precision : "
        f"{stunting_best_precision['Precision']:.4f}"
    )

    print(
        f"Recall    : "
        f"{stunting_best_precision['Recall']:.4f}"
    )

    print(
        f"F1        : "
        f"{stunting_best_precision['F1']:.4f}"
    )

    print(
        f"FPR       : "
        f"{stunting_best_precision['FPR']:.4f}"
    )

    print(
        f"FP        : "
        f"{int(stunting_best_precision['FP'])}"
    )


if len(underweight_eligible) == 0:

    underweight_best_precision = None

    print(
        "\nUNDERWEIGHT: No class weight achieved recall >= 30%."
    )

else:

    underweight_best_precision = (
        underweight_eligible
        .sort_values(
            ["Precision", "F1"],
            ascending=False
        )
        .iloc[0]
    )

    print("\nUNDERWEIGHT")
    print("-" * 100)

    print(
        f"Weight    : "
        f"{underweight_best_precision['Scale_Pos_Weight']:.2f}"
    )

    print(
        f"Precision : "
        f"{underweight_best_precision['Precision']:.4f}"
    )

    print(
        f"Recall    : "
        f"{underweight_best_precision['Recall']:.4f}"
    )

    print(
        f"F1        : "
        f"{underweight_best_precision['F1']:.4f}"
    )

    print(
        f"FPR       : "
        f"{underweight_best_precision['FPR']:.4f}"
    )

    print(
        f"FP        : "
        f"{int(underweight_best_precision['FP'])}"
    )


# =============================================================================
# 16. BEST F1 WITH RECALL >= 30%
# =============================================================================

print("\n" + "=" * 100)
print("12. BEST F1 WITH RECALL >= 30%")
print("=" * 100)


if len(stunting_eligible) > 0:

    stunting_best_f1 = (
        stunting_eligible
        .sort_values(
            ["F1", "Precision"],
            ascending=False
        )
        .iloc[0]
    )

    print("\nSTUNTING")
    print("-" * 100)

    print(
        f"Weight    : "
        f"{stunting_best_f1['Scale_Pos_Weight']:.2f}"
    )

    print(
        f"Precision : "
        f"{stunting_best_f1['Precision']:.4f}"
    )

    print(
        f"Recall    : "
        f"{stunting_best_f1['Recall']:.4f}"
    )

    print(
        f"F1        : "
        f"{stunting_best_f1['F1']:.4f}"
    )

    print(
        f"FP        : "
        f"{int(stunting_best_f1['FP'])}"
    )


if len(underweight_eligible) > 0:

    underweight_best_f1 = (
        underweight_eligible
        .sort_values(
            ["F1", "Precision"],
            ascending=False
        )
        .iloc[0]
    )

    print("\nUNDERWEIGHT")
    print("-" * 100)

    print(
        f"Weight    : "
        f"{underweight_best_f1['Scale_Pos_Weight']:.2f}"
    )

    print(
        f"Precision : "
        f"{underweight_best_f1['Precision']:.4f}"
    )

    print(
        f"Recall    : "
        f"{underweight_best_f1['Recall']:.4f}"
    )

    print(
        f"F1        : "
        f"{underweight_best_f1['F1']:.4f}"
    )

    print(
        f"FP        : "
        f"{int(underweight_best_f1['FP'])}"
    )


# =============================================================================
# 17. MODEL SELECTION RECOMMENDATION
# =============================================================================

print("\n" + "=" * 100)
print("13. STEP 7K MODEL SELECTION")
print("=" * 100)

print(
    "\nSelection principle:"
)

print(
    "1. Recall must be >= 30%"
)

print(
    "2. Among eligible models, prioritize precision"
)

print(
    "3. Use F1 as secondary criterion"
)

print(
    "4. Examine FP and FPR before final selection"
)

print(
    "5. Test set remains untouched"
)


# =============================================================================
# 18. SAVE STEP 7K RESULTS IN MEMORY
# =============================================================================

step7k_results = {
    "stunting": stunting_weight_results.copy(),
    "underweight": underweight_weight_results.copy()
}


# =============================================================================
# 19. INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("14. STEP 7K INTEGRITY CHECK")
print("=" * 100)

print("✓ Fixed 15-predictor set retained")
print("✓ Step 7I winning model structures retained")
print("✓ Only class weight was varied")
print("✓ Preprocessing fitted on training data only")
print("✓ Models fitted on training data only")
print("✓ Validation data used only for evaluation")
print("✓ Threshold fixed at 0.50 for this class-weight experiment")
print("✓ Test predictors NOT accessed")
print("✓ Test targets NOT accessed")
print("✓ Test probabilities NOT generated")
print("✓ Test performance NOT calculated")
print("✓ No test-based optimization performed")

print("\n" + "=" * 100)
print("STEP 7K COMPLETE")
print("=" * 100)

print(
    "\nNEXT STEP:"
)

print(
    "STEP 7L — Combine the best class weight with "
    "validation threshold optimization."
)
```

```python
# =============================================================================
# STEP 7L: CAR MICS6 — JOINT CLASS-WEIGHT + THRESHOLD OPTIMIZATION
# =============================================================================
#
# OBJECTIVE:
#   Find the best combination of:
#       1. scale_pos_weight
#       2. decision threshold
#
#   while prioritizing precision and reducing false positives.
#
# DATA RULE:
#   TRAINING  -> model fitting
#   VALIDATION -> weight/threshold optimization
#   TEST       -> COMPLETELY UNTOUCHED
#
# SELECTION RULE:
#   Primary   : highest Precision
#   Constraint: Recall >= 30%
#   Secondary : highest F1
#   Tie-break : lowest FP
#
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

print("=" * 100)
print("STEP 7L: CAR MICS6 — JOINT CLASS-WEIGHT + THRESHOLD OPTIMIZATION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)

required_objects = [
    "X_train_stunt",
    "X_val_stunt",
    "y_train_stunt",
    "y_val_stunt",
    "X_train_under",
    "X_val_under",
    "y_train_under",
    "y_val_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    print("Missing objects:")
    for obj in missing_objects:
        print(" -", obj)

    raise NameError(
        "Step 7L cannot continue because required "
        "training/validation objects are missing."
    )

print("✓ All required training/validation objects are available")


# =============================================================================
# 2. FIXED 15-PREDICTOR SET
# =============================================================================

predictors_15 = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cdisability",
    "CA1",
    "CA14",
    "BD2",
    "IM2",
    "CA31"
]

print("\n2. FIXED PREDICTOR SET")
print("-" * 100)

print("Number of predictors:", len(predictors_15))

for i, p in enumerate(predictors_15, 1):
    print(f"{i:2d}. {p}")


# =============================================================================
# 3. TEST-SET PROTECTION
# =============================================================================

print("\n" + "=" * 100)
print("3. TEST-SET PROTECTION")
print("=" * 100)

print("✓ X_test_stunt is NOT accessed")
print("✓ y_test_stunt is NOT accessed")
print("✓ X_test_under is NOT accessed")
print("✓ y_test_under is NOT accessed")
print("✓ No test probabilities generated")
print("✓ No test thresholds selected")
print("✓ No test performance calculated")


# =============================================================================
# 4. TRAINING-ONLY PREPROCESSING
# =============================================================================
#
# Refit preprocessing independently for each outcome using TRAINING data only.
# This prevents validation information from entering the preprocessing stage.
# =============================================================================

print("\n" + "=" * 100)
print("4. TRAINING-ONLY PREPROCESSING")
print("=" * 100)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder


numeric_predictors = [
    "CAGE"
]

categorical_predictors = [
    p for p in predictors_15
    if p != "CAGE"
]


def create_preprocessor_7l():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            numeric_predictors
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_predictors
        )
    ])


# -----------------------------
# STUNTING
# -----------------------------

preprocessor_stunt_7l = create_preprocessor_7l()

X_train_stunt_7l = (
    preprocessor_stunt_7l
    .fit_transform(
        X_train_stunt[predictors_15]
    )
)

X_val_stunt_7l = (
    preprocessor_stunt_7l
    .transform(
        X_val_stunt[predictors_15]
    )
)


# -----------------------------
# UNDERWEIGHT
# -----------------------------

preprocessor_under_7l = create_preprocessor_7l()

X_train_under_7l = (
    preprocessor_under_7l
    .fit_transform(
        X_train_under[predictors_15]
    )
)

X_val_under_7l = (
    preprocessor_under_7l
    .transform(
        X_val_under[predictors_15]
    )
)


print(
    "Stunting training   :",
    X_train_stunt_7l.shape
)

print(
    "Stunting validation :",
    X_val_stunt_7l.shape
)

print(
    "Underweight training   :",
    X_train_under_7l.shape
)

print(
    "Underweight validation :",
    X_val_under_7l.shape
)

print("✓ Preprocessing fitted on training data only")


# =============================================================================
# 5. MODEL CONFIGURATIONS
# =============================================================================

stunting_params = {
    "n_estimators": 700,
    "learning_rate": 0.02,
    "max_depth": 3,
    "min_child_weight": 8,
    "subsample": 0.80,
    "colsample_bytree": 0.80,
    "reg_alpha": 0.5,
    "reg_lambda": 5.0
}


underweight_params = {
    "n_estimators": 700,
    "learning_rate": 0.02,
    "max_depth": 3,
    "min_child_weight": 8,
    "subsample": 0.80,
    "colsample_bytree": 0.80,
    "reg_alpha": 1.0,
    "reg_lambda": 3.0
}


# =============================================================================
# 6. WEIGHT SEARCH GRID
# =============================================================================

# We focus around the promising regions discovered in Step 7K.

stunting_weights = [
    1.00,
    1.10,
    1.20,
    1.25,
    1.30,
    1.40,
    1.50
]

underweight_weights = [
    1.75,
    1.85,
    2.00,
    2.15,
    2.25
]


# =============================================================================
# 7. THRESHOLD SEARCH GRID
# =============================================================================
#
# Fine-grained threshold search.
#
# We deliberately do NOT optimize directly for maximum precision without
# a recall constraint because that can produce the pathological situation
# observed earlier:
#
#   Precision = 100%
#   Recall    ≈ 0%
#
# Such a model is not useful for identifying undernourished children.
# =============================================================================

thresholds = np.round(
    np.arange(0.30, 0.91, 0.01),
    2
)

minimum_recall = 0.30


print("\n" + "=" * 100)
print("5. SEARCH DESIGN")
print("=" * 100)

print(
    "Stunting weights   :",
    stunting_weights
)

print(
    "Underweight weights:",
    underweight_weights
)

print(
    "Threshold range   :",
    f"{thresholds.min():.2f} – {thresholds.max():.2f}"
)

print(
    "Minimum recall    :",
    f"{minimum_recall:.0%}"
)


# =============================================================================
# 8. METRIC FUNCTION
# =============================================================================

def calculate_metrics(
    y_true,
    probabilities,
    threshold
):

    predictions = (
        probabilities >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    ).ravel()

    accuracy = accuracy_score(
        y_true,
        predictions
    )

    precision = precision_score(
        y_true,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        predictions,
        zero_division=0
    )

    fpr = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0
    )

    return {
        "Threshold": threshold,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }


# =============================================================================
# 9. STUNTING — JOINT SEARCH
# =============================================================================

print("\n" + "=" * 100)
print("6. STUNTING — JOINT CLASS-WEIGHT + THRESHOLD SEARCH")
print("=" * 100)

stunting_joint_results = []

for weight in stunting_weights:

    print(
        f"\nTraining stunting model "
        f"with scale_pos_weight={weight:.2f}"
    )

    model = XGBClassifier(
        **stunting_params,
        scale_pos_weight=weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_stunt_7l,
        y_train_stunt
    )

    probabilities = model.predict_proba(
        X_val_stunt_7l
    )[:, 1]

    roc_auc = roc_auc_score(
        y_val_stunt,
        probabilities
    )

    pr_auc = average_precision_score(
        y_val_stunt,
        probabilities
    )

    for threshold in thresholds:

        metrics = calculate_metrics(
            y_val_stunt,
            probabilities,
            threshold
        )

        metrics["Scale_Pos_Weight"] = weight
        metrics["ROC_AUC"] = roc_auc
        metrics["PR_AUC"] = pr_auc

        stunting_joint_results.append(
            metrics
        )

    print(
        f"ROC-AUC={roc_auc:.4f} | "
        f"PR-AUC={pr_auc:.4f}"
    )


stunting_joint_results = pd.DataFrame(
    stunting_joint_results
)


# =============================================================================
# 10. UNDERWEIGHT — JOINT SEARCH
# =============================================================================

print("\n" + "=" * 100)
print("7. UNDERWEIGHT — JOINT CLASS-WEIGHT + THRESHOLD SEARCH")
print("=" * 100)

underweight_joint_results = []

for weight in underweight_weights:

    print(
        f"\nTraining underweight model "
        f"with scale_pos_weight={weight:.2f}"
    )

    model = XGBClassifier(
        **underweight_params,
        scale_pos_weight=weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_under_7l,
        y_train_under
    )

    probabilities = model.predict_proba(
        X_val_under_7l
    )[:, 1]

    roc_auc = roc_auc_score(
        y_val_under,
        probabilities
    )

    pr_auc = average_precision_score(
        y_val_under,
        probabilities
    )

    for threshold in thresholds:

        metrics = calculate_metrics(
            y_val_under,
            probabilities,
            threshold
        )

        metrics["Scale_Pos_Weight"] = weight
        metrics["ROC_AUC"] = roc_auc
        metrics["PR_AUC"] = pr_auc

        underweight_joint_results.append(
            metrics
        )

    print(
        f"ROC-AUC={roc_auc:.4f} | "
        f"PR-AUC={pr_auc:.4f}"
    )


underweight_joint_results = pd.DataFrame(
    underweight_joint_results
)


# =============================================================================
# 11. STUNTING — ALL ELIGIBLE OPERATING POINTS
# =============================================================================

print("\n" + "=" * 100)
print("8. STUNTING — ELIGIBLE OPERATING POINTS")
print("=" * 100)

stunting_eligible = (
    stunting_joint_results[
        stunting_joint_results["Recall"] >= minimum_recall
    ]
    .sort_values(
        ["Precision", "F1", "FP"],
        ascending=[False, False, True]
    )
)


print(
    "\nTop 20 stunting operating points:"
)

display(
    stunting_eligible.head(20).round(4)
)


# =============================================================================
# 12. UNDERWEIGHT — ALL ELIGIBLE OPERATING POINTS
# =============================================================================

print("\n" + "=" * 100)
print("9. UNDERWEIGHT — ELIGIBLE OPERATING POINTS")
print("=" * 100)

underweight_eligible = (
    underweight_joint_results[
        underweight_joint_results["Recall"] >= minimum_recall
    ]
    .sort_values(
        ["Precision", "F1", "FP"],
        ascending=[False, False, True]
    )
)


print(
    "\nTop 20 underweight operating points:"
)

display(
    underweight_eligible.head(20).round(4)
)


# =============================================================================
# 13. BEST PRECISION OPERATING POINT
# =============================================================================

print("\n" + "=" * 100)
print("10. BEST PRECISION WITH RECALL >= 30%")
print("=" * 100)


if len(stunting_eligible) > 0:

    best_stunting_precision = (
        stunting_eligible.iloc[0]
    )

    print("\nSTUNTING")
    print("-" * 100)

    print(
        f"Scale_Pos_Weight : "
        f"{best_stunting_precision['Scale_Pos_Weight']:.2f}"
    )

    print(
        f"Threshold        : "
        f"{best_stunting_precision['Threshold']:.2f}"
    )

    print(
        f"Precision        : "
        f"{best_stunting_precision['Precision']:.4f}"
    )

    print(
        f"Recall           : "
        f"{best_stunting_precision['Recall']:.4f}"
    )

    print(
        f"F1               : "
        f"{best_stunting_precision['F1']:.4f}"
    )

    print(
        f"FPR              : "
        f"{best_stunting_precision['FPR']:.4f}"
    )

    print(
        f"FP               : "
        f"{int(best_stunting_precision['FP'])}"
    )

else:

    best_stunting_precision = None

    print(
        "STUNTING: No operating point achieved "
        "recall >= 30%."
    )


if len(underweight_eligible) > 0:

    best_underweight_precision = (
        underweight_eligible.iloc[0]
    )

    print("\nUNDERWEIGHT")
    print("-" * 100)

    print(
        f"Scale_Pos_Weight : "
        f"{best_underweight_precision['Scale_Pos_Weight']:.2f}"
    )

    print(
        f"Threshold        : "
        f"{best_underweight_precision['Threshold']:.2f}"
    )

    print(
        f"Precision        : "
        f"{best_underweight_precision['Precision']:.4f}"
    )

    print(
        f"Recall           : "
        f"{best_underweight_precision['Recall']:.4f}"
    )

    print(
        f"F1               : "
        f"{best_underweight_precision['F1']:.4f}"
    )

    print(
        f"FPR              : "
        f"{best_underweight_precision['FPR']:.4f}"
    )

    print(
        f"FP               : "
        f"{int(best_underweight_precision['FP'])}"
    )

else:

    best_underweight_precision = None

    print(
        "UNDERWEIGHT: No operating point achieved "
        "recall >= 30%."
    )


# =============================================================================
# 14. BEST F1 OPERATING POINT
# =============================================================================

print("\n" + "=" * 100)
print("11. BEST F1 WITH RECALL >= 30%")
print("=" * 100)


if len(stunting_eligible) > 0:

    best_stunting_f1 = (
        stunting_eligible
        .sort_values(
            ["F1", "Precision", "FP"],
            ascending=[False, False, True]
        )
        .iloc[0]
    )

    print("\nSTUNTING")
    print("-" * 100)

    print(
        f"Scale_Pos_Weight : "
        f"{best_stunting_f1['Scale_Pos_Weight']:.2f}"
    )

    print(
        f"Threshold        : "
        f"{best_stunting_f1['Threshold']:.2f}"
    )

    print(
        f"Precision        : "
        f"{best_stunting_f1['Precision']:.4f}"
    )

    print(
        f"Recall           : "
        f"{best_stunting_f1['Recall']:.4f}"
    )

    print(
        f"F1               : "
        f"{best_stunting_f1['F1']:.4f}"
    )

    print(
        f"FP               : "
        f"{int(best_stunting_f1['FP'])}"
    )


if len(underweight_eligible) > 0:

    best_underweight_f1 = (
        underweight_eligible
        .sort_values(
            ["F1", "Precision", "FP"],
            ascending=[False, False, True]
        )
        .iloc[0]
    )

    print("\nUNDERWEIGHT")
    print("-" * 100)

    print(
        f"Scale_Pos_Weight : "
        f"{best_underweight_f1['Scale_Pos_Weight']:.2f}"
    )

    print(
        f"Threshold        : "
        f"{best_underweight_f1['Threshold']:.2f}"
    )

    print(
        f"Precision        : "
        f"{best_underweight_f1['Precision']:.4f}"
    )

    print(
        f"Recall           : "
        f"{best_underweight_f1['Recall']:.4f}"
    )

    print(
        f"F1               : "
        f"{best_underweight_f1['F1']:.4f}"
    )

    print(
        f"FP               : "
        f"{int(best_underweight_f1['FP'])}"
    )


# =============================================================================
# 15. LOWEST FALSE POSITIVES WHILE RETAINING RECALL >= 30%
# =============================================================================

print("\n" + "=" * 100)
print("12. MINIMUM FALSE POSITIVES WITH RECALL >= 30%")
print("=" * 100)


if len(stunting_eligible) > 0:

    best_stunting_fp = (
        stunting_eligible
        .sort_values(
            ["FP", "Precision", "F1"],
            ascending=[True, False, False]
        )
        .iloc[0]
    )

    print("\nSTUNTING")
    print("-" * 100)

    print(
        f"Scale_Pos_Weight : "
        f"{best_stunting_fp['Scale_Pos_Weight']:.2f}"
    )

    print(
        f"Threshold        : "
        f"{best_stunting_fp['Threshold']:.2f}"
    )

    print(
        f"Precision        : "
        f"{best_stunting_fp['Precision']:.4f}"
    )

    print(
        f"Recall           : "
        f"{best_stunting_fp['Recall']:.4f}"
    )

    print(
        f"F1               : "
        f"{best_stunting_fp['F1']:.4f}"
    )

    print(
        f"FP               : "
        f"{int(best_stunting_fp['FP'])}"
    )


if len(underweight_eligible) > 0:

    best_underweight_fp = (
        underweight_eligible
        .sort_values(
            ["FP", "Precision", "F1"],
            ascending=[True, False, False]
        )
        .iloc[0]
    )

    print("\nUNDERWEIGHT")
    print("-" * 100)

    print(
        f"Scale_Pos_Weight : "
        f"{best_underweight_fp['Scale_Pos_Weight']:.2f}"
    )

    print(
        f"Threshold        : "
        f"{best_underweight_fp['Threshold']:.2f}"
    )

    print(
        f"Precision        : "
        f"{best_underweight_fp['Precision']:.4f}"
    )

    print(
        f"Recall           : "
        f"{best_underweight_fp['Recall']:.4f}"
    )

    print(
        f"F1               : "
        f"{best_underweight_fp['F1']:.4f}"
    )

    print(
        f"FP               : "
        f"{int(best_underweight_fp['FP'])}"
    )


# =============================================================================
# 16. SAVE RESULTS IN MEMORY
# =============================================================================

step7l_results = {
    "stunting": stunting_joint_results.copy(),
    "underweight": underweight_joint_results.copy()
}


# =============================================================================
# 17. INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("13. STEP 7L INTEGRITY CHECK")
print("=" * 100)

print("✓ 15-predictor set retained")
print("✓ XGBoost architecture fixed")
print("✓ Class weight varied using validation experiments")
print("✓ Threshold varied using validation data")
print("✓ Preprocessing fitted on training data only")
print("✓ Models fitted on training data only")
print("✓ Validation data used for optimization")
print("✓ Test predictors NOT accessed")
print("✓ Test targets NOT accessed")
print("✓ Test probabilities NOT generated")
print("✓ Test performance NOT calculated")
print("✓ No test-based model selection performed")

print("\n" + "=" * 100)
print("STEP 7L COMPLETE")
print("=" * 100)

print(
    "\nNEXT:"
)

print(
    "Review the Step 7L results before proceeding to "
    "final model fitting."
)
```

```python
# =============================================================================
# STEP 7M: CAR MICS6 — PRECISION-FOCUSED STABILITY / SENSITIVITY ANALYSIS
# =============================================================================
#
# PURPOSE:
#   Assess whether the selected precision-focused operating points are stable
#   across nearby thresholds and class weights.
#
# SELECTED CANDIDATES FROM STEP 7L:
#   STUNTING    : weight = 1.25, threshold = 0.53
#   UNDERWEIGHT : weight = 2.15, threshold = 0.54
#
# IMPORTANT:
#   TRAINING DATA    -> model fitting
#   VALIDATION DATA  -> sensitivity analysis
#   TEST DATA        -> COMPLETELY UNTOUCHED
#
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

print("=" * 100)
print("STEP 7M: CAR MICS6 — PRECISION-FOCUSED STABILITY / SENSITIVITY ANALYSIS")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)

required_objects = [
    "X_train_stunt",
    "X_val_stunt",
    "y_train_stunt",
    "y_val_stunt",
    "X_train_under",
    "X_val_under",
    "y_train_under",
    "y_val_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:

    print("Missing objects:")
    for obj in missing_objects:
        print(" -", obj)

    raise NameError(
        "Step 7M cannot continue because required "
        "training/validation objects are missing."
    )

print("✓ All required training/validation objects are available")


# =============================================================================
# 2. FIXED 15-PREDICTOR SET
# =============================================================================

print("\n2. FIXED PREDICTOR SET")
print("-" * 100)

predictors_15 = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cdisability",
    "CA1",
    "CA14",
    "BD2",
    "IM2",
    "CA31"
]

print("Number of predictors:", len(predictors_15))

for i, predictor in enumerate(predictors_15, 1):
    print(f"{i:2d}. {predictor}")


# =============================================================================
# 3. TEST-SET PROTECTION
# =============================================================================

print("\n" + "=" * 100)
print("3. TEST-SET PROTECTION")
print("=" * 100)

print("✓ X_test_stunt is NOT accessed")
print("✓ y_test_stunt is NOT accessed")
print("✓ X_test_under is NOT accessed")
print("✓ y_test_under is NOT accessed")
print("✓ No test probabilities generated")
print("✓ No test thresholds evaluated")
print("✓ No test performance calculated")


# =============================================================================
# 4. CHECK STEP 7L PREPROCESSORS
# =============================================================================

print("\n" + "=" * 100)
print("4. PREPROCESSING CHECK")
print("=" * 100)

if (
    "preprocessor_stunt_7l" not in globals()
    or
    "preprocessor_under_7l" not in globals()
):

    raise NameError(
        "Step 7M requires preprocessor_stunt_7l and "
        "preprocessor_under_7l from Step 7L."
    )

print("✓ Step 7L training-only preprocessors available")


# =============================================================================
# 5. REUSE TRAINING-ONLY TRANSFORMATIONS
# =============================================================================

print("\n" + "=" * 100)
print("5. TRANSFORMING DEVELOPMENT DATA")
print("=" * 100)

X_train_stunt_7m = (
    preprocessor_stunt_7l.transform(
        X_train_stunt[predictors_15]
    )
)

X_val_stunt_7m = (
    preprocessor_stunt_7l.transform(
        X_val_stunt[predictors_15]
    )
)

X_train_under_7m = (
    preprocessor_under_7l.transform(
        X_train_under[predictors_15]
    )
)

X_val_under_7m = (
    preprocessor_under_7l.transform(
        X_val_under[predictors_15]
    )
)

print(
    "Stunting training   :",
    X_train_stunt_7m.shape
)

print(
    "Stunting validation :",
    X_val_stunt_7m.shape
)

print(
    "Underweight training   :",
    X_train_under_7m.shape
)

print(
    "Underweight validation :",
    X_val_under_7m.shape
)

print("✓ Existing training-only preprocessing reused")


# =============================================================================
# 6. MODEL CONFIGURATION
# =============================================================================

stunting_params = {
    "n_estimators": 700,
    "learning_rate": 0.02,
    "max_depth": 3,
    "min_child_weight": 8,
    "subsample": 0.80,
    "colsample_bytree": 0.80,
    "reg_alpha": 0.5,
    "reg_lambda": 5.0
}

underweight_params = {
    "n_estimators": 700,
    "learning_rate": 0.02,
    "max_depth": 3,
    "min_child_weight": 8,
    "subsample": 0.80,
    "colsample_bytree": 0.80,
    "reg_alpha": 1.0,
    "reg_lambda": 3.0
}


# =============================================================================
# 7. LOCAL SENSITIVITY GRID
# =============================================================================
#
# Rather than searching the entire parameter space again, examine nearby
# values around the Step 7L candidates.
#
# =============================================================================

stunting_weights_local = [
    1.20,
    1.25,
    1.30
]

stunting_thresholds_local = [
    0.50,
    0.51,
    0.52,
    0.53,
    0.54,
    0.55,
    0.56
]

underweight_weights_local = [
    2.00,
    2.10,
    2.15,
    2.20,
    2.25
]

underweight_thresholds_local = [
    0.51,
    0.52,
    0.53,
    0.54,
    0.55,
    0.56,
    0.57
]

minimum_recall = 0.30


# =============================================================================
# 8. METRIC FUNCTION
# =============================================================================

def evaluate_operating_point(
    y_true,
    probabilities,
    threshold
):

    predictions = (
        probabilities >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    ).ravel()

    precision = precision_score(
        y_true,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        predictions,
        zero_division=0
    )

    accuracy = accuracy_score(
        y_true,
        predictions
    )

    fpr = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0
    )

    return {
        "Threshold": threshold,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }


# =============================================================================
# 9. STUNTING SENSITIVITY ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("6. STUNTING — LOCAL SENSITIVITY ANALYSIS")
print("=" * 100)

stunting_sensitivity = []

for weight in stunting_weights_local:

    print(
        f"\nTraining stunting model "
        f"with scale_pos_weight={weight:.2f}"
    )

    model = XGBClassifier(
        **stunting_params,
        scale_pos_weight=weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_stunt_7m,
        y_train_stunt
    )

    probabilities = model.predict_proba(
        X_val_stunt_7m
    )[:, 1]

    roc_auc = roc_auc_score(
        y_val_stunt,
        probabilities
    )

    pr_auc = average_precision_score(
        y_val_stunt,
        probabilities
    )

    for threshold in stunting_thresholds_local:

        result = evaluate_operating_point(
            y_val_stunt,
            probabilities,
            threshold
        )

        result["Scale_Pos_Weight"] = weight
        result["ROC_AUC"] = roc_auc
        result["PR_AUC"] = pr_auc

        stunting_sensitivity.append(result)


stunting_sensitivity = pd.DataFrame(
    stunting_sensitivity
)


# =============================================================================
# 10. UNDERWEIGHT SENSITIVITY ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("7. UNDERWEIGHT — LOCAL SENSITIVITY ANALYSIS")
print("=" * 100)

underweight_sensitivity = []

for weight in underweight_weights_local:

    print(
        f"\nTraining underweight model "
        f"with scale_pos_weight={weight:.2f}"
    )

    model = XGBClassifier(
        **underweight_params,
        scale_pos_weight=weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_under_7m,
        y_train_under
    )

    probabilities = model.predict_proba(
        X_val_under_7m
    )[:, 1]

    roc_auc = roc_auc_score(
        y_val_under,
        probabilities
    )

    pr_auc = average_precision_score(
        y_val_under,
        probabilities
    )

    for threshold in underweight_thresholds_local:

        result = evaluate_operating_point(
            y_val_under,
            probabilities,
            threshold
        )

        result["Scale_Pos_Weight"] = weight
        result["ROC_AUC"] = roc_auc
        result["PR_AUC"] = pr_auc

        underweight_sensitivity.append(result)


underweight_sensitivity = pd.DataFrame(
    underweight_sensitivity
)


# =============================================================================
# 11. STUNTING — ELIGIBLE RESULTS
# =============================================================================

print("\n" + "=" * 100)
print("8. STUNTING — ELIGIBLE OPERATING POINTS")
print("=" * 100)

stunting_eligible_7m = (
    stunting_sensitivity[
        stunting_sensitivity["Recall"] >= minimum_recall
    ]
    .sort_values(
        ["Precision", "F1", "FP"],
        ascending=[False, False, True]
    )
)

print(
    "\nTop 20 stunting operating points:"
)

display(
    stunting_eligible_7m.head(20).round(4)
)


# =============================================================================
# 12. UNDERWEIGHT — ELIGIBLE RESULTS
# =============================================================================

print("\n" + "=" * 100)
print("9. UNDERWEIGHT — ELIGIBLE OPERATING POINTS")
print("=" * 100)

underweight_eligible_7m = (
    underweight_sensitivity[
        underweight_sensitivity["Recall"] >= minimum_recall
    ]
    .sort_values(
        ["Precision", "F1", "FP"],
        ascending=[False, False, True]
    )
)

print(
    "\nTop 20 underweight operating points:"
)

display(
    underweight_eligible_7m.head(20).round(4)
)


# =============================================================================
# 13. STABILITY AROUND SELECTED STUNTING POINT
# =============================================================================

print("\n" + "=" * 100)
print("10. STABILITY AROUND SELECTED STUNTING POINT")
print("=" * 100)

stunting_selected_region = stunting_sensitivity[
    (
        stunting_sensitivity["Scale_Pos_Weight"].between(
            1.20,
            1.30
        )
    )
    &
    (
        stunting_sensitivity["Threshold"].between(
            0.51,
            0.55
        )
    )
    &
    (
        stunting_sensitivity["Recall"] >= minimum_recall
    )
]

if len(stunting_selected_region) > 0:

    print(
        "\nSelected region summary:"
    )

    print(
        stunting_selected_region[
            [
                "Scale_Pos_Weight",
                "Threshold",
                "Precision",
                "Recall",
                "F1",
                "FPR",
                "FP"
            ]
        ]
        .sort_values(
            ["Precision", "F1"],
            ascending=[False, False]
        )
        .round(4)
        .to_string(index=False)
    )

else:

    print(
        "No operating points in the selected region "
        "satisfy recall >= 30%."
    )


# =============================================================================
# 14. STABILITY AROUND SELECTED UNDERWEIGHT POINT
# =============================================================================

print("\n" + "=" * 100)
print("11. STABILITY AROUND SELECTED UNDERWEIGHT POINT")
print("=" * 100)

underweight_selected_region = underweight_sensitivity[
    (
        underweight_sensitivity["Scale_Pos_Weight"].between(
            2.00,
            2.25
        )
    )
    &
    (
        underweight_sensitivity["Threshold"].between(
            0.52,
            0.56
        )
    )
    &
    (
        underweight_sensitivity["Recall"] >= minimum_recall
    )
]

if len(underweight_selected_region) > 0:

    print(
        "\nSelected region summary:"
    )

    print(
        underweight_selected_region[
            [
                "Scale_Pos_Weight",
                "Threshold",
                "Precision",
                "Recall",
                "F1",
                "FPR",
                "FP"
            ]
        ]
        .sort_values(
            ["Precision", "F1"],
            ascending=[False, False]
        )
        .round(4)
        .to_string(index=False)
    )

else:

    print(
        "No operating points in the selected region "
        "satisfy recall >= 30%."
    )


# =============================================================================
# 15. SUMMARY STATISTICS
# =============================================================================

print("\n" + "=" * 100)
print("12. STABILITY SUMMARY")
print("=" * 100)


def stability_summary(
    df,
    name
):

    eligible = df[
        df["Recall"] >= minimum_recall
    ]

    if len(eligible) == 0:

        print(
            f"\n{name}: No eligible operating points."
        )

        return None

    summary = pd.DataFrame({
        "Metric": [
            "Precision",
            "Recall",
            "F1",
            "FPR",
            "FP"
        ],
        "Minimum": [
            eligible["Precision"].min(),
            eligible["Recall"].min(),
            eligible["F1"].min(),
            eligible["FPR"].min(),
            eligible["FP"].min()
        ],
        "Maximum": [
            eligible["Precision"].max(),
            eligible["Recall"].max(),
            eligible["F1"].max(),
            eligible["FPR"].max(),
            eligible["FP"].max()
        ]
    })

    print(f"\n{name}")
    print("-" * 100)
    print(
        summary.round(4).to_string(index=False)
    )

    return summary


stunting_stability_summary = stability_summary(
    stunting_selected_region,
    "STUNTING SELECTED REGION"
)

underweight_stability_summary = stability_summary(
    underweight_selected_region,
    "UNDERWEIGHT SELECTED REGION"
)


# =============================================================================
# 16. FINAL RECOMMENDATION FROM STABILITY ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("13. STEP 7M INTERPRETATION")
print("=" * 100)

print("""
The purpose of this analysis is NOT to maximize precision at any cost.

The preferred operating point should:

1. Maintain Recall >= 30%
2. Minimize false positives
3. Maintain reasonably high precision
4. Avoid a threshold that is highly sensitive to small changes
5. Preserve acceptable F1 performance
6. Be selected using validation data only

A point that achieves high precision but collapses recall is NOT preferred.
""")


# =============================================================================
# 17. SAVE RESULTS IN MEMORY
# =============================================================================

step7m_results = {
    "stunting": stunting_sensitivity.copy(),
    "underweight": underweight_sensitivity.copy(),
    "stunting_eligible": stunting_eligible_7m.copy(),
    "underweight_eligible": underweight_eligible_7m.copy()
}


# =============================================================================
# 18. INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("14. STEP 7M INTEGRITY CHECK")
print("=" * 100)

print("✓ 15-predictor set retained")
print("✓ Step 7L preprocessing reused")
print("✓ Preprocessing remains training-only")
print("✓ Class weights varied only in development")
print("✓ Thresholds evaluated only on validation data")
print("✓ Recall constraint = 30%")
print("✓ Test predictors NOT accessed")
print("✓ Test targets NOT accessed")
print("✓ Test probabilities NOT generated")
print("✓ Test performance NOT calculated")

print("\n" + "=" * 100)
print("STEP 7M COMPLETE")
print("=" * 100)

print(
    "\nNEXT STEP:"
)

print(
    "Review the complete Step 7M output before locking "
    "the final model configurations."
)
```

```python
# =============================================================================
# STEP 8: CAR MICS6 — FINAL MODEL FITTING AND VALIDATION CONFIRMATION
# =============================================================================

import numpy as np
import pandas as pd

from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

print("=" * 100)
print("STEP 8: CAR MICS6 — FINAL MODEL FITTING AND VALIDATION CONFIRMATION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)

required_objects = [
    "X_train_stunt",
    "X_val_stunt",
    "y_train_stunt",
    "y_val_stunt",
    "X_train_under",
    "X_val_under",
    "y_train_under",
    "y_val_under",
    "preprocessor_stunt_7l",
    "preprocessor_under_7l"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required objects: "
        + ", ".join(missing_objects)
    )

print("✓ All required training/validation objects are available")


# =============================================================================
# 2. LOCKED PREDICTOR SET
# =============================================================================

print("\n2. LOCKED PREDICTOR SET")
print("-" * 100)

predictors_15 = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cdisability",
    "CA1",
    "CA14",
    "BD2",
    "IM2",
    "CA31"
]

print("Number of predictors:", len(predictors_15))

for i, predictor in enumerate(predictors_15, 1):
    print(f"{i:2d}. {predictor}")

print("✓ Predictor set locked")


# =============================================================================
# 3. LOCKED MODEL CONFIGURATIONS
# =============================================================================

print("\n3. LOCKED MODEL CONFIGURATIONS")
print("-" * 100)

STUNTING_WEIGHT = 1.25
STUNTING_THRESHOLD = 0.53

UNDERWEIGHT_WEIGHT = 2.00
UNDERWEIGHT_THRESHOLD = 0.53

print(
    f"Stunting    : scale_pos_weight={STUNTING_WEIGHT}, "
    f"threshold={STUNTING_THRESHOLD}"
)

print(
    f"Underweight : scale_pos_weight={UNDERWEIGHT_WEIGHT}, "
    f"threshold={UNDERWEIGHT_THRESHOLD}"
)


# =============================================================================
# 4. TEST-SET PROTECTION
# =============================================================================

print("\n4. TEST-SET PROTECTION")
print("-" * 100)

print("✓ X_test_stunt NOT accessed")
print("✓ y_test_stunt NOT accessed")
print("✓ X_test_under NOT accessed")
print("✓ y_test_under NOT accessed")
print("✓ No test probabilities generated")
print("✓ No test threshold selection")
print("✓ No test performance calculation")


# =============================================================================
# 5. TRAINING-ONLY PREPROCESSING
# =============================================================================

print("\n5. TRAINING-ONLY PREPROCESSING")
print("-" * 100)

X_train_stunt_final = preprocessor_stunt_7l.transform(
    X_train_stunt[predictors_15]
)

X_val_stunt_final = preprocessor_stunt_7l.transform(
    X_val_stunt[predictors_15]
)

X_train_under_final = preprocessor_under_7l.transform(
    X_train_under[predictors_15]
)

X_val_under_final = preprocessor_under_7l.transform(
    X_val_under[predictors_15]
)

print(
    "Stunting training   :",
    X_train_stunt_final.shape
)

print(
    "Stunting validation :",
    X_val_stunt_final.shape
)

print(
    "Underweight training   :",
    X_train_under_final.shape
)

print(
    "Underweight validation :",
    X_val_under_final.shape
)

print("✓ Preprocessing fitted previously using training data only")
print("✓ Validation data transformed only")


# =============================================================================
# 6. FINAL STUNTING MODEL
# =============================================================================

print("\n6. FITTING FINAL STUNTING MODEL")
print("-" * 100)

final_stunting_model = XGBClassifier(
    n_estimators=700,
    learning_rate=0.02,
    max_depth=3,
    min_child_weight=8,
    subsample=0.80,
    colsample_bytree=0.80,
    reg_alpha=0.5,
    reg_lambda=5.0,
    scale_pos_weight=STUNTING_WEIGHT,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

final_stunting_model.fit(
    X_train_stunt_final,
    y_train_stunt
)

print("✓ Final stunting model fitted")
print("✓ Training data only")


# =============================================================================
# 7. FINAL UNDERWEIGHT MODEL
# =============================================================================

print("\n7. FITTING FINAL UNDERWEIGHT MODEL")
print("-" * 100)

final_underweight_model = XGBClassifier(
    n_estimators=700,
    learning_rate=0.02,
    max_depth=3,
    min_child_weight=8,
    subsample=0.80,
    colsample_bytree=0.80,
    reg_alpha=1.0,
    reg_lambda=3.0,
    scale_pos_weight=UNDERWEIGHT_WEIGHT,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

final_underweight_model.fit(
    X_train_under_final,
    y_train_under
)

print("✓ Final underweight model fitted")
print("✓ Training data only")


# =============================================================================
# 8. VALIDATION PREDICTIONS
# =============================================================================

print("\n8. VALIDATION PREDICTIONS")
print("-" * 100)

stunting_val_prob_final = (
    final_stunting_model
    .predict_proba(X_val_stunt_final)[:, 1]
)

underweight_val_prob_final = (
    final_underweight_model
    .predict_proba(X_val_under_final)[:, 1]
)

stunting_val_pred_final = (
    stunting_val_prob_final >= STUNTING_THRESHOLD
).astype(int)

underweight_val_pred_final = (
    underweight_val_prob_final >= UNDERWEIGHT_THRESHOLD
).astype(int)

print("✓ Stunting validation predictions generated")
print("✓ Underweight validation predictions generated")


# =============================================================================
# 9. METRIC FUNCTION
# =============================================================================

def calculate_final_metrics(
    y_true,
    probabilities,
    predictions
):

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    ).ravel()

    return {
        "Accuracy": accuracy_score(
            y_true,
            predictions
        ),

        "Precision": precision_score(
            y_true,
            predictions,
            zero_division=0
        ),

        "Recall": recall_score(
            y_true,
            predictions,
            zero_division=0
        ),

        "F1": f1_score(
            y_true,
            predictions,
            zero_division=0
        ),

        "ROC_AUC": roc_auc_score(
            y_true,
            probabilities
        ),

        "PR_AUC": average_precision_score(
            y_true,
            probabilities
        ),

        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp,

        "FPR": (
            fp / (fp + tn)
            if (fp + tn) > 0
            else 0
        )
    }


# =============================================================================
# 10. STUNTING VALIDATION CONFIRMATION
# =============================================================================

stunting_validation_final = calculate_final_metrics(
    y_val_stunt,
    stunting_val_prob_final,
    stunting_val_pred_final
)

print("\n" + "=" * 100)
print("9. FINAL VALIDATION CONFIRMATION — STUNTING")
print("=" * 100)

for metric, value in stunting_validation_final.items():

    if metric in ["TN", "FP", "FN", "TP"]:
        print(f"{metric:10s}: {int(value)}")
    else:
        print(f"{metric:10s}: {value:.4f}")

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_val_stunt,
        stunting_val_pred_final
    )
)

print("\nClassification Report:")
print(
    classification_report(
        y_val_stunt,
        stunting_val_pred_final,
        target_names=[
            "Not Stunted",
            "Stunted"
        ],
        zero_division=0
    )
)


# =============================================================================
# 11. UNDERWEIGHT VALIDATION CONFIRMATION
# =============================================================================

underweight_validation_final = calculate_final_metrics(
    y_val_under,
    underweight_val_prob_final,
    underweight_val_pred_final
)

print("\n" + "=" * 100)
print("10. FINAL VALIDATION CONFIRMATION — UNDERWEIGHT")
print("=" * 100)

for metric, value in underweight_validation_final.items():

    if metric in ["TN", "FP", "FN", "TP"]:
        print(f"{metric:10s}: {int(value)}")
    else:
        print(f"{metric:10s}: {value:.4f}")

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_val_under,
        underweight_val_pred_final
    )
)

print("\nClassification Report:")
print(
    classification_report(
        y_val_under,
        underweight_val_pred_final,
        target_names=[
            "Not Underweight",
            "Underweight"
        ],
        zero_division=0
    )
)


# =============================================================================
# 12. FINAL VALIDATION SUMMARY
# =============================================================================

final_validation_summary = pd.DataFrame([
    {
        "Target": "Stunting",
        "Predictors": 15,
        "Scale_Pos_Weight": STUNTING_WEIGHT,
        "Threshold": STUNTING_THRESHOLD,
        **stunting_validation_final
    },
    {
        "Target": "Underweight",
        "Predictors": 15,
        "Scale_Pos_Weight": UNDERWEIGHT_WEIGHT,
        "Threshold": UNDERWEIGHT_THRESHOLD,
        **underweight_validation_final
    }
])

print("\n" + "=" * 100)
print("11. FINAL VALIDATION PERFORMANCE SUMMARY")
print("=" * 100)

print(
    final_validation_summary.round(4).to_string(index=False)
)


# =============================================================================
# 13. LOCKED-CONFIGURATION CHECK
# =============================================================================

print("\n" + "=" * 100)
print("12. LOCKED-CONFIGURATION CHECK")
print("=" * 100)

print("Stunting:")
print("  Predictors : 15")
print("  Weight     :", STUNTING_WEIGHT)
print("  Threshold  :", STUNTING_THRESHOLD)

print("\nUnderweight:")
print("  Predictors : 15")
print("  Weight     :", UNDERWEIGHT_WEIGHT)
print("  Threshold  :", UNDERWEIGHT_THRESHOLD)


# =============================================================================
# 14. TEST-SET INTEGRITY
# =============================================================================

print("\n" + "=" * 100)
print("13. FINAL TEST-SET INTEGRITY")
print("=" * 100)

print("✓ Test predictors were not accessed")
print("✓ Test targets were not accessed")
print("✓ Test probabilities were not generated")
print("✓ Test thresholds were not selected")
print("✓ Test performance was not calculated")
print("✓ No predictor selection was performed")
print("✓ No hyperparameter optimization was performed")
print("✓ No threshold optimization was performed")
print("✓ Validation data used only for confirmation")
print("✓ Final configurations remain locked")


# =============================================================================
# 15. SAVE FINAL DEVELOPMENT OBJECTS
# =============================================================================

final_models_7n = {
    "stunting_model": final_stunting_model,
    "underweight_model": final_underweight_model,
    "stunting_preprocessor": preprocessor_stunt_7l,
    "underweight_preprocessor": preprocessor_under_7l,
    "predictors": predictors_15,
    "stunting_weight": STUNTING_WEIGHT,
    "underweight_weight": UNDERWEIGHT_WEIGHT,
    "stunting_threshold": STUNTING_THRESHOLD,
    "underweight_threshold": UNDERWEIGHT_THRESHOLD
}

print("\n" + "=" * 100)
print("STEP 8 COMPLETE")
print("=" * 100)

print("\n✓ Final stunting model fitted")
print("✓ Final underweight model fitted")
print("✓ Locked configurations confirmed")
print("✓ Validation performance confirmed")
print("✓ Test set remains completely unseen")

print("\nNEXT STEP:")
print("STEP 9 — FINAL UNSEEN TEST-SET EVALUATION")
```

```python
# =============================================================================
# STEP 9: CAR MICS6 — FINAL UNSEEN TEST-SET EVALUATION
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

print("=" * 100)
print("STEP 9: CAR MICS6 — FINAL UNSEEN TEST-SET EVALUATION")
print("=" * 100)


# =============================================================================
# 1. REQUIRED OBJECT CHECK
# =============================================================================

print("\n1. REQUIRED OBJECT CHECK")
print("-" * 100)

required_objects = [
    "final_stunting_model",
    "final_underweight_model",
    "preprocessor_stunt_7l",
    "preprocessor_under_7l",
    "X_test_stunt",
    "y_test_stunt",
    "X_test_under",
    "y_test_under"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required Step 8/Step 4.3 objects: "
        + ", ".join(missing_objects)
    )

print("✓ Final models available")
print("✓ Test datasets available")
print("✓ Required preprocessors available")


# =============================================================================
# 2. LOCKED PREDICTORS
# =============================================================================

print("\n2. LOCKED PREDICTOR SET")
print("-" * 100)

predictors_15 = [
    "CAGE",
    "HL4",
    "HH6",
    "HH7",
    "windex5",
    "melevel",
    "caretakerdis",
    "religion",
    "ethnicity",
    "cdisability",
    "CA1",
    "CA14",
    "BD2",
    "IM2",
    "CA31"
]

print("Number of predictors:", len(predictors_15))

for i, predictor in enumerate(predictors_15, 1):
    print(f"{i:2d}. {predictor}")

print("✓ Predictor set unchanged")


# =============================================================================
# 3. LOCKED THRESHOLDS
# =============================================================================

STUNTING_THRESHOLD = 0.53
UNDERWEIGHT_THRESHOLD = 0.53

print("\n3. LOCKED THRESHOLDS")
print("-" * 100)

print("Stunting threshold    :", STUNTING_THRESHOLD)
print("Underweight threshold :", UNDERWEIGHT_THRESHOLD)

print("✓ Thresholds were fixed before test evaluation")
print("✓ No test-based threshold optimization")


# =============================================================================
# 4. TEST-SET SHAPES
# =============================================================================

print("\n4. TEST-SET SHAPES")
print("-" * 100)

print("Stunting:")
print("  X_test :", X_test_stunt.shape)
print("  y_test :", y_test_stunt.shape)

print("\nUnderweight:")
print("  X_test :", X_test_under.shape)
print("  y_test :", y_test_under.shape)


# =============================================================================
# 5. TEST-SET PREPROCESSING
# =============================================================================
#
# IMPORTANT:
# The preprocessors were already fitted using training data.
# Here they are ONLY used to transform the unseen test data.
#
# =============================================================================

print("\n5. TEST-SET PREPROCESSING")
print("-" * 100)

X_test_stunt_final = preprocessor_stunt_7l.transform(
    X_test_stunt[predictors_15]
)

X_test_under_final = preprocessor_under_7l.transform(
    X_test_under[predictors_15]
)

print(
    "Stunting test transformed   :",
    X_test_stunt_final.shape
)

print(
    "Underweight test transformed:",
    X_test_under_final.shape
)

print("✓ No test statistics used to fit preprocessing")
print("✓ Test data only transformed")


# =============================================================================
# 6. DATA INTEGRITY CHECK
# =============================================================================

print("\n6. TEST MATRIX INTEGRITY")
print("-" * 100)

stunt_nan = np.isnan(X_test_stunt_final).sum()
under_nan = np.isnan(X_test_under_final).sum()

stunt_inf = np.isinf(X_test_stunt_final).sum()
under_inf = np.isinf(X_test_under_final).sum()

print("Stunting test NaN       :", stunt_nan)
print("Stunting test Inf       :", stunt_inf)
print("Underweight test NaN    :", under_nan)
print("Underweight test Inf    :", under_inf)

if stunt_nan > 0 or stunt_inf > 0:
    raise ValueError(
        "Invalid values detected in transformed stunting test data."
    )

if under_nan > 0 or under_inf > 0:
    raise ValueError(
        "Invalid values detected in transformed underweight test data."
    )

print("✓ No NaN values")
print("✓ No infinite values")


# =============================================================================
# 7. GENERATE FINAL STUNTING TEST PREDICTIONS
# =============================================================================

print("\n7. FINAL UNSEEN TEST EVALUATION — STUNTING")
print("-" * 100)

stunting_test_prob = (
    final_stunting_model
    .predict_proba(X_test_stunt_final)[:, 1]
)

stunting_test_pred = (
    stunting_test_prob >= STUNTING_THRESHOLD
).astype(int)

print("✓ Stunting predictions generated")


# =============================================================================
# 8. GENERATE FINAL UNDERWEIGHT TEST PREDICTIONS
# =============================================================================

print("\n8. FINAL UNSEEN TEST EVALUATION — UNDERWEIGHT")
print("-" * 100)

underweight_test_prob = (
    final_underweight_model
    .predict_proba(X_test_under_final)[:, 1]
)

underweight_test_pred = (
    underweight_test_prob >= UNDERWEIGHT_THRESHOLD
).astype(int)

print("✓ Underweight predictions generated")


# =============================================================================
# 9. METRIC FUNCTION
# =============================================================================

def calculate_test_metrics(
    y_true,
    probabilities,
    predictions
):

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    ).ravel()

    fpr = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0
    )

    return {
        "Accuracy": accuracy_score(
            y_true,
            predictions
        ),

        "Precision": precision_score(
            y_true,
            predictions,
            zero_division=0
        ),

        "Recall": recall_score(
            y_true,
            predictions,
            zero_division=0
        ),

        "F1": f1_score(
            y_true,
            predictions,
            zero_division=0
        ),

        "ROC_AUC": roc_auc_score(
            y_true,
            probabilities
        ),

        "PR_AUC": average_precision_score(
            y_true,
            probabilities
        ),

        "FPR": fpr,

        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }


# =============================================================================
# 10. STUNTING FINAL TEST METRICS
# =============================================================================

stunting_test_results = calculate_test_metrics(
    y_test_stunt,
    stunting_test_prob,
    stunting_test_pred
)

print("\n" + "=" * 100)
print("9. FINAL UNSEEN TEST PERFORMANCE — STUNTING")
print("=" * 100)

for metric, value in stunting_test_results.items():

    if metric in ["TN", "FP", "FN", "TP"]:
        print(f"{metric:10s}: {int(value)}")
    else:
        print(f"{metric:10s}: {value:.4f}")

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test_stunt,
        stunting_test_pred
    )
)

print("\nClassification Report:")
print(
    classification_report(
        y_test_stunt,
        stunting_test_pred,
        target_names=[
            "Not Stunted",
            "Stunted"
        ],
        zero_division=0
    )
)


# =============================================================================
# 11. UNDERWEIGHT FINAL TEST METRICS
# =============================================================================

underweight_test_results = calculate_test_metrics(
    y_test_under,
    underweight_test_prob,
    underweight_test_pred
)

print("\n" + "=" * 100)
print("10. FINAL UNSEEN TEST PERFORMANCE — UNDERWEIGHT")
print("=" * 100)

for metric, value in underweight_test_results.items():

    if metric in ["TN", "FP", "FN", "TP"]:
        print(f"{metric:10s}: {int(value)}")
    else:
        print(f"{metric:10s}: {value:.4f}")

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test_under,
        underweight_test_pred
    )
)

print("\nClassification Report:")
print(
    classification_report(
        y_test_under,
        underweight_test_pred,
        target_names=[
            "Not Underweight",
            "Underweight"
        ],
        zero_division=0
    )
)


# =============================================================================
# 12. FINAL TEST PERFORMANCE SUMMARY
# =============================================================================

final_test_summary = pd.DataFrame([
    {
        "Target": "Stunting",
        "Predictors": 15,
        "Scale_Pos_Weight": 1.25,
        "Threshold": STUNTING_THRESHOLD,
        **stunting_test_results
    },
    {
        "Target": "Underweight",
        "Predictors": 15,
        "Scale_Pos_Weight": 2.00,
        "Threshold": UNDERWEIGHT_THRESHOLD,
        **underweight_test_results
    }
])

print("\n" + "=" * 100)
print("11. FINAL UNSEEN TEST-SET PERFORMANCE SUMMARY")
print("=" * 100)

print(
    final_test_summary.round(4).to_string(index=False)
)


# =============================================================================
# 13. VALIDATION VS TEST COMPARISON
# =============================================================================

print("\n" + "=" * 100)
print("12. VALIDATION VS FINAL UNSEEN TEST")
print("=" * 100)

comparison = pd.DataFrame([
    {
        "Target": "Stunting",

        "Val_Accuracy":
            stunting_validation_final["Accuracy"],

        "Test_Accuracy":
            stunting_test_results["Accuracy"],

        "Val_Precision":
            stunting_validation_final["Precision"],

        "Test_Precision":
            stunting_test_results["Precision"],

        "Val_Recall":
            stunting_validation_final["Recall"],

        "Test_Recall":
            stunting_test_results["Recall"],

        "Val_F1":
            stunting_validation_final["F1"],

        "Test_F1":
            stunting_test_results["F1"],

        "Val_ROC_AUC":
            stunting_validation_final["ROC_AUC"],

        "Test_ROC_AUC":
            stunting_test_results["ROC_AUC"],

        "Val_PR_AUC":
            stunting_validation_final["PR_AUC"],

        "Test_PR_AUC":
            stunting_test_results["PR_AUC"]
    },

    {
        "Target": "Underweight",

        "Val_Accuracy":
            underweight_validation_final["Accuracy"],

        "Test_Accuracy":
            underweight_test_results["Accuracy"],

        "Val_Precision":
            underweight_validation_final["Precision"],

        "Test_Precision":
            underweight_test_results["Precision"],

        "Val_Recall":
            underweight_validation_final["Recall"],

        "Test_Recall":
            underweight_test_results["Recall"],

        "Val_F1":
            underweight_validation_final["F1"],

        "Test_F1":
            underweight_test_results["F1"],

        "Val_ROC_AUC":
            underweight_validation_final["ROC_AUC"],

        "Test_ROC_AUC":
            underweight_test_results["ROC_AUC"],

        "Val_PR_AUC":
            underweight_validation_final["PR_AUC"],

        "Test_PR_AUC":
            underweight_test_results["PR_AUC"]
    }
])

print(
    comparison.round(4).to_string(index=False)
)


# =============================================================================
# 14. GENERALIZATION GAPS
# =============================================================================

print("\n" + "=" * 100)
print("13. GENERALIZATION GAPS")
print("=" * 100)

gap_table = pd.DataFrame([
    {
        "Target": "Stunting",

        "Accuracy_Gap":
            stunting_test_results["Accuracy"]
            - stunting_validation_final["Accuracy"],

        "Precision_Gap":
            stunting_test_results["Precision"]
            - stunting_validation_final["Precision"],

        "Recall_Gap":
            stunting_test_results["Recall"]
            - stunting_validation_final["Recall"],

        "F1_Gap":
            stunting_test_results["F1"]
            - stunting_validation_final["F1"],

        "ROC_AUC_Gap":
            stunting_test_results["ROC_AUC"]
            - stunting_validation_final["ROC_AUC"],

        "PR_AUC_Gap":
            stunting_test_results["PR_AUC"]
            - stunting_validation_final["PR_AUC"]
    },

    {
        "Target": "Underweight",

        "Accuracy_Gap":
            underweight_test_results["Accuracy"]
            - underweight_validation_final["Accuracy"],

        "Precision_Gap":
            underweight_test_results["Precision"]
            - underweight_validation_final["Precision"],

        "Recall_Gap":
            underweight_test_results["Recall"]
            - underweight_validation_final["Recall"],

        "F1_Gap":
            underweight_test_results["F1"]
            - underweight_validation_final["F1"],

        "ROC_AUC_Gap":
            underweight_test_results["ROC_AUC"]
            - underweight_validation_final["ROC_AUC"],

        "PR_AUC_Gap":
            underweight_test_results["PR_AUC"]
            - underweight_validation_final["PR_AUC"]
    }
])

print(
    gap_table.round(4).to_string(index=False)
)


# =============================================================================
# 15. FINAL INTEGRITY CHECK
# =============================================================================

print("\n" + "=" * 100)
print("14. FINAL EVALUATION INTEGRITY")
print("=" * 100)

print("✓ Predictor set fixed at 15")
print("✓ Stunting weight fixed at 1.25")
print("✓ Underweight weight fixed at 2.00")
print("✓ Stunting threshold fixed at 0.53")
print("✓ Underweight threshold fixed at 0.53")
print("✓ Preprocessors fitted before test transformation")
print("✓ No test-based imputation")
print("✓ No test-based feature selection")
print("✓ No test-based hyperparameter tuning")
print("✓ No test-based threshold tuning")
print("✓ Test set used only for final evaluation")
print("✓ Final test results are unbiased with respect to model selection")


# =============================================================================
# 16. SAVE FINAL RESULTS IN MEMORY
# =============================================================================

final_test_results = {
    "stunting": stunting_test_results,
    "underweight": underweight_test_results,
    "summary": final_test_summary.copy(),
    "comparison": comparison.copy(),
    "generalization_gaps": gap_table.copy()
}


print("\n" + "=" * 100)
print("STEP 9 COMPLETE")
print("=" * 100)

print("""
✓ Final stunting model evaluated on unseen test data
✓ Final underweight model evaluated on unseen test data
✓ Precision-focused thresholds remained fixed
✓ No test-set optimization performed
✓ Final generalization performance obtained
""")
```
