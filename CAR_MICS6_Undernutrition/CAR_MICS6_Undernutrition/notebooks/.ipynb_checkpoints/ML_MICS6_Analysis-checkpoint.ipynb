```python
files = {
    "CH": DATA_DIR / "ch.sav",
    "HH": DATA_DIR / "hh.sav",
    "WM": DATA_DIR / "wm.sav",
    "BH": DATA_DIR / "bh.sav"
}

for name, path in files.items():
    print(
        f"{'✓' if path.exists() else '✗'} {name}: {path}"
    )
```

```python
# =============================================================================
# STEP 1: LOAD CAR MICS6 DATASETS
# =============================================================================

print("=" * 90)
print("STEP 1: LOADING CAR MICS6 DATASETS")
print("=" * 90)

CH = pd.read_spss(files["CH"])
HH = pd.read_spss(files["HH"])
WM = pd.read_spss(files["WM"])
BH = pd.read_spss(files["BH"])

print("\nDATASET DIMENSIONS")
print("-" * 90)

print(f"CH (Child)          : {CH.shape}")
print(f"HH (Household)      : {HH.shape}")
print(f"WM (Women/Mothers)  : {WM.shape}")
print(f"BH (Birth History)  : {BH.shape}")

print("\n✓ All four CAR MICS6 datasets loaded successfully")
print("=" * 90)
```

```python
# =============================================================================
# STEP 1A: BASIC DATASET INSPECTION
# =============================================================================

for name, df in {
    "CH": CH,
    "HH": HH,
    "WM": WM,
    "BH": BH
}.items():

    print("\n" + "=" * 90)
    print(f"{name} DATASET")
    print("=" * 90)

    print("Shape:", df.shape)
    print("\nFirst 5 rows:")
    display(df.head())
```

```python
# =============================================================================
# STEP 2: IDENTIFY MICS6 DATASET LINKAGE VARIABLES
# =============================================================================

print("=" * 90)
print("STEP 2: MICS6 LINKAGE VARIABLE INSPECTION")
print("=" * 90)


# Variables commonly relevant for MICS6 linkage
possible_keys = [
    "HH1",
    "HH2",
    "LN",
    "LNWM",
    "LNBH",
    "UF1",
    "UF2",
    "UF3",
    "UF4",
    "UF5",
    "UF6",
    "WM1",
    "WM2",
    "WM3",
    "WM4",
    "WM5",
    "WM6",
    "BH0",
    "BH2"
]


# =============================================================================
# 1. CHECK VARIABLES IN EACH DATASET
# =============================================================================

datasets = {
    "CH": CH,
    "HH": HH,
    "WM": WM,
    "BH": BH
}

for dataset_name, df in datasets.items():

    print("\n" + "=" * 90)
    print(f"{dataset_name} DATASET")
    print("=" * 90)

    found = [
        variable
        for variable in possible_keys
        if variable in df.columns
    ]

    print("\nLinkage-related variables found:")

    for variable in found:
        print(f"✓ {variable}")

    print("\nNumber found:", len(found))


# =============================================================================
# 2. DISPLAY FIRST RECORDS OF KEY VARIABLES
# =============================================================================

print("\n" + "=" * 90)
print("KEY VARIABLE VALUES")
print("=" * 90)

for dataset_name, df in datasets.items():

    found = [
        variable
        for variable in possible_keys
        if variable in df.columns
    ]

    print(f"\n--- {dataset_name} ---")

    if found:
        display(df[found].head(10))
    else:
        print("No expected linkage variables found.")


# =============================================================================
# 3. CHECK DUPLICATION OF POTENTIAL KEYS
# =============================================================================

print("\n" + "=" * 90)
print("DUPLICATE KEY CHECK")
print("=" * 90)

for dataset_name, df in datasets.items():

    print(f"\n{dataset_name}")

    if "HH1" in df.columns and "HH2" in df.columns:

        household_key = (
            df[["HH1", "HH2"]]
            .astype(str)
            .agg("_".join, axis=1)
        )

        print(
            "HH1 + HH2 unique:",
            household_key.nunique(),
            "/",
            len(df)
        )

    if "LN" in df.columns:

        print(
            "LN unique:",
            df["LN"].nunique(),
            "/",
            len(df)
        )

print("\n" + "=" * 90)
print("STEP 2 COMPLETE")
print("=" * 90)
```

```python
# =============================================================================
# STEP 2B: INSPECT CHILD–MOTHER LINKAGE VARIABLES
# =============================================================================

print("=" * 90)
print("STEP 2B: CHILD–MOTHER LINKAGE INSPECTION")
print("=" * 90)

# -------------------------------------------------------------------------
# 1. Search CH for variables related to mother identification
# -------------------------------------------------------------------------

mother_keywords = [
    "mother",
    "mom",
    "moth",
    "LN",
    "UF"
]

mother_vars = []

for col in CH.columns:

    col_upper = str(col).upper()

    if any(
        keyword.upper() in col_upper
        for keyword in mother_keywords
    ):
        mother_vars.append(col)

print("\nCH variables potentially related to mother linkage:")
print("-" * 90)

for col in mother_vars:
    print(col)

print("\nNumber of potentially relevant variables:", len(mother_vars))


# -------------------------------------------------------------------------
# 2. Display exact CH variable names around the UF variables
# -------------------------------------------------------------------------

print("\n" + "=" * 90)
print("CH UF VARIABLES")
print("=" * 90)

uf_vars = [
    col for col in CH.columns
    if str(col).upper().startswith("UF")
]

for col in uf_vars:
    print(col)


# -------------------------------------------------------------------------
# 3. Display CH records with linkage variables
# -------------------------------------------------------------------------

ch_link_vars = [
    col for col in [
        "HH1",
        "HH2",
        "LN",
        "UF1",
        "UF2",
        "UF3",
        "UF4",
        "UF5",
        "UF6"
    ]
    if col in CH.columns
]

print("\n" + "=" * 90)
print("CH LINKAGE RECORDS")
print("=" * 90)

display(
    CH[ch_link_vars].head(20)
)


# -------------------------------------------------------------------------
# 4. Check whether UF variables identify the mother
# -------------------------------------------------------------------------

print("\n" + "=" * 90)
print("UF VARIABLE UNIQUENESS")
print("=" * 90)

for col in uf_vars:

    print(
        f"{col:10s} | "
        f"Unique: {CH[col].nunique(dropna=True):5d} | "
        f"Missing: {CH[col].isna().sum():5d}"
    )


print("\n" + "=" * 90)
print("STEP 2B COMPLETE")
print("=" * 90)
```

```python
# =============================================================================
# STEP 2C: INSPECT MICS6 VARIABLE LABELS
# =============================================================================

print("=" * 90)
print("STEP 2C: MICS6 VARIABLE LABEL INSPECTION")
print("=" * 90)


# Variables we need to understand
inspect_vars = [
    "HH1", "HH2", "LN",
    "UF1", "UF2", "UFINT", "UF3", "UF4", "UF5", "UF6",
    "WM1", "WM2", "WM3", "WM4", "WM5",
    "BH0", "BH2"
]


# =============================================================================
# 1. CH VARIABLE LABELS
# =============================================================================

print("\n1. CH VARIABLE LABELS")
print("-" * 90)

for var in inspect_vars:

    if var in CH.columns:

        label = CH[var].attrs.get("label", "No label stored")

        print(
            f"{var:10s} | "
            f"{label}"
        )


# =============================================================================
# 2. WM VARIABLE LABELS
# =============================================================================

print("\n2. WM VARIABLE LABELS")
print("-" * 90)

for var in inspect_vars:

    if var in WM.columns:

        label = WM[var].attrs.get("label", "No label stored")

        print(
            f"{var:10s} | "
            f"{label}"
        )


# =============================================================================
# 3. BH VARIABLE LABELS
# =============================================================================

print("\n3. BH VARIABLE LABELS")
print("-" * 90)

for var in inspect_vars:

    if var in BH.columns:

        label = BH[var].attrs.get("label", "No label stored")

        print(
            f"{var:10s} | "
            f"{label}"
        )


# =============================================================================
# 4. DISPLAY ALL CH VARIABLES STARTING WITH UF
# =============================================================================

print("\n4. CH UF VARIABLES AND LABELS")
print("-" * 90)

for var in CH.columns:

    if str(var).upper().startswith("UF"):

        label = CH[var].attrs.get("label", "No label stored")

        print(
            f"{var:10s} | {label}"
        )


print("\n" + "=" * 90)
print("STEP 2C COMPLETE")
print("=" * 90)
```

```python
# =============================================================================
# STEP 2D: COMPARE CHILD AND WOMEN LINKAGE STRUCTURE
# =============================================================================

print("=" * 90)
print("STEP 2D: CH–WM LINKAGE STRUCTURE")
print("=" * 90)


# =============================================================================
# 1. CH LINKAGE COLUMNS
# =============================================================================

print("\n1. CH LINKAGE VARIABLES")
print("-" * 90)

ch_keys = [
    "HH1",
    "HH2",
    "LN",
    "UF1",
    "UF2",
    "UF3",
    "UF4",
    "UF5",
    "UF6"
]

display(CH[ch_keys].head(20))


# =============================================================================
# 2. WM LINKAGE COLUMNS
# =============================================================================

print("\n2. WM LINKAGE VARIABLES")
print("-" * 90)

wm_keys = [
    "HH1",
    "HH2",
    "LN",
    "WM1",
    "WM2",
    "WM3",
    "WM4",
    "WM5"
]

display(WM[wm_keys].head(20))


# =============================================================================
# 3. CHECK WHETHER CH UF1/UF2 MATCH HOUSEHOLD IDENTIFIERS
# =============================================================================

print("\n3. CHECK CH UF1/UF2 AGAINST HH1/HH2")
print("-" * 90)

print(
    "UF1 equals HH1:",
    (CH["UF1"] == CH["HH1"]).mean()
)

print(
    "UF2 equals HH2:",
    (CH["UF2"] == CH["HH2"]).mean()
)


# =============================================================================
# 4. CHECK CH UF3 AGAINST CHILD LINE NUMBER
# =============================================================================

print("\n4. CHECK CH UF3 AGAINST LN")
print("-" * 90)

print(
    "UF3 equals LN:",
    (CH["UF3"] == CH["LN"]).mean()
)


# =============================================================================
# 5. CHECK WM IDENTIFIER STRUCTURE
# =============================================================================

print("\n5. WM IDENTIFIER UNIQUENESS")
print("-" * 90)

for key_set in [
    ["HH1", "HH2"],
    ["HH1", "HH2", "LN"],
    ["HH1", "HH2", "WM1", "WM2", "WM3"],
    ["HH1", "HH2", "LN", "WM1", "WM2", "WM3"]
]:

    temp = WM[key_set].astype(str).agg(
        "_".join,
        axis=1
    )

    print(
        f"{'+'.join(key_set):40s} | "
        f"Unique: {temp.nunique():5d} | "
        f"Rows: {len(temp):5d}"
    )


# =============================================================================
# 6. CHECK BH IDENTIFIER STRUCTURE
# =============================================================================

print("\n6. BH IDENTIFIER UNIQUENESS")
print("-" * 90)

for key_set in [
    ["HH1", "HH2", "LN"],
    ["HH1", "HH2", "LN", "WM1", "WM2", "WM3"],
    ["HH1", "HH2", "LN", "WM1", "WM2", "WM3", "BH0"]
]:

    temp = BH[key_set].astype(str).agg(
        "_".join,
        axis=1
    )

    print(
        f"{'+'.join(key_set):50s} | "
        f"Unique: {temp.nunique():5d} | "
        f"Rows: {len(temp):5d}"
    )


print("\n" + "=" * 90)
print("STEP 2D COMPLETE")
print("=" * 90)
```

```python
# =============================================================================
# STEP 2E: VERIFY CH–WM LINKAGE
# =============================================================================

print("=" * 90)
print("STEP 2E: VERIFYING CHILD–WOMAN LINKAGE")
print("=" * 90)


# =============================================================================
# 1. CREATE HOUSEHOLD + LINE KEYS
# =============================================================================

CH_test = CH[
    ["HH1", "HH2", "LN"]
].copy()

WM_test = WM[
    ["HH1", "HH2", "LN"]
].copy()


CH_test["CH_KEY"] = (
    CH_test["HH1"].astype(str)
    + "_"
    + CH_test["HH2"].astype(str)
    + "_"
    + CH_test["LN"].astype(str)
)

WM_test["WM_KEY"] = (
    WM_test["HH1"].astype(str)
    + "_"
    + WM_test["HH2"].astype(str)
    + "_"
    + WM_test["LN"].astype(str)
)


# =============================================================================
# 2. CHECK HOW MANY CH RECORDS HAVE A WM MATCH
# =============================================================================

wm_keys = set(WM_test["WM_KEY"])

CH_test["WM_MATCH"] = CH_test["CH_KEY"].isin(wm_keys)

print("\n1. CHILD RECORDS MATCHING WM")
print("-" * 90)

print(
    "CH records:",
    len(CH_test)
)

print(
    "Matched:",
    CH_test["WM_MATCH"].sum()
)

print(
    "Not matched:",
    (~CH_test["WM_MATCH"]).sum()
)

print(
    "Match rate:",
    f"{CH_test['WM_MATCH'].mean() * 100:.2f}%"
)


# =============================================================================
# 3. CHECK WHETHER WM MATCHES ARE UNIQUE
# =============================================================================

print("\n2. WM KEY UNIQUENESS")
print("-" * 90)

wm_duplicate_keys = (
    WM_test["WM_KEY"]
    .duplicated()
    .sum()
)

print(
    "Duplicate WM linkage keys:",
    wm_duplicate_keys
)


# =============================================================================
# 4. DISPLAY MATCHED AND UNMATCHED EXAMPLES
# =============================================================================

print("\n3. MATCHED CH RECORDS")
print("-" * 90)

display(
    CH_test[
        CH_test["WM_MATCH"]
    ].head(10)
)


print("\n4. UNMATCHED CH RECORDS")
print("-" * 90)

display(
    CH_test[
        ~CH_test["WM_MATCH"]
    ].head(10)
)


# =============================================================================
# 5. CHECK NUMBER OF CH CHILDREN PER POTENTIAL WM KEY
# =============================================================================

print("\n5. CH RECORDS PER HH1 + HH2 + LN")
print("-" * 90)

ch_counts = (
    CH_test
    .groupby(
        ["HH1", "HH2", "LN"]
    )
    .size()
    .reset_index(
        name="CH_COUNT"
    )
)

print(
    "Maximum CH records per key:",
    ch_counts["CH_COUNT"].max()
)

print(
    "Keys with more than one CH record:",
    (ch_counts["CH_COUNT"] > 1).sum()
)


print("\n" + "=" * 90)
print("STEP 2E COMPLETE")
print("=" * 90)
```

```python
# =============================================================================
# STEP 2F: READ ORIGINAL SPSS METADATA
# =============================================================================

import pyreadstat

print("=" * 90)
print("STEP 2F: READING ORIGINAL SPSS VARIABLE METADATA")
print("=" * 90)


# =============================================================================
# 1. READ CH METADATA
# =============================================================================

CH_meta_data, CH_meta = pyreadstat.read_sav(
    files["CH"],
    metadataonly=True
)

print("\nCH METADATA")
print("-" * 90)

for var in [
    "HH1", "HH2", "LN",
    "UF1", "UF2", "UFINT",
    "UF3", "UF4", "UF5", "UF6"
]:

    if var in CH_meta.column_names:

        label = CH_meta.column_names_to_labels.get(
            var,
            "No label"
        )

        print(
            f"{var:10s} | {label}"
        )


# =============================================================================
# 2. READ WM METADATA
# =============================================================================

WM_meta_data, WM_meta = pyreadstat.read_sav(
    files["WM"],
    metadataonly=True
)

print("\nWM METADATA")
print("-" * 90)

for var in [
    "HH1", "HH2", "LN",
    "WM1", "WM2", "WM3", "WM4", "WM5"
]:

    if var in WM_meta.column_names:

        label = WM_meta.column_names_to_labels.get(
            var,
            "No label"
        )

        print(
            f"{var:10s} | {label}"
        )


# =============================================================================
# 3. READ BH METADATA
# =============================================================================

BH_meta_data, BH_meta = pyreadstat.read_sav(
    files["BH"],
    metadataonly=True
)

print("\nBH METADATA")
print("-" * 90)

for var in [
    "HH1", "HH2", "LN",
    "WM1", "WM2", "WM3",
    "BH0", "BH2"
]:

    if var in BH_meta.column_names:

        label = BH_meta.column_names_to_labels.get(
            var,
            "No label"
        )

        print(
            f"{var:10s} | {label}"
        )


print("\n" + "=" * 90)
print("STEP 2F COMPLETE")
print("=" * 90)
```

```python
# =============================================================================
# STEP 2G: VERIFY CORRECT CH → WM MOTHER LINKAGE
# =============================================================================

print("=" * 90)
print("STEP 2G: VERIFYING CHILD–MOTHER LINKAGE")
print("=" * 90)


# =============================================================================
# 1. CREATE THE CORRECT MOTHER LINKAGE KEY
# =============================================================================

CH_mother = CH[
    ["HH1", "HH2", "LN", "UF1", "UF2", "UF4"]
].copy()

WM_mother = WM[
    ["WM1", "WM2", "WM3"]
].copy()


CH_mother["MOTHER_KEY"] = (
    CH_mother["UF1"].astype(str)
    + "_"
    + CH_mother["UF2"].astype(str)
    + "_"
    + CH_mother["UF4"].astype(str)
)

WM_mother["MOTHER_KEY"] = (
    WM_mother["WM1"].astype(str)
    + "_"
    + WM_mother["WM2"].astype(str)
    + "_"
    + WM_mother["WM3"].astype(str)
)


# =============================================================================
# 2. CHECK WM KEY UNIQUENESS
# =============================================================================

print("\n1. WM MOTHER KEY")
print("-" * 90)

print(
    "WM records:",
    len(WM_mother)
)

print(
    "Unique WM mother keys:",
    WM_mother["MOTHER_KEY"].nunique()
)

print(
    "Duplicate WM mother keys:",
    WM_mother["MOTHER_KEY"].duplicated().sum()
)


# =============================================================================
# 3. CHECK CH MATCH RATE
# =============================================================================

wm_keys = set(
    WM_mother["MOTHER_KEY"]
)

CH_mother["WM_MATCH"] = (
    CH_mother["MOTHER_KEY"].isin(wm_keys)
)


print("\n2. CH → WM MATCHING")
print("-" * 90)

matched = CH_mother["WM_MATCH"].sum()
total = len(CH_mother)

print("CH records       :", total)
print("Matched          :", matched)
print("Not matched      :", total - matched)
print(
    "Match rate       :",
    f"{matched / total * 100:.2f}%"
)


# =============================================================================
# 4. DISPLAY MATCHED EXAMPLES
# =============================================================================

print("\n3. MATCHED CHILD–MOTHER EXAMPLES")
print("-" * 90)

display(
    CH_mother[
        CH_mother["WM_MATCH"]
    ].head(20)
)


# =============================================================================
# 5. DISPLAY UNMATCHED EXAMPLES
# =============================================================================

print("\n4. UNMATCHED CHILD–MOTHER EXAMPLES")
print("-" * 90)

display(
    CH_mother[
        ~CH_mother["WM_MATCH"]
    ].head(20)
)


print("\n" + "=" * 90)
print("STEP 2G COMPLETE")
print("=" * 90)
```

```python
# =============================================================================
# STEP 2H: FINAL LINKAGE VALIDATION BEFORE INTEGRATION
# =============================================================================

print("=" * 90)
print("STEP 2H: FINAL CH–HH–WM LINKAGE VALIDATION")
print("=" * 90)


# =============================================================================
# 1. CREATE HOUSEHOLD KEYS
# =============================================================================

CH_hh = CH[["HH1", "HH2"]].copy()
HH_hh = HH[["HH1", "HH2"]].copy()

CH_hh["HH_KEY"] = (
    CH_hh["HH1"].astype(str)
    + "_"
    + CH_hh["HH2"].astype(str)
)

HH_hh["HH_KEY"] = (
    HH_hh["HH1"].astype(str)
    + "_"
    + HH_hh["HH2"].astype(str)
)


# =============================================================================
# 2. CHECK CH → HH MATCHING
# =============================================================================

hh_keys = set(HH_hh["HH_KEY"])

CH_hh["HH_MATCH"] = CH_hh["HH_KEY"].isin(hh_keys)

print("\n1. CH → HH MATCHING")
print("-" * 90)

print("CH records:", len(CH_hh))
print("Matched:", CH_hh["HH_MATCH"].sum())
print("Not matched:", (~CH_hh["HH_MATCH"]).sum())

print(
    "Match rate:",
    f"{CH_hh['HH_MATCH'].mean() * 100:.2f}%"
)


# =============================================================================
# 3. VERIFY HH KEY IS UNIQUE
# =============================================================================

print("\n2. HOUSEHOLD KEY UNIQUENESS")
print("-" * 90)

print(
    "HH records:",
    len(HH_hh)
)

print(
    "Unique HH keys:",
    HH_hh["HH_KEY"].nunique()
)

print(
    "Duplicate HH keys:",
    HH_hh["HH_KEY"].duplicated().sum()
)


# =============================================================================
# 4. VERIFY WM MOTHER KEY
# =============================================================================

print("\n3. WM MOTHER KEY")
print("-" * 90)

print(
    "WM records:",
    len(WM)
)

print(
    "Unique WM mother keys:",
    WM_mother["MOTHER_KEY"].nunique()
)

print(
    "Duplicate WM mother keys:",
    WM_mother["MOTHER_KEY"].duplicated().sum()
)


# =============================================================================
# 5. CHECK NUMBER OF CHILDREN PER MOTHER
# =============================================================================

print("\n4. CHILDREN PER MOTHER")
print("-" * 90)

children_per_mother = (
    CH_mother[
        CH_mother["WM_MATCH"]
    ]
    .groupby("MOTHER_KEY")
    .size()
)

print(
    "Matched mothers:",
    children_per_mother.shape[0]
)

print(
    "Maximum children per mother:",
    children_per_mother.max()
)

print(
    "Mean children per mother:",
    round(
        children_per_mother.mean(),
        2
    )
)

print(
    "Mothers with multiple children:",
    (children_per_mother > 1).sum()
)


# =============================================================================
# 6. FINAL VALIDATION
# =============================================================================

print("\n5. INTEGRATION SAFETY CHECK")
print("-" * 90)

if HH_hh["HH_KEY"].duplicated().sum() == 0:
    print("✓ HH is safe for many-to-one household merge")
else:
    print("✗ HH contains duplicate household keys")

if WM_mother["MOTHER_KEY"].duplicated().sum() == 0:
    print("✓ WM is safe for many-to-one mother merge")
else:
    print("✗ WM contains duplicate mother keys")

if CH_mother["WM_MATCH"].sum() > 0:
    print("✓ CH → WM linkage established")

print("\n" + "=" * 90)
print("STEP 2H COMPLETE")
print("=" * 90)
```

```python
# =============================================================================
# STEP 2I: BUILD CHILD-LEVEL CH + HH + WM INTEGRATED DATASET
# =============================================================================

print("=" * 90)
print("STEP 2I: BUILDING CHILD-LEVEL INTEGRATED DATASET")
print("=" * 90)


# =============================================================================
# 1. CREATE WORKING COPIES
# =============================================================================

CH_int = CH.copy()
HH_int = HH.copy()
WM_int = WM.copy()

print("\n1. ORIGINAL DATASET SIZES")
print("-" * 90)

print("CH:", CH_int.shape)
print("HH:", HH_int.shape)
print("WM:", WM_int.shape)


# =============================================================================
# 2. CREATE HOUSEHOLD MERGE KEY
# =============================================================================

CH_int["_HH_KEY"] = (
    CH_int["HH1"].astype(str)
    + "_"
    + CH_int["HH2"].astype(str)
)

HH_int["_HH_KEY"] = (
    HH_int["HH1"].astype(str)
    + "_"
    + HH_int["HH2"].astype(str)
)


# =============================================================================
# 3. REMOVE DUPLICATE HH VARIABLES BEFORE MERGE
# =============================================================================

hh_columns_to_add = [
    col for col in HH_int.columns
    if col not in ["HH1", "HH2", "_HH_KEY"]
]

HH_for_merge = HH_int[
    ["_HH_KEY"] + hh_columns_to_add
].copy()


# =============================================================================
# 4. MERGE HOUSEHOLD DATA INTO CHILD DATA
# =============================================================================

print("\n2. MERGING HOUSEHOLD DATA")
print("-" * 90)

integrated = CH_int.merge(
    HH_for_merge,
    on="_HH_KEY",
    how="left",
    validate="many_to_one",
    indicator="_HH_MERGE"
)

print("✓ Household merge completed")
print("Rows before:", len(CH_int))
print("Rows after :", len(integrated))


# Validate row preservation
if len(integrated) != len(CH_int):
    raise ValueError(
        "ERROR: Household merge changed the number of child records."
    )

print("✓ Child-level row count preserved")


# =============================================================================
# 5. CREATE MOTHER MERGE KEY
# =============================================================================

integrated["_MOTHER_KEY"] = (
    integrated["UF1"].astype(str)
    + "_"
    + integrated["UF2"].astype(str)
    + "_"
    + integrated["UF4"].astype(str)
)

WM_int["_MOTHER_KEY"] = (
    WM_int["WM1"].astype(str)
    + "_"
    + WM_int["WM2"].astype(str)
    + "_"
    + WM_int["WM3"].astype(str)
)


# =============================================================================
# 6. SELECT WM VARIABLES
# =============================================================================

wm_columns_to_add = [
    col for col in WM_int.columns
    if col not in [
        "WM1",
        "WM2",
        "WM3",
        "_MOTHER_KEY"
    ]
]

WM_for_merge = WM_int[
    ["_MOTHER_KEY"] + wm_columns_to_add
].copy()


# =============================================================================
# 7. MERGE MATERNAL DATA
# =============================================================================

print("\n3. MERGING MATERNAL DATA")
print("-" * 90)

integrated = integrated.merge(
    WM_for_merge,
    on="_MOTHER_KEY",
    how="left",
    validate="many_to_one",
    indicator="_WM_MERGE"
)

print("✓ Maternal merge completed")
print("Rows after WM merge:", len(integrated))


# =============================================================================
# 8. VERIFY ROW PRESERVATION
# =============================================================================

print("\n4. ROW PRESERVATION CHECK")
print("-" * 90)

print("Original CH rows :", len(CH))
print("Integrated rows  :", len(integrated))

if len(integrated) == len(CH):
    print("✓ EXACTLY 9,037 child records preserved")
else:
    raise ValueError(
        "ERROR: Integrated dataset does not preserve CH row count."
    )


# =============================================================================
# 9. MERGE QUALITY
# =============================================================================

print("\n5. MERGE QUALITY")
print("-" * 90)

print("Household matches:")

display(
    integrated["_HH_MERGE"]
    .value_counts()
)

print("\nMaternal matches:")

display(
    integrated["_WM_MERGE"]
    .value_counts()
)


# =============================================================================
# 10. CHECK MATERNAL MISSINGNESS
# =============================================================================

print("\n6. MATERNAL LINKAGE RESULT")
print("-" * 90)

wm_matched = (
    integrated["_WM_MERGE"] == "both"
).sum()

wm_unmatched = (
    integrated["_WM_MERGE"] != "both"
).sum()

print("Children with WM match    :", wm_matched)
print("Children without WM match:", wm_unmatched)

print(
    "WM match rate:",
    f"{wm_matched / len(integrated) * 100:.2f}%"
)


# =============================================================================
# 11. REMOVE TEMPORARY MERGE COLUMNS
# =============================================================================

integrated.drop(
    columns=[
        "_HH_KEY",
        "_MOTHER_KEY",
        "_HH_MERGE",
        "_WM_MERGE"
    ],
    inplace=True
)


# =============================================================================
# 12. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 90)
print("STEP 2I COMPLETE")
print("=" * 90)

print("Integrated dataset:", integrated.shape)
print("Child records     :", len(integrated))
print("✓ CH retained as master dataset")
print("✓ HH merged using HH1 + HH2")
print("✓ WM merged using UF1 + UF2 + UF4 → WM1 + WM2 + WM3")
print("✓ No child records duplicated")
print("✓ Unmatched maternal records retained as missing")
print("=" * 90)
```

```python
# =============================================================================
# STEP 3A: CONSTRUCT STUNTING AND UNDERWEIGHT TARGETS
# =============================================================================

print("=" * 100)
print("STEP 3A: CONSTRUCTING STUNTING AND UNDERWEIGHT TARGETS")
print("=" * 100)


# =============================================================================
# 1. CREATE WORKING DATASET
# =============================================================================

analysis = integrated.copy()

print("\n1. INITIAL DATA")
print("-" * 100)
print("Children:", len(analysis))


# =============================================================================
# 2. VALID ANTHROPOMETRIC OBSERVATIONS
# =============================================================================

analysis["HAZ_VALID"] = (
    analysis["HAZ"].notna()
    & analysis["HAZFLAG"].astype(str).str.lower().eq("no error")
)

analysis["WAZ_VALID"] = (
    analysis["WAZ"].notna()
    & analysis["WAZFLAG"].astype(str).str.lower().eq("no error")
)

print("\n2. ANTHROPOMETRIC VALIDITY")
print("-" * 100)

print("Valid HAZ:", analysis["HAZ_VALID"].sum())
print("Invalid/missing HAZ:", (~analysis["HAZ_VALID"]).sum())

print("Valid WAZ:", analysis["WAZ_VALID"].sum())
print("Invalid/missing WAZ:", (~analysis["WAZ_VALID"]).sum())


# =============================================================================
# 3. CREATE STUNTING TARGET
# =============================================================================
# Stunting: HAZ < -2

analysis["stunted"] = np.nan

analysis.loc[
    analysis["HAZ_VALID"],
    "stunted"
] = (
    analysis.loc[
        analysis["HAZ_VALID"],
        "HAZ"
    ] < -2
).astype(int)


# =============================================================================
# 4. CREATE UNDERWEIGHT TARGET
# =============================================================================
# Underweight: WAZ < -2

analysis["underweight"] = np.nan

analysis.loc[
    analysis["WAZ_VALID"],
    "underweight"
] = (
    analysis.loc[
        analysis["WAZ_VALID"],
        "WAZ"
    ] < -2
).astype(int)


# =============================================================================
# 5. STUNTING DISTRIBUTION
# =============================================================================

print("\n" + "=" * 100)
print("3. STUNTING TARGET")
print("=" * 100)

print(
    analysis["stunted"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nPercentages:")

display(
    (
        analysis["stunted"]
        .value_counts(normalize=True, dropna=False)
        .sort_index()
        * 100
    ).round(2)
)


# =============================================================================
# 6. UNDERWEIGHT DISTRIBUTION
# =============================================================================

print("\n" + "=" * 100)
print("4. UNDERWEIGHT TARGET")
print("=" * 100)

print(
    analysis["underweight"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nPercentages:")

display(
    (
        analysis["underweight"]
        .value_counts(normalize=True, dropna=False)
        .sort_index()
        * 100
    ).round(2)
)


# =============================================================================
# 7. FINAL TARGET SAMPLE SIZES
# =============================================================================

print("\n" + "=" * 100)
print("5. FINAL TARGET SAMPLE SIZES")
print("=" * 100)

stunting_n = analysis["stunted"].notna().sum()
underweight_n = analysis["underweight"].notna().sum()

print("Total CH children :", len(analysis))
print("Stunting sample   :", stunting_n)
print("Underweight sample:", underweight_n)


# =============================================================================
# 8. CHECK TARGET VALUES
# =============================================================================

print("\n" + "=" * 100)
print("6. TARGET VALIDATION")
print("=" * 100)

stunting_values = set(
    analysis["stunted"]
    .dropna()
    .astype(int)
    .unique()
)

underweight_values = set(
    analysis["underweight"]
    .dropna()
    .astype(int)
    .unique()
)

print(
    "Stunting values:",
    sorted(stunting_values)
)

print(
    "Underweight values:",
    sorted(underweight_values)
)

if stunting_values <= {0, 1}:
    print("✓ Stunting is binary")

if underweight_values <= {0, 1}:
    print("✓ Underweight is binary")


print("\n" + "=" * 100)
print("STEP 3A COMPLETE")
print("=" * 100)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# ============================================================
# STEP 3K: TRAIN/TEST SPLIT AND PREPROCESSING DESIGN
# ============================================================

print("=" * 100)
print("STEP 3K: TRAIN/TEST SPLIT AND PREPROCESSING DESIGN")
print("=" * 100)

# ------------------------------------------------------------
# FINAL FEATURES
# ------------------------------------------------------------

numeric_features = [
    "CAGE",
    "WB4",
    "WAGEM",
    "CEB",
    "CSURV",
    "CDEAD",
    "CM11"
]

categorical_features = [
    "HL4",
    "HH6",
    "windex5",
    "WS1",
    "WS11",
    "WS15",
    "welevel",
    "MSTATUS",
    "CM17",
    "insurance"
]

predictors = numeric_features + categorical_features


# ------------------------------------------------------------
# PREPROCESSING PIPELINE
# ------------------------------------------------------------

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(
        strategy="constant",
        fill_value="Missing"
    )),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])


# ------------------------------------------------------------
# FUNCTION FOR TARGET-SPECIFIC SPLIT
# ------------------------------------------------------------

def prepare_target_data(data, target):

    print("\n" + "=" * 100)
    print(f"PREPARING TARGET: {target.upper()}")
    print("=" * 100)

    # Keep only children with a valid target
    target_data = data[
        data[target].notna()
    ].copy()

    X = target_data[predictors]
    y = target_data[target].astype(int)

    print(f"\nValid records: {len(target_data)}")

    print("\nTarget distribution:")
    print(y.value_counts())

    print("\nTarget proportions:")
    print(y.value_counts(normalize=True).round(4))

    # --------------------------------------------------------
    # STRATIFIED TRAIN/TEST SPLIT
    # --------------------------------------------------------

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    print("\nTrain/Test split:")
    print(f"Training records: {len(X_train)}")
    print(f"Testing records : {len(X_test)}")

    print("\nTraining target distribution:")
    print(y_train.value_counts(normalize=True).round(4))

    print("\nTesting target distribution:")
    print(y_test.value_counts(normalize=True).round(4))

    return X_train, X_test, y_train, y_test


# ------------------------------------------------------------
# STUNTING
# ------------------------------------------------------------

X_train_stunt, X_test_stunt, y_train_stunt, y_test_stunt = \
    prepare_target_data(
        clean_data,
        "stunted"
    )


# ------------------------------------------------------------
# UNDERWEIGHT
# ------------------------------------------------------------

X_train_under, X_test_under, y_train_under, y_test_under = \
    prepare_target_data(
        clean_data,
        "underweight"
    )


print("\n" + "=" * 100)
print("STEP 3K COMPLETE")
print("=" * 100)

print("✓ Stunting data split separately")
print("✓ Underweight data split separately")
print("✓ Stratification applied")
print("✓ 80/20 train/test split")
print("✓ Preprocessing pipeline defined")
print("✓ Median imputation restricted to training data")
print("✓ Categorical missingness represented explicitly")
print("✓ One-hot encoding configured")
print("✓ Unknown test categories handled safely")
print("✓ Ready for model training")
```

```python
# ============================================================
# STEP 3L: FITTING AND VALIDATING PREPROCESSING PIPELINE
# ============================================================

print("=" * 100)
print("STEP 3L: FITTING AND VALIDATING PREPROCESSING PIPELINE")
print("=" * 100)


# ------------------------------------------------------------
# 1. FIT PREPROCESSOR ON STUNTING TRAINING DATA
# ------------------------------------------------------------

print("\n1. STUNTING PREPROCESSING")
print("-" * 100)

preprocessor_stunt = preprocessor.fit(X_train_stunt)

X_train_stunt_processed = preprocessor_stunt.transform(
    X_train_stunt
)

X_test_stunt_processed = preprocessor_stunt.transform(
    X_test_stunt
)

print("✓ Preprocessor fitted using training data only")

print(
    f"Training matrix shape: "
    f"{X_train_stunt_processed.shape}"
)

print(
    f"Testing matrix shape : "
    f"{X_test_stunt_processed.shape}"
)


# ------------------------------------------------------------
# 2. FIT SEPARATE PREPROCESSOR ON UNDERWEIGHT TRAINING DATA
# ------------------------------------------------------------

print("\n2. UNDERWEIGHT PREPROCESSING")
print("-" * 100)

preprocessor_under = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

preprocessor_under.fit(X_train_under)

X_train_under_processed = preprocessor_under.transform(
    X_train_under
)

X_test_under_processed = preprocessor_under.transform(
    X_test_under
)

print("✓ Preprocessor fitted using training data only")

print(
    f"Training matrix shape: "
    f"{X_train_under_processed.shape}"
)

print(
    f"Testing matrix shape : "
    f"{X_test_under_processed.shape}"
)


# ------------------------------------------------------------
# 3. CHECK FOR REMAINING MISSING VALUES
# ------------------------------------------------------------

print("\n3. REMAINING MISSING VALUES")
print("-" * 100)

print(
    "Stunting training missing:",
    np.isnan(X_train_stunt_processed).sum()
)

print(
    "Stunting testing missing:",
    np.isnan(X_test_stunt_processed).sum()
)

print(
    "Underweight training missing:",
    np.isnan(X_train_under_processed).sum()
)

print(
    "Underweight testing missing:",
    np.isnan(X_test_under_processed).sum()
)


# ------------------------------------------------------------
# 4. CHECK TARGET ALIGNMENT
# ------------------------------------------------------------

print("\n4. TARGET ALIGNMENT")
print("-" * 100)

print(
    "Stunting:",
    len(X_train_stunt_processed),
    len(y_train_stunt)
)

print(
    "Stunting test:",
    len(X_test_stunt_processed),
    len(y_test_stunt)
)

print(
    "Underweight:",
    len(X_train_under_processed),
    len(y_train_under)
)

print(
    "Underweight test:",
    len(X_test_under_processed),
    len(y_test_under)
)


# ------------------------------------------------------------
# 5. FINAL VALIDATION
# ------------------------------------------------------------

print("\n5. PREPROCESSING VALIDATION")
print("-" * 100)

if (
    not np.isnan(X_train_stunt_processed).any()
    and
    not np.isnan(X_test_stunt_processed).any()
    and
    not np.isnan(X_train_under_processed).any()
    and
    not np.isnan(X_test_under_processed).any()
):
    print("✓ No missing values remain after preprocessing")

print("✓ Numeric variables imputed")
print("✓ Categorical variables imputed")
print("✓ Categorical variables one-hot encoded")
print("✓ Numeric variables standardized")
print("✓ Test data transformed using training-fitted preprocessing")
print("✓ No target variables included in preprocessing")

print("\n" + "=" * 100)
print("STEP 3L COMPLETE")
print("=" * 100)

# ------------------------------------------------------------
# 6. SAVE CHECKPOINT (allows resuming after a kernel restart
#    without rerunning STEP 0-3K)
# ------------------------------------------------------------

import joblib

checkpoint = {

    "numeric_features": numeric_features,
    "categorical_features": categorical_features,
    "predictors": predictors,

    "X_train_stunt": X_train_stunt,
    "X_test_stunt": X_test_stunt,
    "y_train_stunt": y_train_stunt,
    "y_test_stunt": y_test_stunt,

    "X_train_under": X_train_under,
    "X_test_under": X_test_under,
    "y_train_under": y_train_under,
    "y_test_under": y_test_under,

    "preprocessor_stunt": preprocessor_stunt,
    "preprocessor_under": preprocessor_under,

    "X_train_stunt_processed": X_train_stunt_processed,
    "X_test_stunt_processed": X_test_stunt_processed,

    "X_train_under_processed": X_train_under_processed,
    "X_test_under_processed": X_test_under_processed
}

joblib.dump(
    checkpoint,
    DATA_DIR / "CAR_MICS6_STEP3L_CHECKPOINT.pkl"
)

print("\n✓ Checkpoint saved to CAR_MICS6_STEP3L_CHECKPOINT.pkl")
print("  (run the STEP 3L-RESTORE cell below to resume from here")
print("   after a kernel restart, instead of rerunning STEP 0–3K)")
```

```python
# =============================================================================
# STEP 4H: CLASS-WEIGHTED RANDOM FOREST + XGBOOST
# =============================================================================

import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
from xgboost import XGBClassifier

print("=" * 100)
print("STEP 4H: CLASS-WEIGHTED RANDOM FOREST AND XGBOOST")
print("=" * 100)


# =============================================================================
# 1. WEIGHTED RANDOM FOREST — STUNTING
# =============================================================================

print("\n1. WEIGHTED RANDOM FOREST — STUNTING")
print("-" * 100)

rf_stunt_weighted = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_stunt_weighted.fit(
    X_train_stunt_processed,
    y_train_stunt
)

rf_stunt_weighted_pred = rf_stunt_weighted.predict(
    X_test_stunt_processed
)

rf_stunt_weighted_prob = rf_stunt_weighted.predict_proba(
    X_test_stunt_processed
)[:, 1]

rf_stunt_weighted_accuracy = accuracy_score(
    y_test_stunt,
    rf_stunt_weighted_pred
)

rf_stunt_weighted_precision = precision_score(
    y_test_stunt,
    rf_stunt_weighted_pred,
    zero_division=0
)

rf_stunt_weighted_recall = recall_score(
    y_test_stunt,
    rf_stunt_weighted_pred,
    zero_division=0
)

rf_stunt_weighted_f1 = f1_score(
    y_test_stunt,
    rf_stunt_weighted_pred,
    zero_division=0
)

rf_stunt_weighted_auc = roc_auc_score(
    y_test_stunt,
    rf_stunt_weighted_prob
)

print(f"Accuracy : {rf_stunt_weighted_accuracy:.4f}")
print(f"Precision: {rf_stunt_weighted_precision:.4f}")
print(f"Recall   : {rf_stunt_weighted_recall:.4f}")
print(f"F1-score : {rf_stunt_weighted_f1:.4f}")
print(f"ROC-AUC  : {rf_stunt_weighted_auc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_stunt, rf_stunt_weighted_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test_stunt,
        rf_stunt_weighted_pred,
        target_names=["Not Stunted", "Stunted"],
        zero_division=0
    )
)


# =============================================================================
# 2. WEIGHTED RANDOM FOREST — UNDERWEIGHT
# =============================================================================

print("\n2. WEIGHTED RANDOM FOREST — UNDERWEIGHT")
print("-" * 100)

rf_under_weighted = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_under_weighted.fit(
    X_train_under_processed,
    y_train_under
)

rf_under_weighted_pred = rf_under_weighted.predict(
    X_test_under_processed
)

rf_under_weighted_prob = rf_under_weighted.predict_proba(
    X_test_under_processed
)[:, 1]

rf_under_weighted_accuracy = accuracy_score(
    y_test_under,
    rf_under_weighted_pred
)

rf_under_weighted_precision = precision_score(
    y_test_under,
    rf_under_weighted_pred,
    zero_division=0
)

rf_under_weighted_recall = recall_score(
    y_test_under,
    rf_under_weighted_pred,
    zero_division=0
)

rf_under_weighted_f1 = f1_score(
    y_test_under,
    rf_under_weighted_pred,
    zero_division=0
)

rf_under_weighted_auc = roc_auc_score(
    y_test_under,
    rf_under_weighted_prob
)

print(f"Accuracy : {rf_under_weighted_accuracy:.4f}")
print(f"Precision: {rf_under_weighted_precision:.4f}")
print(f"Recall   : {rf_under_weighted_recall:.4f}")
print(f"F1-score : {rf_under_weighted_f1:.4f}")
print(f"ROC-AUC  : {rf_under_weighted_auc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_under, rf_under_weighted_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test_under,
        rf_under_weighted_pred,
        target_names=["Not Underweight", "Underweight"],
        zero_division=0
    )
)


# =============================================================================
# 3. WEIGHTED XGBOOST — STUNTING
# =============================================================================

print("\n3. WEIGHTED XGBOOST — STUNTING")
print("-" * 100)

scale_pos_weight_stunt = (
    (y_train_stunt == 0).sum() /
    (y_train_stunt == 1).sum()
)

print(
    f"scale_pos_weight: {scale_pos_weight_stunt:.4f}"
)

xgb_stunt_weighted = XGBClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight_stunt,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_stunt_weighted.fit(
    X_train_stunt_processed,
    y_train_stunt
)

xgb_stunt_weighted_pred = xgb_stunt_weighted.predict(
    X_test_stunt_processed
)

xgb_stunt_weighted_prob = xgb_stunt_weighted.predict_proba(
    X_test_stunt_processed
)[:, 1]

xgb_stunt_weighted_accuracy = accuracy_score(
    y_test_stunt,
    xgb_stunt_weighted_pred
)

xgb_stunt_weighted_precision = precision_score(
    y_test_stunt,
    xgb_stunt_weighted_pred,
    zero_division=0
)

xgb_stunt_weighted_recall = recall_score(
    y_test_stunt,
    xgb_stunt_weighted_pred,
    zero_division=0
)

xgb_stunt_weighted_f1 = f1_score(
    y_test_stunt,
    xgb_stunt_weighted_pred,
    zero_division=0
)

xgb_stunt_weighted_auc = roc_auc_score(
    y_test_stunt,
    xgb_stunt_weighted_prob
)

print(f"Accuracy : {xgb_stunt_weighted_accuracy:.4f}")
print(f"Precision: {xgb_stunt_weighted_precision:.4f}")
print(f"Recall   : {xgb_stunt_weighted_recall:.4f}")
print(f"F1-score : {xgb_stunt_weighted_f1:.4f}")
print(f"ROC-AUC  : {xgb_stunt_weighted_auc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_stunt, xgb_stunt_weighted_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test_stunt,
        xgb_stunt_weighted_pred,
        target_names=["Not Stunted", "Stunted"],
        zero_division=0
    )
)


# =============================================================================
# 4. WEIGHTED XGBOOST — UNDERWEIGHT
# =============================================================================

print("\n4. WEIGHTED XGBOOST — UNDERWEIGHT")
print("-" * 100)

scale_pos_weight_under = (
    (y_train_under == 0).sum() /
    (y_train_under == 1).sum()
)

print(
    f"scale_pos_weight: {scale_pos_weight_under:.4f}"
)

xgb_under_weighted = XGBClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight_under,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_under_weighted.fit(
    X_train_under_processed,
    y_train_under
)

xgb_under_weighted_pred = xgb_under_weighted.predict(
    X_test_under_processed
)

xgb_under_weighted_prob = xgb_under_weighted.predict_proba(
    X_test_under_processed
)[:, 1]

xgb_under_weighted_accuracy = accuracy_score(
    y_test_under,
    xgb_under_weighted_pred
)

xgb_under_weighted_precision = precision_score(
    y_test_under,
    xgb_under_weighted_pred,
    zero_division=0
)

xgb_under_weighted_recall = recall_score(
    y_test_under,
    xgb_under_weighted_pred,
    zero_division=0
)

xgb_under_weighted_f1 = f1_score(
    y_test_under,
    xgb_under_weighted_pred,
    zero_division=0
)

xgb_under_weighted_auc = roc_auc_score(
    y_test_under,
    xgb_under_weighted_prob
)

print(f"Accuracy : {xgb_under_weighted_accuracy:.4f}")
print(f"Precision: {xgb_under_weighted_precision:.4f}")
print(f"Recall   : {xgb_under_weighted_recall:.4f}")
print(f"F1-score : {xgb_under_weighted_f1:.4f}")
print(f"ROC-AUC  : {xgb_under_weighted_auc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_under, xgb_under_weighted_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test_under,
        xgb_under_weighted_pred,
        target_names=["Not Underweight", "Underweight"],
        zero_division=0
    )
)


print("\n" + "=" * 100)
print("STEP 4H COMPLETE")
print("=" * 100)

print("✓ Weighted Random Forest trained for stunting")
print("✓ Weighted Random Forest trained for underweight")
print("✓ Weighted XGBoost trained for stunting")
print("✓ Weighted XGBoost trained for underweight")
print("✓ Minority-class performance evaluated")
print("✓ Test sets remain untouched for model fitting")
```

```python
# =============================================================================
# STEP 4I: THRESHOLD OPTIMIZATION FOR THE WEIGHTED XGBOOST MODELS
# TRAINING DATA ONLY — TEST SET REMAINS UNTOUCHED
# =============================================================================
#
# PURPOSE
#   STEP 4H fits xgb_stunt_weighted and xgb_under_weighted using the
#   default 0.5 decision threshold. This step selects a better decision
#   threshold for each, using out-of-fold predictions on the training data
#   only, so that the test set used in STEP 4J remains completely unseen
#   until final evaluation.
# =============================================================================

import numpy as np

from sklearn.base import clone
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import f1_score, precision_score, recall_score

print("=" * 100)
print("STEP 4I: THRESHOLD OPTIMIZATION (TRAINING DATA ONLY)")
print("=" * 100)


# =============================================================================
# 1. THRESHOLD GRID AND CROSS-VALIDATION SETUP
# =============================================================================

candidate_thresholds = np.arange(0.20, 0.81, 0.05)

cv_threshold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


def select_best_threshold(fitted_model, X_train, y_train):

    oof_probability = cross_val_predict(
        clone(fitted_model),
        X_train,
        y_train,
        cv=cv_threshold,
        method="predict_proba",
        n_jobs=-1
    )[:, 1]

    rows = []

    for threshold in candidate_thresholds:

        prediction = (
            oof_probability >= threshold
        ).astype(int)

        rows.append({
            "Threshold": threshold,
            "F1": f1_score(y_train, prediction, zero_division=0),
            "Precision": precision_score(y_train, prediction, zero_division=0),
            "Recall": recall_score(y_train, prediction, zero_division=0)
        })

    table = pd.DataFrame(rows)

    best_row = table.sort_values(
        ["F1", "Recall"],
        ascending=False
    ).iloc[0]

    return float(best_row["Threshold"]), table


# =============================================================================
# 2. STUNTING
# =============================================================================

print("\n1. STUNTING — THRESHOLD SELECTION")
print("-" * 100)

best_stunt_threshold, stunt_threshold_table = select_best_threshold(
    xgb_stunt_weighted,
    X_train_stunt_processed,
    y_train_stunt
)

xgb_stunt_threshold = xgb_stunt_weighted

print("Selected threshold:", round(best_stunt_threshold, 2))

print(
    "Training (out-of-fold) F1 at selected threshold:",
    round(
        stunt_threshold_table.loc[
            stunt_threshold_table["Threshold"] == best_stunt_threshold,
            "F1"
        ].iloc[0],
        4
    )
)


# =============================================================================
# 3. UNDERWEIGHT
# =============================================================================

print("\n2. UNDERWEIGHT — THRESHOLD SELECTION")
print("-" * 100)

best_under_threshold, under_threshold_table = select_best_threshold(
    xgb_under_weighted,
    X_train_under_processed,
    y_train_under
)

xgb_under_threshold = xgb_under_weighted

print("Selected threshold:", round(best_under_threshold, 2))

print(
    "Training (out-of-fold) F1 at selected threshold:",
    round(
        under_threshold_table.loc[
            under_threshold_table["Threshold"] == best_under_threshold,
            "F1"
        ].iloc[0],
        4
    )
)


print("\n" + "=" * 100)
print("STEP 4I COMPLETE")
print("=" * 100)

print("✓ Thresholds selected using out-of-fold training predictions only")
print("✓ Test set not used for threshold selection")
print("✓ xgb_stunt_threshold and xgb_under_threshold ready for STEP 4J")
print("=" * 100)
```

```python
# ================================================================
# STEP 4J: FINAL TEST-SET EVALUATION USING OPTIMIZED THRESHOLDS
# ================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

print("=" * 100)
print("STEP 4J: FINAL TEST-SET EVALUATION")
print("=" * 100)


# ================================================================
# 1. STUNTING
# ================================================================

print("\n1. STUNTING")
print("-" * 100)

# Probability predictions on untouched test set
stunt_test_prob_xgb = xgb_stunt_threshold.predict_proba(
    X_test_stunt_processed
)[:, 1]

# Apply threshold selected from validation data
stunt_test_pred_xgb = (
    stunt_test_prob_xgb >= best_stunt_threshold
).astype(int)

# Metrics
stunt_test_accuracy = accuracy_score(
    y_test_stunt,
    stunt_test_pred_xgb
)

stunt_test_precision = precision_score(
    y_test_stunt,
    stunt_test_pred_xgb,
    zero_division=0
)

stunt_test_recall = recall_score(
    y_test_stunt,
    stunt_test_pred_xgb,
    zero_division=0
)

stunt_test_f1 = f1_score(
    y_test_stunt,
    stunt_test_pred_xgb,
    zero_division=0
)

stunt_test_auc = roc_auc_score(
    y_test_stunt,
    stunt_test_prob_xgb
)

stunt_test_cm = confusion_matrix(
    y_test_stunt,
    stunt_test_pred_xgb
)


print(f"Selected threshold: {best_stunt_threshold:.2f}")
print(f"Accuracy : {stunt_test_accuracy:.4f}")
print(f"Precision: {stunt_test_precision:.4f}")
print(f"Recall   : {stunt_test_recall:.4f}")
print(f"F1-score : {stunt_test_f1:.4f}")
print(f"ROC-AUC  : {stunt_test_auc:.4f}")

print("\nConfusion Matrix:")
print(stunt_test_cm)

print("\nClassification Report:")
print(
    classification_report(
        y_test_stunt,
        stunt_test_pred_xgb,
        target_names=["Not Stunted", "Stunted"],
        zero_division=0
    )
)


# ================================================================
# 2. UNDERWEIGHT
# ================================================================

print("\n2. UNDERWEIGHT")
print("-" * 100)

# Probability predictions on untouched test set
under_test_prob_xgb = xgb_under_threshold.predict_proba(
    X_test_under_processed
)[:, 1]

# Apply threshold selected from validation data
under_test_pred_xgb = (
    under_test_prob_xgb >= best_under_threshold
).astype(int)

# Metrics
under_test_accuracy = accuracy_score(
    y_test_under,
    under_test_pred_xgb
)

under_test_precision = precision_score(
    y_test_under,
    under_test_pred_xgb,
    zero_division=0
)

under_test_recall = recall_score(
    y_test_under,
    under_test_pred_xgb,
    zero_division=0
)

under_test_f1 = f1_score(
    y_test_under,
    under_test_pred_xgb,
    zero_division=0
)

under_test_auc = roc_auc_score(
    y_test_under,
    under_test_prob_xgb
)

under_test_cm = confusion_matrix(
    y_test_under,
    under_test_pred_xgb
)


print(f"Selected threshold: {best_under_threshold:.2f}")
print(f"Accuracy : {under_test_accuracy:.4f}")
print(f"Precision: {under_test_precision:.4f}")
print(f"Recall   : {under_test_recall:.4f}")
print(f"F1-score : {under_test_f1:.4f}")
print(f"ROC-AUC  : {under_test_auc:.4f}")

print("\nConfusion Matrix:")
print(under_test_cm)

print("\nClassification Report:")
print(
    classification_report(
        y_test_under,
        under_test_pred_xgb,
        target_names=["Not Underweight", "Underweight"],
        zero_division=0
    )
)


# ================================================================
# 3. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 100)
print("FINAL OPTIMIZED XGBOOST TEST PERFORMANCE")
print("=" * 100)

summary = pd.DataFrame({
    "Target": [
        "Stunting",
        "Underweight"
    ],
    "Threshold": [
        best_stunt_threshold,
        best_under_threshold
    ],
    "Accuracy": [
        stunt_test_accuracy,
        under_test_accuracy
    ],
    "Precision": [
        stunt_test_precision,
        under_test_precision
    ],
    "Recall": [
        stunt_test_recall,
        under_test_recall
    ],
    "F1": [
        stunt_test_f1,
        under_test_f1
    ],
    "ROC_AUC": [
        stunt_test_auc,
        under_test_auc
    ]
})

print(summary.round(4).to_string(index=False))


print("\n" + "=" * 100)
print("STEP 4J COMPLETE")
print("=" * 100)

print("✓ Validation-selected thresholds applied")
print("✓ Test set remained untouched during threshold selection")
print("✓ Final test performance calculated")
print("✓ ROC-AUC calculated using test probabilities")
```
