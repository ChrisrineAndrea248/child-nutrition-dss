```python
from pathlib import Path
import pandas as pd

# Current notebook location
NOTEBOOK_DIR = Path.cwd()

# Project folder
PROJECT_DIR = NOTEBOOK_DIR.parent

# Data folder
DATA_DIR = PROJECT_DIR / "data"

print("Notebook:", NOTEBOOK_DIR)
print("Project :", PROJECT_DIR)
print("Data    :", DATA_DIR)
```

```python
files = {
    "CH": DATA_DIR / "ch.sav",
    "HH": DATA_DIR / "hh.sav",
    "WM": DATA_DIR / "wm.sav",
    "BH": DATA_DIR / "bh.sav"
}

for name, path in files.items():
    print(
        f"{'✓' if path.exists() else '✗'} {name}: {path}"
    )
```

```python
# =============================================================================
# STEP 1: LOAD CAR MICS6 DATASETS
# =============================================================================

print("=" * 90)
print("STEP 1: LOADING CAR MICS6 DATASETS")
print("=" * 90)

CH = pd.read_spss(files["CH"])
HH = pd.read_spss(files["HH"])
WM = pd.read_spss(files["WM"])
BH = pd.read_spss(files["BH"])

print("\nDATASET DIMENSIONS")
print("-" * 90)

print(f"CH (Child)          : {CH.shape}")
print(f"HH (Household)      : {HH.shape}")
print(f"WM (Women/Mothers)  : {WM.shape}")
print(f"BH (Birth History)  : {BH.shape}")

print("\n✓ All four CAR MICS6 datasets loaded successfully")
print("=" * 90)
```

```python
# =============================================================================
# STEP 2I: BUILD CHILD-LEVEL CH + HH + WM INTEGRATED DATASET
# =============================================================================

print("=" * 90)
print("STEP 2I: BUILDING CHILD-LEVEL INTEGRATED DATASET")
print("=" * 90)


# =============================================================================
# 1. CREATE WORKING COPIES
# =============================================================================

CH_int = CH.copy()
HH_int = HH.copy()
WM_int = WM.copy()

print("\n1. ORIGINAL DATASET SIZES")
print("-" * 90)

print("CH:", CH_int.shape)
print("HH:", HH_int.shape)
print("WM:", WM_int.shape)


# =============================================================================
# 2. CREATE HOUSEHOLD MERGE KEY
# =============================================================================

CH_int["_HH_KEY"] = (
    CH_int["HH1"].astype(str)
    + "_"
    + CH_int["HH2"].astype(str)
)

HH_int["_HH_KEY"] = (
    HH_int["HH1"].astype(str)
    + "_"
    + HH_int["HH2"].astype(str)
)


# =============================================================================
# 3. REMOVE DUPLICATE HH VARIABLES BEFORE MERGE
# =============================================================================

hh_columns_to_add = [
    col for col in HH_int.columns
    if col not in ["HH1", "HH2", "_HH_KEY"]
]

HH_for_merge = HH_int[
    ["_HH_KEY"] + hh_columns_to_add
].copy()


# =============================================================================
# 4. MERGE HOUSEHOLD DATA INTO CHILD DATA
# =============================================================================

print("\n2. MERGING HOUSEHOLD DATA")
print("-" * 90)

integrated = CH_int.merge(
    HH_for_merge,
    on="_HH_KEY",
    how="left",
    validate="many_to_one",
    indicator="_HH_MERGE"
)

print("✓ Household merge completed")
print("Rows before:", len(CH_int))
print("Rows after :", len(integrated))


# Validate row preservation
if len(integrated) != len(CH_int):
    raise ValueError(
        "ERROR: Household merge changed the number of child records."
    )

print("✓ Child-level row count preserved")


# =============================================================================
# 5. CREATE MOTHER MERGE KEY
# =============================================================================

integrated["_MOTHER_KEY"] = (
    integrated["UF1"].astype(str)
    + "_"
    + integrated["UF2"].astype(str)
    + "_"
    + integrated["UF4"].astype(str)
)

WM_int["_MOTHER_KEY"] = (
    WM_int["WM1"].astype(str)
    + "_"
    + WM_int["WM2"].astype(str)
    + "_"
    + WM_int["WM3"].astype(str)
)


# =============================================================================
# 6. SELECT WM VARIABLES
# =============================================================================

wm_columns_to_add = [
    col for col in WM_int.columns
    if col not in [
        "WM1",
        "WM2",
        "WM3",
        "_MOTHER_KEY"
    ]
]

WM_for_merge = WM_int[
    ["_MOTHER_KEY"] + wm_columns_to_add
].copy()


# =============================================================================
# 7. MERGE MATERNAL DATA
# =============================================================================

print("\n3. MERGING MATERNAL DATA")
print("-" * 90)

integrated = integrated.merge(
    WM_for_merge,
    on="_MOTHER_KEY",
    how="left",
    validate="many_to_one",
    indicator="_WM_MERGE"
)

print("✓ Maternal merge completed")
print("Rows after WM merge:", len(integrated))


# =============================================================================
# 8. VERIFY ROW PRESERVATION
# =============================================================================

print("\n4. ROW PRESERVATION CHECK")
print("-" * 90)

print("Original CH rows :", len(CH))
print("Integrated rows  :", len(integrated))

if len(integrated) == len(CH):
    print("✓ EXACTLY 9,037 child records preserved")
else:
    raise ValueError(
        "ERROR: Integrated dataset does not preserve CH row count."
    )


# =============================================================================
# 9. MERGE QUALITY
# =============================================================================

print("\n5. MERGE QUALITY")
print("-" * 90)

print("Household matches:")

display(
    integrated["_HH_MERGE"]
    .value_counts()
)

print("\nMaternal matches:")

display(
    integrated["_WM_MERGE"]
    .value_counts()
)


# =============================================================================
# 10. CHECK MATERNAL MISSINGNESS
# =============================================================================

print("\n6. MATERNAL LINKAGE RESULT")
print("-" * 90)

wm_matched = (
    integrated["_WM_MERGE"] == "both"
).sum()

wm_unmatched = (
    integrated["_WM_MERGE"] != "both"
).sum()

print("Children with WM match    :", wm_matched)
print("Children without WM match:", wm_unmatched)

print(
    "WM match rate:",
    f"{wm_matched / len(integrated) * 100:.2f}%"
)


# =============================================================================
# 11. REMOVE TEMPORARY MERGE COLUMNS
# =============================================================================

integrated.drop(
    columns=[
        "_HH_KEY",
        "_MOTHER_KEY",
        "_HH_MERGE",
        "_WM_MERGE"
    ],
    inplace=True
)


# =============================================================================
# 12. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 90)
print("STEP 2I COMPLETE")
print("=" * 90)

print("Integrated dataset:", integrated.shape)
print("Child records     :", len(integrated))
print("✓ CH retained as master dataset")
print("✓ HH merged using HH1 + HH2")
print("✓ WM merged using UF1 + UF2 + UF4 → WM1 + WM2 + WM3")
print("✓ No child records duplicated")
print("✓ Unmatched maternal records retained as missing")
print("=" * 90)
```

```python
# =============================================================================
# STEP 3: CAR MICS6 STUNTING AND UNDERWEIGHT TARGET VALIDATION
# =============================================================================

print("=" * 100)
print("STEP 3: STUNTING AND UNDERWEIGHT TARGET VALIDATION")
print("=" * 100)


# =============================================================================
# 1. IDENTIFY ANTHROPOMETRIC VARIABLES
# =============================================================================

print("\n1. ANTHROPOMETRIC VARIABLES IN CH")
print("-" * 100)

anthro_candidates = [
    col for col in CH.columns
    if any(
        term in str(col).upper()
        for term in [
            "HAZ",
            "WHZ",
            "WAZ",
            "HEIGHT",
            "WEIGHT",
            "STUNT",
            "UNDER",
            "AN1",
            "AN2",
            "AN3",
            "AN4",
            "AN5"
        ]
    )
]

for col in anthro_candidates:
    print(col)

print("\nNumber found:", len(anthro_candidates))


# =============================================================================
# 2. CHECK EXISTING TARGET VARIABLES
# =============================================================================

print("\n" + "=" * 100)
print("2. POSSIBLE EXISTING TARGET VARIABLES")
print("=" * 100)

target_candidates = [
    col for col in CH.columns
    if any(
        term in str(col).lower()
        for term in [
            "stunted",
            "stunting",
            "underweight",
            "waz",
            "haz"
        ]
    )
]

for col in target_candidates:
    print(col)

print("\nNumber found:", len(target_candidates))


# =============================================================================
# 3. DISPLAY RELEVANT VARIABLES
# =============================================================================

print("\n" + "=" * 100)
print("3. SAMPLE OF RELEVANT VARIABLES")
print("=" * 100)

if target_candidates:
    display(CH[target_candidates].head(20))

elif anthro_candidates:
    display(CH[anthro_candidates].head(20))

else:
    print("No obvious anthropometric/target variables found.")


print("\n" + "=" * 100)
print("STEP 3 COMPLETE")
print("=" * 100)
```

```python
# =============================================================================
# STEP 3A: CONSTRUCT STUNTING AND UNDERWEIGHT TARGETS
# =============================================================================

print("=" * 100)
print("STEP 3A: CONSTRUCTING STUNTING AND UNDERWEIGHT TARGETS")
print("=" * 100)


# =============================================================================
# 1. CREATE WORKING DATASET
# =============================================================================

analysis = integrated.copy()

print("\n1. INITIAL DATA")
print("-" * 100)
print("Children:", len(analysis))


# =============================================================================
# 2. VALID ANTHROPOMETRIC OBSERVATIONS
# =============================================================================

analysis["HAZ_VALID"] = (
    analysis["HAZ"].notna()
    & analysis["HAZFLAG"].astype(str).str.lower().eq("no error")
)

analysis["WAZ_VALID"] = (
    analysis["WAZ"].notna()
    & analysis["WAZFLAG"].astype(str).str.lower().eq("no error")
)

print("\n2. ANTHROPOMETRIC VALIDITY")
print("-" * 100)

print("Valid HAZ:", analysis["HAZ_VALID"].sum())
print("Invalid/missing HAZ:", (~analysis["HAZ_VALID"]).sum())

print("Valid WAZ:", analysis["WAZ_VALID"].sum())
print("Invalid/missing WAZ:", (~analysis["WAZ_VALID"]).sum())


# =============================================================================
# 3. CREATE STUNTING TARGET
# =============================================================================
# Stunting: HAZ < -2

analysis["stunted"] = np.nan

analysis.loc[
    analysis["HAZ_VALID"],
    "stunted"
] = (
    analysis.loc[
        analysis["HAZ_VALID"],
        "HAZ"
    ] < -2
).astype(int)


# =============================================================================
# 4. CREATE UNDERWEIGHT TARGET
# =============================================================================
# Underweight: WAZ < -2

analysis["underweight"] = np.nan

analysis.loc[
    analysis["WAZ_VALID"],
    "underweight"
] = (
    analysis.loc[
        analysis["WAZ_VALID"],
        "WAZ"
    ] < -2
).astype(int)


# =============================================================================
# 5. STUNTING DISTRIBUTION
# =============================================================================

print("\n" + "=" * 100)
print("3. STUNTING TARGET")
print("=" * 100)

print(
    analysis["stunted"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nPercentages:")

display(
    (
        analysis["stunted"]
        .value_counts(normalize=True, dropna=False)
        .sort_index()
        * 100
    ).round(2)
)


# =============================================================================
# 6. UNDERWEIGHT DISTRIBUTION
# =============================================================================

print("\n" + "=" * 100)
print("4. UNDERWEIGHT TARGET")
print("=" * 100)

print(
    analysis["underweight"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nPercentages:")

display(
    (
        analysis["underweight"]
        .value_counts(normalize=True, dropna=False)
        .sort_index()
        * 100
    ).round(2)
)


# =============================================================================
# 7. FINAL TARGET SAMPLE SIZES
# =============================================================================

print("\n" + "=" * 100)
print("5. FINAL TARGET SAMPLE SIZES")
print("=" * 100)

stunting_n = analysis["stunted"].notna().sum()
underweight_n = analysis["underweight"].notna().sum()

print("Total CH children :", len(analysis))
print("Stunting sample   :", stunting_n)
print("Underweight sample:", underweight_n)


# =============================================================================
# 8. CHECK TARGET VALUES
# =============================================================================

print("\n" + "=" * 100)
print("6. TARGET VALIDATION")
print("=" * 100)

stunting_values = set(
    analysis["stunted"]
    .dropna()
    .astype(int)
    .unique()
)

underweight_values = set(
    analysis["underweight"]
    .dropna()
    .astype(int)
    .unique()
)

print(
    "Stunting values:",
    sorted(stunting_values)
)

print(
    "Underweight values:",
    sorted(underweight_values)
)

if stunting_values <= {0, 1}:
    print("✓ Stunting is binary")

if underweight_values <= {0, 1}:
    print("✓ Underweight is binary")


print("\n" + "=" * 100)
print("STEP 3A COMPLETE")
print("=" * 100)
```

```python
# =============================================================================
# STEP 3D SETUP: DEFINE PREDICTOR GROUPS
# =============================================================================

predictors = [
    "CAGE",
    "WB4",
    "WAGEM",
    "HL4",
    "HH6",
    "windex5",
    "WS1",
    "WS11",
    "WS15",
    "welevel",
    "MSTATUS",
    "CEB",
    "CSURV",
    "CDEAD",
    "CM11",
    "CM17",
    "insurance"
]

numeric_features = [
    "CAGE",
    "WB4",
    "WAGEM",
    "CEB",
    "CSURV",
    "CDEAD",
    "CM11"
]

categorical_features = [
    "HL4",
    "HH6",
    "windex5",
    "WS1",
    "WS11",
    "WS15",
    "welevel",
    "MSTATUS",
    "CM17",
    "insurance"
]

print("Predictors:", len(predictors))
print("Numeric predictors:", len(numeric_features))
print("Categorical predictors:", len(categorical_features))

print("\n✓ Predictor definitions restored")
```

```python
# =============================================================================
# STEP 3D: DATA CLEANING AND QUALITY CONTROL — CORRECTED
# =============================================================================

print("=" * 100)
print("STEP 3D: DATA CLEANING AND QUALITY CONTROL")
print("=" * 100)


# =============================================================================
# 1. WORKING COPY
# =============================================================================

clean_data = analysis.copy()

print("\n1. INITIAL DATASET")
print("-" * 100)

print("Rows:", len(clean_data))
print("Columns:", len(clean_data.columns))


# =============================================================================
# 2. CHECK DUPLICATE COLUMN NAMES
# =============================================================================

print("\n2. DUPLICATE COLUMN-NAME CHECK")
print("-" * 100)

duplicate_column_names = clean_data.columns[
    clean_data.columns.duplicated()
].tolist()

print(
    "Duplicate column names:",
    len(duplicate_column_names)
)

if duplicate_column_names:
    print("Duplicated names:")
    for col in duplicate_column_names:
        print("✗", col)
else:
    print("✓ No duplicate column names")


# =============================================================================
# 3. FIND CHILD IDENTIFIER COLUMNS
# =============================================================================

print("\n3. CHILD IDENTIFIER CHECK")
print("-" * 100)

print("Checking for HH1, HH2 and LN...")

for col in ["HH1", "HH2", "LN"]:
    matches = [
        c for c in clean_data.columns
        if str(c).upper() == col
    ]

    print(
        f"{col}:",
        "FOUND" if matches else "NOT FOUND",
        matches
    )


# =============================================================================
# 4. DUPLICATE CHILD RECORD CHECK
# =============================================================================

print("\n4. DUPLICATE CHILD RECORD CHECK")
print("-" * 100)

# Because CH is the master dataset, its original identifiers
# are available in the original CH dataframe.

child_id_columns = ["HH1", "HH2", "LN"]

duplicate_children = CH[
    child_id_columns
].duplicated(
    keep=False
)

duplicate_count = duplicate_children.sum()

print(
    "Duplicate child records:",
    duplicate_count
)

if duplicate_count == 0:
    print("✓ No duplicate child records found in CH")
else:
    print("⚠ Duplicate child records detected")


# =============================================================================
# 5. CHECK CONSTANT PREDICTORS
# =============================================================================

print("\n5. CONSTANT PREDICTOR CHECK")
print("-" * 100)

constant_predictors = []

for col in predictors:

    if col not in clean_data.columns:
        continue

    if clean_data[col].nunique(
        dropna=True
    ) <= 1:

        constant_predictors.append(col)


if len(constant_predictors) == 0:

    print("✓ No constant predictors found")

else:

    print("Constant predictors:")

    for col in constant_predictors:
        print("✗", col)


# =============================================================================
# 6. NUMERIC VARIABLE RANGE CHECK
# =============================================================================

print("\n6. NUMERIC VARIABLE RANGE CHECK")
print("-" * 100)

range_summary = pd.DataFrame({

    "Variable": numeric_features,

    "Minimum": [
        clean_data[col].min()
        for col in numeric_features
    ],

    "Maximum": [
        clean_data[col].max()
        for col in numeric_features
    ],

    "Median": [
        clean_data[col].median()
        for col in numeric_features
    ],

    "Missing": [
        clean_data[col].isna().sum()
        for col in numeric_features
    ]
})

display(range_summary)


# =============================================================================
# 7. CATEGORICAL VARIABLE CHECK
# =============================================================================

print("\n7. CATEGORICAL VARIABLE LEVELS")
print("-" * 100)

for col in categorical_features:

    print("\n" + col)
    print("-" * 60)

    display(
        clean_data[col]
        .value_counts(dropna=False)
        .to_frame("Count")
    )


# =============================================================================
# 8. TARGET QUALITY CHECK
# =============================================================================

print("\n8. TARGET QUALITY CHECK")
print("-" * 100)

print("\nStunting:")
print(
    clean_data["stunted"]
    .value_counts(dropna=False)
)

print("\nUnderweight:")
print(
    clean_data["underweight"]
    .value_counts(dropna=False)
)


# =============================================================================
# 9. TARGET LEAKAGE CHECK
# =============================================================================

print("\n9. TARGET LEAKAGE CHECK")
print("-" * 100)

leakage_variables = [
    "HAZ",
    "WAZ",
    "HAZ2",
    "WAZ2",
    "HAZFLAG",
    "WAZFLAG",
    "WHZ",
    "WHZ2",
    "WHZFLAG",
    "stunted",
    "underweight"
]

leakage_found = [
    col for col in predictors
    if col in leakage_variables
]

if len(leakage_found) == 0:

    print(
        "✓ No target leakage variables in predictor set"
    )

else:

    print("✗ Leakage variables detected:")

    for col in leakage_found:
        print("  ", col)


# =============================================================================
# 10. MISSINGNESS SUMMARY
# =============================================================================

print("\n10. MISSINGNESS SUMMARY")
print("-" * 100)

missing_summary = pd.DataFrame({

    "Variable": predictors,

    "Missing": [
        clean_data[col].isna().sum()
        for col in predictors
    ],

    "Missing_%": [
        round(
            clean_data[col].isna().mean() * 100,
            2
        )
        for col in predictors
    ]
})

display(
    missing_summary.sort_values(
        "Missing_%",
        ascending=False
    )
)


# =============================================================================
# 11. COMPLETE-CASE CHECK
# =============================================================================

print("\n11. COMPLETE-CASE CHECK")
print("-" * 100)

complete_predictors = (
    clean_data[predictors]
    .notna()
    .all(axis=1)
)

print(
    "Rows with all 17 predictors:",
    complete_predictors.sum()
)

print(
    "Rows with at least one missing predictor:",
    (~complete_predictors).sum()
)


# =============================================================================
# 12. FINAL CLEANING DECISION
# =============================================================================

print("\n" + "=" * 100)
print("12. CLEANING DECISION")
print("=" * 100)

print("✓ CH remains the master child dataset")
print("✓ Duplicate child check performed using original CH identifiers")
print("✓ Constant-variable check completed")
print("✓ Numeric range check completed")
print("✓ Categorical-level check completed")
print("✓ Target validation completed")
print("✓ Target leakage check completed")
print("✓ Missingness assessment completed")

print("\nNo observations have been deleted.")
print("Missing values will be handled during ML preprocessing.")

print("\n" + "=" * 100)
print("STEP 3D COMPLETE")
print("=" * 100)
```

```python
# =============================================================================
# STEP 3F: FINAL MISSING-VALUE CLEANING
# =============================================================================

print("=" * 100)
print("STEP 3F: FINAL MISSING-VALUE CLEANING")
print("=" * 100)

# Work from the cleaned integrated dataset
preprocessed_data = clean_data.copy()


# =============================================================================
# 1. CONVERT EXPLICIT UNKNOWN/MISSING CATEGORIES TO NaN
# =============================================================================

print("\n1. CONVERTING EXPLICIT UNKNOWN/MISSING RESPONSES")
print("-" * 100)

missing_categories = {
    "WS15": ["NON REPONSE"],
    "MSTATUS": ["Manquant"],
    "insurance": ["Manquant/NSP"]
}

for col, values in missing_categories.items():

    if col in preprocessed_data.columns:

        before = preprocessed_data[col].isna().sum()

        preprocessed_data[col] = (
            preprocessed_data[col]
            .replace(values, np.nan)
        )

        after = preprocessed_data[col].isna().sum()

        print(
            f"{col}: {after - before} explicit "
            f"missing responses converted to NaN"
        )


# =============================================================================
# 2. CHECK THAT LEGITIMATE NUMERIC ZEROES ARE PRESERVED
# =============================================================================

print("\n2. PRESERVING LEGITIMATE ZERO VALUES")
print("-" * 100)

zero_check_variables = [
    "CEB",
    "CSURV",
    "CDEAD",
    "CM11"
]

for col in zero_check_variables:

    zero_count = (
        preprocessed_data[col] == 0
    ).sum()

    print(
        f"{col}: {zero_count} zero values preserved"
    )


# =============================================================================
# 3. WINDEX5 INVESTIGATION
# =============================================================================

print("\n3. WINDEX5 = 0.0 INVESTIGATION")
print("-" * 100)

print(
    "Number of windex5 = 0.0:",
    (preprocessed_data["windex5"] == 0.0).sum()
)

print("\nDistribution of windex5:")
display(
    preprocessed_data["windex5"]
    .value_counts(dropna=False)
)


# =============================================================================
# 4. CHECK WINDEX5 SOURCE VARIABLES
# =============================================================================

print("\n4. CHECKING RELATED WEALTH VARIABLES")
print("-" * 100)

wealth_variables = [
    "windex5",
    "windex10",
    "wscore",
    "windex5u",
    "windex10u",
    "wscoreu",
    "windex5r",
    "windex10r",
    "wscorer"
]

available_wealth = [
    col for col in wealth_variables
    if col in preprocessed_data.columns
]

print("Available wealth variables:")
print(available_wealth)

display(
    preprocessed_data[
        preprocessed_data["windex5"] == 0.0
    ][available_wealth]
    .head(20)
)


# =============================================================================
# 5. UPDATED MISSINGNESS
# =============================================================================

print("\n5. UPDATED PREDICTOR MISSINGNESS")
print("-" * 100)

updated_missingness = pd.DataFrame({

    "Variable": predictors,

    "Missing": [
        preprocessed_data[col].isna().sum()
        for col in predictors
    ],

    "Missing_%": [
        round(
            preprocessed_data[col].isna().mean() * 100,
            2
        )
        for col in predictors
    ]
})

display(
    updated_missingness.sort_values(
        "Missing_%",
        ascending=False
    )
)


# =============================================================================
# 6. DATASET SIZE CHECK
# =============================================================================

print("\n6. DATASET SIZE CHECK")
print("-" * 100)

print(
    "Original integrated rows:",
    len(clean_data)
)

print(
    "Current rows:",
    len(preprocessed_data)
)

if len(clean_data) == len(preprocessed_data):
    print("✓ No child records removed")
else:
    print("⚠ Row count changed")


# =============================================================================
# 7. TARGET CHECK
# =============================================================================

print("\n7. TARGET CHECK")
print("-" * 100)

print("Stunting:")
display(
    preprocessed_data["stunted"]
    .value_counts(dropna=False)
)

print("\nUnderweight:")
display(
    preprocessed_data["underweight"]
    .value_counts(dropna=False)
)


# =============================================================================
# 8. FINAL STATUS
# =============================================================================

print("\n" + "=" * 100)
print("STEP 3F COMPLETE")
print("=" * 100)

print("""
✓ Explicit missing/unknown categorical responses handled
✓ Legitimate numeric zero values preserved
✓ windex5 = 0.0 retained pending source-code verification
✓ No child records deleted
✓ Target variables unchanged
✓ No imputation performed yet
✓ Dataset ready for final missing-data strategy
""")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# ============================================================
# STEP 3K: TRAIN/TEST SPLIT AND PREPROCESSING DESIGN
# ============================================================

print("=" * 100)
print("STEP 3K: TRAIN/TEST SPLIT AND PREPROCESSING DESIGN")
print("=" * 100)

# ------------------------------------------------------------
# FINAL FEATURES
# ------------------------------------------------------------

numeric_features = [
    "CAGE",
    "WB4",
    "WAGEM",
    "CEB",
    "CSURV",
    "CDEAD",
    "CM11"
]

categorical_features = [
    "HL4",
    "HH6",
    "windex5",
    "WS1",
    "WS11",
    "WS15",
    "welevel",
    "MSTATUS",
    "CM17",
    "insurance"
]

predictors = numeric_features + categorical_features


# ------------------------------------------------------------
# PREPROCESSING PIPELINE
# ------------------------------------------------------------

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(
        strategy="constant",
        fill_value="Missing"
    )),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])


# ------------------------------------------------------------
# FUNCTION FOR TARGET-SPECIFIC SPLIT
# ------------------------------------------------------------

def prepare_target_data(data, target):

    print("\n" + "=" * 100)
    print(f"PREPARING TARGET: {target.upper()}")
    print("=" * 100)

    # Keep only children with a valid target
    target_data = data[
        data[target].notna()
    ].copy()

    X = target_data[predictors]
    y = target_data[target].astype(int)

    print(f"\nValid records: {len(target_data)}")

    print("\nTarget distribution:")
    print(y.value_counts())

    print("\nTarget proportions:")
    print(y.value_counts(normalize=True).round(4))

    # --------------------------------------------------------
    # STRATIFIED TRAIN/TEST SPLIT
    # --------------------------------------------------------

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    print("\nTrain/Test split:")
    print(f"Training records: {len(X_train)}")
    print(f"Testing records : {len(X_test)}")

    print("\nTraining target distribution:")
    print(y_train.value_counts(normalize=True).round(4))

    print("\nTesting target distribution:")
    print(y_test.value_counts(normalize=True).round(4))

    return X_train, X_test, y_train, y_test


# ------------------------------------------------------------
# STUNTING
# ------------------------------------------------------------

X_train_stunt, X_test_stunt, y_train_stunt, y_test_stunt = \
    prepare_target_data(
        clean_data,
        "stunted"
    )


# ------------------------------------------------------------
# UNDERWEIGHT
# ------------------------------------------------------------

X_train_under, X_test_under, y_train_under, y_test_under = \
    prepare_target_data(
        clean_data,
        "underweight"
    )


print("\n" + "=" * 100)
print("STEP 3K COMPLETE")
print("=" * 100)

print("✓ Stunting data split separately")
print("✓ Underweight data split separately")
print("✓ Stratification applied")
print("✓ 80/20 train/test split")
print("✓ Preprocessing pipeline defined")
print("✓ Median imputation restricted to training data")
print("✓ Categorical missingness represented explicitly")
print("✓ One-hot encoding configured")
print("✓ Unknown test categories handled safely")
print("✓ Ready for model training")
```

```python
# ==========================================================================================
# STEP 3L: FITTING AND VALIDATING PREPROCESSING PIPELINE — FINAL CORRECTION
# ==========================================================================================

print("\n" + "=" * 100)
print("STEP 3L: FITTING AND VALIDATING PREPROCESSING PIPELINE")
print("=" * 100)

# ==========================================================================================
# 1. CREATE COPIES
# ==========================================================================================

X_train_stunt_prep = X_train_stunt.copy()
X_test_stunt_prep  = X_test_stunt.copy()

X_train_under_prep = X_train_under.copy()
X_test_under_prep  = X_test_under.copy()


# ==========================================================================================
# 2. FIX MIXED CATEGORICAL TYPES
# ==========================================================================================

print("\n1. CLEANING CATEGORICAL VARIABLES")
print("-" * 100)

for col in categorical_features:

    for df in [
        X_train_stunt_prep,
        X_test_stunt_prep,
        X_train_under_prep,
        X_test_under_prep
    ]:

        # Convert missing values to ordinary np.nan first
        df[col] = df[col].where(df[col].notna(), np.nan)

        # Convert non-missing values to ordinary Python strings
        df[col] = df[col].apply(
            lambda x: str(x) if pd.notna(x) else np.nan
        )

print("✓ Categorical variables converted to ordinary Python strings")
print("✓ Missing values represented as np.nan")
print("✓ No pandas StringDtype / pd.NA values introduced")


# ==========================================================================================
# 3. CREATE PREPROCESSOR
# ==========================================================================================

print("\n2. CREATING PREPROCESSING PIPELINE")
print("-" * 100)

def create_preprocessor():

    numeric_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            )
        ]
    )

    categorical_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(strategy="most_frequent")
            ),
            (
                "encoder",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                )
            )
        ]
    )

    return ColumnTransformer(
        transformers=[
            (
                "numeric",
                numeric_pipeline,
                numeric_features
            ),
            (
                "categorical",
                categorical_pipeline,
                categorical_features
            )
        ],
        remainder="drop"
    )

print("✓ Numeric pipeline created")
print("✓ Categorical pipeline created")
print("✓ OneHotEncoder configured")


# ==========================================================================================
# 4. STUNTING
# ==========================================================================================

print("\n" + "=" * 100)
print("3. STUNTING PREPROCESSING")
print("=" * 100)

preprocessor_stunt = create_preprocessor()

X_train_stunt_processed = preprocessor_stunt.fit_transform(
    X_train_stunt_prep
)

X_test_stunt_processed = preprocessor_stunt.transform(
    X_test_stunt_prep
)

print("✓ Stunting preprocessor fitted on training data only")
print("✓ Stunting training data transformed")
print("✓ Stunting test data transformed")


# ==========================================================================================
# 5. UNDERWEIGHT
# ==========================================================================================

print("\n" + "=" * 100)
print("4. UNDERWEIGHT PREPROCESSING")
print("=" * 100)

preprocessor_under = create_preprocessor()

X_train_under_processed = preprocessor_under.fit_transform(
    X_train_under_prep
)

X_test_under_processed = preprocessor_under.transform(
    X_test_under_prep
)

print("✓ Underweight preprocessor fitted on training data only")
print("✓ Underweight training data transformed")
print("✓ Underweight test data transformed")


# ==========================================================================================
# 6. SHAPE CHECK
# ==========================================================================================

print("\n" + "=" * 100)
print("5. PROCESSED MATRIX SHAPES")
print("=" * 100)

print("Stunting training   :", X_train_stunt_processed.shape)
print("Stunting test       :", X_test_stunt_processed.shape)

print("Underweight training:", X_train_under_processed.shape)
print("Underweight test    :", X_test_under_processed.shape)


# ==========================================================================================
# 7. FINAL INTEGRITY CHECK
# ==========================================================================================

print("\n" + "=" * 100)
print("6. PREPROCESSING INTEGRITY CHECK")
print("=" * 100)

checks = {
    "Stunting train NaN":
        np.isnan(X_train_stunt_processed).sum(),

    "Stunting test NaN":
        np.isnan(X_test_stunt_processed).sum(),

    "Underweight train NaN":
        np.isnan(X_train_under_processed).sum(),

    "Underweight test NaN":
        np.isnan(X_test_under_processed).sum(),

    "Stunting train Inf":
        np.isinf(X_train_stunt_processed).sum(),

    "Stunting test Inf":
        np.isinf(X_test_stunt_processed).sum(),

    "Underweight train Inf":
        np.isinf(X_train_under_processed).sum(),

    "Underweight test Inf":
        np.isinf(X_test_under_processed).sum()
}

for name, value in checks.items():
    print(f"{name:<30}: {value}")

print("\n" + "=" * 100)
print("STEP 3L COMPLETE")
print("=" * 100)
```

```python
# =============================================================================
# STEP 4I: THRESHOLD OPTIMIZATION FOR THE WEIGHTED XGBOOST MODELS
# TRAINING DATA ONLY — TEST SET REMAINS UNTOUCHED
# =============================================================================
#
# PURPOSE
#   STEP 4H fits xgb_stunt_weighted and xgb_under_weighted using the
#   default 0.5 decision threshold. This step selects a better decision
#   threshold for each, using out-of-fold predictions on the training data
#   only, so that the test set used in STEP 4J remains completely unseen
#   until final evaluation.
# =============================================================================

import numpy as np

from sklearn.base import clone
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import f1_score, precision_score, recall_score

print("=" * 100)
print("STEP 4I: THRESHOLD OPTIMIZATION (TRAINING DATA ONLY)")
print("=" * 100)


# =============================================================================
# 1. THRESHOLD GRID AND CROSS-VALIDATION SETUP
# =============================================================================

candidate_thresholds = np.arange(0.20, 0.81, 0.05)

cv_threshold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


def select_best_threshold(fitted_model, X_train, y_train):

    oof_probability = cross_val_predict(
        clone(fitted_model),
        X_train,
        y_train,
        cv=cv_threshold,
        method="predict_proba",
        n_jobs=-1
    )[:, 1]

    rows = []

    for threshold in candidate_thresholds:

        prediction = (
            oof_probability >= threshold
        ).astype(int)

        rows.append({
            "Threshold": threshold,
            "F1": f1_score(y_train, prediction, zero_division=0),
            "Precision": precision_score(y_train, prediction, zero_division=0),
            "Recall": recall_score(y_train, prediction, zero_division=0)
        })

    table = pd.DataFrame(rows)

    best_row = table.sort_values(
        ["F1", "Recall"],
        ascending=False
    ).iloc[0]

    return float(best_row["Threshold"]), table


# =============================================================================
# 2. STUNTING
# =============================================================================

print("\n1. STUNTING — THRESHOLD SELECTION")
print("-" * 100)

best_stunt_threshold, stunt_threshold_table = select_best_threshold(
    xgb_stunt_weighted,
    X_train_stunt_processed,
    y_train_stunt
)

xgb_stunt_threshold = xgb_stunt_weighted

print("Selected threshold:", round(best_stunt_threshold, 2))

print(
    "Training (out-of-fold) F1 at selected threshold:",
    round(
        stunt_threshold_table.loc[
            stunt_threshold_table["Threshold"] == best_stunt_threshold,
            "F1"
        ].iloc[0],
        4
    )
)


# =============================================================================
# 3. UNDERWEIGHT
# =============================================================================

print("\n2. UNDERWEIGHT — THRESHOLD SELECTION")
print("-" * 100)

best_under_threshold, under_threshold_table = select_best_threshold(
    xgb_under_weighted,
    X_train_under_processed,
    y_train_under
)

xgb_under_threshold = xgb_under_weighted

print("Selected threshold:", round(best_under_threshold, 2))

print(
    "Training (out-of-fold) F1 at selected threshold:",
    round(
        under_threshold_table.loc[
            under_threshold_table["Threshold"] == best_under_threshold,
            "F1"
        ].iloc[0],
        4
    )
)


print("\n" + "=" * 100)
print("STEP 4I COMPLETE")
print("=" * 100)

print("✓ Thresholds selected using out-of-fold training predictions only")
print("✓ Test set not used for threshold selection")
print("✓ xgb_stunt_threshold and xgb_under_threshold ready for STEP 4J")
print("=" * 100)
```

```python
# ==================================================================================================
# STEP 4M: CONTROLLED PRECISION REFINEMENT AROUND THE 54% STUNTING RESULT
# ==================================================================================================

import numpy as np
import pandas as pd
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    confusion_matrix
)

print("=" * 100)
print("STEP 4M: CONTROLLED PRECISION REFINEMENT")
print("=" * 100)

# -----------------------------------------------------------------------------------------------
# 1. CHECK ORIGINAL TUNED MODEL
# -----------------------------------------------------------------------------------------------

if "best_stunting_model" not in globals():
    raise NameError(
        "best_stunting_model is not available. "
        "Run the ORIGINAL STEP 4K first."
    )

print("\n✓ Original tuned stunting model available")
print("✓ No new hyperparameter search")
print("✓ Test set remains untouched")


# -----------------------------------------------------------------------------------------------
# 2. GENERATE VALIDATION PROBABILITIES
# -----------------------------------------------------------------------------------------------

stunt_val_prob = best_stunting_model.predict_proba(
    X_val_stunt
)[:, 1]

print("\nValidation probabilities generated:", len(stunt_val_prob))


# -----------------------------------------------------------------------------------------------
# 3. CONTROLLED THRESHOLD SEARCH
# -----------------------------------------------------------------------------------------------

thresholds = np.arange(
    0.50,
    0.601,
    0.005
)

results = []

for threshold in thresholds:

    y_pred = (
        stunt_val_prob >= threshold
    ).astype(int)

    precision = precision_score(
        y_val_stunt,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_val_stunt,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_val_stunt,
        y_pred,
        zero_division=0
    )

    accuracy = accuracy_score(
        y_val_stunt,
        y_pred
    )

    tn, fp, fn, tp = confusion_matrix(
        y_val_stunt,
        y_pred,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn)

    results.append({
        "Threshold": threshold,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })

threshold_results = pd.DataFrame(results)


# -----------------------------------------------------------------------------------------------
# 4. REQUIRE RECALL >= 70%
# -----------------------------------------------------------------------------------------------

eligible = threshold_results[
    threshold_results["Recall"] >= 0.70
].copy()


# -----------------------------------------------------------------------------------------------
# 5. RANK BY PRECISION
# -----------------------------------------------------------------------------------------------

eligible = eligible.sort_values(
    by=[
        "Precision",
        "F1",
        "Recall",
        "FPR"
    ],
    ascending=[
        False,
        False,
        False,
        True
    ]
)


# -----------------------------------------------------------------------------------------------
# 6. DISPLAY RESULTS
# -----------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("THRESHOLDS WITH RECALL >= 70%")
print("=" * 100)

print(
    eligible.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# -----------------------------------------------------------------------------------------------
# 7. BEST PRECISION OPERATING POINT
# -----------------------------------------------------------------------------------------------

best_precision_row = eligible.iloc[0]

print("\n" + "=" * 100)
print("BEST PRECISION OPERATING POINT")
print("=" * 100)

print(
    f"""
Threshold : {best_precision_row['Threshold']:.3f}
Precision : {best_precision_row['Precision']:.4f}
Recall    : {best_precision_row['Recall']:.4f}
F1        : {best_precision_row['F1']:.4f}
Accuracy  : {best_precision_row['Accuracy']:.4f}
FPR       : {best_precision_row['FPR']:.4f}

TN        : {int(best_precision_row['TN'])}
FP        : {int(best_precision_row['FP'])}
FN        : {int(best_precision_row['FN'])}
TP        : {int(best_precision_row['TP'])}
"""
)


# -----------------------------------------------------------------------------------------------
# 8. COMPARE WITH ORIGINAL 0.53 OPERATING POINT
# -----------------------------------------------------------------------------------------------

original_threshold = 0.53

original_row = threshold_results[
    np.isclose(
        threshold_results["Threshold"],
        original_threshold
    )
]

print("\n" + "=" * 100)
print("ORIGINAL 0.53 VS REFINED THRESHOLD")
print("=" * 100)

if len(original_row) > 0:

    original_row = original_row.iloc[0]

    comparison = pd.DataFrame({
        "Metric": [
            "Precision",
            "Recall",
            "F1",
            "Accuracy",
            "FPR",
            "FP"
        ],

        "Original_0.53": [
            original_row["Precision"],
            original_row["Recall"],
            original_row["F1"],
            original_row["Accuracy"],
            original_row["FPR"],
            original_row["FP"]
        ],

        "Refined": [
            best_precision_row["Precision"],
            best_precision_row["Recall"],
            best_precision_row["F1"],
            best_precision_row["Accuracy"],
            best_precision_row["FPR"],
            best_precision_row["FP"]
        ]
    })

    print(
        comparison.to_string(
            index=False,
            float_format=lambda x: f"{x:.4f}"
        )
    )


# -----------------------------------------------------------------------------------------------
# 9. DECISION
# -----------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("FINAL DEVELOPMENT DECISION")
print("=" * 100)

if (
    best_precision_row["Precision"] > original_row["Precision"]
    and
    best_precision_row["Recall"] >= 0.70
):

    print("✓ REFINED THRESHOLD IMPROVES PRECISION")
    print(
        f"✓ New threshold: "
        f"{best_precision_row['Threshold']:.3f}"
    )

    selected_stunting_threshold = float(
        best_precision_row["Threshold"]
    )

else:

    print("✓ ORIGINAL THRESHOLD REMAINS BETTER")
    print("✓ Keep threshold = 0.53")

    selected_stunting_threshold = 0.53


print("\nSelected stunting threshold:",
      selected_stunting_threshold)

print("\n✓ Test set was NOT accessed")
print("✓ Test probabilities were NOT generated")
print("✓ Test performance was NOT calculated")

print("\n" + "=" * 100)
print("STEP 4M COMPLETE")
print("=" * 100)
```

```python
# =============================================================================
# STEP 4L: PRECISION–RECALL FRONTIER ANALYSIS
# =============================================================================
#
# Purpose:
#   Determine whether the current tuned XGBoost models can achieve:
#
#       Precision: 60%+
#       Recall:    70–80%+
#       F1:        60%+
#
# IMPORTANT:
#   - Test data is NOT used here.
#   - Analysis is performed on validation data only.
#   - No threshold is selected from the test set.
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    roc_auc_score,
    precision_recall_curve
)

print("=" * 100)
print("STEP 4L: PRECISION–RECALL FRONTIER ANALYSIS")
print("=" * 100)


# =============================================================================
# 1. VERIFY REQUIRED VARIABLES
# =============================================================================

required_variables = [
    "final_xgb_stunting",
    "final_xgb_underweight",
    "X_val_s",
    "y_val_s",
    "X_val_u",
    "y_val_u"
]

missing_variables = [
    v for v in required_variables
    if v not in globals()
]

if missing_variables:
    raise NameError(
        "The following variables are missing:\n"
        + "\n".join(missing_variables)
        + "\n\nPlease run STEP 4K completely before running STEP 4L."
    )

print("\n✓ Tuned XGBoost models and validation sets are available.")


# =============================================================================
# 2. GENERATE VALIDATION PROBABILITIES
# =============================================================================

print("\n" + "=" * 100)
print("1. GENERATING VALIDATION PROBABILITIES")
print("=" * 100)

stunt_val_prob = final_xgb_stunting.predict_proba(X_val_s)[:, 1]

under_val_prob = final_xgb_underweight.predict_proba(X_val_u)[:, 1]

print(f"\nStunting validation probabilities    : {len(stunt_val_prob):,}")
print(f"Underweight validation probabilities: {len(under_val_prob):,}")

print("✓ Validation probabilities generated")


# =============================================================================
# 3. ROC-AUC
# =============================================================================

stunt_val_auc = roc_auc_score(
    y_val_s,
    stunt_val_prob
)

under_val_auc = roc_auc_score(
    y_val_u,
    under_val_prob
)

print("\n" + "=" * 100)
print("2. VALIDATION ROC-AUC")
print("=" * 100)

print(f"Stunting    ROC-AUC: {stunt_val_auc:.4f}")
print(f"Underweight ROC-AUC: {under_val_auc:.4f}")


# =============================================================================
# 4. FUNCTION FOR PRECISION–RECALL FRONTIER
# =============================================================================

def build_frontier(y_true, probabilities):

    rows = []

    thresholds = np.arange(
        0.10,
        0.91,
        0.01
    )

    for threshold in thresholds:

        predictions = (
            probabilities >= threshold
        ).astype(int)

        precision = precision_score(
            y_true,
            predictions,
            zero_division=0
        )

        recall = recall_score(
            y_true,
            predictions,
            zero_division=0
        )

        f1 = f1_score(
            y_true,
            predictions,
            zero_division=0
        )

        accuracy = accuracy_score(
            y_true,
            predictions
        )

        rows.append({
            "Threshold": round(float(threshold), 2),
            "Accuracy": accuracy,
            "Precision": precision,
            "Recall": recall,
            "F1": f1
        })

    return pd.DataFrame(rows)


# =============================================================================
# 5. STUNTING PRECISION–RECALL FRONTIER
# =============================================================================

stunting_frontier = build_frontier(
    y_val_s,
    stunt_val_prob
)

print("\n" + "=" * 100)
print("3. STUNTING PRECISION–RECALL FRONTIER")
print("=" * 100)


# -------------------------------------------------------------------------
# Find closest thresholds to recall targets
# -------------------------------------------------------------------------

recall_targets = [
    0.70,
    0.75,
    0.80,
    0.85
]

print("\nSTUNTING — THRESHOLDS CLOSEST TO TARGET RECALL")
print("-" * 100)

for target in recall_targets:

    idx = (
        stunting_frontier["Recall"] - target
    ).abs().idxmin()

    row = stunting_frontier.loc[idx]

    print(
        f"Target Recall {target:.0%} → "
        f"Threshold={row['Threshold']:.2f}, "
        f"Precision={row['Precision']:.4f}, "
        f"Recall={row['Recall']:.4f}, "
        f"F1={row['F1']:.4f}, "
        f"Accuracy={row['Accuracy']:.4f}"
    )


# =============================================================================
# 6. STUNTING BEST F1
# =============================================================================

best_stunt_f1 = stunting_frontier.loc[
    stunting_frontier["F1"].idxmax()
]

print("\nSTUNTING — BEST F1")
print("-" * 100)

print(
    f"Threshold : {best_stunt_f1['Threshold']:.2f}"
)

print(
    f"Precision : {best_stunt_f1['Precision']:.4f}"
)

print(
    f"Recall    : {best_stunt_f1['Recall']:.4f}"
)

print(
    f"F1        : {best_stunt_f1['F1']:.4f}"
)

print(
    f"Accuracy  : {best_stunt_f1['Accuracy']:.4f}"
)


# =============================================================================
# 7. STUNTING — DESIRED REGION
# =============================================================================
#
# Desired:
#   Precision >= 0.60
#   Recall    >= 0.70
#   F1        >= 0.60
#
# =============================================================================

stunt_desired = stunting_frontier[
    (stunting_frontier["Precision"] >= 0.60) &
    (stunting_frontier["Recall"] >= 0.70) &
    (stunting_frontier["F1"] >= 0.60)
].copy()

print("\nSTUNTING — DESIRED PERFORMANCE REGION")
print("-" * 100)

if len(stunt_desired) > 0:

    print(
        f"✓ {len(stunt_desired)} threshold(s) satisfy "
        f"Precision>=60%, Recall>=70%, F1>=60%"
    )

    print(
        stunt_desired[
            [
                "Threshold",
                "Accuracy",
                "Precision",
                "Recall",
                "F1"
            ]
        ].to_string(index=False)
    )

else:

    print(
        "✗ No threshold simultaneously satisfies "
        "Precision>=60%, Recall>=70%, and F1>=60%."
    )


# =============================================================================
# 8. STUNTING — BEST PRECISION WITH RECALL >=70%
# =============================================================================

stunt_recall70 = stunting_frontier[
    stunting_frontier["Recall"] >= 0.70
].copy()

if len(stunt_recall70) > 0:

    best_stunt_precision = stunt_recall70.loc[
        stunt_recall70["Precision"].idxmax()
    ]

    print("\nSTUNTING — HIGHEST PRECISION WHILE RECALL >=70%")
    print("-" * 100)

    print(
        f"Threshold : {best_stunt_precision['Threshold']:.2f}"
    )

    print(
        f"Precision : {best_stunt_precision['Precision']:.4f}"
    )

    print(
        f"Recall    : {best_stunt_precision['Recall']:.4f}"
    )

    print(
        f"F1        : {best_stunt_precision['F1']:.4f}"
    )

    print(
        f"Accuracy  : {best_stunt_precision['Accuracy']:.4f}"
    )


# =============================================================================
# 9. UNDERWEIGHT PRECISION–RECALL FRONTIER
# =============================================================================

underweight_frontier = build_frontier(
    y_val_u,
    under_val_prob
)

print("\n" + "=" * 100)
print("4. UNDERWEIGHT PRECISION–RECALL FRONTIER")
print("=" * 100)


print("\nUNDERWEIGHT — THRESHOLDS CLOSEST TO TARGET RECALL")
print("-" * 100)

for target in recall_targets:

    idx = (
        underweight_frontier["Recall"] - target
    ).abs().idxmin()

    row = underweight_frontier.loc[idx]

    print(
        f"Target Recall {target:.0%} → "
        f"Threshold={row['Threshold']:.2f}, "
        f"Precision={row['Precision']:.4f}, "
        f"Recall={row['Recall']:.4f}, "
        f"F1={row['F1']:.4f}, "
        f"Accuracy={row['Accuracy']:.4f}"
    )


# =============================================================================
# 10. UNDERWEIGHT BEST F1
# =============================================================================

best_under_f1 = underweight_frontier.loc[
    underweight_frontier["F1"].idxmax()
]

print("\nUNDERWEIGHT — BEST F1")
print("-" * 100)

print(
    f"Threshold : {best_under_f1['Threshold']:.2f}"
)

print(
    f"Precision : {best_under_f1['Precision']:.4f}"
)

print(
    f"Recall    : {best_under_f1['Recall']:.4f}"
)

print(
    f"F1        : {best_under_f1['F1']:.4f}"
)

print(
    f"Accuracy  : {best_under_f1['Accuracy']:.4f}"
)


# =============================================================================
# 11. UNDERWEIGHT — DESIRED REGION
# =============================================================================

under_desired = underweight_frontier[
    (underweight_frontier["Precision"] >= 0.60) &
    (underweight_frontier["Recall"] >= 0.70) &
    (underweight_frontier["F1"] >= 0.60)
].copy()

print("\nUNDERWEIGHT — DESIRED PERFORMANCE REGION")
print("-" * 100)

if len(under_desired) > 0:

    print(
        f"✓ {len(under_desired)} threshold(s) satisfy "
        "Precision>=60%, Recall>=70%, F1>=60%"
    )

    print(
        under_desired[
            [
                "Threshold",
                "Accuracy",
                "Precision",
                "Recall",
                "F1"
            ]
        ].to_string(index=False)
    )

else:

    print(
        "✗ No threshold simultaneously satisfies "
        "Precision>=60%, Recall>=70%, and F1>=60%."
    )


# =============================================================================
# 12. UNDERWEIGHT — BEST PRECISION WITH RECALL >=70%
# =============================================================================

under_recall70 = underweight_frontier[
    underweight_frontier["Recall"] >= 0.70
].copy()

if len(under_recall70) > 0:

    best_under_precision = under_recall70.loc[
        under_recall70["Precision"].idxmax()
    ]

    print("\nUNDERWEIGHT — HIGHEST PRECISION WHILE RECALL >=70%")
    print("-" * 100)

    print(
        f"Threshold : {best_under_precision['Threshold']:.2f}"
    )

    print(
        f"Precision : {best_under_precision['Precision']:.4f}"
    )

    print(
        f"Recall    : {best_under_precision['Recall']:.4f}"
    )

    print(
        f"F1        : {best_under_precision['F1']:.4f}"
    )

    print(
        f"Accuracy  : {best_under_precision['Accuracy']:.4f}"
    )


# =============================================================================
# 13. TABLE FOR RECALL 70–90%
# =============================================================================

print("\n" + "=" * 100)
print("5. SUMMARY: PERFORMANCE ACROSS RECALL RANGE")
print("=" * 100)


def recall_range_summary(frontier):

    selected = frontier[
        (frontier["Recall"] >= 0.70) &
        (frontier["Recall"] <= 0.90)
    ].copy()

    if len(selected) == 0:
        return pd.DataFrame()

    return selected[
        [
            "Threshold",
            "Accuracy",
            "Precision",
            "Recall",
            "F1"
        ]
    ].sort_values(
        by="F1",
        ascending=False
    ).head(15)


print("\nSTUNTING")
print("-" * 100)

stunt_summary = recall_range_summary(
    stunting_frontier
)

print(
    stunt_summary.to_string(index=False)
)


print("\nUNDERWEIGHT")
print("-" * 100)

under_summary = recall_range_summary(
    underweight_frontier
)

print(
    under_summary.to_string(index=False)
)


# =============================================================================
# 14. FINAL INTERPRETATION
# =============================================================================

print("\n" + "=" * 100)
print("STEP 4L COMPLETE")
print("=" * 100)

print("""
✓ Precision–recall frontier calculated
✓ Stunting thresholds evaluated
✓ Underweight thresholds evaluated
✓ Recall 70–90% region examined
✓ Best F1 identified
✓ Highest precision at Recall>=70% identified
✓ Desired Precision/Recall/F1 region checked
✓ TEST SET REMAINS UNTOUCHED

Next:
STEP 4M = Select final validation-based model/thresholds
          and then perform ONE final test-set evaluation.
""")
```

```python
# ==================================================================================================
# STEP 4M: FINE PRECISION–RECALL THRESHOLD REFINEMENT
# CAR MICS6 — STUNTING
# ==================================================================================================

import numpy as np
import pandas as pd
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    confusion_matrix
)

print("=" * 100)
print("STEP 4M: FINE PRECISION–RECALL THRESHOLD REFINEMENT")
print("=" * 100)

# --------------------------------------------------------------------------------------------------
# 1. CHECK REQUIRED OBJECTS
# --------------------------------------------------------------------------------------------------

required_objects = [
    "best_stunting_model",
    "X_val_stunt",
    "y_val_stunt"
]

missing = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing:
    raise NameError(
        f"Missing required objects: {missing}. "
        "Run the original STEP 4K before STEP 4M."
    )

print("\n✓ Original tuned stunting model available")
print("✓ Validation predictors available")
print("✓ Validation target available")
print("✓ Test set will NOT be accessed")


# --------------------------------------------------------------------------------------------------
# 2. GENERATE VALIDATION PROBABILITIES
# --------------------------------------------------------------------------------------------------

stunt_val_prob = best_stunting_model.predict_proba(
    X_val_stunt
)[:, 1]

print("\nValidation observations :", len(stunt_val_prob))
print("✓ Validation probabilities generated")


# --------------------------------------------------------------------------------------------------
# 3. FINE THRESHOLD GRID
# --------------------------------------------------------------------------------------------------

thresholds = np.array([
    0.525,
    0.530,
    0.535,
    0.540,
    0.545,
    0.550,
    0.555,
    0.560
])

print("\nThresholds tested:")
print(thresholds)


# --------------------------------------------------------------------------------------------------
# 4. EVALUATE EACH THRESHOLD
# --------------------------------------------------------------------------------------------------

results = []

for threshold in thresholds:

    y_pred = (
        stunt_val_prob >= threshold
    ).astype(int)

    precision = precision_score(
        y_val_stunt,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_val_stunt,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_val_stunt,
        y_pred,
        zero_division=0
    )

    accuracy = accuracy_score(
        y_val_stunt,
        y_pred
    )

    tn, fp, fn, tp = confusion_matrix(
        y_val_stunt,
        y_pred,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn)

    results.append({
        "Threshold": threshold,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })


results_df = pd.DataFrame(results)


# --------------------------------------------------------------------------------------------------
# 5. DISPLAY ALL RESULTS
# --------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("FINE THRESHOLD RESULTS")
print("=" * 100)

print(
    results_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# --------------------------------------------------------------------------------------------------
# 6. FILTER: RECALL >= 70%
# --------------------------------------------------------------------------------------------------

eligible = results_df[
    results_df["Recall"] >= 0.70
].copy()

print("\n" + "=" * 100)
print("ELIGIBLE THRESHOLDS — RECALL >= 70%")
print("=" * 100)

if len(eligible) == 0:

    print("No threshold satisfies Recall >= 70%.")

else:

    eligible = eligible.sort_values(
        by=[
            "Precision",
            "F1",
            "Recall",
            "FPR"
        ],
        ascending=[
            False,
            False,
            False,
            True
        ]
    )

    print(
        eligible.to_string(
            index=False,
            float_format=lambda x: f"{x:.4f}"
        )
    )


# --------------------------------------------------------------------------------------------------
# 7. ORIGINAL BENCHMARK
# --------------------------------------------------------------------------------------------------

benchmark_threshold = 0.53

benchmark = results_df[
    np.isclose(
        results_df["Threshold"],
        benchmark_threshold
    )
]

print("\n" + "=" * 100)
print("ORIGINAL BENCHMARK — THRESHOLD 0.53")
print("=" * 100)

if len(benchmark) > 0:

    benchmark = benchmark.iloc[0]

    print(
        f"""
Precision : {benchmark['Precision']:.4f}
Recall    : {benchmark['Recall']:.4f}
F1        : {benchmark['F1']:.4f}
Accuracy  : {benchmark['Accuracy']:.4f}
FPR       : {benchmark['FPR']:.4f}

TN        : {int(benchmark['TN'])}
FP        : {int(benchmark['FP'])}
FN        : {int(benchmark['FN'])}
TP        : {int(benchmark['TP'])}
"""
    )


# --------------------------------------------------------------------------------------------------
# 8. SELECT BEST PRECISION WITH RECALL >= 70%
# --------------------------------------------------------------------------------------------------

if len(eligible) > 0:

    best = eligible.iloc[0]

    print("\n" + "=" * 100)
    print("BEST REFINED OPERATING POINT")
    print("=" * 100)

    print(
        f"""
Threshold : {best['Threshold']:.3f}

Precision : {best['Precision']:.4f}
Recall    : {best['Recall']:.4f}
F1        : {best['F1']:.4f}
Accuracy  : {best['Accuracy']:.4f}
FPR       : {best['FPR']:.4f}

TN        : {int(best['TN'])}
FP        : {int(best['FP'])}
FN        : {int(best['FN'])}
TP        : {int(best['TP'])}
"""
    )


    # ----------------------------------------------------------------------------------------------
    # 9. COMPARE REFINED RESULT AGAINST BENCHMARK
    # ----------------------------------------------------------------------------------------------

    precision_change = (
        best["Precision"] -
        benchmark["Precision"]
    )

    recall_change = (
        best["Recall"] -
        benchmark["Recall"]
    )

    f1_change = (
        best["F1"] -
        benchmark["F1"]
    )

    fp_change = (
        best["FP"] -
        benchmark["FP"]
    )

    print("\n" + "=" * 100)
    print("IMPROVEMENT OVER 0.53 BENCHMARK")
    print("=" * 100)

    print(
        f"""
Precision change : {precision_change:+.4f}
Recall change    : {recall_change:+.4f}
F1 change        : {f1_change:+.4f}
FP change        : {int(fp_change):+d}
"""
    )


    # ----------------------------------------------------------------------------------------------
    # 10. FINAL DEVELOPMENT DECISION
    # ----------------------------------------------------------------------------------------------

    print("\n" + "=" * 100)
    print("DEVELOPMENT DECISION")
    print("=" * 100)

    if precision_change > 0:

        print("✓ PRECISION IMPROVED")
        print(
            f"✓ Refined threshold = {best['Threshold']:.3f}"
        )
        print(
            f"✓ Precision increased to "
            f"{best['Precision']:.4f}"
        )

        selected_stunting_threshold = float(
            best["Threshold"]
        )

    else:

        print("✓ NO PRECISION IMPROVEMENT")
        print("✓ Keep original threshold = 0.53")

        selected_stunting_threshold = 0.53

else:

    selected_stunting_threshold = 0.53

    print("✓ Keep original threshold = 0.53")


# --------------------------------------------------------------------------------------------------
# 11. PROTECTION CHECK
# --------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("DATA-INTEGRITY CHECK")
print("=" * 100)

print("✓ Training data was not modified")
print("✓ Model was not retrained")
print("✓ Validation data used only for threshold evaluation")
print("✓ Test predictors NOT accessed")
print("✓ Test targets NOT accessed")
print("✓ Test probabilities NOT generated")
print("✓ Test performance NOT calculated")

print("\nSelected development threshold:",
      selected_stunting_threshold)

print("\n" + "=" * 100)
print("STEP 4M COMPLETE")
print("=" * 100)
```

```python
# =============================================================================
# STEP 4O: MODEL INTERPRETATION AND FEATURE IMPORTANCE
# =============================================================================
#
# Purpose:
#   Identify which of the 17 predictors contribute most strongly to:
#       1. Stunting prediction
#       2. Underweight prediction
#
# Models:
#   final_cv_xgb_stunting
#   final_cv_xgb_underweight
#
# IMPORTANT:
#   - No model retraining
#   - No test-set model selection
#   - Final test results remain unchanged
#
# =============================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.inspection import permutation_importance

print("=" * 100)
print("STEP 4O: MODEL INTERPRETATION AND FEATURE IMPORTANCE")
print("=" * 100)


# =============================================================================
# 1. VERIFY REQUIRED OBJECTS
# =============================================================================

required_variables = [
    "final_cv_xgb_stunting",
    "final_cv_xgb_underweight",
    "X_train_stunt_processed",
    "X_test_stunt_processed",
    "y_test_stunt",
    "X_train_under_processed",
    "X_test_under_processed",
    "y_test_under",
    "preprocessor_stunt",
    "preprocessor_under"
]

missing_variables = [
    v for v in required_variables
    if v not in globals()
]

if missing_variables:
    raise NameError(
        "The following required variables are missing:\n"
        + "\n".join(missing_variables)
        + "\n\nPlease make sure Steps 3L, 4M and 4N have been executed."
    )

print("\n✓ All required models and preprocessing objects are available.")


# =============================================================================
# 2. ORIGINAL 17 PREDICTOR NAMES
# =============================================================================

predictor_names = [
    "CAGE",
    "WB4",
    "WAGEM",
    "CEB",
    "CSURV",
    "CDEAD",
    "CM11",
    "HL4",
    "HH6",
    "windex5",
    "WS1",
    "WS11",
    "WS15",
    "welevel",
    "MSTATUS",
    "CM17",
    "insurance"
]

print("\n" + "=" * 100)
print("1. ORIGINAL PREDICTOR SET")
print("=" * 100)

print(f"Total predictors: {len(predictor_names)}")

for i, variable in enumerate(predictor_names, start=1):
    print(f"{i:2d}. {variable}")


# =============================================================================
# 3. EXTRACT TRANSFORMED FEATURE NAMES
# =============================================================================
#
# XGBoost sees the transformed 66/67 variables, not the original 17.
# Therefore we first obtain the one-hot encoded feature names.
# =============================================================================

print("\n" + "=" * 100)
print("2. EXTRACTING TRANSFORMED FEATURE NAMES")
print("=" * 100)


try:

    stunt_feature_names = (
        preprocessor_stunt
        .get_feature_names_out()
    )

    under_feature_names = (
        preprocessor_under
        .get_feature_names_out()
    )

except Exception as e:

    print("\n⚠ Could not obtain transformed feature names.")
    print("Error:", e)

    raise


print(
    f"\nStunting transformed features   : "
    f"{len(stunt_feature_names)}"
)

print(
    f"Underweight transformed features: "
    f"{len(under_feature_names)}"
)


# =============================================================================
# 4. XGBOOST GAIN IMPORTANCE
# =============================================================================
#
# XGBoost provides feature_importances_.
# This represents the relative contribution of transformed features.
# =============================================================================

print("\n" + "=" * 100)
print("3. XGBOOST FEATURE IMPORTANCE")
print("=" * 100)


# -----------------------------------------------------------------------------
# STUNTING
# -----------------------------------------------------------------------------

stunting_importance = pd.DataFrame({
    "Feature": stunt_feature_names,
    "Importance": final_cv_xgb_stunting.feature_importances_
})

stunting_importance = (
    stunting_importance
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

stunting_importance["Rank"] = (
    np.arange(len(stunting_importance)) + 1
)

stunting_importance = stunting_importance[
    [
        "Rank",
        "Feature",
        "Importance"
    ]
]


print("\nSTUNTING — TOP 20 TRANSFORMED FEATURES")
print("-" * 100)

print(
    stunting_importance
    .head(20)
    .to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# -----------------------------------------------------------------------------
# UNDERWEIGHT
# -----------------------------------------------------------------------------

underweight_importance = pd.DataFrame({
    "Feature": under_feature_names,
    "Importance": final_cv_xgb_underweight.feature_importances_
})

underweight_importance = (
    underweight_importance
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

underweight_importance["Rank"] = (
    np.arange(len(underweight_importance)) + 1
)

underweight_importance = underweight_importance[
    [
        "Rank",
        "Feature",
        "Importance"
    ]
]


print("\nUNDERWEIGHT — TOP 20 TRANSFORMED FEATURES")
print("-" * 100)

print(
    underweight_importance
    .head(20)
    .to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# =============================================================================
# 5. MAP ONE-HOT FEATURES BACK TO ORIGINAL PREDICTORS
# =============================================================================
#
# For categorical variables, one original predictor can generate several
# one-hot encoded variables.
#
# Example:
#   windex5_Riche
#   windex5_Moyen
#   windex5_Pauvre
#
# These are all components of the original predictor windex5.
#
# We therefore aggregate the importance of transformed variables belonging
# to the same original predictor.
# =============================================================================


def map_feature_to_original(feature_name):

    # Remove transformer prefixes where present
    clean_name = feature_name

    # Numeric transformer
    if "__" in clean_name:
        clean_name = clean_name.split("__", 1)[1]

    # Match longest predictor name first
    sorted_predictors = sorted(
        predictor_names,
        key=len,
        reverse=True
    )

    for predictor in sorted_predictors:

        if (
            clean_name == predictor
            or clean_name.startswith(predictor + "_")
        ):
            return predictor

    # Fallback
    return clean_name


stunting_importance["Original_Predictor"] = (
    stunting_importance["Feature"]
    .apply(map_feature_to_original)
)

underweight_importance["Original_Predictor"] = (
    underweight_importance["Feature"]
    .apply(map_feature_to_original)
)


# =============================================================================
# 6. AGGREGATED ORIGINAL-PREDICTOR IMPORTANCE
# =============================================================================

print("\n" + "=" * 100)
print("4. AGGREGATED IMPORTANCE BY ORIGINAL PREDICTOR")
print("=" * 100)


stunting_original_importance = (
    stunting_importance
    .groupby(
        "Original_Predictor",
        as_index=False
    )["Importance"]
    .sum()
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

stunting_original_importance["Rank"] = (
    np.arange(
        len(stunting_original_importance)
    ) + 1
)

stunting_original_importance = (
    stunting_original_importance[
        [
            "Rank",
            "Original_Predictor",
            "Importance"
        ]
    ]
)


underweight_original_importance = (
    underweight_importance
    .groupby(
        "Original_Predictor",
        as_index=False
    )["Importance"]
    .sum()
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

underweight_original_importance["Rank"] = (
    np.arange(
        len(underweight_original_importance)
    ) + 1
)

underweight_original_importance = (
    underweight_original_importance[
        [
            "Rank",
            "Original_Predictor",
            "Importance"
        ]
    ]
)


# =============================================================================
# 7. STUNTING ORIGINAL-PREDICTOR RESULTS
# =============================================================================

print("\nSTUNTING — ORIGINAL PREDICTOR IMPORTANCE")
print("-" * 100)

print(
    stunting_original_importance
    .to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# =============================================================================
# 8. UNDERWEIGHT ORIGINAL-PREDICTOR RESULTS
# =============================================================================

print("\nUNDERWEIGHT — ORIGINAL PREDICTOR IMPORTANCE")
print("-" * 100)

print(
    underweight_original_importance
    .to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# =============================================================================
# 9. TOP 10 STUNTING PREDICTORS — PLOT
# =============================================================================

print("\n" + "=" * 100)
print("5. STUNTING FEATURE IMPORTANCE VISUALIZATION")
print("=" * 100)

top_stunting = (
    stunting_original_importance
    .head(10)
    .sort_values(
        "Importance",
        ascending=True
    )
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_stunting["Original_Predictor"],
    top_stunting["Importance"]
)

plt.xlabel("Aggregated XGBoost Feature Importance")
plt.ylabel("Predictor")
plt.title(
    "Top 10 Predictors of Child Stunting"
)

plt.tight_layout()
plt.show()


# =============================================================================
# 10. TOP 10 UNDERWEIGHT PREDICTORS — PLOT
# =============================================================================

print("\n" + "=" * 100)
print("6. UNDERWEIGHT FEATURE IMPORTANCE VISUALIZATION")
print("=" * 100)

top_underweight = (
    underweight_original_importance
    .head(10)
    .sort_values(
        "Importance",
        ascending=True
    )
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_underweight["Original_Predictor"],
    top_underweight["Importance"]
)

plt.xlabel("Aggregated XGBoost Feature Importance")
plt.ylabel("Predictor")
plt.title(
    "Top 10 Predictors of Child Underweight"
)

plt.tight_layout()
plt.show()


# =============================================================================
# 11. COMPARE PREDICTOR RANKINGS
# =============================================================================

print("\n" + "=" * 100)
print("7. COMPARISON OF PREDICTOR IMPORTANCE")
print("=" * 100)


comparison = pd.merge(
    stunting_original_importance[
        [
            "Original_Predictor",
            "Importance"
        ]
    ].rename(
        columns={
            "Importance": "Stunting_Importance"
        }
    ),

    underweight_original_importance[
        [
            "Original_Predictor",
            "Importance"
        ]
    ].rename(
        columns={
            "Importance": "Underweight_Importance"
        }
    ),

    on="Original_Predictor",
    how="outer"
)


comparison = comparison.fillna(0)

comparison["Average_Importance"] = (
    comparison["Stunting_Importance"]
    +
    comparison["Underweight_Importance"]
) / 2


comparison = comparison.sort_values(
    "Average_Importance",
    ascending=False
).reset_index(drop=True)

comparison["Overall_Rank"] = (
    np.arange(
        len(comparison)
    ) + 1
)


comparison = comparison[
    [
        "Overall_Rank",
        "Original_Predictor",
        "Stunting_Importance",
        "Underweight_Importance",
        "Average_Importance"
    ]
]


print(
    comparison.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# =============================================================================
# 12. TOP PREDICTORS SHARED BY BOTH OUTCOMES
# =============================================================================

print("\n" + "=" * 100)
print("8. TOP PREDICTORS SHARED BY BOTH OUTCOMES")
print("=" * 100)

top_stunting_names = set(
    stunting_original_importance
    .head(10)["Original_Predictor"]
)

top_underweight_names = set(
    underweight_original_importance
    .head(10)["Original_Predictor"]
)

shared_predictors = (
    top_stunting_names
    .intersection(
        top_underweight_names
    )
)

if shared_predictors:

    print("\nPredictors appearing in the top 10 for BOTH outcomes:")

    for predictor in sorted(shared_predictors):
        print("✓", predictor)

else:

    print(
        "\nNo predictors appear in the top 10 for both outcomes."
    )


# =============================================================================
# 13. SAVE RESULTS IN MEMORY
# =============================================================================

feature_importance_stunting = (
    stunting_original_importance.copy()
)

feature_importance_underweight = (
    underweight_original_importance.copy()
)

feature_importance_comparison = (
    comparison.copy()
)


# =============================================================================
# 14. FINAL STATUS
# =============================================================================

print("\n" + "=" * 100)
print("STEP 4O COMPLETE")
print("=" * 100)

print("""
✓ XGBoost feature importance calculated
✓ Transformed features examined
✓ One-hot encoded variables aggregated
✓ Importance mapped back to original predictors
✓ Stunting predictors ranked
✓ Underweight predictors ranked
✓ Predictor comparison completed
✓ Feature-importance plots generated
✓ No model retraining performed
✓ Final test results remain unchanged

""")
```
